## 1 · A computational biology question
Imagine measuring the area (in square micrometers, µm²) of several cell nuclei. We want to calculate their average and identify unusually large nuclei. The measurements below are **invented for practice**. A list holds values in order; each element has an index starting at zero.

```python
areas_um2 = [34, 42, 38, 55, 31, 47, 61, 44]
print('Number of nuclei:', len(areas_um2))
print('First measurement:', areas_um2[0])
print('Third measurement:', areas_um2[2])
assert len(areas_um2) == 8
```

## 2 · Variables, arithmetic and types
Python can store integers (`int`), decimal numbers (`float`), text (`str`) and truth values (`bool`). A variable is a name attached to a value; the `=` sign assigns a value.

```python
sample_name = 'practice_slide_A'
cell_count = len(areas_um2)
mean_area = sum(areas_um2) / cell_count
large_threshold_um2 = 50
print(sample_name, '| mean nuclear area:', round(mean_area, 1), 'µm²')
print('Type of mean:', type(mean_area).__name__)
print('Is mean above threshold?', mean_area > large_threshold_um2)
```

## 3 · Loops and conditions
A `for` loop repeats work. An `if` statement lets a program make a decision. Indentation is meaningful in Python.

```python
large_ids = []
for index, area in enumerate(areas_um2):
    if area > large_threshold_um2:
        large_ids.append(index)  # index of nucleus in the list
print('Indices above 50 µm²:', large_ids)
assert large_ids == [3, 6]
```

## 4 · Write a reusable function
A function accepts inputs (arguments), calculates something, and returns a result. We check for an empty list to avoid dividing by zero.

```python
def average(values):
    if len(values) == 0:
        raise ValueError('Provide at least one measurement.')
    return sum(values) / len(values)

print('Average:', average(areas_um2))
assert abs(average([2, 4, 6]) - 4) < 1e-9
```

## 5 · Organize individual observations
A dictionary stores named properties. Here one dictionary corresponds to one measured cell.

```python
observations = [
    {'cell_id': f'cell_{i+1:02d}', 'area_um2': area, 'large': area > large_threshold_um2}
    for i, area in enumerate(areas_um2)
]
for item in observations[:3]:
    print(item)
print('Large nuclei:', sum(item['large'] for item in observations))
```

## Your turn (5–10 minutes)
1. Change `large_threshold_um2` to 40 and rerun the counting cells. How many nuclei exceed the threshold?
2. Create a function `fraction_above(values, threshold)` that returns the proportion *strictly above* a threshold.
3. Why should we report the threshold and measurement units when comparing experiments?

**Hint:** Start with `sum(value > threshold for value in values) / len(values)`. Handle the empty-list case.

```python
# Example solution / self-check (read after trying)
def fraction_above(values, threshold):
    if not values:
        raise ValueError('Need at least one value.')
    return sum(value > threshold for value in values) / len(values)

assert fraction_above([1, 4, 5, 8], 4) == 0.5
print('Fraction above 50:', fraction_above(areas_um2, 50))
```
