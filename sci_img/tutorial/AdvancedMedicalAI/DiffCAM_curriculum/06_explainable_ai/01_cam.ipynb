## Concept and motivation

In our CNN, the classifier follows global average pooling. This allows **CAM** to use the final classifier weights directly as feature-map weights. These weights explain a class score, while DiffCAM will use target-versus-reference feature distributions instead.

**Key vocabulary:** class weight, spatial attribution, weighted channel sum, signed relevance.

**Learning goal:** Describe where the CAM weights come from and what they explain.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 2

```python
from diffcam_edu import make_cell_images
from diffcam_edu.math import diffcam_map,normalize_map
from diffcam_edu.models import fit_cnn,feature_maps
images,y,masks=make_cell_images(220,seed=40)
model,_=fit_cnn(images[:170],y[:170],epochs=7,seed=3)
j=int(np.flatnonzero(y[170:]==1)[0]+170)
A=feature_maps(model,images[j])[0]
W=model.classifier.weight.detach().numpy()
heat=diffcam_map(A,W[1])
print('Channels:',len(W[1]),'Heatmap shape:',heat.shape)
```

### Experiment 2 of 2

```python
fig,ax=plt.subplots(1,3,figsize=(9,3))
ax[0].imshow(images[j],cmap='gray');ax[0].set_title('Target')
ax[1].imshow(heat,cmap='inferno');ax[1].set_title('CAM (low resolution)')
from scipy.ndimage import zoom
resized=zoom(normalize_map(heat),np.array(images[j].shape)/np.array(heat.shape),order=1)
ax[2].imshow(images[j],cmap='gray');ax[2].imshow(resized,cmap='inferno',alpha=.5);ax[2].set_title('Overlay')
for a in ax:a.axis('off')
plt.show()
```
