# Covariance and Mahalanobis geometry

**Weeks:** 12–13 · **Core question:** Why is diagonal variance not enough?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

If two features change together, treating them as independent exaggerates how much new information their joint change contains. Covariance describes the orientation of the reference cloud. Mahalanobis geometry discounts high-variance directions.

**Key vocabulary:** covariance matrix, correlation, ellipse, Mahalanobis distance.

**Learning goal:** Explain why scatter and covariance use different normalization.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_tabular_cells
X,y=make_tabular_cells(130,seed=11)
ref=X[y==0]; target=np.array([6.5,3.6]);mu=ref.mean(axis=0)
cov=np.cov(ref,rowvar=False)
print('Covariance:',cov,'Correlation:',np.corrcoef(ref.T))
d=target-mu
print('Euclidean:',np.linalg.norm(d),'Mahalanobis:',np.sqrt(d@np.linalg.solve(cov,d)))
```

### Experiment 2 of 3

```python
eig,V=np.linalg.eigh(cov); theta=np.linspace(0,2*np.pi,150)
ellipse=mu[:,None] + V@np.diag(2*np.sqrt(eig))@np.vstack([np.cos(theta),np.sin(theta)])
plt.scatter(*ref.T,alpha=.25);plt.plot(*ellipse,color='black');plt.scatter(*target,c='red',marker='*',s=130)
plt.axis('equal');plt.xlabel('Area');plt.ylabel('Intensity');plt.show()
```

### Experiment 3 of 3

```python
from diffcam_edu.math import reference_scatter
mu2,S=reference_scatter(ref)
np.testing.assert_allclose(S,(len(ref)-1)*cov)
print('Covariance vs scatter differ by a factor of N-1.')
```
