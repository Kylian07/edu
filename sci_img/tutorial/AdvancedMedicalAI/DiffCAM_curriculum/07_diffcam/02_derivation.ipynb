# Derive DiffCAM weights from the Rayleigh quotient

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

Let $d=z-\mu$, $A=S_v+\lambda I$, and $\lambda>0$. The regularized objective is $J_\lambda(w)=(w^Td)^2/(w^TAw)$. Since $A$ is positive definite, use $u=A^{1/2}w$ and $v=A^{-1/2}d$. Cauchy–Schwarz gives $(u^Tv)^2\leq\|u\|^2\|v\|^2$, with equality for $w\propto A^{-1}d$.

Important: the optimum **direction** is unique up to scaling (unless $d=0$). Using a linear solver is preferable to explicitly computing an inverse.

**Learning goal:** Derive and implement the regularized direction without using a matrix inverse.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu.math import diffcam_weights, reference_scatter
rng=np.random.default_rng(7)
refs=rng.normal(size=(40,5));target=np.array([2.,-.8,1.2,0.,.5]);ridge=.4
mu,S=reference_scatter(refs);d=target-mu
w=diffcam_weights(target,refs,ridge=ridge)
A=S+ridge*np.eye(5)
np.testing.assert_allclose(A@w,d)
print('Closed-form solve residual:',np.linalg.norm(A@w-d))
```

### Experiment 2 of 3

```python
def quotient(v):return (v@d)**2/(v@A@v)
random_directions=rng.normal(size=(5000,5))
values=np.array([quotient(v) for v in random_directions])
print('Analytic J:',quotient(w),'Best among random candidates:',values.max())
assert quotient(w)>=values.max()-1e-8
```

### Experiment 3 of 3

```python
for a in [0.1,1.,10.]:
    print(a,quotient(a*w))
print('Mean difference only:',d.round(3),'Covariance-aware:',w.round(3))
```
