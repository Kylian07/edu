## Concept and motivation

Medical-image interpretation requires anatomy, annotations, and external validation. Here we use **geometric chest-like phantoms** to practice masks and metrics; these are NOT diagnostic radiographs. The actual paper includes chest radiograph benchmarks, but those datasets are not bundled.

**Key vocabulary:** segmentation mask, threshold, overlap, class prevalence, patient-level split.

**Learning goal:** Distinguish pixel localization from validated clinical explanation.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_chest_phantoms
from diffcam_edu.evaluation import mask_iou,pointing_game,area_under_precision_recall
x,y,truth=make_chest_phantoms(12)
i=1
fig,ax=plt.subplots(1,2,figsize=(7,3))
ax[0].imshow(x[i],cmap='gray');ax[0].set_title('Geometric phantom')
ax[1].imshow(truth[i],cmap='gray');ax[1].set_title('Known simulated opacity')
for a in ax:a.axis('off')
plt.show()
```

### Experiment 2 of 3

```python
from scipy.ndimage import gaussian_filter
s=gaussian_filter(x[i],sigma=1)
pred=s>.35
print('IoU:',mask_iou(pred,truth[i]),'Pointing:',pointing_game(s,truth[i]))
print('AUPRC:',area_under_precision_recall(s,truth[i]))
```

### Experiment 3 of 3

```python
for t in [.25,.35,.45,.55]:
    print(t,mask_iou(s>t,truth[i]))
```
