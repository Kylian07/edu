## Concept and motivation

A reference set is part of the *definition* of the explanation, not an incidental hyperparameter. All negatives, class-specific negatives, and within-class examples ask different questions. Restrict reference groups using legitimate biological factors rather than cherry-picking a desired saliency map.

**Key vocabulary:** cohort, matching, reference selection, within-class atypicality, selection bias.

**Learning goal:** Report reference counts, inclusion criteria, and scientific question alongside each result.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_cell_images
from diffcam_edu.math import diffcam_weights,diffcam_map,normalize_map
from diffcam_edu.models import fit_cnn,feature_maps,embeddings_from_maps
x,y,m=make_cell_images(240,seed=15)
model,_=fit_cnn(x[:180],y[:180],epochs=6)
A=feature_maps(model,x);z=embeddings_from_maps(A)
i=int(np.flatnonzero(y[180:]==1)[0]+180)
rng=np.random.default_rng(10)
choices={'class 0':np.flatnonzero(y[:180]==0),
         'within class 1':np.flatnonzero(y[:180]==1),
         'balanced mixed':np.arange(180)}
```

### Experiment 2 of 3

```python
fig,axes=plt.subplots(1,4,figsize=(12,3))
axes[0].imshow(x[i],cmap='gray');axes[0].set_title('Fixed target');axes[0].axis('off')
results={}
for ax,(name,indices) in zip(axes[1:],choices.items()):
    w=diffcam_weights(z[i],z[indices],ridge=.1)
    hm=normalize_map(diffcam_map(A[i],w));results[name]=hm
    ax.imshow(hm,cmap='inferno');ax.set_title(name);ax.axis('off')
plt.show()
```

### Experiment 3 of 3

```python
for a in results:
    for b in results:
        print(f'{a:14s} vs {b:14s}: {np.corrcoef(results[a].ravel(),results[b].ravel())[0,1]:.3f}')
```
