## Concept and motivation

The last convolutional layer produces one image-like map per channel. Global average pooling converts these maps into an embedding vector. DiffCAM combines statistics of those *vectors* to obtain weights for the *maps*.

**Key vocabulary:** feature map, channel, embedding, global average pooling, PCA.

**Learning goal:** Explain why one importance weight is assigned to each spatial channel.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from sklearn.decomposition import PCA
from diffcam_edu import make_cell_images
from diffcam_edu.models import fit_cnn,feature_maps,embeddings_from_maps
images,y,_=make_cell_images(160,seed=31)
model,loss=fit_cnn(images[:120],y[:120],epochs=6)
maps=feature_maps(model,images)
z=embeddings_from_maps(maps)
print('Maps:',maps.shape,'Embeddings:',z.shape)
```

### Experiment 2 of 3

```python
fig,ax=plt.subplots(2,4,figsize=(9,5))
for j,a in enumerate(ax.flat):
    a.imshow(maps[0,j],cmap='viridis');a.set_title(f'Channel {j}');a.axis('off')
plt.tight_layout();plt.show()
```

### Experiment 3 of 3

```python
Z=PCA(n_components=2).fit_transform(z)
plt.scatter(Z[:,0],Z[:,1],c=y,cmap='coolwarm',alpha=.75)
plt.xlabel('PCA 1');plt.ylabel('PCA 2');plt.title('Learned embeddings');plt.show()
assert z.shape[1]==maps.shape[1]
```
