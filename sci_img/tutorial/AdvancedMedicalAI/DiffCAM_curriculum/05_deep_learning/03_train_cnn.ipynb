## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
import torch
from diffcam_edu import make_cell_images
from diffcam_edu.models import fit_cnn
images,y,masks=make_cell_images(320,seed=12)
tr=np.arange(240);te=np.arange(240,320)
model,history=fit_cnn(images[tr],y[tr],epochs=7,seed=4)
plt.plot(history);plt.xlabel('Epoch');plt.ylabel('Training loss');plt.show()
```

### Experiment 2 of 3

```python
with torch.no_grad():
    logits=model(torch.tensor(images[te])[:,None])
pred=logits.argmax(1).numpy()
print('Held-out accuracy:',np.mean(pred==y[te]))
print('Classifier output shape:',logits.shape)
```

### Experiment 3 of 3

```python
fig,axs=plt.subplots(1,5,figsize=(12,3))
for j,ax in enumerate(axs):
    idx=te[j];ax.imshow(images[idx],cmap='gray');ax.set_title(f'true {y[idx]}, pred {pred[j]}');ax.axis('off')
plt.show()
```
