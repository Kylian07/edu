## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_tabular_cells
X,y=make_tabular_cells(60,seed=0)
X=(X-X.mean(axis=0))/(X.std(axis=0)+1e-7)
Y=np.eye(2)[y];rng=np.random.default_rng(2)
W1=rng.normal(0,.3,(2,8));W2=rng.normal(0,.3,(8,2));lr=.1
losses=[]
```

### Experiment 2 of 3

```python
for step in range(250):
    z1=X@W1;h=np.tanh(z1);logits=h@W2
    e=np.exp(logits-logits.max(axis=1,keepdims=True));p=e/e.sum(axis=1,keepdims=True)
    losses.append(-np.mean(np.log(p[np.arange(len(y)),y]+1e-12)))
    delta=(p-Y)/len(y)
    grad2=h.T@delta
    grad1=X.T@((delta@W2.T)*(1-h*h))
    W2-=lr*grad2;W1-=lr*grad1
print('First/last loss:',losses[0],losses[-1])
```

### Experiment 3 of 3

```python
plt.plot(losses);plt.xlabel('Update');plt.ylabel('Cross entropy');plt.show()
print('Training accuracy:',np.mean(p.argmax(axis=1)==y))
assert losses[-1]<losses[0]
```
