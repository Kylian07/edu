## Concept and motivation

Pixels are numeric measurements, not diagnoses. Microscopy intensity is affected by staining, illumination, sensor noise, focus, and biology. A saliency overlay might highlight an artifact instead of morphology. We start with simulated cells so their true changes are known.

**Key vocabulary:** pixel, channel, grayscale, intensity distribution, ground-truth region, nuisance variation.

**Learning goal:** Explain image measurement, label, and simulated ground-truth region in your own words.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_cell_images
x,y,region=make_cell_images(n=12, seed=2)
fig,axs=plt.subplots(2,4,figsize=(10,5))
for j,ax in enumerate(axs.flat):
    ax.imshow(x[j],cmap='gray',vmin=0,vmax=1); ax.set_title(f'Class {y[j]}'); ax.axis('off')
plt.tight_layout(); plt.show()
print('Array shape (images, rows, columns):',x.shape)
```

### Experiment 2 of 3

```python
j=int(np.flatnonzero(y==1)[0]); img=x[j]
fig,axs=plt.subplots(1,3,figsize=(10,3))
axs[0].imshow(img,cmap='gray'); axs[0].set_title('Original')
axs[1].hist(img.ravel(),bins=25); axs[1].set_title('Pixel histogram')
axs[2].imshow(img,cmap='gray'); axs[2].imshow(region[j],alpha=.4,cmap='autumn'); axs[2].set_title('Known synthetic change')
for ax in [axs[0],axs[2]]: ax.axis('off')
plt.tight_layout(); plt.show()
```

### Experiment 3 of 3

```python
print('Mean intensity class 0:',x[y==0].mean())
print('Mean intensity class 1:',x[y==1].mean())
assert x.ndim==3 and region.dtype==bool and region[y==1].any()
```
