## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from scipy.ndimage import gaussian_filter
from diffcam_edu import make_cell_images
images,y,masks=make_cell_images(20,seed=2)
j=int(np.flatnonzero(y==1)[0]);img=images[j]
blur=gaussian_filter(img,sigma=1)
threshold=blur>.55
fig,ax=plt.subplots(1,3,figsize=(10,3))
for a,A,t in zip(ax,[img,blur,threshold],['Raw','Gaussian smoothing','Threshold']):
    a.imshow(A,cmap='gray');a.set_title(t);a.axis('off')
plt.show()
```

### Experiment 2 of 3

```python
from diffcam_edu.evaluation import mask_iou
print('Threshold IoU vs simulated change:',mask_iou(threshold,masks[j]))
for t in [.35,.45,.55,.65]:
    print(f't={t}: detected={int((blur>t).sum())} pixels')
```

### Experiment 3 of 3

```python
fig,ax=plt.subplots(figsize=(4,4))
ax.imshow(img,cmap='gray');ax.contour(masks[j],[.5],colors='lime');ax.contour(threshold,[.5],colors='red');ax.set_title('Green: known change, red: threshold');ax.axis('off');plt.show()
```
