A regularized scatter matrix is especially sensitive to reference count and covariance scale. Compare repeated reference subsamples while fixing the target and encoder. ## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu.math import diffcam_weights
rng=np.random.default_rng(5)
refs=rng.multivariate_normal(np.zeros(6),np.diag([2,1,.5,.3,.1,.05]),size=300)
target=np.array([1.,.4,1.,-.3,.5,0.])
base=diffcam_weights(target,refs,ridge=.5,unit_norm=True)
scans=[]
for n in [5,10,20,50,100,200]:
    cs=[]
    for _ in range(30):
        sel=rng.choice(len(refs),size=n,replace=False)
        w=diffcam_weights(target,refs[sel],ridge=.5,unit_norm=True)
        cs.append(w@base)
    scans.append([n,np.mean(cs),np.std(cs)])
print('N, mean cosine, SD cosine:\n',np.round(scans,3))
```

### Experiment 2 of 3

```python
v=np.asarray(scans)
plt.errorbar(v[:,0],v[:,1],yerr=v[:,2],capsize=3,marker='o')
plt.setp(plt.gca(),xscale='log',ylim=(-1,1.05),xlabel='Reference count',ylabel='Cosine vs full reference')
plt.show()
```

### Experiment 3 of 3

```python
for lam in [1e-3,.1,1.,10.]:
    w=diffcam_weights(target,refs[:10],ridge=lam,unit_norm=True)
    print('Ridge:',lam,'cosine to full-direction:',round(w@base,3))
```
