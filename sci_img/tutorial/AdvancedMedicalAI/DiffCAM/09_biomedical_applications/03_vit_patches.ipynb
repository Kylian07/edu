## Concept and motivation

A Vision Transformer divides an image into patches represented as tokens, often followed by a learned encoder. Here we **only simulate patch tokenization**. This does not implement the paper's ViT attribution procedure or a pretrained ViT. The key caution is that class and special tokens must not be naively reshaped into a spatial grid.

**Key vocabulary:** patch size, token, embedding, spatial grid, class token.

**Learning goal:** Explain token-to-grid reshaping and the special-token pitfall.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

### Experiment 1 of 3

```python
from diffcam_edu import make_cell_images
x,y,m=make_cell_images(16,size=32,seed=7)
im=x[1];patch=4
patches=im.reshape(8,4,8,4).transpose(0,2,1,3).reshape(64,16)
print('64 patches, each 16 pixels:',patches.shape)
mean_token=patches.mean(axis=1).reshape(8,8)
fig,ax=plt.subplots(1,2,figsize=(7,3))
ax[0].imshow(im,cmap='gray');ax[0].set_title('Image')
ax[1].imshow(mean_token,cmap='gray');ax[1].set_title('Patch intensities')
for a in ax:a.axis('off')
plt.show()
```

### Experiment 2 of 3

```python
# A spatial map can be mapped back from the grid of patch scores.
upscaled=np.repeat(np.repeat(mean_token,patch,axis=0),patch,axis=1)
assert upscaled.shape==im.shape
plt.imshow(im,cmap='gray');plt.imshow(upscaled,alpha=.35,cmap='magma');plt.title('Illustrative token-to-pixel overlay');plt.axis('off');plt.show()
```
