# The same image, different reference questions

**Weeks:** 2 · **Core question:** Why is a comparison meaningless without a reference?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

DiffCAM is explicitly *reference-relative*. Describing a cell as “large” requires a group against which to compare it. In this experiment, the target remains fixed while the reference group changes. You should anticipate different explanations without changing any classifier.

**Key vocabulary:** target, reference population, baseline, contrastive question, distribution shift.

**Learning goal:** Explain why no absolute saliency statement follows from a reference-relative difference.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 2

```python
from diffcam_edu import make_tabular_cells
X,y=make_tabular_cells(70,seed=4)
target=np.array([6.6,4.1]); refs_0=X[y==0]; refs_1=X[y==1]
fig,ax=plt.subplots(figsize=(6,5))
ax.scatter(*refs_0.T,alpha=.4,label='Reference population 0')
ax.scatter(*refs_1.T,alpha=.4,label='Reference population 1')
ax.scatter(*target,c='black',s=130,marker='*',label='Fixed target')
ax.legend();ax.set(xlabel='Cell area (arbitrary units)',ylabel='Intensity (arbitrary units)');plt.show()
```

### Experiment 2 of 2

```python
for name,ref in [('Population 0',refs_0),('Population 1',refs_1)]:
    diff=target-ref.mean(axis=0)
    print(name,'mean:',ref.mean(axis=0).round(2),'target minus mean:',diff.round(2))
assert not np.allclose(refs_0.mean(axis=0),refs_1.mean(axis=0))
```
