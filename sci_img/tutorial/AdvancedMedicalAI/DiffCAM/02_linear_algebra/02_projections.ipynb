# Projection directions and separability

**Weeks:** 8–9 · **Core question:** Which direction best distinguishes a target from a population?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

The projection of a feature vector onto a direction reduces multidimensional information to a scalar. Different directions produce different group separation and within-group variability. These are the competing ingredients of the DiffCAM objective.

**Key vocabulary:** unit direction, projection, projected mean, projected variance.

**Learning goal:** Explain the tradeoff between large target difference and low reference variance.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_tabular_cells
X,y=make_tabular_cells(90,seed=6)
angles=np.linspace(0,np.pi,181)
score=[]
for t in angles:
    w=np.array([np.cos(t),np.sin(t)])
    groups=[X[y==k]@w for k in [0,1]]
    score.append(abs(np.mean(groups[1])-np.mean(groups[0]))/(np.std(groups[0])+1e-9))
best=angles[np.argmax(score)]
print('Best grid angle in degrees:',np.degrees(best))
```

### Experiment 2 of 3

```python
fig,ax=plt.subplots(1,2,figsize=(10,4))
ax[0].scatter(*X.T,c=y,alpha=.35);ax[0].arrow(4,4,2*np.cos(best),2*np.sin(best),width=.02,color='black')
ax[0].set(xlabel='Area',ylabel='Intensity',title='Candidate direction')
ax[1].plot(np.degrees(angles),score);ax[1].axvline(np.degrees(best),c='red',ls='--')
ax[1].set(xlabel='Angle (degrees)',ylabel='Projected standardized separation');plt.show()
```

### Experiment 3 of 3

```python
w=np.array([np.cos(best),np.sin(best)])
proj=X@w
print('Projected reference variance:',proj[y==0].var())
assert abs(np.linalg.norm(w)-1)<1e-10
```
