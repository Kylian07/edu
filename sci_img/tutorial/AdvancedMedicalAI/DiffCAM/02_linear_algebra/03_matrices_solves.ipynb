## Concept and motivation

DiffCAM uses a regularized linear system. Matrix inversion notation is useful for a derivation, but numerical code should call a solver such as `numpy.linalg.solve`. Singular or poorly conditioned matrices require special care.

**Key vocabulary:** matrix product, transpose, eigenvalue, rank, singularity, ridge regularization.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
A=np.array([[3.,1.],[1.,2.]])
b=np.array([1.,4.])
w=np.linalg.solve(A,b)
print('Solution:',w,'Residual:',np.linalg.norm(A@w-b))
```

### Experiment 2 of 3

```python
singular=np.array([[1.,2.],[2.,4.]])
print('Rank:',np.linalg.matrix_rank(singular))
for ridge in [.001,.1,1.0]:
    w=np.linalg.solve(singular+ridge*np.eye(2),b)
    print('Ridge:',ridge,'weights:',w.round(3),'norm:',round(np.linalg.norm(w),3))
```

### Experiment 3 of 3

```python
eigvals=np.linalg.eigvalsh(A)
print('Eigenvalues:',eigvals)
assert np.all(eigvals>0)
```
