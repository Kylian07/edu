### Experiment 1 of 3

```python
fluorescence=[12.1,11.8,15.0,14.2,9.6]
total=0
for measurement in fluorescence:
    total+=measurement
print('Mean intensity:',total/len(fluorescence))
```

### Experiment 2 of 3

```python
def above_threshold(values,threshold):
    return [v for v in values if v>threshold]
selected=above_threshold(fluorescence,12.0)
print('High-intensity cells:',selected)
assert len(selected)==3
```
