# NumPy arrays and image pixels

**Weeks:** 4 · **Core question:** How can an image be represented as a matrix?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

A grayscale image is a height-by-width matrix. A dataset is a larger array whose first axis indexes examples. NumPy broadcasts operations across all pixels and makes feature extraction compact.

**Key vocabulary:** shape, axis, indexing, dtype, broadcasting, normalization.

**Learning goal:** Explain the dimensions of an `[N,H,W]` image collection.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

### Experiment 1 of 3

```python
from diffcam_edu import make_cell_images
images,labels,masks=make_cell_images(20,seed=5)
print('Shape:',images.shape,'dtype:',images.dtype)
print('Top-left patch:',images[0,:4,:4])
print('Mean intensity per image:',images.mean(axis=(1,2))[:5])
```

### Experiment 2 of 3

```python
center=images[:,12:20,12:20]
center_mean=center.mean(axis=(1,2))
whole_mean=images.mean(axis=(1,2))
plt.scatter(whole_mean,center_mean,c=labels,cmap='coolwarm')
plt.xlabel('Mean whole-image intensity');plt.ylabel('Mean center intensity');plt.show()
```

### Experiment 3 of 3

```python
bright_pixels=(images>.65).sum(axis=(1,2))
print('Bright-pixel counts:',bright_pixels[:8]);assert len(bright_pixels)==20
```
