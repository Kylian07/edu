# Vectors and distances for cell measurements

## Concept and motivation

A point in feature space can represent a cell without retaining its raw pixels. Choosing features and scales changes our notion of similarity; this foreshadows learned features in neural networks.

**Key vocabulary:** embedding, vector, norm, dot product, Euclidean distance, scaling.

**Learning goal:** Calculate Euclidean distance and explain sensitivity to scaling.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_tabular_cells
X,y=make_tabular_cells(60)
a,b=X[0],X[1]
print('a=',a,'b=',b,'difference=',a-b)
print('Distance=',np.linalg.norm(a-b),'Dot product=',a@b)
```

### Experiment 2 of 3

```python
ax=plt.figure(figsize=(5,5)).gca();ax.scatter(X[:,0],X[:,1],c=y,alpha=.5)
ax.arrow(0,0,a[0],a[1],head_width=.18,color='black',length_includes_head=True)
ax.set(xlim=(0,10),ylim=(0,10),xlabel='Cell area',ylabel='Intensity');plt.show()
```

### Experiment 3 of 3

```python
X_rescaled=X.copy();X_rescaled[:,0]*=100
print('Original Euclidean distance:',np.linalg.norm(X[0]-X[1]))
print('After changing the area units:',np.linalg.norm(X_rescaled[0]-X_rescaled[1]))
```
