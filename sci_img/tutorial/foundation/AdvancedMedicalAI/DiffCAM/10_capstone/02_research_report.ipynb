## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
# Example analysis template: quantify reference-group sensitivity using simulated data.
from diffcam_edu import make_tabular_cells
from diffcam_edu.math import diffcam_weights
X,y=make_tabular_cells(100,seed=123);target=np.array([6.9,4.1]);rng=np.random.default_rng(1)
reference=X[y==0]
W=np.stack([diffcam_weights(target,reference[rng.choice(len(reference),30,replace=False)],ridge=1.,unit_norm=True) for _ in range(50)])
print('Mean direction:',W.mean(axis=0).round(3))
print('Direction SD:',W.std(axis=0).round(3))
```

### Experiment 2 of 3

```python
fig,ax=plt.subplots(figsize=(5,5))
ax.quiver(np.zeros(len(W)),np.zeros(len(W)),W[:,0],W[:,1],angles='xy',scale_units='xy',scale=1,alpha=.25)
ax.set(xlim=(-1.2,1.2),ylim=(-1.2,1.2),title='Reference-resampling weight uncertainty',xlabel='Feature 1',ylabel='Feature 2');ax.grid();plt.show()
```

### Experiment 3 of 3

```python
hypotheses=[
 'H1: Explanation changes with reference-group composition.',
 'H2: Small reference groups yield less stable directions.',
 'H3: Better localization need not imply model faithfulness.'
]
for h in hypotheses:print(h)
print('Next step: propose measurements and controls for each hypothesis.')
```
