# Convolutions reveal local image structure

**Weeks:** 20 · **Core question:** Why are convolutional filters natural for images?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

Convolution detects spatial patterns using the same weights at different positions. Filters can react to edges, bright spots, or texture. Feature maps preserve *where* the pattern was found, unlike a single image-wide label.

**Key vocabulary:** kernel, convolution, feature map, padding, stride, translation equivariance.

**Learning goal:** Explain why convolution produces a spatial feature map.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from scipy.ndimage import convolve
from diffcam_edu import make_cell_images
x,y,m=make_cell_images(10,seed=3);im=x[0]
vertical=np.array([[1.,0,-1],[2,0,-2],[1,0,-1]])
horizontal=vertical.T
edge_x=convolve(im,vertical)
edge_y=convolve(im,horizontal)
```

### Experiment 2 of 3

```python
fig,axs=plt.subplots(1,3,figsize=(10,3))
for ax,A,t in zip(axs,[im,edge_x,edge_y],['Image','Vertical response','Horizontal response']):
    ax.imshow(A,cmap='gray');ax.set_title(t);ax.axis('off')
plt.show()
```

### Experiment 3 of 3

```python
print('Filter shape:',vertical.shape,'map shape:',edge_x.shape)
assert im.shape==edge_x.shape
```
