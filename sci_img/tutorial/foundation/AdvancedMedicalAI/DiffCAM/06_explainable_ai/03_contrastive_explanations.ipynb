## Concept and motivation

Contrastive explanations answer a specific question: “What separates this example from *these* examples?” We construct a simple visualization using directly measured features. Do not call it a causal explanation of a decision.

**Key vocabulary:** contrast class, counterfactual question, class-conditional reference, exemplar selection.

**Learning goal:** Describe target and reference groups explicitly before interpreting a heatmap.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 2

```python
from diffcam_edu import make_tabular_cells
from diffcam_edu.math import diffcam_weights
X,y=make_tabular_cells(80,seed=1)
target=np.array([7.0,4.1])
choices={'group 0':X[y==0],'group 1':X[y==1],'all':X}
weights={name:diffcam_weights(target,r,ridge=1.) for name,r in choices.items()}
for k,v in weights.items():print(k, v.round(3))
```

### Experiment 2 of 2

```python
fig,ax=plt.subplots(figsize=(5,5))
ax.scatter(*X.T,c=y,cmap='coolwarm',alpha=.35)
ax.scatter(*target,c='black',s=100,marker='*')
for name,w in weights.items():
    v=w/(np.linalg.norm(w)+1e-12)
    ax.arrow(*target,*(v*1.8),label=name,width=.025,length_includes_head=True)
    ax.text(*(target+v*2),name)
ax.set(xlabel='Area',ylabel='Intensity');plt.show()
```
