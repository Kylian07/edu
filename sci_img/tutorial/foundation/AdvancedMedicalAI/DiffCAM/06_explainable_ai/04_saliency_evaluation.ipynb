## Concept and motivation

A visually convincing map may not track relevant regions or model behavior. Localization requires a region annotation; faithfulness asks whether changing highlighted pixels changes a model output; sanity checks test dependence on model parameters. These evaluate different claims.

**Key vocabulary:** IoU, pointing game, precision-recall, perturbation, model randomization.

**Learning goal:** Choose the right metric for a stated explanation claim and note its limitations.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_cell_images
from diffcam_edu.evaluation import mask_iou,pointing_game,area_under_precision_recall
x,y,m=make_cell_images(16,seed=12)
j=int(np.flatnonzero(y==1)[0]);truth=m[j]
rng=np.random.default_rng(1)
good=truth.astype(float)+rng.normal(0,.05,truth.shape)
bad=rng.random(truth.shape)
for name,s in [('region-linked',good),('random',bad)]:
    chosen=s>np.quantile(s,.9)
    print(name,'IoU',round(mask_iou(chosen,truth),3),'pointing',pointing_game(s,truth),'AUPRC',round(area_under_precision_recall(s,truth),3))
```

### Experiment 2 of 3

```python
fig,ax=plt.subplots(1,3,figsize=(9,3))
for a,A,t in zip(ax,[x[j],good,bad],['Target','Region-linked','Random']):
    a.imshow(A,cmap='inferno');a.set_title(t);a.axis('off')
plt.show()
```

### Experiment 3 of 3

```python
print('Lesion fraction:',truth.mean())
print('AUPRC random baseline depends strongly on positive fraction.')
```
