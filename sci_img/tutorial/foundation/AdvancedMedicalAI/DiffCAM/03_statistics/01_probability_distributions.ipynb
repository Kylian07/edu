# Distributions and variation among cells

**Weeks:** 11 · **Core question:** What does a reference population look like statistically?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

Two cells can differ for reasons unrelated to condition. We model this variability with distributions and distinguish group means from variability. Do not conflate biological variance with labeling noise or technical batch effects.

**Key vocabulary:** sample, distribution, histogram, sampling bias, mean, standard deviation.

**Learning goal:** Explain why individual reference images cannot substitute automatically for a population distribution.

> All images and patient-like examples in this repository are synthetic teaching artifacts, not clinical data.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu import make_tabular_cells
X,y=make_tabular_cells(80,seed=9)
fig,ax=plt.subplots(1,2,figsize=(10,3))
for k in (0,1):
    ax[0].hist(X[y==k,0],bins=15,alpha=.5,label=f'Group {k}')
    ax[1].hist(X[y==k,1],bins=15,alpha=.5,label=f'Group {k}')
ax[0].set_title('Area distributions');ax[1].set_title('Intensity distributions');ax[0].legend();plt.show()
```

### Experiment 2 of 3

```python
for k in (0,1):
    print('Group',k,'mean=',X[y==k].mean(axis=0).round(2),'SD=',X[y==k].std(axis=0).round(2))
```

### Experiment 3 of 3

```python
rng=np.random.default_rng(8)
means=np.array([rng.choice(X[y==0,0],size=20,replace=True).mean() for _ in range(500)])
print('Bootstrap 95% interval:',np.quantile(means,[.025,.975]))
```
