# Estimating covariance with few references

**Weeks:** 14 · **Core question:** What happens when we have more features than reference samples?

**Prerequisites:** complete preceding notebooks in the curriculum. Each lesson targets 45–90 minutes including exercises.

## Concept and motivation

A reference scatter matrix built from N observations has rank at most N−1. If features outnumber references, it cannot be inverted without regularization. This is common for image encoders with hundreds of channels.

**Key vocabulary:** high dimensionality, rank bound, condition number, regularization, stability.

**Learning goal:** Explain what the sample-rank bound means for DiffCAM reference-set design.

## Guided experiment
Execute the code cells in order. Before running each cell, predict what it will show; afterward, explain what changed and why.

### Experiment 1 of 3

```python
from diffcam_edu.math import reference_scatter,diffcam_weights
rng=np.random.default_rng(8);refs=rng.normal(size=(6,20));target=rng.normal(size=20)
mu,S=reference_scatter(refs)
print('Scatter rank:',np.linalg.matrix_rank(S),'feature dimension:',S.shape[0])
```

### Experiment 2 of 3

```python
for lam in [1e-5,1e-3,.1,10.]:
    w=diffcam_weights(target,refs,ridge=lam)
    print(f'ridge={lam:g}  norm={np.linalg.norm(w):.2f}  residual={np.linalg.norm((S+lam*np.eye(20))@w-(target-mu)):.2e}')
```

### Experiment 3 of 3

```python
fig,ax=plt.subplots(figsize=(6,3))
for n in (5,10,50):
    sample=rng.normal(size=(n,20));t=rng.normal(size=20)
    values=[np.linalg.norm(diffcam_weights(t,sample,ridge=float(r))) for r in np.logspace(-3,2,20)]
    ax.plot(np.logspace(-3,2,20),values,label=f'N={n}')
ax.set_xscale('log');ax.set_yscale('log');ax.set(xlabel='Ridge',ylabel='Weight norm');ax.legend();plt.show()
```
