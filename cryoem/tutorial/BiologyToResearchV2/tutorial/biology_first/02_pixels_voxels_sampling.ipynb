## 1. Why this matters

A detector samples a continuous signal on a grid. A pixel is an area sample in a 2D image; a voxel is a volume sample in a 3D reconstruction. Coordinates may be written x–y–z by microscopists but indexed z–y–x by NumPy. Confusing these conventions can put a picked particle at the wrong location or make a segmentation appear transposed.


## 2. The idea in plain language

If the pixel spacing is `a` Å per pixel, the shortest representable period is `2a` Å by Nyquist sampling. Achieved resolution is usually worse because the signal may be weak, blurred, or absent. A 64³ subtomogram with 4 Å voxels has a nominal side length of 256 Å (25.6 nm). State both pixel size and box size whenever comparing structures.

```python
voxel_A = np.array([4.0, 4.0, 6.0])  # z, y, x spacing
shape = np.array([64, 64, 64])
physical_nm = shape * voxel_A / 10
print('Box extent (z,y,x) in nm:', physical_nm)
print('Nyquist period in Å:', 2 * voxel_A)
point_zyx = np.array([10, 20, 30])
print('Point position in nm:', point_zyx * voxel_A / 10)
assert np.allclose(physical_nm, [25.6,25.6,38.4])
```
