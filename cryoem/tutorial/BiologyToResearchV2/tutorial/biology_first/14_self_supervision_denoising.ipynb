## 1. Why this matters

For many tomograms there is no clean ground-truth image. Self-supervised methods create a training signal from the data itself, such as predicting held-out pixels from neighbors. Generative methods model plausible image distributions and can synthesize examples or remove noise. A denoised image may be easier to inspect, but plausible features may not be measured features. Structural claims still need raw-data and independent-map support.


## 2. The idea in plain language

A J-invariant denoiser predicts a held-out value without seeing that value. If noise is independent across positions, minimizing prediction error can favor the underlying signal. Correlated noise or strong neighboring artifacts violate the simple argument. The learner should compare denoising quality with preservation of small structures and downstream biological conclusions.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
rng=np.random.default_rng(SEED)
x=np.linspace(-2,2,101)
clean=np.exp(-x*x/0.2)
noisy=clean+rng.normal(0,.25,len(x))
# Neighbor-only predictor never reads the target sample at i.
pred=(noisy[:-2]+noisy[2:])/2
rmse_raw=np.sqrt(np.mean((noisy[1:-1]-clean[1:-1])**2))
rmse_pred=np.sqrt(np.mean((pred-clean[1:-1])**2))
print('raw RMSE',round(rmse_raw,3),'neighbor predictor RMSE',round(rmse_pred,3))
assert len(pred)==99
```

The neighbor average is only an analogy for a blind-spot denoiser. It can blur a narrow peak. Check whether denoising changes localization, segmentation, or map features you care about.
