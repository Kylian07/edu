## 1. Why this matters

Machine learning extracts patterns from examples. In cryo-ET, images from the same tomogram share noise, acquisition conditions, and biological context. Randomly splitting crops from one tomogram into both training and test sets can make evaluation look better than performance on a new specimen. A paper should state the split unit, class counts, baselines, and uncertainty.


## 2. The idea in plain language

For binary labels, logistic regression models `p(y=1|x)=sigmoid(w·x+b)`. Training minimizes cross-entropy. A more expressive network can fit complex visual patterns, but the same evaluation rules apply. Accuracy alone is misleading when most crops are background; report per-class precision and recall or a confusion matrix.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
rng=np.random.default_rng(SEED)
X=np.vstack([rng.normal(-1,.5,(50,2)),rng.normal(1,.5,(50,2))])
y=np.array([0]*50+[1]*50)
w=np.zeros(2); b=0.0
for _ in range(300):
    p=1/(1+np.exp(-np.clip(X@w+b,-30,30)))
    w-=.1*(X.T@(p-y)/len(y))
    b-=.1*float(np.mean(p-y))
pred=(1/(1+np.exp(-(X@w+b)))>.5).astype(int)
print('toy training accuracy',np.mean(pred==y),'weights',w.round(2))
assert np.mean(pred==y)>.9
```
