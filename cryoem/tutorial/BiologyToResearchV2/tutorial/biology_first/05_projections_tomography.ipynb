## 1. Why this matters

In cryo-ET, a specimen is tilted and imaged repeatedly. Each tilt provides a 2D projection of the same 3D region. Mechanical and thickness limits prevent full 180° sampling. The resulting unsampled region in Fourier space is called the missing wedge; reconstructions often have direction-dependent detail. This is a measurement limitation, not a biological shape.


## 2. The idea in plain language

The Fourier slice theorem says that the 2D Fourier transform of a projection corresponds to a central slice through the 3D Fourier transform of the object. To make that statement concrete here, we use a 2D phantom and 1D projections. This is a conceptual analogue, not a production tomographic reconstruction.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
n=65
y,x=np.mgrid[-32:33,-32:33]
phantom=np.exp(-((x-10)**2+(y+4)**2)/35)+0.8*np.exp(-((x+9)**2+(y-5)**2)/55)
for angle in [0,30,60,90]:
    t=np.deg2rad(angle)
    coordinate=x*np.cos(t)+y*np.sin(t)
    bins=np.clip(np.rint(coordinate).astype(int)+32,0,64)
    projection=np.bincount(bins.ravel(),weights=phantom.ravel(),minlength=n)
    print(f'{angle:2d}° projection mass={projection.sum():.2f}; peak bin={projection.argmax()}')
assert np.isclose(phantom.sum(),projection.sum())
```
