## 1. Why this matters

A 3D density map is an intermediate result. To make a biological claim, researchers may compare known structures, fit components, examine spatial context, test perturbations, or connect to biochemical data. A good interpretation distinguishes what the map directly supports from what comes from prior knowledge. If multiple models fit the same low-resolution density, report the ambiguity.


## 2. The idea in plain language

A useful evidence chain is: raw images → corrected measurements → reconstruction → validated features → candidate identity/conformation → biological interpretation. At every arrow, write a possible alternative explanation and a check. Fitting is optimization under a scoring function; a high fit score may be non-unique in crowded or low-resolution maps.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
# Compare two candidate fits to a coarse, noisy 1D density.
rng=np.random.default_rng(SEED)
x=np.linspace(-2,2,100)
measured=np.exp(-(x+.5)**2/.15)+.8*np.exp(-(x-.5)**2/.18)+rng.normal(0,.06,100)
candidates={
 'two-domain model':np.exp(-(x+.5)**2/.15)+.8*np.exp(-(x-.5)**2/.18),
 'single broad model':1.2*np.exp(-x*x/.7)}
for name,model in candidates.items():
    print(name,'mean squared residual',round(float(np.mean((measured-model)**2)),4))
assert len(measured)==100
```

## 4. Interpret, then test a change

A better residual for one toy model is evidence in this setup, not definitive molecular identity. Model complexity, resolution, prior structures, and independent biochemical checks matter.
