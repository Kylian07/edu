## 1. Why this matters

Particle picking asks where potential molecules are. In a crowded tomogram, neighboring structures, ice features, and noise create false positives. A strict threshold gives fewer, more reliable picks; a loose threshold finds more true objects but also more mistakes. A picker must be evaluated on held-out regions and paired with downstream validation.


## 2. The idea in plain language

With TP, FP, and FN counts, `precision=TP/(TP+FP)` and `recall=TP/(TP+FN)`. An apparently good picker can still miss rare states or perform poorly in a new cell type. A detection threshold should therefore be chosen for the biological use case and reported with a test set independent of training.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
truth={5,14,29,44,60}
scores=np.zeros(70)
scores[[5,14,29,44,60]]=[.94,.82,.61,.74,.48]
scores[[12,23,52,67]]=[.69,.53,.81,.42]
for threshold in [.4,.6,.8]:
    predicted=set(np.flatnonzero(scores>=threshold))
    tp=len(predicted & truth); fp=len(predicted-truth); fn=len(truth-predicted)
    precision=tp/(tp+fp) if tp+fp else 0
    recall=tp/(tp+fn) if tp+fn else 0
    print(threshold, 'precision',round(precision,2),'recall',round(recall,2))
assert len(truth)==5
```
