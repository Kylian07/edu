## 1. Why this matters

A cell is a crowded place, but its jobs are carried out by molecular assemblies. A ribosome translates RNA; a membrane separates compartments; a protein complex can switch conformations. A structural image can therefore answer several different questions: **what** is present, **where** it is, **how** parts fit, and **which state** an assembly occupies. Those questions require different evidence. A blob of density is not yet a named protein. An apparent overlap is not yet a molecular interaction.


## 2. The idea in plain language

Use a scale ladder: 1 nm = 10 Å; a protein is often a few nanometres across, a ribosome tens of nanometres, and a cell thousands of nanometres. If the pixel size is 4 Å, a 20 nm object spans about 50 pixels. Size is a useful first check, but similarly sized objects can have different identities. We will repeatedly separate **observation** (a visible density) from **interpretation** (a biological label).


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
# Scale ladder: convert nanometres to pixels at a chosen sampling rate.
objects_nm = {'protein complex': 8, 'ribosome': 25, 'bacterium width': 500}
pixel_size_A = 4.0
for name, diameter_nm in objects_nm.items():
    diameter_A = diameter_nm * 10
    print(f'{name:18s}: {diameter_A/pixel_size_A:7.1f} pixels across')
assert objects_nm['ribosome'] * 10 / pixel_size_A == 62.5
```
