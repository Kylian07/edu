## 1. Why this matters

Fine molecular features correspond to rapidly varying image patterns. Fourier analysis expresses an image as a combination of spatial frequencies. Microscope defocus changes which frequencies are transferred strongly, weakly, or with reversed sign. A contrast transfer function (CTF) is part of the forward model, so computational correction must respect frequencies that were never measured.


## 2. The idea in plain language

For a toy 1D signal, write `Y(k)=H(k)S(k)+N(k)`. Here `H` is the transfer function; at `H(k)=0`, the signal component is absent from that image. Dividing by a tiny `H` amplifies noise. Real electron-microscope CTFs depend on electron wavelength, defocus, spherical aberration and other effects; our numerical exercise illustrates transfer and zeros, not a calibrated microscope.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
x = np.arange(128)
s = np.sin(2*np.pi*3*x/128) + 0.5*np.sin(2*np.pi*16*x/128)
S = np.fft.rfft(s)
f = np.fft.rfftfreq(len(s))
H = np.cos(2*np.pi*8*f)
Y = H*S
recovered = np.fft.irfft(Y, n=len(s))
print('Original RMS:', np.sqrt(np.mean(s*s)))
print('Transferred RMS:', np.sqrt(np.mean(recovered*recovered)))
print('Near-zero transfer bins:', np.where(np.abs(H)<0.05)[0][:10])
assert len(S) == 65
```
