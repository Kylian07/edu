## 1. Biological and computational motivation

A foundation or generative model can supply plausible structure when measurements are noisy or incomplete. That can help denoising and segmentation, but plausibility is not evidence that a feature was measured. For example, a promptable model can make a coherent 3D mask from a few 2D cues, while an image generator may fill a missing pattern based on its training distribution. A careful paper separates model prior, raw-data agreement, and independent biological validation.


## 2. Theory and assumptions

Think of a reconstruction as minimizing `data_misfit(V) + λ prior_penalty(V)`. When `λ` rises, the solution follows the prior more strongly. This is only a conceptual analogy to score-based or diffusion methods, not a derivation of their training objective. In practice, examine output stability across prompts and random seeds, fidelity to raw projections, cross-plane consistency, topology, and rare-feature preservation.


## 3. Runnable experiment

Predict a result before running. In Colab, NumPy is included; elsewhere install it with `%pip install numpy` if needed.


```python
import numpy as np
SEED=7
print("NumPy",np.__version__,"| device: CPU | seed:",SEED)
```

```python
measurement=np.array([0.,1.,0.,.4,0.])
prior=np.array([0.,.8,.8,.8,0.])
for lam in [0,.25,1,4]:
    estimate=(measurement+lam*prior)/(1+lam)
    print('prior weight',lam,'estimate',estimate.round(2),
          'residual',round(float(np.linalg.norm(estimate-measurement)),3))
assert np.array_equal((measurement+0*prior)/(1+0),measurement)
```

## 4. Interpretation and challenge

The middle value rises although its measured value is zero. In real data, zero may mean missing information, low contrast, or true absence; the distinction must be tested rather than assumed. A plausible output should be compared with original images and independent measurements.

**Change one thing:** Invent a tiny rare structure that the prior does not expect. Increase the prior weight and show when it disappears. Design a blinded evaluation that would detect this loss.
