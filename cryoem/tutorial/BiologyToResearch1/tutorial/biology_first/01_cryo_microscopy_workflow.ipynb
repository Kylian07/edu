## 1. Why this matters

A biological sample is rapidly frozen into vitreous ice so its molecules remain embedded in a hydrated environment. Electrons passing through it carry information about the specimen, but the detector records a noisy, optics-modulated image. In single-particle analysis, many isolated copies of a target are averaged. In tomography, the same cell region is imaged at successive tilts to recover a 3D volume. Tomography preserves cellular context; repeated-particle averaging can sharpen recurring structures. These are complementary experiments.


## 2. The idea in plain language

A useful forward model is `image = imaging_operator(structure, orientation, optics) + noise`. It reminds us that reconstruction is an **inverse problem**: we infer an unobserved structure from indirect observations. A reconstruction is a model supported by data, not a photograph of atoms. Each processing step should document its inputs, outputs, and assumptions.
