## 1. Why this matters

After picking, particles can be aligned and averaged to strengthen recurring features. A single average may hide multiple conformations or contamination. Resolution assessment therefore uses two independently refined half sets. Their Fourier shell correlation (FSC) compares agreement by spatial frequency. A nominal FSC threshold is meaningful only with appropriate independent processing and masking practices.


## 2. The idea in plain language

At each frequency shell, `FSC = Re(sum(F1*conj(F2))) / sqrt(sum|F1|² sum|F2|²)`. The common 0.143 cutoff is used for **gold-standard FSC of independent half maps**. It is not a universal Fourier ring correlation or “half-bit” threshold. Global resolution can also hide local variation and anisotropy.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
rng=np.random.default_rng(SEED)
y,x=np.mgrid[-16:16,-16:16]
true=np.exp(-(x*x+y*y)/35)
a=true+rng.normal(0,.7,size=true.shape)
b=true+rng.normal(0,.7,size=true.shape)
F1=np.fft.fftshift(np.fft.fft2(a)); F2=np.fft.fftshift(np.fft.fft2(b))
r=np.sqrt(x*x+y*y)
for lo in range(0,16,3):
    m=(r>=lo)&(r<lo+3)
    numerator=np.real(np.sum(F1[m]*np.conj(F2[m])))
    denom=np.sqrt(np.sum(np.abs(F1[m])**2)*np.sum(np.abs(F2[m])**2))
    print(f'shell {lo:2d}-{lo+3:2d}: corr={numerator/denom:.3f}')
assert true.shape==(32,32)
```
