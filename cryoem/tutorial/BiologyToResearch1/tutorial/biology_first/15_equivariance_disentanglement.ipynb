## 1. Why this matters

A molecule can rotate without changing its identity. A useful identity representation should therefore ignore pose, while a pose predictor should change predictably when the input rotates. This is the distinction between invariance and equivariance. A different conformation may be the biological result we seek, so a model that removes all variation can erase discovery. The design must say which factors it preserves and which it discounts.


## 2. The idea in plain language

If `g` is a rotation and `f` an identity encoder, invariance means `f(gx)=f(x)`. For a pose predictor `h`, equivariance means `h(gx)=g h(x)` under a suitable pose representation. Disentanglement aims to represent content and transformation separately; the two are not identifiable from arbitrary data without assumptions or constraints.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
image=np.arange(9).reshape(3,3)
rotated=np.rot90(image)
identity_feature=lambda a: np.sort(a.ravel())
pose_feature=lambda a: np.unravel_index(np.argmax(a),a.shape)
print('identity unchanged:',np.array_equal(identity_feature(image),identity_feature(rotated)))
print('pose before/after:',pose_feature(image),pose_feature(rotated))
assert np.array_equal(identity_feature(image),identity_feature(rotated))
```
