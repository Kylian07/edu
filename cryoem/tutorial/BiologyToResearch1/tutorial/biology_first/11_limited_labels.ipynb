## 1. Why this matters

Expert labels are expensive, particularly in 3D crowded volumes. Few-shot methods learn from a handful of labeled examples; one-shot methods use one; weak supervision uses incomplete or coarse labels; active learning asks which sample should be labeled next. Open-set recognition adds a crucial restraint: the correct answer may be “unknown” rather than one of the known classes.


## 2. The idea in plain language

For two-class probabilities `(p,1-p)`, uncertainty is highest near `p=0.5`. An active learner may select uncertain samples, but uncertainty alone can favor noisy or out-of-distribution crops. Selection should consider diversity, the labeling budget, and a held-out test set. A few-shot result must report the number of examples **per class** and the number of repeated draws.

```python
probabilities=np.array([.02,.18,.49,.51,.79,.96])
entropy=-(probabilities*np.log(probabilities)+(1-probabilities)*np.log(1-probabilities))
order=np.argsort(-entropy)
print('Query order (most uncertain first):',order)
print('Probabilities:',probabilities[order])
assert set(order[:2])=={2,3}
```
