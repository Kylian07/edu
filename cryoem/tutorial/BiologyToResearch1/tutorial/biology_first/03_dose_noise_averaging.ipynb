## 1. Why this matters

Electrons reveal structure but also damage biological material. Low-dose images can be hard to interpret individually. If repeated objects can be aligned, their common signal adds while independent zero-mean noise partly cancels. Real data complicate this ideal result: motion, contamination, heterogeneity, and correlated noise can prevent perfect gains.


## 2. The idea in plain language

For aligned observations `X_i = S + ε_i` with noise standard deviation `σ`, the mean has noise standard deviation `σ/√N`. Thus **amplitude** signal-to-noise ratio improves as `√N`; signal-power/noise-power ratio improves as `N` under this simple model. The distinction matters when reading papers because “SNR” is defined in different ways. Averaging misaligned or nonidentical particles can erase features even if the output looks smooth.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
rng = np.random.default_rng(SEED)
x = np.linspace(-1, 1, 64)
signal = np.exp(-(x/0.18)**2)
stack = signal + rng.normal(0, 1, size=(1024,64))
for n in [1,4,16,64,256,1024]:
    estimate = stack[:n].mean(axis=0)
    error = np.sqrt(np.mean((estimate-signal)**2))
    print(f'N={n:4d}, RMS error={error:.3f}')
assert stack.shape == (1024,64)
```
