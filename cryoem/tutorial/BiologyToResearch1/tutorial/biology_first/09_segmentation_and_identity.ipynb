## 1. Why this matters

Segmentation marks the voxels belonging to a structure; classification assigns a category to a crop or particle. The two can support each other but answer different questions. A membrane boundary may be segmented correctly without identifying a nearby complex, while a ribosome classifier might identify a crop without tracing its exact surface. Biological identity is strongest when morphology, location, molecular labels, and independent data agree.


## 2. The idea in plain language

For a predicted mask P and reference mask T, `Dice=2|P∩T|/(|P|+|T|)`. Dice is sensitive to class size and does not reveal every topology error. For sparse biological objects, inspect boundary quality, connected components, and downstream measurements too.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
truth=np.zeros((20,20),bool); truth[4:15,5:15]=True
pred=np.zeros_like(truth); pred[5:16,6:16]=True
intersection=np.count_nonzero(truth&pred)
dice=2*intersection/(truth.sum()+pred.sum())
print('truth voxels',truth.sum(),'predicted',pred.sum(),'Dice',round(dice,3))
print('false positives',np.count_nonzero(pred&~truth))
assert 0<dice<1
```
