## 1. Why this matters

Suppose a tomogram contains many unknown complexes. Instead of training on named examples, we can crop candidates, align them, encode their shapes, group similar encodings, and average each group. Iteration can improve the representation and the groups. Yet clusters may reflect orientation, missing-wedge direction, specimen type, or noise rather than protein identity. Biological interpretation is a separate validation stage.


## 2. The idea in plain language

A simple clustering objective is `sum_i ||z_i - μ_(cluster_i)||²` for feature vectors `z_i`. The objective only rewards compact groups under a chosen representation. It does not know biology. The central research question is how to learn features that preserve molecular content while discounting nuisance changes such as pose and acquisition.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
rng=np.random.default_rng(SEED)
Z=np.vstack([rng.normal([-1,0],.25,(30,2)),rng.normal([1,0],.25,(30,2))])
centers=Z[[0,30]].copy()
for _ in range(10):
    labels=np.argmin(((Z[:,None,:]-centers[None,:,:])**2).sum(2),axis=1)
    centers=np.vstack([Z[labels==k].mean(0) for k in range(2)])
print('cluster sizes:',np.bincount(labels),'centers:',centers.round(2))
assert min(np.bincount(labels))>0
```
