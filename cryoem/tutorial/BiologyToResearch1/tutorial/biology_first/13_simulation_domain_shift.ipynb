## 1. Why this matters

Simulated tomograms give known particle positions and identities, which are useful for measuring detection and classification errors. But simulation can accidentally make a task easier than reality: textures, noise, crowding, missing-wedge artifacts, and biology may differ. Domain adaptation tries to bridge this gap. A strong evaluation still checks real held-out specimens and reports the simulator settings.


## 2. The idea in plain language

Think of data as `X = F(structure, pose, imaging, noise, crowding)`. If training and test distributions differ in any input to F, performance may fall. Domain randomization deliberately varies acquisition and biological nuisance factors during training. It is useful only if the chosen variation covers realistic conditions.


## 3. Run the experiment

Run the setup cell, then the activity. Before running the activity, predict one quantity or pattern you expect. The numbers and shapes are illustrative; they are not published data.


```python
# ── Shared setup (runs on free Colab CPU) ──
# Colab includes NumPy. If running elsewhere, install with: %pip install numpy
import numpy as np
SEED = 7
print('NumPy', np.__version__, '| device: CPU | seed:', SEED)
```

```python
rng=np.random.default_rng(SEED)
true=np.r_[rng.normal(-1,.25,100),rng.normal(1,.25,100)]
labels=np.r_[np.zeros(100,int),np.ones(100,int)]
threshold=0.0
for shift in [0,.4,1.0]:
    measured=true+shift+rng.normal(0,.15,len(true))
    predicted=(measured>threshold).astype(int)
    print('test shift',shift,'accuracy',round(np.mean(predicted==labels),3))
assert len(labels)==200
```
