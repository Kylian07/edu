## 🧫 The biological question

A pathogen — say a bacterium — gets into the body. Within hours, immune cells sense it and switch genes on.
Cytokines such as **TNF**, **IL1B** and **IL6** call in reinforcements; interferon-stimulated genes such as **MX1**
and **ISG15** put neighbouring cells on alert; the liver releases acute-phase proteins such as **SAA1** and the
complement factor **C3**. Meanwhile "housekeeping" genes such as **ACTB** (β-actin) and **GAPDH** carry on as before.

A research team wants to understand this response. They take **spleen, kidney and liver** from **four control and
four infected** animals and measure the activity (the *expression*) of **400 genes** in each of the 24 samples. As
soon as the numbers arrive, the questions start — and every single one of them is a mathematical question:

### 🧰 Course toolkit

The next cell copies the shared helpers this notebook needs from [`course_toolkit.py`](../course_toolkit.py). `truth_table` and `implies` evaluate logical rules (Module 02), `draw_venn` draws Venn diagrams
           of two or three sets (Module 03), and `simulate_tissue_expression` generates the course's infection
           experiment — with a fixed seed, so that everyone gets the same numbers, and together with the hidden
           truth. You do not need to read this code today. Just run it.

````python
# ── Course toolkit (identical to ../course_toolkit.py v1.0; copied so this notebook runs alone) ──
import itertools

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def implies(p: bool, q: bool) -> bool:
    """Material implication p ⇒ q: false only when p is true and q is false."""
    return (not p) or q


def truth_table(formulas: dict, variables: list, letters: bool = True) -> pd.DataFrame:
    """Truth table of one or more Boolean formulas.

    ``variables`` is a list of names such as ``["p", "q"]``; ``formulas`` maps a column
    title to a Python function of those variables, e.g. ``{"p ⇒ q": lambda p, q: implies(p, q)}``.
    Rows run through every assignment in the textbook order (T…T first, F…F last).
    With ``letters=True`` the table shows "T"/"F"; otherwise it holds Python booleans.
    """
    rows = []
    for values in itertools.product([True, False], repeat=len(variables)):
        env = dict(zip(variables, values))
        row = dict(env)
        for title, f in formulas.items():
            row[title] = bool(f(**env))
        rows.append(row)
    table = pd.DataFrame(rows)
    if letters:
        table = table.replace({True: "T", False: "F"})
    return table


def _venn_order(x):
    """Sort key for Venn-diagram labels: numbers in numeric order first, then everything else as text."""
    if isinstance(x, (int, float, np.integer, np.floating)) and not isinstance(x, bool):
        return (0, float(x), "")
    return (1, 0.0, str(x))


def draw_venn(sets: dict, universe=None, ax=None, show_elements: bool = True, max_items: int = 6,
              title: str | None = None, highlight=None):
    """Draw a Venn diagram of two or three sets with matplotlib only.

    ``sets`` maps a label to a Python set, e.g. ``{"S": {1, 2, 3}, "T": {3, 4}}``.  Each region
    shows its elements (or ``n = …`` if it has more than ``max_items``).  If ``universe`` is given,
    a rectangle is drawn and the elements outside every set are listed in a corner.
    ``highlight`` is a function of the membership tuple, e.g. ``lambda s, t: s and not t`` for
    S \\ T; the regions where it is true are shaded.  Returns the Axes.
    """
    labels = list(sets)
    if len(labels) not in (2, 3):
        raise ValueError("draw_venn draws two or three sets")
    if ax is None:
        _, ax = plt.subplots(figsize=(4.6, 3.8))
    if len(labels) == 2:
        centres = [(-0.55, 0.0), (0.55, 0.0)]
    else:
        centres = [(-0.55, 0.32), (0.55, 0.32), (0.0, -0.6)]
    radius = 1.0
    colours = ["tab:blue", "tab:orange", "tab:green"]
    # classify a fine grid of points to find each region (used for shading and label positions)
    gx, gy = np.meshgrid(np.linspace(-2.2, 2.2, 441), np.linspace(-2.0, 1.9, 391))
    inside = [((gx - cx) ** 2 + (gy - cy) ** 2) <= radius ** 2 for cx, cy in centres]
    if highlight is not None:
        mask = np.zeros_like(gx, dtype=bool)
        for member in itertools.product([True, False], repeat=len(labels)):
            if any(member) and highlight(*member):
                region = np.ones_like(gx, dtype=bool)
                for k, m in enumerate(member):
                    region &= inside[k] if m else ~inside[k]
                mask |= region
        ax.contourf(gx, gy, mask.astype(float), levels=[0.5, 1.5], colors=["#ffd54f"], alpha=0.8)
    for (cx, cy), lab, col in zip(centres, labels, colours):
        ax.add_patch(plt.Circle((cx, cy), radius, fill=False, lw=2, color=col))
        lx = cx + (-1.0 if cx < 0 else 1.0 if cx > 0 else 0.0) * 0.95
        ly = cy + (0.95 if cy >= 0 else -1.15)
        ax.text(lx, ly, lab, ha="center", va="center", fontsize=11, fontweight="bold", color=col)
    elements = set().union(*sets.values())
    for member in itertools.product([True, False], repeat=len(labels)):
        if not any(member):
            continue
        region_items = [x for x in elements if all((x in sets[l]) == m for l, m in zip(labels, member))]
        region = np.ones_like(gx, dtype=bool)
        for k, m in enumerate(member):
            region &= inside[k] if m else ~inside[k]
        if not region.any():
            continue
        px, py = gx[region].mean(), gy[region].mean()
        listing = ", ".join(str(x) for x in sorted(region_items, key=_venn_order))
        if show_elements and len(region_items) <= max_items and len(listing) <= 16:
            text = listing if region_items else "∅"
        else:
            text = f"n = {len(region_items)}"
        ax.text(px, py, text, ha="center", va="center", fontsize=8, wrap=True)
    if universe is not None:
        ax.add_patch(plt.Rectangle((-2.1, -1.9), 4.2, 3.7, fill=False, lw=1, color="0.4"))
        outside = sorted(set(universe) - elements, key=_venn_order)
        text = ", ".join(map(str, outside)) if len(outside) <= max_items else f"n = {len(outside)}"
        ax.text(-2.0, -1.8, "U: " + (text if outside else "∅"), fontsize=8, va="bottom", color="0.3")
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.0, 1.9)
    ax.set_aspect("equal")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax


# Named genes of the simulated infection experiment.  The infection-response genes are real genes
# known to be induced during infection (cytokines, interferon-stimulated genes, acute-phase
# proteins; lysozyme, LYZ, is an antimicrobial enzyme rather than a classic acute-phase protein and is
# grouped with them only for simplicity); the reference genes are classic "housekeeping" genes used
# to normalise qPCR.
INFECTION_PATHWAYS = {
    "cytokine signalling": ["IL1B", "TNF", "IL6", "CXCL8"],
    "interferon response": ["IRF1", "STAT1", "MX1", "ISG15"],
    "acute phase": ["SAA1", "HAMP", "C3", "LYZ"],
}


HOUSEKEEPING_GENES = ["ACTB", "GAPDH", "EEF1A1", "RPL13A", "TBP", "HPRT1",
                      "PPIA", "UBC", "YWHAZ", "SDHA", "TUBA1A", "B2M"]


def simulate_tissue_expression(n_genes: int = 400, tissues=("spleen", "kidney", "liver"), n_reps: int = 4,
                               seed: int = 2680) -> dict:
    """SIMULATED log2 gene expression from an infection experiment in several tissues.

    Every tissue has ``n_reps`` control and ``n_reps`` infected samples.  The 12 genes of
    ``INFECTION_PATHWAYS`` respond to infection in every tissue (up by 1.5–3 log2 units); further
    "g…" genes respond in one tissue only or in two tissues, so the sets of truly changed genes
    overlap like the differentially-expressed-gene sets of a real multi-tissue study.  The 12
    ``HOUSEKEEPING_GENES`` are stable (tiny noise, no response).  About a third of all genes differ in
    baseline level between tissues.

    Returns a dict with ``X`` (samples × genes, log2 expression), ``samples`` (DataFrame with
    columns sample, tissue, condition), ``genes`` (list), ``true_lfc`` (DataFrame tissues × genes,
    true infected − control difference), ``true_de`` (dict tissue → set of truly changed genes),
    ``pathways`` (dict pathway → genes) and ``source``.
    """
    rng = np.random.default_rng(seed)
    named = [g for genes in INFECTION_PATHWAYS.values() for g in genes]
    n_other = n_genes - len(named) - len(HOUSEKEEPING_GENES)
    if n_other < 60:
        raise ValueError("n_genes must be at least 84")
    genes = named + HOUSEKEEPING_GENES + [f"g{i:03d}" for i in range(1, n_other + 1)]
    J, T = len(genes), len(tissues)
    base = rng.normal(7.0, 1.8, J)
    base[len(named):len(named) + len(HOUSEKEEPING_GENES)] = rng.uniform(9.0, 12.0, len(HOUSEKEEPING_GENES))
    tissue_shift = rng.normal(0.0, 1.0, (T, J)) * (rng.random(J) < 0.35)
    tissue_shift[:, len(named):len(named) + len(HOUSEKEEPING_GENES)] = 0.0
    lfc = np.zeros((T, J))
    lfc[:, :len(named)] = rng.uniform(1.5, 3.0, (T, len(named)))
    other = list(range(len(named) + len(HOUSEKEEPING_GENES), J))
    rng.shuffle(other)
    sizes_single = [24, 18, 14, 12, 10][:T]
    pairs = list(itertools.combinations(range(T), 2))
    sizes_pair = [10, 5, 4, 3, 3, 2, 2, 2, 2, 2][:len(pairs)]
    k = 0

    def effect(m):
        return rng.uniform(1.0, 2.5, m) * np.where(rng.random(m) < 0.7, 1.0, -1.0)

    for t, m in enumerate(sizes_single):
        idx = other[k:k + m]
        lfc[t, idx] = effect(m)
        k += m
    for (t1, t2), m in zip(pairs, sizes_pair):
        idx = other[k:k + m]
        e = effect(m)
        lfc[t1, idx] = e
        lfc[t2, idx] = e * rng.uniform(0.8, 1.2, m)
        k += m
    noise_sd = rng.uniform(0.25, 0.55, J)
    noise_sd[len(named):len(named) + len(HOUSEKEEPING_GENES)] = 0.08
    rows, X = [], []
    for t, tissue in enumerate(tissues):
        for condition in ("control", "infected"):
            for r in range(1, n_reps + 1):
                mean = base + tissue_shift[t] + (lfc[t] if condition == "infected" else 0.0)
                X.append(mean + rng.normal(0.0, noise_sd))
                rows.append((f"{tissue}_{condition[:3]}{r}", tissue, condition))
    samples = pd.DataFrame(rows, columns=["sample", "tissue", "condition"])
    true_lfc = pd.DataFrame(lfc, index=list(tissues), columns=genes)
    true_de = {tissue: {g for g, v in zip(genes, lfc[t]) if v != 0} for t, tissue in enumerate(tissues)}
    return {"X": np.array(X), "samples": samples, "genes": genes, "true_lfc": true_lfc,
            "true_de": true_de, "pathways": {k_: list(v) for k_, v in INFECTION_PATHWAYS.items()},
            "source": "SIMULATED log2 expression (BiologyToMath simulate_tissue_expression)"}

print("🧰 toolkit loaded:", ", ".join(sorted(['HOUSEKEEPING_GENES', 'INFECTION_PATHWAYS', '_venn_order', 'draw_venn', 'implies', 'simulate_tissue_expression', 'truth_table'])))
````

````python
# ── 0.1 Generate the course's infection experiment (SIMULATED) ──
expr = simulate_tissue_expression()        # a dict; the fixed seed inside gives everyone the same numbers
X = expr["X"]                              # the data matrix: one row per sample, one column per gene
samples, genes = expr["samples"], expr["genes"]
print(expr["source"])
print("X has shape", X.shape, "→", X.shape[0], "samples (rows) ×", X.shape[1], "genes (columns)")
print("\nsamples per tissue and condition:")
print(samples.groupby(["tissue", "condition"]).size().unstack())
print("\nnumber of genes that TRULY respond to infection (known because we simulated):",
      {tissue: len(changed) for tissue, changed in expr["true_de"].items()})
````

🔍 **What just happened?** The simulator produced a table `X` with **24 rows** (samples: 3 tissues × 2 conditions ×
4 replicates) and **400 columns** (genes). Each number is a **log2 expression level**: an increase of 1 means

## 1 · Why biology needs mathematics

> 📍 **Where we are.** Before any formulas: why should a biologist spend a semester on logic, matrices, derivatives
> and probabilities? Because computational biology — and, increasingly, all of biology — consists of three
> activities, and each of them is built on mathematics.

**1. Modelling biological processes.** Gene regulatory networks, metabolic pathways, protein folding, the dynamics of
a population or an epidemic: to *model* a system is to describe it with equations and logical rules precise enough
for a computer to simulate. That needs **logic** to state the relationships (*gene X is expressed if activator A is
present and repressor R is absent*), **linear algebra** to represent the states of a system and the transitions
between them, and **calculus** to describe how those states change in time.

**2. Analysing large-scale biological data.** High-throughput technologies — **RNA-seq** for gene expression,
**single-cell sequencing**, proteomics, metabolomics — flood biology with numbers. Extracting meaning from them is
not just a matter of running a software package; it needs a solid quantitative framework. **Statistics** identifies
patterns, quantifies variability, assesses significance and separates signal from the noise inherent in every
biological measurement. **Linear algebra** is everywhere too, because virtually every large biological data set is
stored as a **matrix**.

**3. Making predictions.** Will this patient's disease progress? Which protein is a promising drug target? What
structure will this sequence fold into? Prediction rests on **probability**, **statistical inference** and the
mathematics of **machine learning**.

What this course gives you for these three activities:

| Branch of mathematics | What it does for a biologist | A first example |
|---|---|---|
| **Logic** | structures biological rules precisely: conditions, pathways, algorithms | "*lacZ* is transcribed if and only if lactose is present and glucose is absent" |
| **Linear algebra** | the language of data: a data set is a matrix; samples are vectors | 24 samples × 400 genes; the distance between two expression profiles |
| **Calculus** | describes change and finds optimal values | the growth rate $dN/dt$ of a bacterial population; the best-fitting parameter |
| **Probability and statistics** | models uncertainty and draws reliable conclusions from noisy data | the chance of infection after a positive test; is a fold change real? |

🧮 **Math refresher — powers of ten and logarithms.** $10^3 = 10 \times 10 \times 10 = 1000$: the exponent counts
the factors of ten (the zeros). Large numbers are written in **scientific notation**, $3.1 \times 10^9 =
3{,}100{,}000{,}000$; in Python, `3.1e9`. The **logarithm to base 10** undoes the power: $\log_{10}(1000) = 3$
because $10^3 = 1000$, and $\log_{10}(3.1\times 10^9) \approx 9.49$. On a **logarithmic axis** each step of one unit
means "ten times more" — the only way to show a thousand and a billion in the same picture.

How many numbers does modern biology produce? Run the next cell.

````python
# ── 1.1 How many numbers does biology produce? (order-of-magnitude sizes, illustrative) ──
SECONDS_PER_YEAR = 365.25 * 24 * 3600
datasets = pd.DataFrame([
    ("our simulated infection experiment", 24, 400),
    ("a bulk RNA-seq study: 24 samples × all genes", 24, 20_000),
    ("one single-cell RNA-seq experiment", 10_000, 20_000),
    ("a single-cell atlas of a million cells", 1_000_000, 20_000),
    ("one human genome sequence (letters)", 1, 3_100_000_000),
], columns=["data set", "rows", "columns"])
datasets["numbers"] = datasets["rows"] * datasets["columns"]
datasets["log10(numbers)"] = np.log10(datasets["numbers"]).round(2)
datasets["years to read at 1 number/s"] = datasets["numbers"] / SECONDS_PER_YEAR
print(datasets.to_string(index=False, formatters={"numbers": "{:.2e}".format,
                                                  "years to read at 1 number/s": "{:.4g}".format}))

fig, ax = plt.subplots(figsize=(8, 2.9))
ax.barh(datasets["data set"], datasets["numbers"], color="tab:blue")
ax.set_xscale("log")
ax.invert_yaxis()
ax.set_xlabel("how many numbers (log scale: each tick = 10× more)")
ax.set_title("Biology produces far more numbers than anyone can read — illustrative sizes")
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** Our toy experiment holds 24 × 400 = **9,600** numbers ($\log_{10} \approx 3.98$) — reading
them aloud at one per second would take under three hours. A real bulk RNA-seq study with all ~20,000 genes holds
**4.8 × 10⁵** numbers (about 5.6 days of reading), one single-cell experiment **2 × 10⁸** (about **6.3 years**), a
million-cell atlas **2 × 10¹⁰** (over **600 years**), and one human genome **3.1 × 10⁹** letters (about 98 years).
No one reads such data; we *compute* with them, and to compute we need mathematics. Notice the shape too: the
expression tables have far more **columns (genes)** than rows (samples) — a fact that will matter again and again.

Size is only half of the problem. The other half is **noise**: measure the same thing twice and you get two different
numbers.

````python
# ── 1.2 Biology is noisy: four replicates of the same tissue, condition and gene ──
spleen_control = (samples["tissue"] == "spleen") & (samples["condition"] == "control")
spleen_infected = (samples["tissue"] == "spleen") & (samples["condition"] == "infected")
for gene in ["IL6", "ACTB"]:
    j = genes.index(gene)                                   # the column of this gene
    ctrl, inf = X[spleen_control.to_numpy(), j], X[spleen_infected.to_numpy(), j]
    print(f"{gene:5s} control : {np.round(ctrl, 2)}  mean {ctrl.mean():.2f}, spread (SD) {ctrl.std(ddof=1):.2f}")
    print(f"{gene:5s} infected: {np.round(inf, 2)}  mean {inf.mean():.2f}, spread (SD) {inf.std(ddof=1):.2f}")
    print(f"      difference of means (infected − control): {inf.mean() - ctrl.mean():+.2f} log2 units\n")
````

🔍 **What just happened?** Four control spleens gave four *different* IL6 levels (5.10, 3.81, 4.53, 4.16 — a spread,
or standard deviation, of 0.55 log2 units), although they are biological replicates of the same condition. In the
infected spleens IL6 is higher (mean 6.69 vs 4.40, a difference of **+2.30** log2 units, i.e. roughly $2^{2.3} \approx 5$
times more mRNA). The housekeeping gene **ACTB** barely moves: its replicates agree to about 0.07 units and the difference
of means is only +0.02. Is the IL6 difference *real*, or could four noisy replicates produce it by chance? That is a
question of **probability and statistics** — demo 7.3 gives a first answer, and Parts F–G of the course the full one.

### 📘 Lecture notes — why computational biology needs this course

- **Computational biology consists of three activities:** (1) **modelling biological processes** (gene regulation,
  metabolism, protein folding, population dynamics) — describing complex systems with equations and logical rules;
  (2) **analysing large-scale biological data** (RNA-seq, single-cell data, proteomics, metabolomics) — finding
  patterns, quantifying variability and assessing significance in noisy measurements; (3) **making predictions**
  (disease progression, drug targets, protein structures, functional relationships).
- **What 02-680 provides:** **logic** for *structuring* biological rules (formal definitions of interactions,
  conditional statements for pathways, algorithms with a clear logical flow); **linear algebra** for working with
  *data matrices* — nearly every high-throughput data set is a matrix, and vectors, matrix operations, eigenvalues
  and eigenvectors are needed to manipulate it and reduce its dimension; **calculus** for change and optimisation;
  **probability and statistics** for *modelling uncertainty* — biology is inherently noisy (experimental error,
  natural variation), probability quantifies the uncertainty, and statistics draws reliable conclusions (hypothesis
  tests, predictive models, confidence in findings).
- **The order of the course follows the dependencies:** it starts with logic and builds systematically through
  linear algebra, calculus, probability and statistics; each block is a building block for the next (Section 2).
- **Prerequisites of 02-680:** high-school mathematics, single-variable and multivariable calculus, logical and
  mathematical reasoning ability; no advanced statistics or programming. (This course assumes less: high-school
  algebra; Part D teaches the calculus from scratch, and [Module 01](01_python_for_mathematics.ipynb) the programming.)
- **Two facts about biological data** shown in this section: data sets are **huge** (10⁴ to 10¹⁰ numbers) and
  usually **wide** (more genes than samples); and replicate measurements **differ** (noise), so every conclusion
  needs an estimate of its uncertainty.

**Common confusion.** "Computational biology means running software." The software implements mathematics; without
understanding the mathematics you cannot choose a method, check its assumptions or interpret its output.

> 🎓 **Exam tip.** For any biological task, be able to say which activity it is (model, analyse, predict) and which
> branch of mathematics carries it — e.g. "is this gene differentially expressed?" = analysis → statistics (on a
> matrix → linear algebra).

Behind every notebook runs a Python program called the **kernel**. When you run a code cell, the kernel executes it

1. **Run cells in order, from the top.** A cell that uses a name defined further up fails if that cell has not run.
2. **If things get strange, restart.** *Runtime → Restart session and run all* (Colab) or *Kernel → Restart Kernel
   and Run All Cells* (Jupyter) wipes the kernel's memory and runs everything again from a clean state.
3. **Keep your own copy.** On Colab, *File → Save a copy in Drive* — then your changes and notes are yours to keep.

A first exercise in counting, the kind of mathematics you will meet in Module 04: how many different DNA sequences
of length 10 are there? Each position can hold one of 4 bases, so there are $4 \times 4 \times \dots \times 4 = 4^{10}$.

````python
# ── 3.1 Your first code cell: how many DNA sequences of length k? ──
n_bases = 4                      # A, C, G, T
k = 10                           # sequence length
n_sequences = n_bases ** k       # ** means "to the power of"
print(f"number of DNA sequences of length {k}: 4^{k} = {n_sequences:,}")
````

🔍 **What just happened?** The kernel stored three numbers under the names `n_bases`, `k` and `n_sequences` (such
names are **variables**) and printed $4^{10}$ = **1,048,576** — already a million different 10-letter DNA words.
Lines starting with `#` are **comments**: notes for humans that Python ignores.

````python
# ── 3.2 Cells remember: this cell uses names created by the cell above ──
genome_length = 3_100_000_000    # letters in one copy of the human genome (≈ 3.1 × 10^9)
print(f"expected occurrences of one particular {k}-letter word in a random genome of this length: "
      f"{genome_length / n_sequences:,.0f}")
for k_try in [10, 12, 14, 16, 18]:
    print(f"  k = {k_try:2d}: 4^k = {n_bases ** k_try:>17,}   expected occurrences ≈ {genome_length / n_bases ** k_try:10.3f}")
````

🔍 **What just happened?** This cell used `n_bases` and `k` from the *previous* cell — the kernel remembered them. In
a random sequence of 3.1 billion letters, any particular 10-letter word is expected about **2,956** times, but a
16-letter word only **0.722** times and an 18-letter word 0.045 times. That is why a short "barcode" of about 16–20
letters is usually enough to pin down a unique place in a genome — the idea behind sequencing-read mapping and the

### 4.2 Subscripts and indices: naming many numbers at once

A data set contains thousands of numbers; we cannot give each its own letter. Instead we use one letter with an
**index** (subscript):

* $x_i$ ("x sub i", or "x i") is the $i$-th number of a list $x_1, x_2, \dots, x_n$; the index $i$ runs from 1 to $n$.
* $x_{ij}$ (or $x_{i,j}$) has two indices: the entry in **row $i$, column $j$** of a table. In our data,
  $x_{ij}$ = the expression of gene $j$ in sample $i$.
* A superscript is usually a **power**: $x_i^2 = x_i \cdot x_i$. When a superscript is a label instead, it is put in
  parentheses, $x^{(i)}$, to avoid confusion.
* Decorations change the meaning: $\bar x$ ("x bar") is an average, $\hat\theta$ ("theta hat") an estimate of
  $\theta$, $x'$ ("x prime") a second value or a derivative.

⚠️ **Mathematics counts from 1, Python from 0.** The mathematician's $x_{1,1}$ (first row, first column) is `X[0, 0]`
in Python, and $x_{ij}$ is `X[i - 1, j - 1]`.

### 4.3 Sums and products: Σ and Π

🧮 **Math refresher — the summation sign.** Adding up a list is so common that it has its own symbol, a capital
sigma:

$$\sum_{i=1}^{n} x_i = x_1 + x_2 + \dots + x_n .$$

*In words:* "the sum of $x_i$ for $i$ from 1 to $n$" — let the index $i$ take the values $1, 2, \dots, n$ in turn,
write down $x_i$ each time, and add everything up. The letter $i$ is a placeholder (a *dummy* index): $\sum_{i=1}^n x_i$
and $\sum_{k=1}^n x_k$ mean exactly the same.

Worked examples, by hand:

* $\sum_{i=1}^{4} i = 1 + 2 + 3 + 4 = 10$.
* $\sum_{i=1}^{4} i^2 = 1 + 4 + 9 + 16 = 30$.
* $\sum_{i=1}^{3} 5 = 5 + 5 + 5 = 15$ — a constant summed $n$ times gives $n$ times the constant.
* The **mean** of $n$ measurements is $\bar x = \frac{1}{n}\sum_{i=1}^{n} x_i$.

Three rules follow directly from the definition (write out the sums to see why):

$$\sum_{i=1}^{n} c = n\,c, \qquad \sum_{i=1}^{n} (a\,x_i + b\,y_i) = a\sum_{i=1}^{n} x_i + b\sum_{i=1}^{n} y_i, \qquad
\sum_{i=1}^{n} x_i = \sum_{i=1}^{m} x_i + \sum_{i=m+1}^{n} x_i .$$

*In words:* summing a constant $n$ times multiplies it by $n$; constants can be pulled out of a sum and a sum of two
lists can be split into two sums (**linearity**); and a sum can be cut into two consecutive pieces.

⚠️ **But** $\sum_i x_i y_i \neq \left(\sum_i x_i\right)\left(\sum_i y_i\right)$ and $\sum_i x_i^2 \neq \left(\sum_i x_i\right)^2$
in general: try $x = (1, 2)$: $1 + 4 = 5$ but $(1 + 2)^2 = 9$.

The **product sign** (capital pi) does the same with multiplication:

$$\prod_{i=1}^{n} x_i = x_1 \cdot x_2 \cdots x_n .$$

*In words:* "the product of $x_i$ for $i$ from 1 to $n$" — multiply all the $x_i$ together. Example:
$\prod_{i=1}^{4} i = 1 \cdot 2 \cdot 3 \cdot 4 = 24$, which is called $4!$ ("four factorial"). Products appear whenever
independent events must *all* happen: if the bases of a DNA word are independent, its probability is the product of
the base probabilities, $P(x_1 x_2 \dots x_L) = \prod_{t=1}^{L} p(x_t)$. Logarithms turn products into sums,
$\log \prod_t p_t = \sum_t \log p_t$ — a trick used all over statistics (Module 23).

````python
# ── 4.2 Σ and Π by hand (a loop) and by Python's built-ins ──
n = 4
total = 0
for i in range(1, n + 1):                 # i = 1, 2, 3, 4 (range stops just BEFORE n + 1)
    total = total + i ** 2
print("Σ_{i=1}^{4} i²  — loop:", total, "| sum():", sum(i ** 2 for i in range(1, n + 1)))
product = 1
for i in range(1, n + 1):
    product = product * i
print("Π_{i=1}^{4} i   — loop:", product, "| math.prod():", math.prod(range(1, n + 1)),
      "| math.factorial(4):", math.factorial(4))

il6_control = X[spleen_control.to_numpy(), genes.index("IL6")]
print(f"mean x̄ = (1/n) Σ x_i of the IL6 spleen controls: {il6_control.sum() / len(il6_control):.4f}"
      f" | NumPy: {il6_control.mean():.4f}")

base_probability = {"A": 0.3, "C": 0.2, "G": 0.2, "T": 0.3}   # illustrative AT-rich base composition
word = "GATTACA"
p_word = math.prod(base_probability[b] for b in word)
log_p_word = sum(math.log(base_probability[b]) for b in word)
print(f"P({word}) = Π p(x_t) = {p_word:.3e}; Σ log p(x_t) = {log_p_word:.4f} = log of the product: {math.log(p_word):.4f}")
````

🔍 **What just happened?** The loop adds $1 + 4 + 9 + 16$ one term at a time and reaches **30**, exactly like
Python's `sum`; the product loop reaches $1 \cdot 2 \cdot 3 \cdot 4 = $ **24** $= 4!$. A `for` loop *is* a Σ or a Π
written out: the loop variable plays the role of the index $i$, and `range(1, n + 1)` is the range "$i = 1$ to $n$".
The mean of the four IL6 control values, computed literally as $\frac{1}{n}\sum x_i$, is **4.3966**, identical to
NumPy's `mean`. Finally, the probability of the word GATTACA under independent bases is
$0.2 \cdot 0.3 \cdot 0.3 \cdot 0.3 \cdot 0.3 \cdot 0.2 \cdot 0.3 = 0.2^2 \cdot 0.3^5 =$ **9.72 × 10⁻⁵**, and the sum
of the seven logarithms (−9.2387) equals the logarithm of that product.

(In Python the union can also be written with a vertical bar between the two sets.) The next cell runs some of
these statements on our gene sets.

````python
# ── 4.4 ∈, ⊆, ∀ and ∃ evaluated on the infection experiment ──
all_genes = set(genes)
housekeeping = set(HOUSEKEEPING_GENES)
changed_in = expr["true_de"]                        # tissue → set of truly responding genes
lfc_spleen = expr["true_lfc"].loc["spleen"]         # true log2 fold change of every gene in the spleen

print("'IL6' ∈ changed_in['spleen']             :", "IL6" in changed_in["spleen"])
print("'ACTB' ∉ changed_in['spleen']            :", "ACTB" not in changed_in["spleen"])
print("housekeeping ⊆ all_genes                 :", housekeeping <= all_genes)
print("∀ g ∈ housekeeping: g ∉ changed (spleen) :", all(g not in changed_in["spleen"] for g in housekeeping))
print("∃ g ∈ changed (spleen): lfc(g) > 2.5     :", any(lfc_spleen[g] > 2.5 for g in changed_in["spleen"]))
witnesses = {g for g in changed_in["spleen"] if lfc_spleen[g] > 2.5}       # set-builder notation in Python
print("   the witnesses {g ∈ changed : lfc(g) > 2.5} =", sorted(witnesses))
print("∀ g ∈ changed (spleen): lfc(g) > 0       :", all(lfc_spleen[g] > 0 for g in changed_in["spleen"]))
down = sorted(g for g in changed_in["spleen"] if lfc_spleen[g] < 0)
print(f"   counterexamples (genes that go DOWN): {len(down)}, e.g. {down[:3]}")
````

🔍 **What just happened?** Each line is a mathematical statement evaluated to `True` or `False`. IL6 is in the set of
spleen-responding genes and ACTB is not; the housekeeping genes form a subset of all genes; **for all** housekeeping
genes it is true that they do not respond. The **existential** statement "there exists a responding gene with fold
change above 2.5" is true, and Python can list the witnesses: **C3, IL1B, LYZ, SAA1**. The last universal statement,

### 4.6 Functions: $f: S \to T$

A **function** $f$ from a set $S$ (the **domain**, what you may put in) to a set $T$ (the **codomain**, where the
results live) assigns to *every* $x \in S$ exactly *one* value $f(x) \in T$. We write

$$f: S \to T, \qquad x \mapsto f(x) .$$

*In words:* "$f$ maps $S$ to $T$; it sends each $x$ to $f(x)$". The arrow $\to$ goes between *sets*, the barred arrow
$\mapsto$ between *elements*. Examples: $f: \mathbb R \to \mathbb R$, $x \mapsto x^2$ (squaring); the GC content
$\text{GC}: \{\text{DNA strings}\} \to [0, 1]$; the genetic code, which maps each of the 64 codons to one of 20 amino
acids or "stop" ([Module 03](03_sets_and_functions.ipynb)). Two functions can be chained: $(g \circ f)(x) = g(f(x))$ ("g after f") applies $f$ first.
In Python a function is written with `def`, and its *type hints* (`seq: str`, `-> float`) state the domain and the
codomain.

````python
# ── 4.5 Functions f: S → T in Python, and composition g ∘ f ──
def gc_content(seq: str) -> float:
    """GC: DNA strings → [0, 1], the fraction of letters that are G or C."""
    return (seq.count("G") + seq.count("C")) / len(seq)


def as_percent(fraction: float) -> str:
    """[0, 1] → text, e.g. 0.25 ↦ '25.0 %'."""
    return f"{100 * fraction:.1f} %"


for dna in ["GATTACA", "GCGCGCAT", "ATATAT"]:
    print(f"GC({dna}) = {gc_content(dna):.3f}   (as_percent ∘ GC)({dna}) = as_percent(GC({dna})) = "
          f"{as_percent(gc_content(dna))}")
print("squaring is many-to-one: (−3)² =", (-3) ** 2, "= 3² =", 3 ** 2)
````

🔍 **What just happened?** `gc_content` is a function from DNA strings to $[0, 1]$: GATTACA has 2 G/C letters out of
7, so GC = **0.286**; GCGCGCAT has 6 of 8, **0.750**; ATATAT has none, **0.000**. `as_percent(gc_content(dna))` is
the **composition** $\text{as\_percent} \circ \text{GC}$: the inner function is applied first. Squaring sends both
$-3$ and $3$ to 9 — a function may map several inputs to the same output (it is then *many-to-one*); what it may
never do is give two outputs for one input.

### 4.8 How to read a definition, a theorem and a proof

Mathematical texts are built from a few kinds of statements:

* A **definition** gives a precise name to an idea. Example: *"A set $S$ is a **subset** of a set $T$, written
  $S \subseteq T$, iff every element of $S$ is an element of $T$: $\forall x: x \in S \Rightarrow x \in T$."* To read
  it: find the new word (in bold), restate it in your own words, then build one example (housekeeping genes ⊆ all
  genes) and one non-example ($\{\text{IL6}, \text{QQQ}\} \not\subseteq$ our genes).
* A **theorem** (or **lemma** — a helper theorem; **corollary** — an easy consequence) is a statement that has been
  proved. It almost always has the shape **"if hypotheses, then conclusion"**. Example: *"For all finite sets $S$ and
  $T$: $\lvert S \cup T\rvert = \lvert S\rvert + \lvert T\rvert - \lvert S \cap T\rvert$."* To read it: name the
  hypotheses (finite sets), name the conclusion (the counting formula), test it on an example (demo 6.2 does), and ask
  what breaks if you drop a hypothesis.
* A **proof** is an argument that the conclusion follows from the hypotheses *in every case*. Standard strategies,
  several of which 02-680 uses: a **direct proof** (chain of implications), **proof by contrapositive** (show "not q ⇒ not
  p"), **proof by contradiction** (assume the opposite and derive nonsense), **proof by induction** (true for $n = 1$,
  and true for $n$ ⇒ true for $n + 1$), and **disproof by counterexample** (one case where the statement fails).

The crucial asymmetry: **one counterexample disproves a "for all" statement, but no number of examples proves it.**
The next cell shows a famous case where 40 examples in a row would fool you.

````python
# ── 4.6 Examples are not proofs: Euler's prime-generating polynomial n² + n + 41 ──
def is_prime(m: int) -> bool:
    """True if the integer m ≥ 2 has no divisor d with 2 ≤ d ≤ √m."""
    if m < 2:
        return False
    return all(m % d != 0 for d in range(2, math.isqrt(m) + 1))


values = [n * n + n + 41 for n in range(45)]
print("n² + n + 41 for n = 0 … 9:", values[:10])
print("prime for all n = 0 … 39?", all(is_prime(v) for v in values[:40]))
first_failure = next(n for n in range(45) if not is_prime(values[n]))
print(f"first failure: n = {first_failure}: {first_failure}² + {first_failure} + 41 = {values[first_failure]} "
      f"= 41 × {values[first_failure] // 41}")
print("the Gauss formula Σ i = n(n+1)/2 holds for n = 1 … 1000:",
      all(sum(range(1, m + 1)) == m * (m + 1) // 2 for m in range(1, 1001)), "— still not a proof!")
````

🔍 **What just happened?** The polynomial $n^2 + n + 41$ produces primes (41, 43, 47, 53, 61, …) for **all 40** values
$n = 0, 1, \dots, 39$ — and then fails at $n = 40$: $40^2 + 40 + 41 = 1681 = 41 \times 41$. Forty confirmations, then a
counterexample. The Gauss formula, by contrast, passes 1,000 checks *and* has a proof (the staircase picture of
Section 4.3; a proof by induction works too), so we know it holds for every $n$. Biology knows this asymmetry well: "all
swans are white" survived thousands of observations until black swans were found in Australia; a single knockout
mouse that develops normally refutes "gene X is essential for development". Code is a wonderful tool to *check*
mathematics and to *find* counterexamples, but it does not replace proofs.

### 📘 Lecture notes — how to read mathematical notation

**Indices.** $x_i$ is the $i$-th element of a list $x_1, \dots, x_n$; $x_{ij}$ is the entry in row $i$, column $j$
of a table (in this course: sample $i$, feature $j$ of an $N \times J$ data matrix). Mathematics counts from 1,
Python from 0: $x_{ij}$ is `X[i-1, j-1]`. Hats estimate ($\hat\theta$), bars average ($\bar x$).

**Sums and products.**

$$\sum_{i=1}^{n} x_i = x_1 + \dots + x_n, \qquad \prod_{i=1}^{n} x_i = x_1 \cdots x_n .$$

*In words:* run the index from 1 to $n$ and add (resp. multiply) the terms. Rules: $\sum_{i=1}^n c = nc$;
$\sum (a x_i + b y_i) = a\sum x_i + b\sum y_i$ (linearity); but $\sum x_i y_i \ne (\sum x_i)(\sum y_i)$ and
$\sum x_i^2 \ne (\sum x_i)^2$. $\log \prod_i p_i = \sum_i \log p_i$. *Worked examples:* $\sum_{i=1}^4 i^2 = 30$,
$\prod_{i=1}^4 i = 4! = 24$, $\sum_{i=1}^n (2i-1) = n^2$, $\sum_{i=1}^n i = n(n+1)/2$.

**Symbols.** $\in$ (element of), $\subseteq$ (subset of), $\cup\ \cap\ \setminus$ (union, intersection, difference),
$\emptyset$ (empty set), $\lvert S\rvert$ (size), $\{x \in S : P(x)\}$ (set-builder: "such that"); $\neg\ \land\ \lor$
(not, and, or), $\Rightarrow$ (implies), $\Leftrightarrow$ (iff), $\forall$ (for all), $\exists$ (there exists);
$\mathbb N \subset \mathbb Z \subset \mathbb Q \subset \mathbb R$; intervals $[a, b]$ (endpoints included),
$(a, b)$ (excluded); $f: S \to T$, $x \mapsto f(x)$ (domain $S$, codomain $T$, one output per input);
$(g \circ f)(x) = g(f(x))$.

**Reading mathematics.** Definition = a precise name (restate it; build an example and a non-example). Theorem =
"if hypotheses then conclusion" (find both; test on an example; ask what fails without each hypothesis). Proof
strategies: direct, contrapositive, contradiction, induction, counterexample. **Examples can refute a ∀-statement but
never prove it** ($n^2 + n + 41$ is prime for $n = 0, \dots, 39$ but $40^2 + 40 + 41 = 41^2$).

**Common confusions.** $2 \in \{1, 2\}$ but $\{2\} \subseteq \{1, 2\}$ (element vs subset). $p \Rightarrow q$ is not
causation and not the same as $q \Rightarrow p$. $x^2_i$ is a power of $x_i$, not an index. "Iff" is two statements
($\Rightarrow$ and $\Leftarrow$).

> 🎓 **Exam tip.** Read every formula aloud in words before you compute with it — the *In words:* lines model how.
> Most notation errors in exams come from mixing up ∈ and ⊆, from forgetting where an index starts, or from
> "simplifying" $\sum x_i^2$ to $(\sum x_i)^2$.

### 5.5 Derivations, one step per line

A derivation should show **every step on its own line**, with the equals signs aligned. `&` marks the alignment
point and `\\` ends a line. The Hardy–Weinberg identity, for allele frequencies $p$ and $q$ with $p + q = 1$:

```markdown
$$\begin{aligned}
(p + q)^2 &= (p + q)(p + q) \\
          &= p^2 + pq + qp + q^2 \\
          &= p^2 + 2pq + q^2 \\
          &= 1 \qquad \text{because } p + q = 1 .
\end{aligned}$$
```

renders as

$$\begin{aligned}
(p + q)^2 &= (p + q)(p + q) \\
          &= p^2 + pq + qp + q^2 \\
          &= p^2 + 2pq + q^2 \\
          &= 1 \qquad \text{because } p + q = 1 .
\end{aligned}$$

*In words:* expanding $(p+q)^2$ gives the three genotype frequencies $p^2$ (AA), $2pq$ (Aa) and $q^2$ (aa), and they
add up to 1 because $p + q = 1$.

### 5.6 Definitions by cases

```markdown
$$\mathbb{1}_A(x) = \begin{cases} 1 & \text{if } x \in A, \\ 0 & \text{otherwise.} \end{cases}$$
```

renders as

$$\mathbb{1}_A(x) = \begin{cases} 1 & \text{if } x \in A, \\ 0 & \text{otherwise.} \end{cases}$$

*In words:* the **indicator function** of a set $A$ answers the question "is $x$ in $A$?" with 1 (yes) or 0 (no).

### 6.1 Logic: when is a gene switched on? (Part B)

The *lac* operon of *E. coli* is the textbook example of gene regulation (Jacob & Monod, 1961). Simplified, the gene
*lacZ* (which makes the lactose-digesting enzyme β-galactosidase) is transcribed **if and only if lactose is present
and glucose is absent**:

$$\text{lacZ on} \iff \text{lactose} \land \neg\,\text{glucose} .$$

*In words:* the bacterium makes the enzyme exactly when there is lactose to digest and no glucose, its preferred food.
A **truth table** lists every combination of the inputs and the resulting output.

````python
# ── 6.1 A gene-regulation rule as a truth table (simplified lac operon) ──
lac_rule = {"lacZ on = lactose ∧ ¬glucose": lambda lactose, glucose: lactose and not glucose,
            "glucose ⇒ ¬(lacZ on)": lambda lactose, glucose: implies(glucose, not (lactose and not glucose))}
print(truth_table(lac_rule, ["lactose", "glucose"]))
````

🔍 **What just happened?** Two inputs, each true (T) or false (F), give $2^2 = 4$ combinations, one per row. The rule
switches *lacZ* on in **exactly one** row: lactose present, glucose absent. The last column checks a consequence of the
rule — "if glucose is present, then *lacZ* is off" — and it is true in **all four** rows: a statement that is true for
every combination of inputs is a **tautology**, and with only four cases we have *proved* it by checking them all. With 3 inputs the table has 8 rows, with 20 inputs
over a million; Boolean network models of whole regulatory circuits are built from rules like this one.
→ **Explained in [Module 02 · 🔣 Logic: the grammar of precise reasoning](02_logic.ipynb).**

### 6.3 Linear algebra: a data matrix and the distance between two samples (Part C)

Each sample is a list of 400 numbers — a **vector** $\mathbf x_i \in \mathbb R^{400}$, one row of the matrix $X$. How
different are two samples? The **Euclidean distance** generalises Pythagoras to 400 dimensions:

$$d(\mathbf x_a, \mathbf x_b) = \sqrt{\sum_{j=1}^{J} (x_{aj} - x_{bj})^2} .$$

*In words:* for every gene, take the difference between the two samples, square it, add the squares over all $J$
genes, and take the square root — the straight-line distance between two points in a 400-dimensional space.

````python
# ── 6.3 The data matrix and the Euclidean distance between samples ──
def euclidean_distance(a: np.ndarray, b: np.ndarray) -> float:
    """d(a, b) = sqrt(Σ_j (a_j − b_j)²), the straight-line distance between two expression profiles."""
    return float(np.sqrt(np.sum((a - b) ** 2)))


name_to_row = {name: i for i, name in enumerate(samples["sample"])}
for a, b in [("spleen_con1", "spleen_con2"), ("spleen_con1", "spleen_inf1"), ("spleen_con1", "liver_con1")]:
    print(f"d({a}, {b}) = {euclidean_distance(X[name_to_row[a]], X[name_to_row[b]]):6.2f}")
````

🔍 **What just happened?** Two control spleens are **11.38** apart — not zero, because every measurement is noisy
(Section 1.2). A control and an infected spleen are farther apart, **17.07**: infection moved 51 genes. A spleen and a
liver sample are farthest, **19.81**: different tissues express genes at different baseline levels. One number per
pair of samples already captures the structure of the experiment. Let us compute it for *all* $24 \times 24$ pairs.

````python
# ── 6.4 All pairwise distances as a heatmap ──
distance_matrix = np.array([[euclidean_distance(X[a], X[b]) for b in range(len(X))] for a in range(len(X))])
fig, ax = plt.subplots(figsize=(6.6, 5.6))
image = ax.imshow(distance_matrix, cmap="viridis")
ax.set_xticks(range(len(X)))
ax.set_yticks(range(len(X)))
ax.set_xticklabels(samples["sample"], rotation=90, fontsize=6.5)
ax.set_yticklabels(samples["sample"], fontsize=6.5)
fig.colorbar(image, ax=ax, label="Euclidean distance (log2 units)")
ax.set_title("Distances between the 24 samples (SIMULATED infection experiment)")
plt.tight_layout()
plt.show()
groups = samples["tissue"] + " " + samples["condition"]
same_group = groups.to_numpy()[:, None] == groups.to_numpy()[None, :]
same_tissue = samples["tissue"].to_numpy()[:, None] == samples["tissue"].to_numpy()[None, :]
off_diagonal = ~np.eye(len(X), dtype=bool)
print(f"mean distance: same tissue & condition {distance_matrix[same_group & off_diagonal].mean():.2f} | "
      f"same tissue, other condition {distance_matrix[same_tissue & ~same_group].mean():.2f} | "
      f"different tissue {distance_matrix[~same_tissue].mean():.2f}")
````

### 7.1 Calculus: a derivative is a growth rate (Part D)

In a rich medium, a bacterium such as *E. coli* can divide about every 20 minutes. Starting from $N_0 = 100$ cells,
the population after $t$ minutes is

$$N(t) = N_0 \cdot 2^{t/20} = N_0\, e^{rt}, \qquad r = \frac{\ln 2}{20} \approx 0.0347 \text{ per minute}.$$

*In words:* the population doubles every 20 minutes; written with the exponential function, the **growth constant**
$r$ is $\ln 2$ divided by the doubling time. How fast is the population growing at $t = 60$ minutes — how many new
cells per minute? The **average** rate over a short interval $h$ is the difference quotient

$$\frac{N(t + h) - N(t)}{h},$$

*In words:* the number of cells added between $t$ and $t + h$, divided by the length of the interval. As $h$ shrinks
towards 0, this approaches the **derivative** $\frac{dN}{dt}$, the *instantaneous* growth rate. For exponential growth
calculus gives the exact answer $\frac{dN}{dt} = rN$ — the growth rate is proportional to the population.

````python
# ── 7.1 The growth rate as a limit of difference quotients (illustrative E. coli numbers) ──
N0, doubling_time = 100, 20.0                       # cells, minutes
r = math.log(2) / doubling_time                     # growth constant per minute


def population(t: float) -> float:
    """N(t) = N0 · 2^(t / doubling time): exponential growth."""
    return N0 * 2 ** (t / doubling_time)


t0 = 60.0
print(f"N({t0:.0f} min) = {population(t0):.0f} cells")
for h in [30, 10, 1, 0.1, 0.001]:
    print(f"  h = {h:>6}: [N(t0 + h) − N(t0)] / h = {(population(t0 + h) - population(t0)) / h:8.4f} cells/min")
print(f"exact derivative r·N(t0) = {r * population(t0):.4f} cells/min")
````

🔍 **What just happened?** At $t = 60$ min there are $100 \cdot 2^3 =$ **800** cells. Over the next 30 minutes the
population gains on average **48.76** cells per minute, over 10 minutes 33.14, over 1 minute 28.21, and over a
thousandth of a minute **27.73** — the difference quotients settle on the exact derivative $rN = 0.0347 \times 800 =$
**27.73 cells per minute**. Long intervals overestimate the rate because the curve bends upwards. The picture below
shows why, and adds the more realistic **logistic** growth, in which the food runs out.

````python
# ── 7.2 Secants approaching the tangent; logistic growth and its growth rate ──
times = np.linspace(0, 120, 300)
fig, (ax_exp, ax_log) = plt.subplots(1, 2, figsize=(11, 3.9))
ax_exp.plot(times, population(times), color="tab:blue", label="N(t) = 100 · 2^(t/20)")
for h, colour in [(30, "tab:orange"), (10, "tab:green")]:
    slope = (population(t0 + h) - population(t0)) / h
    ax_exp.plot([t0 - 15, t0 + h + 5], population(t0) + slope * (np.array([t0 - 15, t0 + h + 5]) - t0), "--",
                color=colour, label=f"secant, h = {h} min: slope {slope:.1f}")
ax_exp.plot([t0 - 40, t0 + 40], population(t0) + r * population(t0) * np.array([-40, 40]), color="tab:red",
            label=f"tangent: slope dN/dt = {r * population(t0):.1f}")
ax_exp.scatter([t0], [population(t0)], color="k", zorder=3)
ax_exp.set_ylim(0, 3000)
ax_exp.set_xlabel("time t (min)")
ax_exp.set_ylabel("cells N(t)")
ax_exp.set_title("Exponential growth: secants → tangent at t = 60 min")
ax_exp.legend(fontsize=7)

K = 1e5                                              # carrying capacity: the medium supports at most K cells
long_times = np.linspace(0, 450, 500)
logistic = K / (1 + (K - N0) / N0 * np.exp(-r * long_times))
growth_rate = r * logistic * (1 - logistic / K)      # dN/dt = rN(1 − N/K)
t_fastest = long_times[np.argmax(growth_rate)]
ax_log.plot(long_times, logistic, color="tab:blue", label="N(t), logistic")
ax_log.set_xlabel("time t (min)")
ax_log.set_ylabel("cells N(t)", color="tab:blue")
rate_axis = ax_log.twinx()
rate_axis.plot(long_times, growth_rate, color="tab:red", label="dN/dt")
rate_axis.set_ylabel("growth rate dN/dt (cells/min)", color="tab:red")
rate_axis.spines["right"].set_visible(True)
ax_log.axvline(t_fastest, color="0.5", ls=":")
ax_log.set_title(f"Logistic growth (K = {K:,.0f}): fastest at t ≈ {t_fastest:.0f} min, N = K/2")
plt.tight_layout()
plt.show()
print(f"fastest growth at t ≈ {t_fastest:.0f} min, where N ≈ {logistic[np.argmax(growth_rate)]:,.0f} "
      f"and dN/dt ≈ {growth_rate.max():.0f} cells/min (theory: rK/4 = {r * K / 4:.0f})")
````

🔍 **What just happened?** Left: the secant lines through $t = 60$ and $t = 60 + h$ have slopes 48.8 ($h = 30$) and
33.1 ($h = 10$); as $h$ shrinks they turn into the **tangent** line with slope **27.7** — the derivative is the slope of
the tangent. Right: real cultures cannot grow forever. In the **logistic** model $\frac{dN}{dt} = rN(1 - N/K)$ growth
slows as the population approaches the carrying capacity $K$, so the growth rate (red) rises, peaks and falls. It is
largest at **t ≈ 199 min**, when the population is at half capacity ($N \approx$ 50,000), with about **866 cells per
minute** — exactly the value $rK/4$ that calculus predicts by setting the derivative of the growth rate to zero.
→ **Explained in [Module 14 · 📈 Calculus: derivatives, integrals and rates of change](14_calculus_derivatives_integrals.ipynb)** (derivatives, differential equations, optimisation) and [Module 15](15_multivariable_calculus_dynamics.ipynb) (dynamics).

### 7.2 Probability: what does a positive test mean? (Part E)

A rapid test for an infection detects **95 %** of infected people (its *sensitivity*) and correctly gives a negative
result for **90 %** of healthy people (its *specificity*). In the population being tested, **1 %** are infected (the
*prevalence*). Your test is positive. How likely is it that you are infected? **Bayes' rule** answers:

$$P(\text{infected} \mid +) = \frac{P(+ \mid \text{infected})\, P(\text{infected})}
{P(+ \mid \text{infected})\, P(\text{infected}) + P(+ \mid \text{healthy})\, P(\text{healthy})} .$$

*In words:* among all the ways to get a positive test — being infected and detected, or being healthy and wrongly
flagged — what fraction comes from truly infected people? The vertical bar "∣" reads "given".

````python
# ── 7.3 Bayes' rule for a diagnostic test (illustrative numbers) ──
sensitivity, specificity, prevalence = 0.95, 0.90, 0.01
p_positive = sensitivity * prevalence + (1 - specificity) * (1 - prevalence)      # all ways to test positive
p_infected_given_positive = sensitivity * prevalence / p_positive
print(f"P(+) = {p_positive:.4f};  P(infected | +) = {p_infected_given_positive:.4f}")
population_size = 10_000                                                          # the same in natural frequencies
infected = prevalence * population_size
true_pos, false_pos = sensitivity * infected, (1 - specificity) * (population_size - infected)
print(f"of {population_size:,} people: {infected:.0f} infected → {true_pos:.0f} test positive; "
      f"{population_size - infected:,.0f} healthy → {false_pos:.0f} test positive; "
      f"{true_pos:.0f} / {true_pos + false_pos:.0f} = {true_pos / (true_pos + false_pos):.4f}")
````

🔍 **What just happened?** Only **8.76 %** of positive results come from infected people — although the test catches
95 % of infections! The counts explain the paradox: of 10,000 people, 100 are infected and 95 of them test positive;
but 9,900 are healthy, and 10 % of them — **990** — also test positive. The 95 true positives drown among 1,085
positives in total. This **base-rate effect** is one of the most important ideas in medicine and screening, and it is
pure probability. The next cell shows how the answer depends on the prevalence.

````python
# ── 7.4 The answer depends on the prior: P(infected | +) as a function of prevalence ──
prevalences = np.logspace(-4, np.log10(0.5), 200)
fig, ax = plt.subplots(figsize=(6.6, 3.6))
for spec, colour in [(0.90, "tab:red"), (0.99, "tab:blue")]:
    posterior = sensitivity * prevalences / (sensitivity * prevalences + (1 - spec) * (1 - prevalences))
    ax.plot(prevalences, posterior, color=colour, label=f"sensitivity 0.95, specificity {spec}")
    print(f"specificity {spec}: P(infected | +) at prevalence 1 % = "
          f"{sensitivity * 0.01 / (sensitivity * 0.01 + (1 - spec) * 0.99):.3f}, at 20 % = "
          f"{sensitivity * 0.2 / (sensitivity * 0.2 + (1 - spec) * 0.8):.3f}")
ax.axvline(0.01, color="0.5", ls=":")
ax.set_xscale("log")
ax.set_xlabel("prevalence P(infected) before the test (log scale)")
ax.set_ylabel("P(infected | positive test)")
ax.set_title("Bayes' rule: the same test means different things in different populations")
ax.legend()
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** The probability of infection after a positive test (the **posterior**) depends strongly
on the probability before the test (the **prior**, here the prevalence). With specificity 0.90 (red) it is **0.088** at
1 % prevalence but **0.704** at 20 % (say, among patients with symptoms). A more specific test (blue, specificity 0.99)
helps enormously at low prevalence: **0.490** at 1 %, 0.960 at 20 %. The same positive test means different things in
a screening programme and in a clinic. → **Explained in [Module 17 · 🔀 Conditional probability, independence and Bayes' rule](17_conditional_probability_bayes.ipynb);** the same reasoning about parameters becomes
Bayesian estimation in [Module 24](24_bayesian_estimation_map.ipynb).

### 7.3 Statistics: is IL6 really higher after infection? (Parts F–G)

In Section 1.2 the four infected spleens had more IL6 than the four controls (means 6.69 vs 4.40). But replicates
are noisy. A **two-sample t-test** asks: if infection had *no* effect, how surprising would a difference this large
be? Its answer is a **p-value**: the probability, assuming no effect, of a difference at least as extreme as the one
observed.

````python
# ── 7.5 A two-group comparison: Welch's t-test for IL6 and ACTB in the spleen (SIMULATED data) ──
fig, axes = plt.subplots(1, 2, figsize=(8.4, 3.2))
for ax, gene in zip(axes, ["IL6", "ACTB"]):
    j = genes.index(gene)
    ctrl, inf = X[spleen_control.to_numpy(), j], X[spleen_infected.to_numpy(), j]
    test = stats.ttest_ind(inf, ctrl, equal_var=False)                  # Welch's two-sample t-test
    true_change = expr["true_lfc"].loc["spleen", gene]
    print(f"{gene:4s}: difference of means {inf.mean() - ctrl.mean():+.3f}, t = {test.statistic:6.2f}, "
          f"p = {test.pvalue:.4f}  | TRUE change (simulation) {true_change:+.3f}")
    ax.scatter(np.zeros(4) + rng.uniform(-0.05, 0.05, 4), ctrl, color="tab:blue", label="control")
    ax.scatter(np.ones(4) + rng.uniform(-0.05, 0.05, 4), inf, color="tab:red", label="infected")
    ax.hlines([ctrl.mean(), inf.mean()], [-0.2, 0.8], [0.2, 1.2], color="k")
    ax.set_xticks([0, 1], ["control", "infected"])
    ax.set_xlim(-0.5, 1.5)
    centre = (ctrl.mean() + inf.mean()) / 2
    ax.set_ylim(centre - 2, centre + 2)                                    # both panels span 4 log2 units
    ax.set_ylabel("log2 expression")
    ax.set_title(f"{gene} in the spleen: p = {test.pvalue:.3g}")
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** For **IL6** the difference of means is **+2.296** log2 units, the t statistic is **7.03**
and the p-value is **0.0008**: if infection had no effect, a difference this large would occur less than once in a
thousand experiments, so we conclude IL6 responds — and the simulation's truth (+2.349) confirms it. For **ACTB** the
difference is only +0.023, $t = 0.48$, $p = 0.65$: entirely compatible with noise, and indeed the true change is 0.
The black bars are the group means; the dots are the four replicates; both panels span the same 4 log2 units,
so the flatness of ACTB is visible at a glance. But beware: we have 400 genes, and testing each
at the 5 % level would flag about 5 % of the 349 spleen genes that do *not* respond — roughly 17 genes — *by chance
alone*: the **multiple-testing** problem.
→ **Explained in [Module 21](21_expectation_variance_moments.ipynb)–[Module 22](22_sampling_lln_clt_confidence.ipynb)** (means, variances, sampling), **[Module 25 · 🚦 Hypothesis testing: errors, power, z- and t-tests](25_hypothesis_testing.ipynb)** (t-tests) and [Module 26](26_categorical_tests_multiple_testing.ipynb) (testing
thousands of genes).

## 📌 Key takeaways

- Computational biology **models** processes, **analyses** large-scale data and **makes predictions**; it rests on
  **logic** (rules), **linear algebra** (data matrices), **calculus** (change, optimisation) and **probability and
  statistics** (uncertainty).
- Biological data are **huge** (10⁴–10¹⁰ numbers), usually **wide** (more genes than samples) and **noisy**
  (replicates differ), so every conclusion needs mathematics and an estimate of uncertainty.
- 02-680 aims at an overview of mathematical techniques (including proofs and statistical tests), a quantitative
  foundation for machine learning and automation, and preparation for 02-620, 02-710 and 02-750. Its six blocks —
  fundamentals, linear algebra, calculus, probability, statistics, applications — map onto Parts B–G of this course.
- The topics are **cumulative**: logic → sets → tuples → vectors → matrices → … → regression, with long-range links
  (sets → events, eigenvalues → stability and Markov chains, derivatives → maximum likelihood).
- Study actively: run, **redo every calculation by hand**, predict before running, change one thing at a time,
  test yourself, write a lab note; revise from the 📘 boxes.
- Notation is a language: $x_{ij}$ (row $i$, column $j$; `X[i-1, j-1]` in Python), $\sum$ and $\prod$ (loops),
  ∈ ⊆ ∪ ∩ (sets), ∀ ∃ ⇒ ⇔ (logic; `all`, `any`), $f: S \to T$ (functions). **Examples can disprove a ∀-statement,
  never prove it.**
- LaTeX writes it: `$…$`, `$$…$$`, `\frac`, `\sum_{i=1}^{n}`, `bmatrix`, `aligned`, `cases`.
- The guided tour previewed the whole course: a truth table, a Venn diagram and inclusion–exclusion, distances
