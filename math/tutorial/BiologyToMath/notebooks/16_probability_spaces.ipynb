## 🧫 The biological question

In 1865 Gregor Mendel crossed pea plants that were heterozygous for two traits — seed colour (yellow $Y$ dominant
over green $y$) and seed shape (round $R$ dominant over wrinkled $r$). He could not predict the seeds of any single
pod, yet over hundreds of seeds the four phenotypes appeared in the ratio **9 : 3 : 3 : 1**. Why exactly these
numbers? And what does it even *mean* to say that "a seed is green and wrinkled with probability 1/16"?

The same questions run through modern biology. Which of the 3 billion bases of a genome will mutate in the next
generation? Will a new mutation fall into a gene? Why are most genetic variants in a population rare? Will a cell
express a marker gene at all? None of these has a certain answer — but each has a **probability**, and probabilities
obey a small set of rules from which everything else in this course (conditional probability, random variables,
estimation, hypothesis tests, regression) is built.

### 🧰 Course toolkit

The next cell copies the shared helpers this notebook needs from [`course_toolkit.py`](../course_toolkit.py). `draw_venn` draws Venn diagrams of events (Sections 6 and 7), `simulate_dna` generates a random DNA sequence (Section 1) and `simulate_genotypes` draws genotypes of a population in Hardy–Weinberg proportions (Section 9). Just run it.

````python
# ── Course toolkit (identical to ../course_toolkit.py v1.0; copied so this notebook runs alone) ──
import itertools

import matplotlib.pyplot as plt
import numpy as np


def _venn_order(x):
    """Sort key for Venn-diagram labels: numbers in numeric order first, then everything else as text."""
    if isinstance(x, (int, float, np.integer, np.floating)) and not isinstance(x, bool):
        return (0, float(x), "")
    return (1, 0.0, str(x))


def draw_venn(sets: dict, universe=None, ax=None, show_elements: bool = True, max_items: int = 6,
              title: str | None = None, highlight=None):
    """Draw a Venn diagram of two or three sets with matplotlib only.

    ``sets`` maps a label to a Python set, e.g. ``{"S": {1, 2, 3}, "T": {3, 4}}``.  Each region
    shows its elements (or ``n = …`` if it has more than ``max_items``).  If ``universe`` is given,
    a rectangle is drawn and the elements outside every set are listed in a corner.
    ``highlight`` is a function of the membership tuple, e.g. ``lambda s, t: s and not t`` for
    S \\ T; the regions where it is true are shaded.  Returns the Axes.
    """
    labels = list(sets)
    if len(labels) not in (2, 3):
        raise ValueError("draw_venn draws two or three sets")
    if ax is None:
        _, ax = plt.subplots(figsize=(4.6, 3.8))
    if len(labels) == 2:
        centres = [(-0.55, 0.0), (0.55, 0.0)]
    else:
        centres = [(-0.55, 0.32), (0.55, 0.32), (0.0, -0.6)]
    radius = 1.0
    colours = ["tab:blue", "tab:orange", "tab:green"]
    # classify a fine grid of points to find each region (used for shading and label positions)
    gx, gy = np.meshgrid(np.linspace(-2.2, 2.2, 441), np.linspace(-2.0, 1.9, 391))
    inside = [((gx - cx) ** 2 + (gy - cy) ** 2) <= radius ** 2 for cx, cy in centres]
    if highlight is not None:
        mask = np.zeros_like(gx, dtype=bool)
        for member in itertools.product([True, False], repeat=len(labels)):
            if any(member) and highlight(*member):
                region = np.ones_like(gx, dtype=bool)
                for k, m in enumerate(member):
                    region &= inside[k] if m else ~inside[k]
                mask |= region
        ax.contourf(gx, gy, mask.astype(float), levels=[0.5, 1.5], colors=["#ffd54f"], alpha=0.8)
    for (cx, cy), lab, col in zip(centres, labels, colours):
        ax.add_patch(plt.Circle((cx, cy), radius, fill=False, lw=2, color=col))
        lx = cx + (-1.0 if cx < 0 else 1.0 if cx > 0 else 0.0) * 0.95
        ly = cy + (0.95 if cy >= 0 else -1.15)
        ax.text(lx, ly, lab, ha="center", va="center", fontsize=11, fontweight="bold", color=col)
    elements = set().union(*sets.values())
    for member in itertools.product([True, False], repeat=len(labels)):
        if not any(member):
            continue
        region_items = [x for x in elements if all((x in sets[l]) == m for l, m in zip(labels, member))]
        region = np.ones_like(gx, dtype=bool)
        for k, m in enumerate(member):
            region &= inside[k] if m else ~inside[k]
        if not region.any():
            continue
        px, py = gx[region].mean(), gy[region].mean()
        listing = ", ".join(str(x) for x in sorted(region_items, key=_venn_order))
        if show_elements and len(region_items) <= max_items and len(listing) <= 16:
            text = listing if region_items else "∅"
        else:
            text = f"n = {len(region_items)}"
        ax.text(px, py, text, ha="center", va="center", fontsize=8, wrap=True)
    if universe is not None:
        ax.add_patch(plt.Rectangle((-2.1, -1.9), 4.2, 3.7, fill=False, lw=1, color="0.4"))
        outside = sorted(set(universe) - elements, key=_venn_order)
        text = ", ".join(map(str, outside)) if len(outside) <= max_items else f"n = {len(outside)}"
        ax.text(-2.0, -1.8, "U: " + (text if outside else "∅"), fontsize=8, va="bottom", color="0.3")
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.0, 1.9)
    ax.set_aspect("equal")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax


NUCLEOTIDES = ("A", "C", "G", "T")


def simulate_dna(length: int, transitions=None, initial=None, seed: int = 0) -> str:
    """Random DNA string.  Without ``transitions`` the bases are i.i.d. uniform; otherwise they
    follow a first-order Markov chain with the given 4 × 4 transition matrix (order A, C, G, T)
    and initial distribution ``initial`` (default uniform)."""
    rng = np.random.default_rng(seed)
    initial = np.full(4, 0.25) if initial is None else np.asarray(initial, dtype=float)
    if transitions is None:
        return "".join(rng.choice(list(NUCLEOTIDES), size=length, p=initial))
    P = np.asarray(transitions, dtype=float)
    states = [rng.choice(4, p=initial)]
    for _ in range(length - 1):
        states.append(rng.choice(4, p=P[states[-1]]))
    return "".join(NUCLEOTIDES[s] for s in states)


def simulate_genotypes(n: int, p: float, inbreeding: float = 0.0, seed: int = 1866) -> np.ndarray:
    """SIMULATED genotypes of ``n`` diploid individuals at one SNP, coded 0/1/2 = number of copies of
    the allele with frequency ``p``.  With ``inbreeding`` F = 0 the genotype frequencies follow
    Hardy–Weinberg proportions ((1−p)², 2p(1−p), p²); F > 0 moves individuals from heterozygotes to
    homozygotes: P(1) = 2p(1−p)(1−F)."""
    rng = np.random.default_rng(seed)
    q = 1.0 - p
    probs = np.array([q * q + inbreeding * p * q, 2 * p * q * (1 - inbreeding), p * p + inbreeding * p * q])
    return rng.choice(3, size=n, p=probs / probs.sum())

print("🧰 toolkit loaded:", ", ".join(sorted(['NUCLEOTIDES', '_venn_order', 'draw_venn', 'simulate_dna', 'simulate_genotypes'])))
````

````python
# ── 0.1 Two small helpers used throughout the notebook ──
def classical_probability(event: set, sample_space: set) -> Fraction:
    """P(A) = |A| / |Ω| — valid ONLY when all outcomes in Ω are equally likely."""
    return Fraction(len(event & sample_space), len(sample_space))


def relative_frequency(event: set, outcomes: list) -> float:
    """Fraction of simulated outcomes that fall in the event (an estimate of P(event))."""
    return float(np.mean([o in event for o in outcomes]))


Omega_coin = {"H", "T"}
print("P({H}) by counting:", classical_probability({"H"}, Omega_coin))
rng = np.random.default_rng(SEED)
tosses = list(rng.choice(sorted(Omega_coin), size=10))
print("10 simulated tosses:", "".join(tosses), "→ relative frequency of H =", relative_frequency({"H"}, tosses))
````

🔍 **What just happened?** `classical_probability` divides the number of outcomes in an event by the number of
outcomes in the sample space and keeps the answer as an exact fraction (`Fraction(1, 2)` prints as `1/2`).
`relative_frequency` does what an experimentalist does: it counts how often the event happened among the observed
outcomes. Ten simulated tosses gave 6 heads, a relative frequency of 0.6 rather than 0.5 — a first hint that
frequencies from small samples wobble around the true probability. The two helpers embody the two ways of thinking
about probability that Section 2 compares.

### 1.1 Three sources of uncertainty in biology

Biological measurements are uncertain **across samples** (two patients with the same diagnosis have different
genomes and different blood chemistry), **across genes** (in an RNA-seq table of a cancer cohort — rows = genes,
columns = patients — every gene has its own level and its own variability) and **across cells** (single-cell RNA-seq
of a mouse brain region profiles thousands of cells, and even cells of one type differ). The geneticist and
statistician **R. A. Fisher** (1890–1962) built much of modern statistics precisely to reason about such variation in
inherited traits.

### 1.2 A genome is a string — where are the genes?

In 2001 the first draft of the human genome sequence — about 3 billion nucleotides — became available, and the
UCSC Genome Browser (built by David Haussler's group, with Jim Kent) made it explorable. But a genome is just a long
string over $\{A, C, G, T\}$ (Module 04). Nothing in the letters says *"a gene starts here"*.

````python
# ── 1.2 A SIMULATED stretch of DNA, and the hidden gene structure behind it ──
dna = simulate_dna(180, seed=2001)            # i.i.d. uniform bases: a stand-in for real sequence
for start in range(0, 180, 60):
    print(dna[start:start + 60].lower())

# an invented gene model for the same 180 positions (what a gene finder would have to discover)
segments = [(0, 40, "intergenic"), (40, 70, "exon"), (70, 95, "intron"), (95, 120, "exon"),
            (120, 140, "intergenic"), (140, 168, "exon"), (168, 180, "intergenic")]
colours = {"intergenic": "tab:blue", "exon": "tab:red", "intron": "goldenrod"}
fig, (ax_top, ax_seq) = plt.subplots(2, 1, figsize=(10, 3.4), gridspec_kw={"height_ratios": [1, 1.6]})
for a, b, kind in segments:
    height = 0.5 if kind == "exon" else (0.2 if kind == "intron" else 0.08)
    ax_top.add_patch(plt.Rectangle((a, -height / 2), b - a, height, color=colours[kind]))
ax_top.text(55, 0.45, "gene 1 (2 exons, 1 intron)", ha="left", fontsize=8)
ax_top.text(140, 0.45, "gene 2", ha="left", fontsize=8)
ax_top.set_xlim(0, 180); ax_top.set_ylim(-0.5, 0.7); ax_top.axis("off")
ax_top.set_title("the gene model (hidden) — blue: intergenic, red: exon (coding), yellow: intron")
for i, base in enumerate(dna):
    row, col = divmod(i, 60)
    kind = next(k for a, b, k in segments if a <= i < b)
    ax_seq.text(col, -row, base.lower(), color=colours[kind], family="monospace", fontsize=9, ha="center")
ax_seq.set_xlim(-1, 60); ax_seq.set_ylim(-2.6, 0.6); ax_seq.axis("off")
ax_seq.set_title("the same 180 letters coloured by the hidden labels (SIMULATED)")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The three printed lines are all a sequencing machine gives you. The figure colours the very
same letters by a *hidden* annotation: intergenic DNA (blue), two exons and an intron of gene 1, one exon of gene 2.
In this simulation the letters were drawn independently of the labels, so no method could recover them; in real
genomes exons, introns and intergenic DNA differ *statistically* — in base composition, in codon structure, in the
signals at their borders (Module 17 shows one such signal, the stop codons).

Gene finding was solved by **borrowing algorithms from speech recognition**. A recording of the sentence *"I love the
Santa Cruz beach."* is just a wiggly sound wave; the words are hidden behind it. Speech recognisers model the wave as
being *emitted* by a sequence of hidden states (the words, the sounds) with probabilities attached — a **hidden Markov
model (HMM)**. Haussler's group and others realised that a genome is the same kind of object: the hidden states are
"exon", "intron", "intergenic", and each emits bases with its own probabilities. Every modern gene finder rests on that
probabilistic idea. You will meet Markov chains in [Module 20 · ⛓️ Graphical models: Bayesian networks and Markov chains](20_bayesian_networks_markov_chains.ipynb).

### 1.3 Genomes in a population: mutation and recombination are random events

Within a population, genomes differ because of two **probabilistic events**:

* **Mutation** — at each generation, each base has a tiny probability of changing. A mutation can change a phenotype
  (disease susceptibility, a gene's expression level).
* **Recombination** — when a parent makes an egg or sperm, the two copies of each chromosome exchange segments at
  random positions (crossovers). Over many generations recombination **breaks long ancestral haplotypes** (the
  combination of alleles carried together on one chromosome) into a mosaic. Inferring haplotypes from genotypes
  ("phasing") and locating recombination sites are probabilistic inference problems; the International HapMap Project
  mapped haplotype structure and recombination hot spots across the human genome (for example a sharp hot spot next to
  the β-globin gene *HBB*). A recombination event can itself act as a disease-causing mutation (for example when it
  deletes or duplicates a gene).

The next cell simulates the mosaic in a deliberately simple model.

````python
# ── 1.3 Recombination breaks ancestral chromosomes into a mosaic (SIMULATED, simple model) ──
def mosaic_chromosome(generations: int, n_founders: int, rng) -> tuple:
    """One modern chromosome of length 1 Morgan, `generations` generations after a founder population.

    Crossovers along its line of ancestors fall at random positions, about one per Morgan per generation, so we
    draw their number from a Poisson distribution with mean `generations` (Module 18) and their positions uniformly.
    Each segment between crossovers is copied from a random founder chromosome.  Returns (segment edges, founders)."""
    n_crossovers = rng.poisson(generations)
    edges = np.r_[0.0, np.sort(rng.uniform(0.0, 1.0, n_crossovers)), 1.0]
    founders = rng.integers(0, n_founders, len(edges) - 1)
    return edges, founders


rng = np.random.default_rng(SEED + 13)
palette = plt.cm.tab10(np.arange(8))
fig, axes = plt.subplots(1, 3, figsize=(11, 3.0), sharey=True)
for ax, g in zip(axes, [1, 10, 50]):
    for row in range(6):
        edges, founders = mosaic_chromosome(g, 8, rng)
        for a, b, f in zip(edges[:-1], edges[1:], founders):
            ax.add_patch(plt.Rectangle((a, row), b - a, 0.7, color=palette[f]))
    ax.set_xlim(0, 1); ax.set_ylim(-0.3, 6); ax.set_yticks([])
    ax.set_xlabel("position along the chromosome (Morgans)")
    ax.set_title(f"{g} generation{'s' if g > 1 else ''} after the founders")
axes[0].set_ylabel("6 modern chromosomes")
fig.suptitle("each colour = one of 8 founder chromosomes (SIMULATED)", fontsize=10)
plt.tight_layout(); plt.show()

for g in [1, 10, 50]:
    n_segments = [len(mosaic_chromosome(g, 8, rng)[1]) for _ in range(2000)]
    print(f"{g:2d} generations: on average {np.mean(n_segments):5.2f} segments per chromosome "
          f"(mean segment length ≈ {100 / np.mean(n_segments):5.1f} cM)")
````

🔍 **What just happened?** One generation after the founders a chromosome is made of only a few long pieces (two on average); after 10
generations about 11 pieces (each ≈ 9 cM long); after 50 generations about 51 pieces of only ≈ 2 cM. The *number* of
crossovers is itself random (the printed averages are close to $1 + g$, as the model predicts), and so are their
positions. That is why two people who share a stretch of chromosome from a common ancestor share *short* stretches if
the ancestor lived long ago — the principle behind haplotype phasing, identity-by-descent and genetic association
mapping. (Real chromosomes have recombination *hot spots*, so crossovers are not uniform; the model ignores that.)

### 1.4 Allele frequencies: most variants are rare

The probability of seeing a variant depends on its **allele frequency** in the population, and environmental effects
add further uncertainty to how a genotype becomes a phenotype — the reason why "complex" diseases (cancer, diabetes,
arthritis, heart disease) are influenced by many variants, each changing risk only a little. Large sequencing projects
such as the 1000 Genomes Project found that **SNPs (single-nucleotide polymorphisms) are more frequent than any other
kind of polymorphism** (insertions, deletions, larger structural variants), and that most variants are **rare**. The
distribution of variant frequencies in a sample is called the **site-frequency spectrum (SFS)**. Population-genetics
theory predicts its shape under the simplest ("standard neutral") model: in a sample of $n$ chromosomes, the expected
number of variant sites carried by exactly $i$ chromosomes is proportional to $1/i$. The next cell simulates that model
(a coalescent simulation — the genealogy of the sample is traced back in time and mutations are dropped on its
branches; you do not need the details).

````python
# ── 1.4 The site-frequency spectrum under the standard neutral model (SIMULATED coalescent) ──
def simulate_sfs(n_chrom: int, theta: float, n_regions: int, rng) -> np.ndarray:
    """Counts of variant sites carried by i = 1 … n_chrom − 1 of n_chrom sampled chromosomes.

    Kingman coalescent with infinite-sites mutation: while k lineages remain, wait an exponential time with rate
    k(k−1)/2, drop Poisson(theta/2 × time) mutations on every lineage (each is inherited by all sampled chromosomes
    below that lineage), then merge two random lineages.  Repeated for `n_regions` independent regions."""
    sfs = np.zeros(n_chrom - 1)
    for _ in range(n_regions):
        below = np.ones(n_chrom, dtype=int)                    # sampled chromosomes below each lineage
        while len(below) > 1:
            k = len(below)
            t = rng.exponential(2.0 / (k * (k - 1)))
            np.add.at(sfs, below - 1, rng.poisson(theta / 2 * t, size=k))
            i, j = rng.choice(k, size=2, replace=False)
            below = np.r_[np.delete(below, [i, j]), below[i] + below[j]]
    return sfs


rng = np.random.default_rng(SEED + 14)
n_chrom, theta, n_regions = 20, 5.0, 3000
sfs = simulate_sfs(n_chrom, theta, n_regions, rng)
i = np.arange(1, n_chrom)
expected = n_regions * theta / i                             # neutral expectation θ/i per region
fig, ax = plt.subplots(figsize=(7, 3.3))
ax.bar(i, sfs / sfs.sum(), color="tab:blue", alpha=0.7, label="simulated (3,000 regions)")
ax.plot(i, expected / expected.sum(), "o-", color="tab:red", ms=4, label="theory: proportional to 1/i")
ax.set_xticks(i); ax.set_xlabel("number of the 20 sampled chromosomes that carry the variant")
ax.set_ylabel("fraction of variant sites"); ax.set_title("site-frequency spectrum: rare variants dominate")
ax.legend(); plt.tight_layout(); plt.show()
print(f"variant sites: {int(sfs.sum()):,};  singletons (i = 1): {sfs[0] / sfs.sum():.3f}  (theory {1 / np.sum(1 / i):.3f})")
print(f"carried by ≤ 2 of 20 chromosomes: {sfs[:2].sum() / sfs.sum():.3f};  by ≥ 10: {sfs[9:].sum() / sfs.sum():.3f}")
````

🔍 **What just happened?** The bars (simulation) follow the red curve (theory, $\propto 1/i$): **27.9 %** of the 53,628
variant sites are *singletons* (theory: 28.2 %) — carried by just one of the 20 sampled chromosomes — and more than 40 % are carried by
at most two, while only about a fifth (0.21) are carried by 10 or more of the 20 chromosomes. Real human samples have even more
rare variants than this model, because human populations grew explosively in the last few thousand years. For
medicine this means that most of the variants in any one patient's genome are rare, and their effects must be judged
probabilistically.

## 2 · What does a probability mean? Frequentist and Bayesian views

> 📍 **Where we are.** Section 1 argued that biology needs probability. Before writing any rule we must agree what a
> statement such as "$P(\text{heads}) = 1/2$" *means*. There are two classical answers, and both are used in this
> course.

The goal of a **probability space** is a mathematical framework that describes **random outcomes** and allows
automated reasoning **under uncertainty**. The coin toss is the standard example: a *single* toss is unpredictable,
but *many* tosses show a pattern in their average result. Mathematics describes and analyses that pattern.

**Frequentist interpretation.** The probability of an event is its **long-run frequency** over repeated trials,
defined in the limit as the number of trials goes to infinity:

$$P(A) = \lim_{n \to \infty} \frac{n_A}{n},$$

*In words:* repeat the experiment $n$ times, count the number $n_A$ of repetitions in which $A$ happened, and look at
the fraction $n_A / n$ as $n$ grows without bound. "The probability of heads is 0.5" means that in a large number of
tosses about half will be heads. Frequentists see this as an *objective* property of the coin.

🧮 **Math refresher — limits.** $\lim_{n\to\infty} a_n = L$ means that the numbers $a_n$ get (and stay) as close to
$L$ as you like once $n$ is large enough ([Module 14 · 📈 Calculus: derivatives, integrals and rates of change](14_calculus_derivatives_integrals.ipynb)). A frequency $n_A/n$ never *equals* the probability for sure; it
settles down near it.

````python
# ── 2.1 Long-run frequency of heads in three SIMULATED sequences of 5,000 tosses ──
rng = np.random.default_rng(SEED + 21)
n_tosses = 5000
fig, ax = plt.subplots(figsize=(7, 3.3))
for run, colour in zip(range(3), ["tab:blue", "tab:orange", "tab:green"]):
    heads = rng.random(n_tosses) < 0.5                       # True = heads, each with probability 1/2
    running = np.cumsum(heads) / np.arange(1, n_tosses + 1)   # n_A / n after every toss
    ax.plot(np.arange(1, n_tosses + 1), running, color=colour, lw=1, label=f"run {run + 1}")
    print(f"run {run + 1}: after 10 tosses {running[9]:.2f}, after 100 {running[99]:.3f}, "
          f"after 5,000 {running[-1]:.4f}")
ax.axhline(0.5, color="k", ls="--", lw=1)
ax.set_xscale("log"); ax.set_xlabel("number of tosses n (log scale)"); ax.set_ylabel("fraction of heads  n_A / n")
ax.set_title("a single toss is unpredictable; the long-run frequency is not"); ax.legend()
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** After 10 tosses the three runs disagree wildly (0.30, 0.40 and 0.50); after 100 tosses they
are within 0.08 of one half (0.48, 0.58, 0.44); after 5,000 tosses all three are within 0.006 (0.4940, 0.4984,
0.4956). The *pattern* — convergence towards 0.5 — is what the frequentist definition captures. How fast the wobble
shrinks is made precise by the **law of large numbers** in [Module 22 · 📏 Sampling: the law of large numbers, the CLT and confidence intervals](22_sampling_lln_clt_confidence.ipynb).

**Bayesian interpretation.** A probability is a **degree of belief** (or of uncertainty) about an event, based on
current knowledge and personal judgement; it can be **updated** when new information arrives. Example: *"I am 80 %
confident it will rain tomorrow."* Tomorrow happens only once, so no long-run frequency exists — yet the statement is
meaningful. This is also called **subjective probability**. A Bayesian may say $P(\text{heads}) = 1/2$ simply because
they have no reason to favour either side of the coin.

Biology needs both views. "The probability that a child of two carriers is affected is 1/4" is a long-run frequency
over many such children. "The probability that *this* newly found variant in *this* patient is pathogenic is 0.9" is a
degree of belief about a single, unrepeatable case. The two views agree whenever beliefs are **calibrated**: among all
the occasions on which a good forecaster says "80 %", it should rain about 80 % of the time.

````python
# ── 2.2 A calibrated forecaster: beliefs that behave like frequencies (SIMULATED, illustrative) ──
rng = np.random.default_rng(SEED + 22)
n_days = 3000
forecast = rng.choice(np.round(np.arange(0.1, 1.0, 0.1), 1), size=n_days)   # the stated degree of belief
rain = rng.random(n_days) < forecast                                         # a calibrated world: belief = chance
table = (pd.DataFrame({"forecast": forecast, "rain": rain})
         .groupby("forecast")["rain"].agg(days="size", fraction_rainy="mean").round(3))
print(table.to_string())
````

🔍 **What just happened?** Each simulated day the forecaster states a belief between 0.1 and 0.9, and in this
(simulated) world rain then happens with exactly that probability. Grouping the 3,000 days by the stated belief, the
observed fraction of rainy days lands close to the belief in every row (for example 0.81 among the 332 days
announced as "80 %", 0.074 among the days announced as "10 %"). A calibrated degree of belief therefore *behaves like* a long-run frequency. Checking the
calibration of predicted probabilities — of a weather model, a pathogenicity predictor or a clinical risk score — is
an important part of trustworthy machine learning ([BiologyToML 42 · Trustworthy ML](../../../../ml/tutorial/BiologyToML/notebooks/42_trustworthy_ml.ipynb)).

### 📘 Lecture notes — interpretations of probability

**Objective of a probability space:** a mathematical framework to describe random outcomes and to reason
automatically under uncertainty. Coin toss: one toss is unpredictable; many tosses show a pattern in the average.

| | Frequentist | Bayesian |
|---|---|---|
| A probability is … | the **long-run frequency** of the event over repeated trials, $P(A) = \lim_{n\to\infty} n_A/n$ | a **degree of belief** (uncertainty) about the event |
| Based on … | repeatable experiments; "objective" | current knowledge and judgement; "subjective"; **updated** as information arrives |
| $P(\text{heads}) = 1/2$ means … | flip 100 times → about 50 heads | I believe heads will come up half of the time |
| Example | proportion of affected children of two carriers | "I am 80 % confident it will rain tomorrow" |

Both interpretations obey **the same axioms** (Section 5), so all the rules of Part E hold in either view. They
differ when we *estimate* unknown quantities: maximum likelihood (frequentist, [Module 23 · 🎯 Parameter estimation and maximum likelihood](23_maximum_likelihood.ipynb)) vs Bayesian estimation ([Module 24 · 🧭 Bayesian estimation and maximum a posteriori](24_bayesian_estimation_map.ipynb)).

## 3 · The sample space Ω

> 📍 **Where we are.** We now build the formal framework. Modern probability rests on a set of axioms proposed by
> **Andrey Kolmogorov** (1933). They introduce three concepts — the **sample space**, the **event space** and the
> **probability measure** — which together form a **probability space** $(\Omega, \mathcal{A}, P)$. A probability
> space models a real-world process with random outcomes, called an **experiment**. This section is about $\Omega$.

**Definition.** The **sample space** $\Omega$ (sometimes called the **universe**) is the set of all possible
outcomes of the experiment. The outcomes must be **mutually exclusive** (exactly one of them happens) and
**exhaustive** (one of them must happen). A single outcome $\omega \in \Omega$ (also called a *sample outcome* or
*atomic element*) is one of the possible things that can happen.

For one coin toss and for two successive tosses (a **Cartesian product**, [Module 04 · 🔤 Tuples, sequences and strings: when order matters](04_tuples_sequences_strings.ipynb)):

$$\Omega_{\text{coin}} = \{\texttt{Heads}, \texttt{Tails}\}, \qquad
\Omega_{\text{twocoin}} = \Omega_{\text{coin}} \times \Omega_{\text{coin}} =
\{\langle H, H\rangle, \langle H, T\rangle, \langle T, H\rangle, \langle T, T\rangle\}.$$

*In words:* one toss has two possible outcomes; two tosses have $2 \times 2 = 4$ outcomes, each an *ordered* pair
(first toss, second toss). $\langle H, T\rangle$ and $\langle T, H\rangle$ are different outcomes — order matters,
exactly as for the tuples of Module 04. We usually abbreviate $\langle H, T\rangle$ as $HT$.

⚠️ **Other names for $\Omega$.** Books use different words for the same set: *state space* (but that name is also
used for the states of a dynamical system, [Module 15 · ⛰️ Gradients, Jacobians, optimization and dynamics](15_multivariable_calculus_dynamics.ipynb)), *sample description space*, *possibility space* — and some even
call it the *event space*. In this course "event space" means something else (Section 4), so we always say
**sample space** for $\Omega$.

````python
# ── 3.1 Sample spaces as Python sets, built with Cartesian products ──
Omega_twocoin = {"".join(t) for t in itertools.product(sorted(Omega_coin), repeat=2)}
Omega_threecoin = {"".join(t) for t in itertools.product(sorted(Omega_coin), repeat=3)}
Omega_die = {1, 2, 3, 4, 5, 6}
print("Ω_coin      =", sorted(Omega_coin), "  |Ω| =", len(Omega_coin))
print("Ω_twocoin   =", sorted(Omega_twocoin), "  |Ω| =", len(Omega_twocoin))
print("Ω_threecoin =", sorted(Omega_threecoin), "  |Ω| =", len(Omega_threecoin))
print("Ω_die       =", sorted(Omega_die), "  |Ω| =", len(Omega_die))
print("|Ω_coin^n| for n = 1 … 10:", [len(Omega_coin) ** n for n in range(1, 11)])
````

🔍 **What just happened?** `itertools.product` builds the Cartesian product, and joining each tuple into a string
gives the short names $HH, HT, TH, TT$. Two tosses have 4 outcomes and three tosses 8; in general $n$ tosses have
$|\Omega_{\text{coin}}|^n = 2^n$ outcomes — 1,024 for ten tosses. Sample spaces grow exponentially with the number of
repeated steps, which is why we will soon stop listing outcomes and start computing with rules.

Pictures help. A **tree diagram** lists the outcomes step by step: one branch per result of the first toss, then one
branch per result of the second toss; each path from the root to a leaf is one outcome. Alternatively draw $\Omega$ as
a rectangle cut into regions, one region per outcome (the region sizes do not matter yet — Section 5 makes "size"
into probability).

````python
# ── 3.2 Three pictures of a sample space ──
fig, axes = plt.subplots(1, 3, figsize=(11, 3.4))
ax = axes[0]                                                  # (a) one flip
ax.add_patch(plt.Rectangle((0, 0), 1, 1, fc="#b0c8cc", ec="k"))
ax.add_patch(Ellipse((0.62, 0.55), 0.55, 0.62, fc="#c0a097", ec="k"))
ax.text(0.62, 0.55, "H", ha="center", va="center", fontsize=12); ax.text(0.1, 0.1, "T", fontsize=12)
ax.set_title("(a) one flip: Ω = {H, T}")
ax = axes[1]                                                  # (b) two flips
ax.add_patch(plt.Rectangle((0, 0), 1, 1, fc="#ccfafa", ec="k", lw=1.5))
for x0, x1 in [(0.19, 0.25), (0.48, 0.41), (0.83, 0.79)]:
    ax.plot([x0, x1], [1, 0], color="k", lw=1.5)
for x, lab in [(0.1, "HH"), (0.33, "HT"), (0.62, "TH"), (0.9, "TT")]:
    ax.text(x, 0.55, lab, ha="center", fontsize=11)
ax.set_title("(b) two flips: four regions")
for a in axes[:2]:
    a.set_xlim(-0.05, 1.05); a.set_ylim(-0.05, 1.05); a.set_aspect("equal"); a.axis("off")
ax = axes[2]                                                  # (c) tree diagram
ax.annotate("", xy=(1, 1), xytext=(0, 0), arrowprops=dict(arrowstyle="->"))
ax.annotate("", xy=(1, -1), xytext=(0, 0), arrowprops=dict(arrowstyle="->"))
for first, y1 in [("H", 1), ("T", -1)]:
    ax.text(1.1, y1, first, va="center", fontsize=12, color="tab:red" if first == "H" else "tab:blue")
    for second, dy in [("H", 0.5), ("T", -0.5)]:
        ax.annotate("", xy=(2.4, y1 + dy), xytext=(1.35, y1), arrowprops=dict(arrowstyle="->"))
        ax.text(2.5, y1 + dy, second, va="center", fontsize=12, color="tab:red" if second == "H" else "tab:blue")
        ax.annotate("", xy=(3.6, y1 + dy), xytext=(2.8, y1 + dy), arrowprops=dict(arrowstyle="->", color="0.5"))
        ax.text(3.7, y1 + dy, first + second, va="center", fontsize=12)
ax.text(1.1, 1.9, "1st toss", ha="center"); ax.text(2.5, 1.9, "2nd toss", ha="center")
ax.text(3.9, 1.9, "Ω", ha="center", fontsize=12)
ax.set_xlim(-0.2, 4.4); ax.set_ylim(-1.9, 2.2); ax.axis("off"); ax.set_title("(c) tree: one path = one outcome")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** Panel (a) draws $\Omega_{\text{coin}}$ as a rectangle in which the outcome $H$ occupies the
oval and $T$ everything else; panel (b) cuts the rectangle for two flips into four regions $HH, HT, TH, TT$; panel (c)
is the tree: two branches for the first toss, two for the second, so $2 \times 2 = 4$ leaves — one leaf, e.g. $TT$,
is **one sample outcome**. All three pictures show the same thing: the outcomes are mutually exclusive (regions do not
overlap) and exhaustive (they fill the rectangle).

### 3.1 A Punnett square is a sample space

When two heterozygous parents $Aa \times Aa$ have a child, the "experiment" is: the mother passes on one of her two
alleles, the father one of his. The sample space is the Cartesian product of the two **gamete** sets,

$$\Omega_{\text{cross}} = \{A, a\}_{\text{mother}} \times \{A, a\}_{\text{father}} = \{\langle A,A\rangle, \langle A,a\rangle, \langle a,A\rangle, \langle a,a\rangle\},$$

*In words:* four ordered pairs (mother's allele, father's allele) — exactly the four cells of a **Punnett square**. For
two genes on different chromosomes (Mendel's dihybrid cross $YyRr \times YyRr$) each parent makes four kinds of
gametes $\{YR, Yr, yR, yr\}$, so $|\Omega| = 4 \times 4 = 16$.

````python
# ── 3.3 Punnett squares are Cartesian products of gametes ──
gametes = ["A", "a"]
Omega_cross = list(itertools.product(gametes, gametes))          # (mother, father)
print("Ω_cross =", Omega_cross, "→", len(Omega_cross), "outcomes")
punnett = pd.DataFrame([["".join(sorted(m + f, key=lambda c: (c.lower(), c.islower()))) for f in gametes] for m in gametes],
                       index=[f"mother {m}" for m in gametes], columns=[f"father {f}" for f in gametes])
print(punnett)

dihybrid_gametes = ["".join(g) for g in itertools.product("Yy", "Rr")]
Omega_dihybrid = list(itertools.product(dihybrid_gametes, dihybrid_gametes))
print("\ngametes of a YyRr parent:", dihybrid_gametes)
print("dihybrid cross: |Ω| =", len(Omega_dihybrid), "— first five outcomes:", Omega_dihybrid[:5])
````

🔍 **What just happened?** The monohybrid cross has the four outcomes $(A,A), (A,a), (a,A), (a,a)$, which the
Punnett square displays as the genotypes $AA$, $Aa$, $Aa$, $aa$ (we write the dominant allele first). The dihybrid cross
has 16 ordered pairs of gametes, from $(YR, YR)$ to $(yr, yr)$. Notice that $(A,a)$ and $(a,A)$ are *different
outcomes* leading to the *same genotype* $Aa$ — this distinction becomes decisive in Section 4.

## 4 · Events and the event space

> 📍 **Where we are.** The sample space lists what *can* happen. The questions we ask — "is the first toss heads?",
> "is the child affected?" — are about *sets* of outcomes. Sets were Module 03's topic; here they get a new name.

**Definition.** An **event** is a subset of the sample space, usually described by a condition. For instance, the
event that the first of two coin tosses is heads is

$$A_{\text{firstheads}} = \{\langle H, H\rangle, \langle H, T\rangle\} \subset \Omega_{\text{twocoin}}.$$

*In words:* the event collects every outcome in which the condition "first toss = heads" is true — set-builder
notation $\{\omega \in \Omega \mid \omega_1 = H\}$ from Module 03. We say the event **occurs** if the outcome of the
experiment belongs to it.

**Counting rule.** If all outcomes are **equally likely**, the probability of an event is the fraction of outcomes it
covers:

$$P(A) = \frac{|A|}{|\Omega|}, \qquad P(A_{\text{firstheads}}) = \frac{2}{4} = \frac{1}{2}.$$

*In words:* count the outcomes in the event and divide by the number of all outcomes. This is the oldest definition of
probability (the "classical" one).

````python
# ── 4.1 Events as subsets of Ω_twocoin, and the counting rule ──
A_firstheads = {w for w in Omega_twocoin if w[0] == "H"}
A_lastheads = {w for w in Omega_twocoin if w[1] == "H"}
A_allheads = {w for w in Omega_twocoin if w == "HH"}
A_threetails = {w for w in Omega_twocoin if w.count("T") == 3}      # impossible with two tosses
for name, event in [("A_firstheads", A_firstheads), ("A_lastheads", A_lastheads),
                    ("A_allheads", A_allheads), ("A_threetails", A_threetails)]:
    print(f"{name:13s} = {str(sorted(event)) if event else '∅':14s}  P = |A|/|Ω| = {classical_probability(event, Omega_twocoin)}")

# ⚠️ the counting rule needs equally likely outcomes
genotypes_unordered = {"AA", "Aa", "aa"}                      # 3 genotypes — NOT equally likely
child = ["".join(sorted(m + f)) for m, f in Omega_cross]       # genotype of each of the 4 equally likely outcomes
print("\nwrong: P(aa) = 1/|{AA, Aa, aa}| =", classical_probability({"aa"}, genotypes_unordered))
print("right: P(aa) = |{(a,a)}| / |Ω_cross| =", Fraction(child.count("aa"), len(child)))
````

🔍 **What just happened?** Set comprehensions turned each condition into an event: $A_{\text{firstheads}} = \{HH,
HT\}$ and $A_{\text{lastheads}} = \{HH, TH\}$ (probability 1/2 each), $A_{\text{allheads}} = \{HH\}$ (1/4), and
$A_{\text{threetails}} = \emptyset$ (probability 0 — two tosses cannot give three tails).

The last two lines show the classic trap. The three *genotypes* $AA, Aa, aa$ are not equally likely, so counting them
gives the wrong answer 1/3 for $P(aa)$. Counting the four equally likely *ordered* outcomes of the Punnett square gives
the right answer 1/4. The great 18th-century mathematician d'Alembert made exactly this error with coins (he argued
that "at least one head in two tosses" has probability 2/3 by counting $H$, $TH$, $TT$). ⚠️ **The counting rule is only
valid for equally likely outcomes** — choose the sample space so that they are.

### 4.1 The event space 𝒜

The **event space** $\mathcal{A}$ is the set of all events we are allowed to ask about — the potential results of the
experiment. A subset $A \subseteq \Omega$ belongs to $\mathcal{A}$ if, at the end of the experiment, we can observe
whether the outcome $\omega$ is in $A$. For discrete experiments (coins, dice, crosses) $\mathcal{A}$ is usually the
**power set** of $\Omega$ (Module 03): *every* subset is an event. In general, for a non-empty $\Omega$, an event space
must satisfy two closure rules:

$$\text{(1)}\;\; A \in \mathcal{A} \;\Rightarrow\; \overline{A} \in \mathcal{A}, \qquad
\text{(2)}\;\; A_1, A_2, \ldots \in \mathcal{A} \;\Rightarrow\; \bigcup_{i=1}^{\infty} A_i \in \mathcal{A}.$$

*In words:* (1) if we can ask whether $A$ happened, we can ask whether it did **not** happen ($\overline{A} = \Omega
\setminus A$, the complement); (2) if we can ask about each of a (possibly infinite) sequence of events, we can ask
whether **at least one** of them happened. A collection with these properties is called a **σ-field** (or σ-algebra).
Consequences: $\Omega$ and $\emptyset$ are always events, and so are intersections (by De Morgan's law,
$A \cap B = \overline{\overline{A} \cup \overline{B}}$, Modules 02–03).

Examples: one coin toss gives $\mathcal{A} = \{\emptyset, \{H\}, \{T\}, \{H, T\}\}$; a die gives all $2^6 = 64$ subsets
of $\{1, \ldots, 6\}$, such as $\emptyset, \{1\}, \{2, 4\}, \{1, 2, 3\}, \{1, 3, 5\}, \Omega, \ldots$

````python
# ── 4.2 Event spaces as power sets, and a check of the closure rules ──
def power_set(omega: set) -> list:
    """All subsets of a finite set, as frozensets (Module 03: there are 2^|Ω| of them)."""
    items = sorted(omega)
    return [frozenset(c) for r in range(len(items) + 1) for c in itertools.combinations(items, r)]


def is_event_space(events: list, omega: set) -> bool:
    """Closed under complement and under (finite, here = countable) unions, and contains Ω."""
    family = set(events)
    closed_complement = all(frozenset(omega) - A in family for A in family)
    closed_union = all(A | B in family for A in family for B in family)
    return closed_complement and closed_union and frozenset(omega) in family


A_coin = power_set(Omega_coin)
print("𝒜 for one coin:", [set(A) if A else "∅" for A in A_coin])
A_die = power_set(Omega_die)
print("𝒜 for a die has", len(A_die), "events, e.g.", [set(A) if A else "∅" for A in A_die[:4]], "…", set(A_die[-1]))
print("power set of the die is an event space:", is_event_space(A_die, Omega_die))
too_small = [frozenset(), frozenset({1}), frozenset(Omega_die)]
print("{∅, {1}, Ω} is an event space:", is_event_space(too_small, Omega_die), "(the complement {2,…,6} is missing)")
coarse = [frozenset(), frozenset({1, 3, 5}), frozenset({2, 4, 6}), frozenset(Omega_die)]
print("{∅, odd, even, Ω} is an event space:", is_event_space(coarse, Omega_die))
````

🔍 **What just happened?** The coin's event space has the four events $\emptyset, \{H\}, \{T\}, \{H, T\}$; the die's
has $2^6 = 64$, and our checker confirms that the power set satisfies both closure rules. The family $\{\emptyset,
\{1\}, \Omega\}$ is *not* an event space — it contains $\{1\}$ but not its complement $\{2, 3, 4, 5, 6\}$. But the small
family $\{\emptyset, \text{odd}, \text{even}, \Omega\}$ *is* one: it describes an observer who is told only whether the
roll was odd or even. Event spaces can therefore encode **what information is available** — an idea that returns in
conditional probability ([Module 17 · 🔀 Conditional probability, independence and Bayes' rule](17_conditional_probability_bayes.ipynb)).

### 📘 Lecture notes — sample space, events and event space

**Probability space** $(\Omega, \mathcal{A}, P)$ (Kolmogorov): sample space, event space, probability measure; it
models an *experiment* with random outcomes.

* **Sample space** $\Omega$ (also "universe"): the set of all possible, mutually exclusive outcomes.
  $\Omega_{\text{coin}} = \{H, T\}$; $\Omega_{\text{twocoin}} = \Omega_{\text{coin}} \times \Omega_{\text{coin}} =
  \{HH, HT, TH, TT\}$ (tree diagram: one path = one outcome). An outcome (atomic element) $\omega \in \Omega$ is one of
  the things that can happen. Other names: state space, sample description space, possibility space.
* **Event**: a subset of $\Omega$ defined by a condition, e.g. $A_{\text{firstheads}} = \{HH, HT\}$. For **equally
  likely** outcomes $P(A) = |A|/|\Omega|$, so $P(A_{\text{firstheads}}) = 1/2$.
* **Event space** $\mathcal{A}$: the set of all events. Discrete $\Omega$: the power set ($2^{|\Omega|}$ events; coin:
  $\{\emptyset, \{H\}, \{T\}, \{H,T\}\}$; die: 64 events such as $\{2,4\}, \{1,2,3\}, \{1,3,5\}$). Requirements:
  $A \in \mathcal{A} \Rightarrow \overline{A} \in \mathcal{A}$ and $A_1, A_2, \ldots \in \mathcal{A} \Rightarrow
  \bigcup_i A_i \in \mathcal{A}$. Continuous $\Omega$: the **Borel field** (Section 8).

**Common confusions.** An event is a *set* of outcomes, not a single outcome ($\{HH\}$ vs $HH$). The counting rule
fails when outcomes are not equally likely (genotypes $AA, Aa, aa$).

## 5 · The probability measure and Kolmogorov's three axioms

> 📍 **Where we are.** We have outcomes ($\Omega$) and the events we may ask about ($\mathcal{A}$). The last ingredient
> attaches a number to every event. Three short rules decide which assignments are allowed.

With each event $A \in \mathcal{A}$ we associate a number $P(A)$ that measures the probability — or the degree of
belief — that the event will occur. Formally, $P$ is a **function** (Module 03) from the event space to the real
numbers, called a **probability measure**: $P: \mathcal{A} \to [0, 1]$. It must satisfy **Kolmogorov's three axioms**:

$$\textbf{Axiom 1 (non-negativity):}\quad P(A) \ge 0 \;\text{ for every event } A.$$

*In words:* probabilities can be zero but never negative. If $A_{\text{threetails}}$ is the event of three tails in two
tosses, $P(A_{\text{threetails}}) = 0$ — and that is as low as it gets.

$$\textbf{Axiom 2 (total probability):}\quad P(\Omega) = 1.$$

*In words:* *something* in the sample space certainly happens. For one coin flip, $P(\Omega) = P(\{H\}) + P(\{T\}) =
1$: as defined, there are no other possible outcomes.

$$\textbf{Axiom 3 (additivity):}\quad A_1, A_2, \ldots \text{ pairwise disjoint } \;\Rightarrow\;
P(A_1 \cup A_2 \cup \cdots) = P(A_1) + P(A_2) + \cdots$$

*In words:* if events cannot happen together (no outcome in common: $A_i \cap A_j = \emptyset$ for $i \ne j$), the
probability that one of them happens is the sum of their probabilities. For one coin flip, $A_{\text{heads}}$ and
$A_{\text{tails}}$ share no outcome, so $P(A_{\text{heads}} \cup A_{\text{tails}}) = \tfrac12 + \tfrac12 = 1$.
Kolmogorov required this for any *countable* sequence of disjoint events; for finitely many events $A_1, \ldots, A_n$
it reads $P(A_1 \cup \cdots \cup A_n) = P(A_1) + \cdots + P(A_n)$.

⚠️ **Disjointness is essential.** Take $A_{\text{firstheads}} = \{HH, HT\}$ and $A_{\text{lastheads}} = \{HH, TH\}$.
They overlap: $A_{\text{firstheads}} \cap A_{\text{lastheads}} = \{HH\} \ne \emptyset$. So Axiom 3 does **not** apply,
and indeed $P(A_{\text{firstheads}} \cup A_{\text{lastheads}}) \ne P(A_{\text{firstheads}}) + P(A_{\text{lastheads}})$.

````python
# ── 5.1 The axioms on two coin tosses, and the overlapping counterexample ──
P2 = lambda event: classical_probability(set(event), Omega_twocoin)      # the uniform measure on Ω_twocoin
print("Axiom 1: P(A_threetails) =", P2(A_threetails), "≥ 0")
print("Axiom 2: P(Ω_twocoin) =", P2(Omega_twocoin))
A_heads, A_tails = {"H"}, {"T"}
P1 = lambda event: classical_probability(set(event), Omega_coin)
print("Axiom 3: P(A_heads ∪ A_tails) =", P1(A_heads | A_tails), "= P(A_heads) + P(A_tails) =", P1(A_heads) + P1(A_tails))
union = A_firstheads | A_lastheads
print("\noverlapping events: A_firstheads ∩ A_lastheads =", A_firstheads & A_lastheads)
print("P(A_firstheads ∪ A_lastheads) =", P2(union), "  but  P(A_firstheads) + P(A_lastheads) =",
      P2(A_firstheads) + P2(A_lastheads))
````

🔍 **What just happened?** The three axioms hold for the counting measure on coin tosses: the impossible event has
probability 0, the whole sample space has probability 1, and the two disjoint one-toss events add up to 1. For the
overlapping events the union $\{HH, HT, TH\}$ has probability 3/4, while the sum of the separate probabilities is
$1/2 + 1/2 = 1$: adding counts the shared outcome $HH$ twice. Section 6 turns this into the inclusion–exclusion rule.

### 5.1 A measure does not have to be uniform

The axioms say nothing about outcomes being equally likely. On a **finite** (or countable) sample space we can build
*any* probability measure by giving each outcome a non-negative weight, with all weights summing to 1, and defining

$$P(A) = \sum_{\omega \in A} p(\omega).$$

*In words:* the probability of an event is the total weight of the outcomes inside it. Axiom 1 holds because weights
are non-negative, Axiom 2 because they sum to 1, and Axiom 3 because the weights of disjoint events are added without
overlap. Example: the child's **genotype** in $Aa \times Aa$ takes the three values $AA, Aa, aa$ with weights
$\tfrac14, \tfrac12, \tfrac14$. The next cell builds such measures and checks *every* axiom on *every* event.

````python
# ── 5.2 Probability measures from outcome weights, and an exhaustive check of the axioms ──
def make_measure(weights: dict):
    """P(A) = sum of the weights of the outcomes in A (a discrete probability measure if the weights are valid)."""
    return lambda event: sum((weights[w] for w in event), Fraction(0))


def check_axioms(P, omega: set) -> dict:
    """Test the three axioms on every event of the power set (Axiom 3 on every pair of disjoint events)."""
    events = power_set(omega)
    disjoint_pairs = [(A, B) for A in events for B in events if not A & B]
    return {"axiom 1 (all P(A) ≥ 0)": all(P(A) >= 0 for A in events),
            "axiom 2 (P(Ω) = 1)": P(omega) == 1,
            f"axiom 3 ({len(disjoint_pairs)} disjoint pairs)": all(P(A | B) == P(A) + P(B) for A, B in disjoint_pairs)}


genotype_weights = {"AA": Fraction(1, 4), "Aa": Fraction(1, 2), "aa": Fraction(1, 4)}   # Mendel, Aa × Aa
P_genotype = make_measure(genotype_weights)
print("genotype measure:", check_axioms(P_genotype, set(genotype_weights)))
print("P(child carries at least one a) = P({Aa, aa}) =", P_genotype({"Aa", "aa"}))

bad_weights = {"AA": Fraction(3, 10), "Aa": Fraction(1, 2), "aa": Fraction(3, 10)}      # sums to 11/10
print("\ninvalid assignment:", check_axioms(make_measure(bad_weights), set(bad_weights)))
negative_weights = {"AA": Fraction(1, 2), "Aa": Fraction(3, 4), "aa": Fraction(-1, 4)}
print("negative weight:   ", check_axioms(make_measure(negative_weights), set(negative_weights)))
````

🔍 **What just happened?** For the Mendelian genotype weights $(\tfrac14, \tfrac12, \tfrac14)$ all three axioms hold on
all $2^3 = 8$ events — Axiom 3 was checked on all 27 pairs of disjoint events — and the event "at least one $a$
allele" gets $P(\{Aa, aa\}) = 3/4$. The assignment $(0.3, 0.5, 0.3)$ fails Axiom 2 (its total is 11/10), so it is not
a probability measure even though each number lies between 0 and 1. The assignment $(\tfrac12, \tfrac34, -\tfrac14)$
sums to 1 but fails Axiom 1. **Both requirements are needed.**

### 5.2 Interpreting the axioms as areas

A helpful picture: draw $\Omega$ as a rectangle of **area 1** and each event as a region inside it; the probability of
an event is its area ("all possible worlds" fill the rectangle, and $A$ is the set of worlds in which $A$ is true).
Areas are never negative (Axiom 1), the whole rectangle has area 1 (Axiom 2), and the areas of non-overlapping regions
add up (Axiom 3). Two consequences are visible at once: the area of $A$ cannot exceed 1, and an area of 1 would mean
that $A$ is true in *all* worlds. We can even *measure* an area by throwing random points into the rectangle.

````python
# ── 5.3 Probability as area: estimate the area of an event by random points (SIMULATED) ──
rng = np.random.default_rng(SEED + 53)
n_points = 20_000
pts = rng.random((n_points, 2))                                  # uniform points in the unit square Ω
centre, radius = np.array([0.6, 0.5]), 0.3
in_A = ((pts - centre) ** 2).sum(axis=1) <= radius ** 2           # event A = a disc
print(f"fraction of points in A: {in_A.mean():.4f};  exact area π·0.3² = {np.pi * radius ** 2:.4f}")
fig, ax = plt.subplots(figsize=(4.2, 4.0))
ax.scatter(*pts[:2000][~in_A[:2000]].T, s=2, color="0.7")
ax.scatter(*pts[:2000][in_A[:2000]].T, s=2, color="tab:red")
ax.add_patch(plt.Circle(centre, radius, fill=False, lw=1.5)); ax.add_patch(plt.Rectangle((0, 0), 1, 1, fill=False, lw=1.5))
ax.text(0.6, 0.5, "A", fontsize=14, ha="center", va="center"); ax.text(0.03, 0.92, "Ω (area 1)", fontsize=10)
ax.set_aspect("equal"); ax.axis("off")
ax.set_title("P(A) = area of A (first 2,000 of 20,000 points shown)")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** About 28 % of the 20,000 random points fell inside the disc, matching its exact area
$\pi \times 0.3^2 = 0.2827$ to about two decimals. The rectangle has area 1, so the area of the disc *is* a
probability: the probability that a random point lands in $A$. This "probability = area" picture is how we will
think about Venn diagrams of events from now on, and it is the heart of the continuous sample spaces of Section 8. (The
method of estimating an area by random points is called **Monte Carlo** integration.)

### 📘 Lecture notes — the probability measure and the three axioms

With each event $A \in \mathcal{A}$ we associate a number $P(A)$, the probability (or degree of belief) that $A$
occurs; $P: \mathcal{A} \to [0, 1]$ is the **probability measure**. It must satisfy:

1. **Axiom 1 (positive probability):** $P(A) \ge 0$ for every event. ($P(A_{\text{threetails}}) = 0$ for two tosses
   — zero, never negative.)
2. **Axiom 2 (total probability):** $P(\Omega) = 1$. (One flip: $P(\Omega) = P(H) + P(T) = 1$.)
3. **Axiom 3 (disjoint events):** for pairwise disjoint $A_1, A_2, \ldots, A_n$ (Kolmogorov: even countably many),
   $P(A_1 \cup \cdots \cup A_n) = P(A_1) + \cdots + P(A_n)$. (One flip: $P(A_{\text{heads}} \cup A_{\text{tails}}) =
   \tfrac12 + \tfrac12 = 1$.)

**Counterexample.** $A_{\text{lastheads}} = \{HH, TH\}$ overlaps $A_{\text{firstheads}} = \{HH, HT\}$ in $\{HH\}$, so
$P(A_{\text{firstheads}} \cup A_{\text{lastheads}}) = 3/4 \ne 1/2 + 1/2$.

**Areas.** Think of $\Omega$ as a region of area 1 and $P(A)$ as the area of $A$: it can't get bigger than 1, and an
area of 1 means $A$ is true in all worlds.

**Discrete measures** come from outcome weights: $P(A) = \sum_{\omega \in A} p(\omega)$ with $p(\omega) \ge 0$ and
$\sum_\omega p(\omega) = 1$. ⚠️ The axioms do **not** require equally likely outcomes.

> 🎓 **Exam tip.** To show that a proposed $P$ is *not* a probability measure, one failing axiom is enough: find a
> negative value, a total different from 1, or two disjoint events whose probabilities do not add.

## 6 · Consequences of the axioms

> 📍 **Where we are.** Three axioms are all we assume. Everything else must be **proved** from them — and the proofs
> are short, use only the set identities of Module 03, and are favourite exam questions. Each proof below is followed
> by an exhaustive check in code.

**Theorem 1.** $P(\emptyset) = 0$.

*Proof.*
1. $\Omega \cup \emptyset = \Omega$ (adding nothing to a set does not change it).
2. $\Omega$ and $\emptyset$ are disjoint ($\Omega \cap \emptyset = \emptyset$), so by Axiom 3, $P(\Omega \cup \emptyset)
   = P(\Omega) + P(\emptyset)$.
3. Putting 1 and 2 together: $P(\Omega) = P(\Omega) + P(\emptyset)$.
4. By Axiom 2, $P(\Omega) = 1$, so $1 = 1 + P(\emptyset)$, and therefore $P(\emptyset) = 0$. ∎

**Theorem 2 (complement rule).** $P(\overline{A}) = 1 - P(A)$.

*Proof.*
1. By definition $\overline{A} = \Omega \setminus A$, so $A$ and $\overline{A}$ are disjoint and $A \cup \overline{A}
   = \Omega$.
2. Axiom 3: $P(A \cup \overline{A}) = P(A) + P(\overline{A})$.
3. With step 1 and Axiom 2: $P(A) + P(\overline{A}) = P(\Omega) = 1$.
4. Subtract $P(A)$ from both sides: $P(\overline{A}) = 1 - P(A)$. ∎

**Theorem 3.** $0 \le P(A) \le 1$.

*Proof.* $0 \le P(A)$ is Axiom 1. Also $P(\overline{A}) \ge 0$ by Axiom 1, i.e. $-1 \cdot P(\overline{A}) \le 0$. By
Theorem 2, $P(A) = 1 - P(\overline{A}) \le 1 - 0 = 1$. ∎ (So the codomain $[0, 1]$ of $P$ is a *consequence*, not an
extra assumption.)

**Theorem 4 (monotonicity).** If $A \subseteq B$ then $P(A) \le P(B)$.

*Proof.*

$$\begin{aligned}
P(B) &= P\big(A \cup (B \setminus A)\big) && B = A \cup (B \setminus A) \text{ because } A \subseteq B\\
&= P(A) + P(B \setminus A) && A \text{ and } B \setminus A \text{ are disjoint (Axiom 3)}\\
&\ge P(A) && P(B \setminus A) \ge 0 \text{ (Axiom 1)}. \;∎
\end{aligned}$$

*In words:* a bigger event is the smaller one plus an extra, non-overlapping piece $B \setminus A$, and that extra
piece cannot have negative probability. "The child is a carrier *or* affected" can never be less likely than "the
child is affected".

````python
# ── 6.1 Pictures of the complement and of monotonicity ──
fig, axes = plt.subplots(1, 2, figsize=(9, 3.3))
ax = axes[0]
ax.add_patch(plt.Rectangle((0, 0), 1.6, 1, fc="#ffd54f", ec="k"))          # Ω shaded = complement …
ax.add_patch(plt.Circle((0.6, 0.48), 0.3, fc="white", ec="k"))             # … minus A
ax.text(0.6, 0.48, "A", fontsize=14, ha="center", va="center"); ax.text(1.3, 0.8, "Ā = Ω \\ A", fontsize=11)
ax.set_title("complement: P(Ā) = 1 − P(A)")
ax = axes[1]
ax.add_patch(plt.Rectangle((0, 0), 1.6, 1, fc="white", ec="k"))
ax.add_patch(Ellipse((0.8, 0.5), 1.25, 0.88, fc="#ffd54f", ec="k"))       # B \ A shaded
ax.add_patch(Ellipse((0.68, 0.45), 0.75, 0.55, fc="white", ec="k"))         # A inside B
ax.text(0.68, 0.45, "A", fontsize=14, ha="center", va="center"); ax.text(1.18, 0.62, "B \\ A", fontsize=11)
ax.text(1.38, 0.85, "B", fontsize=12, fontweight="bold"); ax.text(1.5, 0.05, "Ω", fontsize=12)
ax.set_title("A ⊆ B: P(B) = P(A) + P(B \\ A) ≥ P(A)")
for a in axes:
    a.set_xlim(-0.05, 1.65); a.set_ylim(-0.05, 1.05); a.set_aspect("equal"); a.axis("off")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** Left: the shaded region is everything in $\Omega$ outside $A$; its area is $1$ minus the
area of $A$ — Theorem 2 as a picture. Right: $A$ sits inside $B$; the shaded ring $B \setminus A$ is what $B$ adds to
$A$, so $B$'s area is $A$'s area plus a non-negative amount — Theorem 4.

Now let us check all four theorems on **every** event of a *loaded* die, so that we are sure nothing depends on
outcomes being equally likely.

````python
# ── 6.2 Exhaustive check of Theorems 1–4 on a loaded die ──
loaded = {1: Fraction(1, 10), 2: Fraction(1, 10), 3: Fraction(1, 10), 4: Fraction(1, 10), 5: Fraction(1, 10), 6: Fraction(1, 2)}
P_die = make_measure(loaded)
events = power_set(Omega_die)
omega = frozenset(Omega_die)
print("Theorem 1: P(∅) =", P_die(frozenset()))
print("Theorem 2: P(Ā) = 1 − P(A) for all", len(events), "events:",
      all(P_die(omega - A) == 1 - P_die(A) for A in events))
print("Theorem 3: 0 ≤ P(A) ≤ 1 for all events:", all(0 <= P_die(A) <= 1 for A in events))
nested = [(A, B) for A in events for B in events if A <= B]
print(f"Theorem 4: P(A) ≤ P(B) for all {len(nested)} pairs with A ⊆ B:", all(P_die(A) <= P_die(B) for A, B in nested))
print("example: P({6}) =", P_die({6}), "≤ P({2, 4, 6}) =", P_die({2, 4, 6}), "≤ P(Ω) =", P_die(omega))
````

🔍 **What just happened?** With the loaded die ($P(\{6\}) = 1/2$, the other faces $1/10$ each) every theorem holds on
all 64 events, and monotonicity holds on all **729** nested pairs $A \subseteq B$ (each of the 6 faces can be in
neither, only $B$, or both: $3^6 = 729$). For example $P(\{6\}) = 1/2 \le P(\{2, 4, 6\}) = 7/10 \le P(\Omega) = 1$.
A computer check over a finite example is not a proof — but it is an excellent way to catch a wrong formula.

### 6.1 Inclusion–exclusion: the probability of "A or B"

**Theorem 5 (inclusion–exclusion).** For any two events,

$$P(A \cup B) = P(A) + P(B) - P(A \cap B).$$

*In words:* add the two probabilities, then subtract the overlap, which was counted twice — the probability version of
Module 03's $|S \cup T| = |S| + |T| - |S \cap T|$.

*Proof.*
1. $A \cup B = A \cup (B \setminus A)$, and $A$, $B \setminus A$ are disjoint, so (Axiom 3) $P(A \cup B) = P(A) +
   P(B \setminus A)$.
2. $B = (A \cap B) \cup (B \setminus A)$, again disjoint, so $P(B) = P(A \cap B) + P(B \setminus A)$, i.e.
   $P(B \setminus A) = P(B) - P(A \cap B)$.
3. Substitute step 2 into step 1: $P(A \cup B) = P(A) + P(B) - P(A \cap B)$. ∎

⚠️ **Common confusion.** $P(A \cup B) = P(A) + P(B)$ is **not always correct**; it holds only when the overlap has
probability zero — in particular when $A$ and $B$ are disjoint. Dropping the last term (the union bound
$P(A \cup B) \le P(A) + P(B)$) always *over*estimates.

**Example (a die).** $A$ = "roll is even" $= \{2, 4, 6\}$, $B$ = "roll is $\ge 4$" $= \{4, 5, 6\}$. Then
$A \cup B = \{2, 4, 5, 6\}$, $A \cap B = \{4, 6\}$ and

$$P(A \cup B) = \tfrac36 + \tfrac36 - \tfrac26 = \tfrac46.$$

*In words:* three even faces plus three high faces would be six, but 4 and 6 are both even and high, so they were
counted twice; four faces remain.

````python
# ── 6.3 Inclusion–exclusion for the fair die, with a Venn diagram ──
A_even, B_high = {2, 4, 6}, {4, 5, 6}
P_fair = make_measure({w: Fraction(1, 6) for w in Omega_die})
lhs = P_fair(A_even | B_high)
rhs = P_fair(A_even) + P_fair(B_high) - P_fair(A_even & B_high)
print("A ∪ B =", A_even | B_high, "  A ∩ B =", A_even & B_high)
print(f"P(A ∪ B) = {lhs}   P(A) + P(B) − P(A ∩ B) = {P_fair(A_even)} + {P_fair(B_high)} − {P_fair(A_even & B_high)} = {rhs}")
print("naive sum P(A) + P(B) =", P_fair(A_even) + P_fair(B_high), "(counts 4 and 6 twice)")
print("inclusion–exclusion holds on all", len(events) ** 2, "pairs of events of the LOADED die:",
      all(P_die(A | B) == P_die(A) + P_die(B) - P_die(A & B) for A in events for B in events))
draw_venn({"A (even)": A_even, "B (≥ 4)": B_high}, universe=Omega_die, highlight=lambda a, b: a or b,
          title="A ∪ B shaded: {2, 4, 5, 6}, probability 4/6")
plt.show()
````

🔍 **What just happened?** The fair die gives $P(A \cup B) = 2/3$ (= 4/6) both directly and by inclusion–exclusion,
while the naive sum is 1 — a 50 % overestimate, because faces 4 and 6 sit in the overlap. The rule also held on all
4,096 ordered pairs of events of the loaded die. In the Venn diagram the shaded union contains 2 (only even), 4 and 6
(both), 5 (only high); the faces outside both circles (1 and 3) are listed in the corner of the universe.

### 6.2 Biology: does at least one of two genes carry a variant?

Suppose that in a population (illustrative numbers) 3 % of people carry a rare loss-of-function variant in gene 1,
5 % in gene 2, and 0.4 % in both. What fraction carries a variant in *at least one* of the two genes? Inclusion–
exclusion gives $0.03 + 0.05 - 0.004 = 0.076$. We check it on a simulated cohort.

````python
# ── 6.4 At least one of two genes carries a variant (SIMULATED cohort, ILLUSTRATIVE frequencies) ──
p_both, p_1, p_2 = 0.004, 0.03, 0.05
cells = {"both": p_both, "gene 1 only": p_1 - p_both, "gene 2 only": p_2 - p_both,
         "neither": 1 - p_1 - p_2 + p_both}                     # a partition of Ω (Module 03)
print("partition probabilities:", {k: round(v, 4) for k, v in cells.items()}, "sum =", round(sum(cells.values()), 12))
rng = np.random.default_rng(SEED + 64)
person = rng.choice(list(cells), size=200_000, p=list(cells.values()))
V1 = np.isin(person, ["both", "gene 1 only"]); V2 = np.isin(person, ["both", "gene 2 only"])
print(f"inclusion–exclusion: P(V1 ∪ V2) = {p_1} + {p_2} − {p_both} = {p_1 + p_2 - p_both:.3f}")
print(f"simulated cohort of 200,000: fraction with a variant in gene 1 or gene 2 = {np.mean(V1 | V2):.4f}")
print(f"naive sum {p_1 + p_2:.3f} overestimates by {p_both}: exactly the overlap")
````

🔍 **What just happened?** We split the population into four disjoint groups — both genes, gene 1 only, gene 2 only,
neither — whose probabilities $0.004, 0.026, 0.046, 0.924$ sum to 1 (a **partition**), and simulated 200,000 people. The
fraction with a variant in at least one gene came out at about 0.076, as inclusion–exclusion predicts; the naive sum
0.08 overestimates by exactly the overlap 0.004. In gene-panel testing, burden tests and "how many patients would a
panel of $k$ genes diagnose?" calculations, this overlap matters.

## 7 · Set operations on events: Mendel's ratios

> 📍 **Where we are.** Events are sets, so the set operations of Module 03 — and the logical connectives of Module 02 —
> build new events from old ones. Here we use them to derive Mendel's ratios exactly.

Let $A, B$ be events in the event space:

* **Union** $A \cup B$ — the event that occurs if $A$ **or** $B$ (or both) occurs;
* **Intersection** $A \cap B$ — occurs if **both** $A$ **and** $B$ occur;
* **Complement** $A^c = \overline{A}$ — occurs if $A$ does **not** occur.

The two rules that go with them are inclusion–exclusion and the complement rule:

$$P(A \cup B) = P(A) + P(B) - P(A \cap B), \qquad P(A^c) = 1 - P(A).$$

*In words:* "or" adds probabilities and removes the double-counted overlap; "not" subtracts from 1. Below, $A = \{1, 3,
5, 7\}$ and $B = \{2, 4, 5, 7\}$ inside the universe $\Omega = \{1, \ldots, 8\}$; the shaded region is the new event.

````python
# ── 7.1 Union, intersection and complement of events, shaded ──
Omega8 = set(range(1, 9))
A_ev, B_ev = {1, 3, 5, 7}, {2, 4, 5, 7}
fig, axes = plt.subplots(1, 3, figsize=(11, 3.4))
for ax, (title, rule, result) in zip(axes, [("union A ∪ B", lambda a, b: a or b, A_ev | B_ev),
                                            ("intersection A ∩ B", lambda a, b: a and b, A_ev & B_ev),
                                            ("complement Aᶜ", lambda a, b: not a, Omega8 - A_ev)]):
    draw_venn({"A": A_ev, "B": B_ev}, universe=Omega8, ax=ax, highlight=None if title.startswith("complement") else rule,
              title=f"{title} = {sorted(result)}")
    if title.startswith("complement"):                         # Aᶜ also contains everything outside both circles
        ax.add_patch(plt.Rectangle((-2.1, -1.9), 4.2, 3.7, color="#ffd54f", alpha=0.8, zorder=0))
        ax.add_patch(plt.Circle((-0.55, 0.0), 1.0, color="white", zorder=0.5))   # … but nothing inside A
plt.tight_layout(); plt.show()
P8 = lambda e: classical_probability(e, Omega8)
print("P(A ∪ B) =", P8(A_ev | B_ev), "=", P8(A_ev), "+", P8(B_ev), "−", P8(A_ev & B_ev))
print("P(Aᶜ) =", P8(Omega8 - A_ev), "= 1 −", P8(A_ev))
````

🔍 **What just happened?** Each panel shades the outcomes of a new event: $A \cup B = \{1, 2, 3, 4, 5, 7\}$,
$A \cap B = \{5, 7\}$ and $A^c = \{2, 4, 6, 8\}$ (the complement includes 6 and 8, which lie outside both circles —
hence the shaded background around the circles). With 8 equally likely outcomes, $P(A \cup B) = 3/4 = 1/2 + 1/2 - 1/4$ and
$P(A^c) = 1/2 = 1 - 1/2$. **"or" ↔ $\cup$, "and" ↔ $\cap$, "not" ↔ complement**: the dictionary between logic
(Module 02), sets (Module 03) and probability.

### 7.1 Mendel's 3 : 1 ratio

For $Aa \times Aa$ the four ordered outcomes of Section 3 are equally likely (each parent passes on either allele with
probability 1/2). The event "the child is affected by a recessive trait" is $\{(a, a)\}$, so $P(aa) = 1/4$. The event
"the child shows the dominant phenotype" is its **complement**, so the complement rule gives $P = 1 - 1/4 = 3/4$ —
Mendel's 3 : 1 ratio.

### 7.2 Mendel's 9 : 3 : 3 : 1 ratio

For the dihybrid cross $YyRr \times YyRr$ the 16 gamete pairs are equally likely (genes on different chromosomes
assort independently — Mendel's second law, which [Module 17 · 🔀 Conditional probability, independence and Bayes' rule](17_conditional_probability_bayes.ipynb) formalises as *independence*). Let $Y$ = "yellow seeds" (at
least one $Y$ allele) and $R$ = "round seeds" (at least one $R$ allele).

````python
# ── 7.2 The dihybrid cross: events, exact probabilities and the 9 : 3 : 3 : 1 ratio ──
def phenotype(mother_gamete: str, father_gamete: str) -> str:
    """Seed phenotype of an offspring: yellow if at least one Y, round if at least one R (Y and R dominant)."""
    alleles = mother_gamete + father_gamete
    return ("yellow" if "Y" in alleles else "green") + "/" + ("round" if "R" in alleles else "wrinkled")


offspring = {(m, f): phenotype(m, f) for m, f in Omega_dihybrid}       # 16 equally likely outcomes
Omega16 = set(offspring)
P16 = lambda e: classical_probability(e, Omega16)
counts = pd.Series(offspring).value_counts()
print(counts.to_string(), "\n→ ratio", " : ".join(str(c) for c in counts.values))

E_yellow = {w for w in Omega16 if offspring[w].startswith("yellow")}
E_round = {w for w in Omega16 if offspring[w].endswith("round")}
print("\nP(yellow) =", P16(E_yellow), "  P(round) =", P16(E_round), "  P(yellow ∩ round) =", P16(E_yellow & E_round))
print("P(yellow ∪ round) =", P16(E_yellow | E_round), "= 3/4 + 3/4 − 9/16 =", Fraction(3, 4) + Fraction(3, 4) - Fraction(9, 16))
print("P(green and wrinkled) = P((yellow ∪ round)ᶜ) =", P16(Omega16 - (E_yellow | E_round)), "= 1 − 15/16")
````

🔍 **What just happened?** Enumerating the 16 equally likely gamete pairs gives 9 yellow/round, 3 yellow/wrinkled,
3 green/round and 1 green/wrinkled — **Mendel's 9 : 3 : 3 : 1**, derived purely from counting outcomes. The set
operations answer every phenotype question: $P(\text{yellow}) = P(\text{round}) = 3/4$, $P(\text{yellow} \cap
\text{round}) = 9/16$, $P(\text{yellow} \cup \text{round}) = 15/16$ by inclusion–exclusion, and the double-recessive
green-wrinkled class is its complement, $1 - 15/16 = 1/16$. (Mendel observed 315 : 101 : 108 : 32 among 556 seeds —
close to, but not exactly, 9 : 3 : 3 : 1. Whether the deviation is just chance is a question for hypothesis testing,
[Module 26 · 📊 Categorical data and multiple testing](26_categorical_tests_multiple_testing.ipynb).)

## 8 · Continuous sample spaces

> 📍 **Where we are.** So far $\Omega$ was finite. Many biological quantities — a position along a chromosome, a
> waiting time, a concentration — can take *any* value in an interval. The axioms still apply, but some intuitions
> must change. (Integrals from [Module 14 · 📈 Calculus: derivatives, integrals and rates of change](14_calculus_derivatives_integrals.ipynb) appear briefly.)

A quantity that can take any value in an interval of real numbers has an **uncountably infinite** sample space —
there are too many outcomes to list them even in an infinite sequence (unlike $\mathbb{N}$):

| | discrete (countable) | continuous (uncountable) |
|---|---|---|
| examples of $\Omega$ | $\{1, 2, 3, 4, 5, 6\}$, $\mathbb{N}$ | $[0, 1]$, $\mathbb{R}$ |
| typical event space | the power set $\mathcal{P}(\Omega)$ | the **Borel field** $\mathcal{B}(\Omega)$ |
| probability of one outcome | can be positive ($P(\{6\}) = 1/6$) | always 0 |

(There are also **mixed** spaces such as $\Omega = [0, 1] \cup \{2\}$ — see Section 8.2.)

Suppose $\Omega = [0, 1]$ and the outcome is "a point chosen at random" — any real number between 0 and 1, with no
region preferred. The natural probability measure is **length**:

$$P([a, b]) = b - a \qquad (0 \le a \le b \le 1).$$

*In words:* the probability that the point lands in an interval equals the interval's length. So $P(0.3 \le X \le
0.4) = 0.1 > 0$, but a single point is an interval of length zero: $P(X = 0.5) = P([0.5, 0.5]) = 0$. **We cannot
assign positive probability to a single value**: with uncountably many values, even a tiny positive probability for
each would make the total infinite, violating $P(\Omega) = 1$.

````python
# ── 8.1 Ω = [0, 1]: single points have probability 0, intervals do not (SIMULATED) ──
rng = np.random.default_rng(SEED + 81)
x = rng.random(1_000_000)                                       # one million random points in [0, 1)
print("fraction exactly equal to 0.5:      ", np.mean(x == 0.5))
print("fraction in [0.3, 0.4]:             ", round(np.mean((x >= 0.3) & (x <= 0.4)), 4), "(length 0.1)")
for half_width in [0.1, 0.01, 0.001, 0.0001]:
    frac = np.mean(np.abs(x - 0.5) <= half_width)
    print(f"fraction within ±{half_width:<6} of 0.5: {frac:.5f}   (length {2 * half_width})")
````

🔍 **What just happened?** Among a million random points not a single one equals 0.5 exactly, while about 10 % fall in
$[0.3, 0.4]$. Shrinking the window around 0.5 shrinks the fraction in proportion to the window's length: about 0.2,
0.02, 0.002, 0.0002. In the limit of zero width the probability is zero. (Strictly, a computer draws from a huge but
*finite* set of floating-point numbers, so exact hits are possible in principle; with $2^{53}$ possible values they
essentially never happen — mirroring the mathematics.)

**Why a new event space?** For discrete sample spaces every subset is measurable, so the power set works. For an
uncountable set such as $[0, 1]$ mathematicians proved (G. Vitali, 1905) that the power set contains bizarre
**non-measurable sets** to which *no* length can be assigned consistently with the axioms. The fix is to use a smaller
event space, the **Borel field** $\mathcal{B}$: all sets that can be built from intervals by complements and countable
unions (and hence intersections). Every set you will ever meet in practice — intervals, unions of intervals, single
points, the rational numbers — is a Borel set, so this is a technicality (the details are beyond the scope of this
course). One surprising consequence of countable additivity is worth knowing: since each single point has probability
0, any *countable* set of points has probability 0 — the probability that a random point of $[0, 1]$ is a rational
number is 0.

### 8.1 Biology: where will a mutation fall?

Treat the position of a new mutation in a 10-kb genomic region as a point chosen uniformly along it (a continuous
approximation of 10,000 discrete bases; illustrative gene structure). The probability that it hits an exon is the total
exon length divided by 10 kb.

````python
# ── 8.2 Mutation positions along a 10-kb region with three exons (SIMULATED, illustrative lengths) ──
region_length = 10_000                                           # bp
exons = [(1_200, 1_700), (4_000, 4_300), (7_100, 7_500)]         # start, end (bp)
exon_length = sum(b - a for a, b in exons)
rng = np.random.default_rng(SEED + 82)
positions = rng.uniform(0, region_length, 5_000)                 # 5,000 independent mutations
in_exon = np.zeros(positions.size, dtype=bool)
for a, b in exons:
    in_exon |= (positions >= a) & (positions <= b)               # union of the three interval events
print(f"P(exonic) = total exon length / region length = {exon_length} / {region_length} = {exon_length / region_length}")
print(f"simulated fraction of 5,000 mutations that hit an exon: {in_exon.mean():.4f}")
fig, ax = plt.subplots(figsize=(9, 2.2))
ax.plot([0, region_length], [0, 0], color="tab:blue", lw=3)
for a, b in exons:
    ax.add_patch(plt.Rectangle((a, -0.25), b - a, 0.5, color="tab:red"))
ax.scatter(positions[:150], np.full(150, 0.6), marker="v", s=12, c=np.where(in_exon[:150], "tab:red", "0.5"))
ax.set_xlim(0, region_length); ax.set_ylim(-0.5, 0.9); ax.set_yticks([])
ax.set_xlabel("position (bp)"); ax.set_title("first 150 mutations (▼, red = exonic); exons in red")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The three exons cover $500 + 300 + 400 = 1{,}200$ bp of 10,000, so the length rule gives
$P(\text{exonic}) = 0.12$, and the 5,000 simulated mutations hit exons about 12 % of the time. The event "exonic" is a
**union of three disjoint intervals**, and by Axiom 3 its probability is the sum of their lengths divided by the total.
(Real mutations are not uniform — CpG sites mutate about ten times faster than other bases — so this is a first model,
not the truth.)

### 8.2 Mixed sample spaces

Some measurements are partly discrete and partly continuous. A normalised single-cell expression value is often
**exactly 0** (the gene's transcripts were not detected in that cell) and otherwise a positive real number. Its sample
space is $\{0\} \cup (0, \infty)$, a **mixed space** like $[0, 1] \cup \{2\}$: the single outcome 0 carries a positive
probability, while every single positive value has probability 0.

````python
# ── 8.3 A mixed sample space: expression values with exact zeros (SIMULATED) ──
rng = np.random.default_rng(SEED + 83)
n_cells = 10_000
detected = rng.random(n_cells) < 0.4                            # transcripts detected in 40 % of cells
expression = np.where(detected, rng.lognormal(mean=0.5, sigma=0.6, size=n_cells), 0.0)
print("P(X = 0) ≈", np.mean(expression == 0.0), " ← one outcome with positive probability")
print("P(X = 1.2345) ≈", np.mean(expression == 1.2345), " ← a single positive value: probability 0")
print("P(1 ≤ X ≤ 2) ≈", round(np.mean((expression >= 1) & (expression <= 2)), 4), " ← an interval: positive")
````

🔍 **What just happened?** About 60 % of the simulated cells have expression exactly 0 — a single outcome with
probability ≈ 0.6 — while no cell has expression exactly 1.2345, and an interval such as $[1, 2]$ has probability of
about 0.17. Neither a purely discrete nor a purely continuous description fits such data, which is why single-cell
models use "zero-inflated" or count-based distributions ([BiologyToML 30 · Single-cell RNA-seq](../../../../ml/tutorial/BiologyToML/notebooks/30_single_cell_rnaseq.ipynb)).

### 📘 Lecture notes — continuous sample spaces

* A **continuous** random quantity can take any value in an interval of real numbers; its sample space is
  **uncountably infinite**. Discrete: $\Omega = \{1, \ldots, 6\}$, $\mathbb{N}$; continuous: $\Omega = [0, 1]$, $\mathbb{R}$;
  mixed: $\Omega = [0, 1] \cup \{2\}$.
* Discrete sample spaces: every subset is measurable, $\mathcal{A} = \mathcal{P}(\Omega)$ (power set). Uncountable sets:
  the power set contains **non-measurable** sets, so we use the **Borel field** $\mathcal{A} = \mathcal{B}(\Omega)$,
  generated by intervals.
* For $\Omega = [0, 1]$ with the uniform ("length") measure: $P(X = 0.5) = 0$ but $P(0.3 \le X \le 0.4) = 0.1 > 0$.
  **We cannot assign positive probability to a single value** of a continuous quantity; probabilities live on intervals
  (and are computed as areas under a density, [Module 18 · 🎰 Random variables and their distributions](18_random_variables_distributions.ipynb)).

## 9 · Probabilities from simulation

> 📍 **Where we are.** Throughout this notebook we checked exact probabilities against simulated frequencies. This
> section looks at that check itself: how close is a relative frequency to the probability it estimates, and how many
> trials do we need?

If an experiment is repeated $n$ times independently, the **relative frequency** of an event,

$$\hat{P}_n(A) = \frac{n_A}{n} = \frac{\text{number of repetitions in which } A \text{ occurred}}{n},$$

*In words:* the fraction of the simulated (or real) repetitions in which $A$ happened — the frequentist definition of
Section 2 without the limit. The hat $\hat{\ }$ marks an **estimate**. A relative frequency satisfies the three axioms
itself (it is non-negative, $\hat{P}_n(\Omega) = 1$, and counts of disjoint events add), so it is a probability measure
on its own — the *empirical* one.

````python
# ── 9.1 Exact probabilities vs relative frequencies for growing n (SIMULATED) ──
rng = np.random.default_rng(SEED + 91)
experiments = {
    "die: even or ≥ 4": (lambda n: rng.integers(1, 7, n), lambda r: (r % 2 == 0) | (r >= 4), Fraction(2, 3)),
    "Aa × Aa: child aa": (lambda n: rng.integers(0, 2, (n, 2)).sum(axis=1), lambda r: r == 2, Fraction(1, 4)),
    "YyRr × YyRr: green wrinkled": (lambda n: rng.integers(0, 4, (n, 2)), lambda r: (r == 3).all(axis=1), Fraction(1, 16)),
}
rows = []
for name, (simulate, in_event, exact) in experiments.items():
    row = {"event": name, "exact": f"{exact} = {float(exact):.4f}"}
    for n in [100, 1_000, 10_000, 100_000]:
        row[f"n = {n:,}"] = round(float(np.mean(in_event(simulate(n)))), 4)
    rows.append(row)
print(pd.DataFrame(rows).set_index("event").to_string())
````

🔍 **What just happened?** Each experiment is simulated in its simplest form: a die roll; two parents each passing
allele $a$ (coded 1) with probability 1/2, so the child is $aa$ when both pass it; and two dihybrid gametes coded
0–3 (code 3 = $yr$). At $n = 100$ and $n = 1{,}000$ the relative frequencies still wander (0.72 instead of 0.667 for the die; 0.218
instead of 0.25 for $aa$ even with 1,000 crosses); at $n = 100{,}000$ all three are within 0.002 of the exact values
(0.6670, 0.2522, 0.0644). Notice that the rare event (1/16) needs more trials for
the same *relative* accuracy.

How fast does the error shrink? Repeating each simulation many times and averaging the absolute error gives the answer.

````python
# ── 9.2 The error of a relative frequency shrinks like 1/√n (SIMULATED) ──
rng = np.random.default_rng(SEED + 92)
p_true = 1 / 16                                                    # green wrinkled
ns = np.unique(np.logspace(1, 5, 25).astype(int))
mean_abs_error = [np.mean(np.abs(rng.binomial(n, p_true, size=2000) / n - p_true)) for n in ns]
fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.loglog(ns, mean_abs_error, "o", ms=4, label="mean |relative frequency − 1/16| (2,000 repeats)")
ax.loglog(ns, np.sqrt(2 / np.pi) * np.sqrt(p_true * (1 - p_true) / ns), "r-", label="theory: 0.8 · √(p(1−p)/n)")
ax.set_xlabel("number of trials n"); ax.set_ylabel("typical error"); ax.legend(fontsize=7)
ax.set_title("ten times more precision costs a hundred times more trials")
plt.tight_layout(); plt.show()
for n in [100, 10_000]:
    k = np.argmin(np.abs(ns - n))
    print(f"n ≈ {ns[k]:>6,}: typical error {mean_abs_error[k]:.4f}")
````

🔍 **What just happened?** `rng.binomial(n, p, size=2000)` simulates 2,000 independent experiments of $n$ trials at
once and returns the number of successes in each (the binomial distribution of [Module 18 · 🎰 Random variables and their distributions](18_random_variables_distributions.ipynb)). On log–log axes the typical
error falls on a straight line of slope $-1/2$: it is proportional to $1/\sqrt{n}$, and the red theory curve
$\sqrt{2/\pi}\sqrt{p(1-p)/n}$ matches the points. From $n \approx 100$ to $n \approx 10{,}000$ the typical error drops by
a factor of about 10 (from about 0.019 to about 0.0019). The $1/\sqrt{n}$ law — the **standard error** — governs every
estimate in Part F ([Module 22 · 📏 Sampling: the law of large numbers, the CLT and confidence intervals](22_sampling_lln_clt_confidence.ipynb)).

Finally, the same idea on a simulated population: genotypes in Hardy–Weinberg proportions with allele frequency
$p = 0.3$, for which the probability of the homozygous genotype "2" (two copies of the allele) is $p^2 = 0.09$.

````python
# ── 9.3 Genotype probabilities estimated from a SIMULATED population ──
p_allele = 0.3
for n in [50, 500, 5_000, 50_000]:
    g = simulate_genotypes(n, p_allele, seed=1866)
    freq = np.bincount(g, minlength=3) / n
    print(f"n = {n:6,}: relative frequencies of genotypes 0/1/2 = {np.round(freq, 4)}")
print("Hardy–Weinberg probabilities (1−p)², 2p(1−p), p² =", np.round([(1 - p_allele) ** 2, 2 * p_allele * (1 - p_allele), p_allele ** 2], 4))
````

🔍 **What just happened?** `simulate_genotypes` (from the toolkit) draws each person's genotype with the
Hardy–Weinberg probabilities $(0.49, 0.42, 0.09)$ — [BiologyToML 02 · Probability for biologists](../../../../ml/tutorial/BiologyToML/notebooks/02_probability_for_biologists.ipynb) derives them from two independent allele draws. With 50
people the estimates are far off (0.34 instead of 0.49 for genotype 0, 0.18 instead of 0.09 for genotype 2); the
500-person sample happens to hit the probabilities exactly, while the 5,000-person sample is off by 0.007 for
genotype 2 — luck goes both ways, and only the *typical* error shrinks like $1/\sqrt{n}$. With 50,000 people every
frequency is within 0.005 of its probability. Every genotype or allele frequency reported in a genetics paper is such a relative frequency,
and its precision depends on the sample size exactly as in cell 9.2.

### 🌉 Bridge — from probability to statistics and to Monte Carlo methods

This module ran probability "forwards": given $P$, predict frequencies. **Statistics** runs it backwards: given
observed frequencies, infer $P$. A genotype frequency in a cohort, the fraction of cells expressing a marker, a
response rate in a trial — each is a relative frequency $\hat{P}_n$ whose error shrinks like $1/\sqrt{n}$ (cell 9.2).
Part F of this course ([Module 21 · ⚖️ Expectation, variance and moments](21_expectation_variance_moments.ipynb)–[Module 24 · 🧭 Bayesian estimation and maximum a posteriori](24_bayesian_estimation_map.ipynb)) turns that observation into estimators, confidence intervals and Bayesian
posteriors; [BiologyToML 02 · Probability for biologists](../../../../ml/tutorial/BiologyToML/notebooks/02_probability_for_biologists.ipynb) and [BiologyToML 05 · Likelihood, MLE and MAP](../../../../ml/tutorial/BiologyToML/notebooks/05_likelihood_mle_map.ipynb) use it for machine learning.

Simulation is also a research tool in its own right: when a probability cannot be computed by hand — the site-frequency
spectrum of Section 1, the null distribution of a test statistic, a posterior distribution — we estimate it from many

([BiologyToResearch (cryo-EM) 04 · Noise, statistics and inference](../../../../cryoem/tutorial/BiologyToResearch/notebooks/04_noise_statistics_inference.ipynb)) are built on that idea.

## 📌 Key takeaways

- Biology is uncertain **across samples, genes, cells and genomes**: mutation and recombination are random events,
  most variants are rare (SFS $\propto 1/i$), and even gene finding is a probabilistic inference problem (HMMs).
- A probability can be read as a **long-run frequency** (frequentist) or a **degree of belief** (Bayesian); both obey
  the same axioms, and calibrated beliefs behave like frequencies.
- A **probability space** $(\Omega, \mathcal{A}, P)$: the **sample space** of all mutually exclusive outcomes (e.g. a
  Cartesian product such as a Punnett square), the **event space** of subsets we may ask about (the power set for
  discrete $\Omega$; closed under complement and countable union; the Borel field for continuous $\Omega$), and the
  **probability measure**.
- **Kolmogorov's axioms:** $P(A) \ge 0$; $P(\Omega) = 1$; probabilities of **disjoint** events add.
- **Consequences (proved):** $P(\emptyset) = 0$; $P(\overline{A}) = 1 - P(A)$; $0 \le P(A) \le 1$; $A \subseteq B
  \Rightarrow P(A) \le P(B)$; $P(A \cup B) = P(A) + P(B) - P(A \cap B)$ — so $P(A \cup B) = P(A) + P(B)$ only for
  disjoint events.
- **Counting** $|A|/|\Omega|$ works only for **equally likely** outcomes — Mendel's 3 : 1 and 9 : 3 : 3 : 1 follow from
  counting ordered gamete pairs.
- On a **continuous** sample space single outcomes have probability 0 and intervals carry the probability (length,
  later area under a density).
- Relative frequencies estimate probabilities with an error proportional to $1/\sqrt{n}$.

4. Prove $P(\emptyset) = 0$ from the axioms, naming each axiom you use.
   <details><summary>Answer</summary>Ω ∪ ∅ = Ω and Ω ∩ ∅ = ∅, so by Axiom 3 P(Ω) = P(Ω ∪ ∅) = P(Ω) + P(∅). By Axiom 2 this reads 1 = 1 + P(∅), hence P(∅) = 0.</details>
5. Prove that $A \subseteq B$ implies $P(A) \le P(B)$.
   <details><summary>Answer</summary>B = A ∪ (B \ A) with A and B \ A disjoint, so P(B) = P(A) + P(B \ A) (Axiom 3) ≥ P(A) because P(B \ A) ≥ 0 (Axiom 1).</details>
6. A die: $A$ = "odd" $= \{1, 3, 5\}$, $B$ = "$\le 2$" $= \{1, 2\}$. Compute $P(A \cup B)$ by inclusion–exclusion.
   <details><summary>Answer</summary>P(A) = 3/6, P(B) = 2/6, A ∩ B = {1} so P(A ∩ B) = 1/6; P(A ∪ B) = 3/6 + 2/6 − 1/6 = 4/6 = 2/3 (indeed A ∪ B = {1, 2, 3, 5}).</details>
7. In $Aa \times Aa$, why is $P(aa) = 1/4$ and not $1/3$?
   <details><summary>Answer</summary>The three genotypes AA, Aa, aa are not equally likely. The equally likely outcomes are the four ordered gamete pairs (A,A), (A,a), (a,A), (a,a); only (a,a) gives aa, so P = 1/4 (and Aa has probability 2/4).</details>
8. In the dihybrid cross, what is $P(\text{green} \cup \text{wrinkled})$?
