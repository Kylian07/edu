## 🧫 The biological question

In 1961 François Jacob and Jacques Monod explained how the bacterium *Escherichia coli* decides whether to make the
enzymes that digest lactose. Their answer, stripped to its skeleton, is a **logical rule**: the *lac* genes are
transcribed strongly **if and only if** lactose is present **and** glucose is **not**. Lactose (through allolactose)
removes the Lac repressor from the DNA; low glucose raises cAMP, which lets the activator protein CAP switch
transcription on. Two inputs, three logical words — *and*, *not*, *if and only if* — and the behaviour of the cell is
captured. Much of biology is written this way, even when nobody writes the symbols down:

| Biological question | Logical tool | Where in this module |
|---|---|---|
| When does *E. coli* transcribe its *lac* genes? | a Boolean formula with ∧, ¬ and ⇔ | Sections 1, 3, 10 |
| Which cells in a blood sample are helper T cells? | a conjunction of marker propositions (a "gate") | Section 2 |
| "Gene X is necessary for phenotype Y" — which experiment tests that? | implication and its contrapositive | Sections 4–5 |
| How do I correctly negate "the cell is a T cell or a B cell"? | De Morgan's laws | Section 7 |
| "All tumours carry a *TP53* mutation" — how many tumours refute it? | ∀, ∃ and counterexamples | Section 8 |
| "Every row of the plate has a hit" vs "some column is a hit in every row" | nested quantifiers | Section 9 |
| Does a loop of three repressors oscillate? | a Boolean network and its attractors | Section 10 |

Logic is also the hidden grammar of the rest of 02-680: sets are built from predicates (Module 03), probability

They supply the three connectives Python has no operator for (`implies` for ⇒, `iff` for ⇔, `xor` for ⊕), print truth tables (`truth_table`), and draw Venn diagrams (`draw_venn`) and small directed graphs (`draw_graph` with its helper `circle_layout`, used for the gene network of Section 10).
````python
# ── Course toolkit (identical to ../course_toolkit.py v1.0; copied so this notebook runs alone) ──
import itertools

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def implies(p: bool, q: bool) -> bool:
    """Material implication p ⇒ q: false only when p is true and q is false."""
    return (not p) or q


def iff(p: bool, q: bool) -> bool:
    """Biconditional p ⇔ q: true when p and q have the same truth value."""
    return p == q


def xor(p: bool, q: bool) -> bool:
    """Exclusive or p ⊕ q: true when exactly one of p, q is true."""
    return p != q


def truth_table(formulas: dict, variables: list, letters: bool = True) -> pd.DataFrame:
    """Truth table of one or more Boolean formulas.

    ``variables`` is a list of names such as ``["p", "q"]``; ``formulas`` maps a column
    title to a Python function of those variables, e.g. ``{"p ⇒ q": lambda p, q: implies(p, q)}``.
    Rows run through every assignment in the textbook order (T…T first, F…F last).
    With ``letters=True`` the table shows "T"/"F"; otherwise it holds Python booleans.
    """
    rows = []
    for values in itertools.product([True, False], repeat=len(variables)):
        env = dict(zip(variables, values))
        row = dict(env)
        for title, f in formulas.items():
            row[title] = bool(f(**env))
        rows.append(row)
    table = pd.DataFrame(rows)
    if letters:
        table = table.replace({True: "T", False: "F"})
    return table


def _venn_order(x):
    """Sort key for Venn-diagram labels: numbers in numeric order first, then everything else as text."""
    if isinstance(x, (int, float, np.integer, np.floating)) and not isinstance(x, bool):
        return (0, float(x), "")
    return (1, 0.0, str(x))


def draw_venn(sets: dict, universe=None, ax=None, show_elements: bool = True, max_items: int = 6,
              title: str | None = None, highlight=None):
    """Draw a Venn diagram of two or three sets with matplotlib only.

    ``sets`` maps a label to a Python set, e.g. ``{"S": {1, 2, 3}, "T": {3, 4}}``.  Each region
    shows its elements (or ``n = …`` if it has more than ``max_items``).  If ``universe`` is given,
    a rectangle is drawn and the elements outside every set are listed in a corner.
    ``highlight`` is a function of the membership tuple, e.g. ``lambda s, t: s and not t`` for
    S \\ T; the regions where it is true are shaded.  Returns the Axes.
    """
    labels = list(sets)
    if len(labels) not in (2, 3):
        raise ValueError("draw_venn draws two or three sets")
    if ax is None:
        _, ax = plt.subplots(figsize=(4.6, 3.8))
    if len(labels) == 2:
        centres = [(-0.55, 0.0), (0.55, 0.0)]
    else:
        centres = [(-0.55, 0.32), (0.55, 0.32), (0.0, -0.6)]
    radius = 1.0
    colours = ["tab:blue", "tab:orange", "tab:green"]
    # classify a fine grid of points to find each region (used for shading and label positions)
    gx, gy = np.meshgrid(np.linspace(-2.2, 2.2, 441), np.linspace(-2.0, 1.9, 391))
    inside = [((gx - cx) ** 2 + (gy - cy) ** 2) <= radius ** 2 for cx, cy in centres]
    if highlight is not None:
        mask = np.zeros_like(gx, dtype=bool)
        for member in itertools.product([True, False], repeat=len(labels)):
            if any(member) and highlight(*member):
                region = np.ones_like(gx, dtype=bool)
                for k, m in enumerate(member):
                    region &= inside[k] if m else ~inside[k]
                mask |= region
        ax.contourf(gx, gy, mask.astype(float), levels=[0.5, 1.5], colors=["#ffd54f"], alpha=0.8)
    for (cx, cy), lab, col in zip(centres, labels, colours):
        ax.add_patch(plt.Circle((cx, cy), radius, fill=False, lw=2, color=col))
        lx = cx + (-1.0 if cx < 0 else 1.0 if cx > 0 else 0.0) * 0.95
        ly = cy + (0.95 if cy >= 0 else -1.15)
        ax.text(lx, ly, lab, ha="center", va="center", fontsize=11, fontweight="bold", color=col)
    elements = set().union(*sets.values())
    for member in itertools.product([True, False], repeat=len(labels)):
        if not any(member):
            continue
        region_items = [x for x in elements if all((x in sets[l]) == m for l, m in zip(labels, member))]
        region = np.ones_like(gx, dtype=bool)
        for k, m in enumerate(member):
            region &= inside[k] if m else ~inside[k]
        if not region.any():
            continue
        px, py = gx[region].mean(), gy[region].mean()
        listing = ", ".join(str(x) for x in sorted(region_items, key=_venn_order))
        if show_elements and len(region_items) <= max_items and len(listing) <= 16:
            text = listing if region_items else "∅"
        else:
            text = f"n = {len(region_items)}"
        ax.text(px, py, text, ha="center", va="center", fontsize=8, wrap=True)
    if universe is not None:
        ax.add_patch(plt.Rectangle((-2.1, -1.9), 4.2, 3.7, fill=False, lw=1, color="0.4"))
        outside = sorted(set(universe) - elements, key=_venn_order)
        text = ", ".join(map(str, outside)) if len(outside) <= max_items else f"n = {len(outside)}"
        ax.text(-2.0, -1.8, "U: " + (text if outside else "∅"), fontsize=8, va="bottom", color="0.3")
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.0, 1.9)
    ax.set_aspect("equal")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax


def circle_layout(nodes: list) -> dict:
    """Place nodes evenly on a circle (first node at the top, clockwise)."""
    n = len(nodes)
    angles = np.pi / 2 - 2 * np.pi * np.arange(n) / max(n, 1)
    return {v: (float(np.cos(a)), float(np.sin(a))) for v, a in zip(nodes, angles)}


def draw_graph(edges: list, nodes=None, directed: bool = False, pos=None, ax=None, node_colors=None,
               edge_labels=None, title: str | None = None, font_size: int = 9):
    """Draw a small graph with matplotlib only (no extra packages needed).

    ``edges`` is a list of (u, v) pairs; ``nodes`` (optional) adds isolated nodes and fixes the
    order; ``pos`` maps node → (x, y) (default: ``circle_layout``).  Directed edges get arrows,
    self-loops are drawn as small rings above the node, ``edge_labels`` maps (u, v) → text (e.g. a
    weight) and ``node_colors`` maps node → colour.  Node circles grow with the length of their
    label.  Returns the Axes.
    """
    from matplotlib.transforms import offset_copy

    if nodes is None:
        nodes = []
        for u, v in edges:
            for w in (u, v):
                if w not in nodes:
                    nodes.append(w)
    pos = circle_layout(list(nodes)) if pos is None else pos
    if ax is None:
        _, ax = plt.subplots(figsize=(4.2, 3.6))
    radius = {v: max(0.62 * font_size * len(str(v)) / 2 + 4, 0.9 * font_size) for v in nodes}  # points
    for u, v in edges:
        if u == v:
            ring = offset_copy(ax.transData, fig=ax.figure, x=0, y=1.25 * radius[u], units="points")
            ax.scatter([pos[u][0]], [pos[u][1]], s=(1.3 * radius[u]) ** 2, facecolors="none",
                       edgecolors="0.35", linewidths=1.3, transform=ring, zorder=1)
            continue
        style = "-|>" if directed else "-"
        ax.annotate("", xy=pos[v], xytext=pos[u],
                    arrowprops=dict(arrowstyle=style, color="0.35", lw=1.3, mutation_scale=12,
                                    shrinkA=radius[u], shrinkB=radius[v],
                                    connectionstyle="arc3,rad=0.08" if directed else "arc3,rad=0"))
        if edge_labels and (u, v) in edge_labels:
            (x1, y1), (x2, y2) = pos[u], pos[v]
            ax.text((x1 + x2) / 2, (y1 + y2) / 2, str(edge_labels[(u, v)]), fontsize=font_size - 1,
                    ha="center", va="center", color="darkred", zorder=4,
                    bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85))
    xs = [pos[v][0] for v in nodes]
    ys = [pos[v][1] for v in nodes]
    colours = [(node_colors or {}).get(v, "#cfe2f3") for v in nodes]
    ax.scatter(xs, ys, s=[(2 * radius[v]) ** 2 for v in nodes], c=colours, edgecolors="0.2",
               linewidths=1.2, zorder=2, clip_on=False)
    for v in nodes:
        ax.text(pos[v][0], pos[v][1], str(v), ha="center", va="center", fontsize=font_size, zorder=3,
                clip_on=False)
    ax.margins(0.22)
    ax.set_aspect("equal", adjustable="datalim")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax

print("🧰 toolkit loaded:", ", ".join(sorted(['_venn_order', 'circle_layout', 'draw_graph', 'draw_venn', 'iff', 'implies', 'truth_table', 'xor'])))
````

````python
# ── 0.1 SIMULATED flow-cytometry panel: 2,000 blood cells × 4 surface markers ──
MARKERS = ["CD3", "CD4", "CD8", "CD19"]
# population: (fraction of cells, expected +/− for CD3, CD4, CD8, CD19) — simplified textbook profiles
POPULATIONS = {
    "CD4 T cell": (0.40, (1, 1, 0, 0)),     # helper T cells
    "CD8 T cell": (0.20, (1, 0, 1, 0)),     # cytotoxic T cells
    "B cell":     (0.12, (0, 0, 0, 1)),
    "NK cell":    (0.10, (0, 0, 0, 0)),     # (many real NK cells are CD8-dim; ignored here)
    "monocyte":   (0.18, (0, 1, 0, 0)),     # human monocytes carry a low ("dim") level of CD4
}
THRESHOLD = 2.0                             # log10 intensity above which a marker is called "positive"


def simulate_flow_cells(n_cells: int = 2000, seed: int = SEED) -> pd.DataFrame:
    """SIMULATED log10 fluorescence intensities of four surface markers on n_cells blood cells.

    Positive markers are centred at 3.0, negative ones at 1.0 and the dim CD4 of monocytes at 2.4;
    every measurement gets Gaussian noise (SD 0.35).  The column 'population' is the hidden truth.
    """
    rng_cells = np.random.default_rng(seed)
    names = list(POPULATIONS)
    population = rng_cells.choice(names, size=n_cells, p=[POPULATIONS[k][0] for k in names])
    profile = np.array([POPULATIONS[k][1] for k in population])
    mean = np.where(profile == 1, 3.0, 1.0)
    mean[population == "monocyte", MARKERS.index("CD4")] = 2.4
    intensity = mean + rng_cells.normal(0.0, 0.35, size=mean.shape)
    table = pd.DataFrame(intensity.round(2), columns=MARKERS)
    table.insert(0, "population", population)
    return table


cells = simulate_flow_cells()
print("SIMULATED cells:", cells.shape[0], "| markers:", MARKERS, "| positive means log10 intensity >", THRESHOLD)
print(cells.head(6).to_string())
print("\n🧪 hidden truth — cells per population:")
print(cells["population"].value_counts().to_string())
````

🔍 **What just happened?** Each row is one SIMULATED cell, each marker column a log10 fluorescence intensity: values
near 3 mean "this protein is on the cell surface", values near 1 mean "it is not". The first cell (a B cell) has
CD19 = 2.76 but CD3 = 1.54, CD4 = 0.63 and CD8 = 0.77 — the profile CD3⁻CD4⁻CD8⁻CD19⁺; the next three are CD4 T cells
with CD3 ≈ 2.9–3.3 and CD4 ≈ 2.4–2.7. The monocyte in row 5 has CD4 = 1.73: its "dim" CD4 happened to fall below 2. The hidden
population column tells us the truth: 815 CD4 T cells, 394 CD8 T cells, 371 monocytes, 228 B cells and 192 NK cells.

markers with logic.

## 1 · Propositions: statements that are true or false

> 📍 **Where we are.** Module 00 showed that 02-680 is written in a precise symbolic language and Module 01 introduced
> Python's two Boolean values, `True` and `False`. Logic is where the two meet. It comes first in 02-680 because every
> later topic — set operations, linear algebra proofs, the events of probability, the decisions of a statistical test —
> is phrased in it.

**Why logic?** Logic is the essential building block of mathematical reasoning. It lets us

* **formalise statements** — turn an English sentence or a lab observation into an expression with exactly one meaning;
* **analyse arguments** — decide whether a conclusion really follows from its premises, or hides a fallacy;
* **lay the groundwork** for the rest of the course — sets are built from logical conditions (Module 03), probability
  combines events with *and*, *or*, *not* and *if–then* (Modules 16–17), and interpreting a test or a confidence
  interval means stating precisely what the data do and do not let us conclude (Modules 22 and 25).

Think of logic as the **grammar of mathematics**: grammar decides which sentences are well formed, logic decides which
arguments are valid. Without it, statements become ambiguous, proofs flawed and algorithms wrong.

**Two kinds of logic.**

* **Propositional logic** treats whole statements as indivisible units and names them with letters $p, q, r$. The
  letters are like variables that have not been given values: you cannot know whether "$p$ and $q$" is true until you
  are told whether $p$ is true and whether $q$ is true.
* **Predicate logic** looks *inside* statements. It has variables for objects ("$x$ is an even number", "gene $g$ is
  up-regulated in tissue $t$") and words such as "for all" and "there exists", so that the full statement can be
  evaluated once we say what the variables range over. It extends propositional logic by adding more detail
  (Sections 8–9).

**Definition (proposition).** A *proposition* is a statement that can be evaluated for truth: it is either **true**
or **false**. Examples: "2 + 2 = 4" (true); "a bear is CMU's mascot" (false — the mascot is Scotty, a Scottish
terrier); "the *TP53* gene lies on human chromosome 17" (true). *Not* propositions: questions ("Is it raining?"),
commands ("Pipette 5 µL into well A1") and sentences that contain an unknown, such as "2 + x = 4" — Section 8 deals
with those.

We usually name atomic propositions with lower-case letters, much like variables:

$$p \leftarrow \text{“}2+2=4\text{”}, \qquad q \leftarrow \text{“a bear is CMU's mascot”}.$$

*In words:* the arrow ← means "is the name for": $p$ stands for the statement "2 + 2 = 4" and $q$ for "a bear is CMU's
mascot". We know that $p$ is true and $q$ is false.

````python
# ── 1.1 Propositions are statements with a truth value: Python evaluates them for us ──
p = (2 + 2 == 4)                         # "2 + 2 = 4"
cmu_mascot = "Scottish terrier"          # CMU's mascot, Scotty, is a Scottish terrier
q = (cmu_mascot == "bear")               # "a bear is CMU's mascot"
tp53_chromosome = 17                     # the tumour-suppressor gene TP53 sits on chromosome 17 (band 17p13.1)
r = (tp53_chromosome == 17)              # "TP53 lies on chromosome 17"
print("p =", p, "| q =", q, "| r =", r, "| Python type of p:", type(p).__name__)
print("not q       =", not q, "   ← negating the false q gives a true statement")
print("p and q     =", p and q)
print("p and not q =", p and not q)
````

🔍 **What just happened?** A comparison such as `2 + 2 == 4` is Python's way of *evaluating a proposition*: it returns
a value of type `bool`, `True` or `False` (note the double `==`, "is equal to", versus the single `=` that stores a
value under a name — the Python mirror of our ←). We got $p$ = `True`, $q$ = `False`, $r$ = `True`. Negating $q$ gives
the true statement "a bear is **not** CMU's mascot", written $\lnot q$. Combining $p$ and $q$ with *and* gives
`False`, because *and* needs both parts to be true; $p$ and $\lnot q$ gives `True`. These are our first
**compound propositions**.

### 1.2 Atomic and compound propositions

* An **atomic proposition** is conceptually indivisible: it cannot be broken into simpler statements. Atomic
  propositions are also called **Boolean variables** (after the English mathematician George Boole, 1815–1864).
* A **compound proposition** is formed by combining simpler propositions with **logical connectives** (*not*, *and*,
  *or*, …). A compound proposition that contains the Boolean variables $p_1, p_2, \dots, p_k$ is called a
  **Boolean expression** or **Boolean formula over** $p_1, \dots, p_k$.

| Statement | Atomic or compound? | Comment |
|---|---|---|
| The University of Minnesota's mascot is the Badger. | atomic | one indivisible claim (a false one: Minnesota's mascot is a gopher; the badger belongs to Wisconsin) |
| The University of Washington's mascot is the Duck **or** the University of Oregon's mascot is the Duck. | compound | two claims joined by *or*; true, because Oregon's mascot is the Duck |
| The sun rises in the east. | atomic | a single simple claim |
| It is raining **or** it is snowing. | compound | "it is raining" ∨ "it is snowing" |
| Lactose is present. | atomic | |
| Lactose is present **and** glucose is **not** present. | compound | "lactose present" ∧ ¬ "glucose present" |

Being atomic or compound is a matter of **structure**, not of truth: atomic propositions can be false, compound ones
true, and vice versa. In code, a Boolean formula is simply a function that takes truth values and returns a truth value.

````python
# ── 1.2 A Boolean formula is a rule that turns truth values into a truth value ──
def lac_genes_on(lactose: bool, glucose: bool) -> bool:
    """Simplified lac-operon rule: transcribed when lactose is present AND glucose is NOT present."""
    return lactose and not glucose


for lactose, glucose in itertools.product([True, False], repeat=2):      # every combination of the two inputs
    print(f"lactose = {lactose!s:5}  glucose = {glucose!s:5}  →  lac genes on: {lac_genes_on(lactose, glucose)}")
````

🔍 **What just happened?** `lac_genes_on` is a Boolean formula over two Boolean variables, written as a Python function.
`itertools.product([True, False], repeat=2)` produced all four combinations of inputs, and the rule switched the genes
on in exactly one of them — lactose present and glucose absent — the condition under which a bacterium growing on a
mixture of sugars turns to lactose only after the glucose is used up (the "diauxic shift" studied by Monod). Listing
*every* combination of inputs and the resulting output is called a **truth table**; it is the main tool of the next
section.

## 2 · The connectives ¬, ∧, ∨ and truth tables

> 📍 **Where we are.** We have propositions and have seen that they can be combined. Here we define the combining
> words — the **logical connectives** — precisely, by listing their output for every input. These definitions never
> change in the rest of the course, and the same three words return as complement, intersection and union of sets
> (Module 03) and of events (Module 16).

**Logical connectives** are the links that combine simple propositions into compound ones — the verbs and
conjunctions of the logical language. The first three:

* **Negation** [*not*, ¬]. $\lnot p$ ("not $p$") is true when $p$ is false and false when $p$ is true. It acts on a
  single proposition, so it is a **unary** operator — like the minus sign that turns 5 into −5. If $p$ = "the sky is
  blue" is true, $\lnot p$ = "the sky is not blue" is false; if $p$ = "the moon is made of cheese" is false, then
  $\lnot p$ is true.
* **Conjunction** [*and*, ∧]. $p \land q$ ("$p$ and $q$") is true **only if both** $p$ and $q$ are true; it is false
  if either one or both are false. "I will give you an A **and** a million dollars" is kept only if you get both.
* **Disjunction** [*or*, ∨]. $p \lor q$ ("$p$ or $q$") is true if **at least one** of $p$, $q$ is true; it is false
  only if both are false. This is the **inclusive** or — "$p$ or $q$, or both": "you pass if you do the homework or
  pass the final" is satisfied by doing both.

Example (two true atomic propositions): let $p$ = "Ohio State's mascot is the Buckeye" and $q$ = "Michigan's mascot is
the Wolverine". Then $\lnot q$ = "Michigan's mascot is not the Wolverine" (false), $p \land q$ = "Ohio State's mascot
is the Buckeye and Michigan's mascot is the Wolverine" (true), and $p \lor q$ (true).

**Definition (truth table).** A *truth table* shows all possible truth assignments for a proposition. Each **row**
corresponds to one truth assignment (one choice of T or F for every atomic proposition); the **last column** gives the
truth value of the whole proposition under that assignment. The truth table of a connective *is* its definition.

🧮 **Math refresher — why a truth table has $2^k$ rows.** Each Boolean variable has two possible values. With two
variables every value of $p$ can be paired with every value of $q$: 2 × 2 = 4 rows. Choices multiply, so $k$ variables
give $2 \times 2 \times \dots \times 2 = 2^k$ rows: 2, 4, 8, 16, … for $k$ = 1, 2, 3, 4. We always list the rows in the
textbook order TT, TF, FT, FF — T before F, the left column changing slowest — so that tables can be compared row by
row.

````python
# ── 2.1 The truth tables that *define* ¬, ∧ and ∨ ──
print(truth_table({"¬p": lambda p: not p}, ["p"]).to_string(index=False))
print()
basic = truth_table({"p ∧ q": lambda p, q: p and q,
                     "p ∨ q": lambda p, q: p or q}, ["p", "q"])
print(basic.to_string(index=False))

# the rows come from itertools.product, which lists every assignment in the order T…T → F…F
print("\nassignments of (p, q):", list(itertools.product([True, False], repeat=2)))
for k in (1, 2, 3, 4):
    n_rows = len(list(itertools.product([True, False], repeat=k)))
    print(f"k = {k} variable(s) → {n_rows:2d} rows = 2**{k}")
````

🔍 **What just happened?** The toolkit's `truth_table` evaluated each formula on every assignment and printed T/F.
Read the $p \land q$ column the way a textbook defines "and": columns $p$ and $q$ hold the atomic propositions, there
is one row for each possible truth assignment — TT, TF, FT, FF — and the third column has a **T only in the first
row**: the truth value of $p \land q$ is false unless both $p$ and $q$ are true. The $p \lor q$ column is the mirror
image: an **F only in the last row**. The counts 2, 4, 8, 16 confirm the $2^k$ rule — a formula in 10 variables would
need 1,024 rows, which is why we let the computer write them.

### 2.2 Gating cells is Boolean algebra

A flow cytometer shines lasers on cells flowing one by one past a detector and measures fluorescent antibodies bound to
surface proteins ("CD" = *cluster of differentiation*). For every cell and marker, "the cell is CD4-positive" (CD4⁺)
is a **proposition**: true if the CD4 signal exceeds a threshold. A **gate** — the immunologist's rule for selecting a
cell population — is then a **Boolean formula** over these propositions. Helper T cells, for example, are

$$\text{helper T} \;=\; \text{CD3}^+ \land \text{CD4}^+ \land \lnot\,\text{CD8}^+ .$$

*In words:* a cell is called a helper T cell when it carries CD3 (the T-cell receptor complex) **and** CD4 **and not**
CD8. In NumPy and pandas the connectives that act on whole columns, element by element, are `&` (∧), `|` (∨) and `~` (¬).

````python
# ── 2.2 Gating SIMULATED cells: every marker call is a proposition, every gate a Boolean formula ──
positive = cells[MARKERS] > THRESHOLD          # True/False table: "cell i is positive for marker m"
print(positive.head(6).to_string())
cd3, cd4, cd8, cd19 = (positive[m] for m in MARKERS)

helper_gate = cd3 & cd4 & ~cd8                 # CD3⁺ ∧ CD4⁺ ∧ ¬CD8⁺, evaluated for every cell at once
naive_gate = cd4                               # the tempting shortcut: CD4⁺ alone
truth = cells["population"] == "CD4 T cell"    # 🧪 hidden truth, known only because we simulated the cells
print()
for name, gate in [("CD3⁺ ∧ CD4⁺ ∧ ¬CD8⁺", helper_gate), ("CD4⁺ alone", naive_gate)]:
    print(f"{name:20s}: {int(gate.sum()):4d} cells pass, {int((gate & truth).sum()):4d} of them truly CD4 T cells")
print("\nwho passes the naive gate CD4⁺?")
print(cells.loc[naive_gate, "population"].value_counts().to_string())
````

🔍 **What just happened?** `cells[MARKERS] > THRESHOLD` turned every intensity into a truth value — 8,000 propositions
at once. The three-marker gate $\text{CD3}^+ \land \text{CD4}^+ \land \lnot\text{CD8}^+$ selected 815 cells, 813 of
which are truly CD4 T cells (the 2 others are measurement-noise accidents). The naive gate "CD4⁺" selected 1,139 cells:
all 813 helper T cells, but also **325 monocytes** (human monocytes carry a little CD4) and one B cell. Adding the
conjunct CD3⁺ is what removes the monocytes: a conjunction can only shrink a selection, never enlarge it, because every
extra "and" is one more condition to meet.

````python
# ── 2.3 Pitfall: `and`/`or`/`not` work on single truth values; `&`/`|`/`~` on whole columns ──
try:
    cd3 and cd4                                  # Python's `and` needs ONE truth value, a column has 2,000
except ValueError as err:
    print("ValueError:", err)
print("cd3 & cd4 works:", int((cd3 & cd4).sum()), "cells are CD3⁺ ∧ CD4⁺")
# comparisons must be wrapped in parentheses, because & binds more tightly than >
print("(cells.CD3 > 2) & (cells.CD4 > 2):", int(((cells["CD3"] > 2) & (cells["CD4"] > 2)).sum()), "cells")
````

🔍 **What just happened?** Python's keywords `and`, `or`, `not` expect *one* truth value; a pandas column holds 2,000,
so Python refuses ("The truth value of a Series is ambiguous") instead of guessing whether you meant "all" or "any".
The element-wise operators `&`, `|`, `~` apply the connective cell by cell and gave 815 CD3⁺ ∧ CD4⁺ cells — the same
cells as the helper gate, because no simulated cell is CD3⁺CD4⁺CD8⁺. The parentheses around each comparison are
essential: in Python `&` binds more tightly than `>`, a first example of **operator precedence** (Section 6).

````python
# ── 2.4 Picture the gates: thresholds turn intensities into propositions (SIMULATED) ──
colours = {"CD4 T cell": "tab:blue", "CD8 T cell": "tab:orange", "B cell": "tab:green",
           "NK cell": "tab:purple", "monocyte": "tab:brown"}
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for population, colour in colours.items():
    sub = cells[cells["population"] == population]
    axes[0].scatter(sub["CD19"], sub["CD3"], s=4, alpha=0.45, color=colour, label=population)
    t_like = sub[sub["CD3"] > THRESHOLD]                      # right panel: only cells that are CD3⁺
    axes[1].scatter(t_like["CD8"], t_like["CD4"], s=4, alpha=0.45, color=colour)
for ax in axes:
    ax.axhline(THRESHOLD, color="k", lw=1, ls="--")
    ax.axvline(THRESHOLD, color="k", lw=1, ls="--")
    ax.set_xlim(-0.2, 4.6)
    ax.set_ylim(-0.2, 4.6)
axes[0].set(xlabel="CD19 intensity (log10)", ylabel="CD3 intensity (log10)", title="all cells: CD3 vs CD19")
axes[0].text(0.0, 4.25, "CD3⁺ ∧ ¬CD19⁺", fontsize=8)
axes[0].text(2.6, 0.0, "¬CD3⁺ ∧ CD19⁺", fontsize=8)
axes[1].set(xlabel="CD8 intensity (log10)", ylabel="CD4 intensity (log10)", title="CD3⁺ cells only: CD4 vs CD8")
axes[1].text(0.0, 4.25, "CD3⁺ ∧ CD4⁺ ∧ ¬CD8⁺ (helper T)", fontsize=8)
axes[1].text(2.4, 0.0, "CD3⁺ ∧ ¬CD4⁺ ∧ CD8⁺", fontsize=8)
axes[0].legend(markerscale=3, fontsize=7, loc="center right")
fig.suptitle("SIMULATED flow cytometry: dashed lines are the thresholds that define the propositions", fontsize=10)
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** Each dashed line is a threshold, so each quadrant is a conjunction of two propositions. In
the left panel T cells (blue, orange) sit in the CD3⁺ ∧ ¬CD19⁺ corner, B cells (green) in ¬CD3⁺ ∧ CD19⁺, and NK cells
and monocytes in the double-negative corner. The right panel keeps only CD3⁺ cells — a gate inside a gate, i.e. one
more "∧" — and splits them cleanly into helper (CD4⁺ ∧ ¬CD8⁺) and cytotoxic (¬CD4⁺ ∧ CD8⁺) T cells. The monocytes'
dim CD4 never shows up on the right, because the CD3 conjunct removed them first.

## 3 · From English to logic: translations, exclusive or, and "if and only if"

> 📍 **Where we are.** We can now evaluate formulas. The harder, everyday skill is **writing** them: turning an English
> sentence — a homework problem, an inclusion criterion of a clinical trial, a gene-regulation rule from a paper — into
> a formula that means exactly what the sentence means. We also meet the remaining connectives ⊕ and ⇔.

**Recipe.** (i) Name the **atomic** propositions with letters, phrasing each one positively; (ii) find the connective
words — *and*, *but* (= and), *or*, *either … or*, *not*, *both*; (iii) add parentheses that follow the grouping of
the sentence. Cover the right-hand column and try the four classic exercises first:

| Sentence | Atomic propositions | Formula |
|---|---|---|
| Alice is majoring in computer science and studio art. | $p$: Alice majors in computer science; $q$: Alice majors in studio art | $p \land q$ |
| Either John took a formal logic class, or he is a quick learner. | $r$: John took a formal logic class; $s$: John is a quick learner | $r \lor s$ |
| Kevin broke his hand and didn't take the test as scheduled. | $t$: Kevin broke his hand; $u$: Kevin took the test as scheduled | $t \land \lnot u$ |
| Jessica knows Python or she has programmed in both C and Java. | $v$: Jessica knows Python; $w$: she has programmed in C; $x$: she has programmed in Java | $v \lor (w \land x)$ |

Notes on the solutions. "Computer science **and** studio art" hides two claims about Alice. "Either … or" is still the
inclusive or unless the context rules out *both*. "Didn't take the test" is the negation of the positive proposition
$u$, so we write $\lnot u$ rather than inventing a negative atom. In Jessica's sentence the word "both" groups C and Java
together, which is exactly what the parentheses $(w \land x)$ say. Biology works the same way: "the gene is expressed in
liver **but not** in kidney" is $\ell \land \lnot k$; "a patient is eligible if she is over 18 **and** has diabetes
**or** hypertension" is $a \land (d \lor h)$ — and the parentheses matter, as the next cell shows.

````python
# ── 3.1 Grouping matters: "v or (w and x)" is not "(v or w) and x" ──
jessica = truth_table({"v ∨ (w ∧ x)": lambda v, w, x: v or (w and x),
                       "(v ∨ w) ∧ x": lambda v, w, x: (v or w) and x}, ["v", "w", "x"])
jessica["agree?"] = np.where(jessica["v ∨ (w ∧ x)"] == jessica["(v ∨ w) ∧ x"], "", "← differ")
print(jessica.to_string(index=False))
print("rows where each reading is true:", int((jessica["v ∨ (w ∧ x)"] == "T").sum()), "vs",
      int((jessica["(v ∨ w) ∧ x"] == "T").sum()), "of", len(jessica))
````

🔍 **What just happened?** With three variables the table has $2^3 = 8$ rows. The correct translation
$v \lor (w \land x)$ is true in 5 rows; the mis-grouped $(v \lor w) \land x$ in only 3. They disagree in the two rows
where $v$ = T and $x$ = F — Jessica knows Python but has never programmed in Java. The sentence says she qualifies; the
mis-grouped formula says she does not. Two formulas with the same letters and the same connectives can mean different
things: parentheses are part of the meaning.

### 3.2 Inclusive or versus exclusive or

**Exclusive or** [⊕, "xor", pronounced "ex-or"]. $p \oplus q$ ("$p$ exclusive or $q$") is true when **exactly one** of
$p$, $q$ is true — one or the other, but not both. So $p \oplus q$ is false when both are true and when both are false.
Everyday "or" is ambiguous: "coffee or tea?" usually allows both (inclusive), "you can enter through door A or door B"
does not (exclusive). In digital circuits an XOR gate outputs 1 exactly when its two input bits differ. In genetics,
if $m$ = "the maternal copy of a gene carries the variant" and $f$ = "the paternal copy carries it", then
"the person is heterozygous" is $m \oplus f$.

**Example (a cover letter).** "You may contact me by email or by phone. I am available for an interview on October 8th
in Minneapolis or Hong Kong." With $p$ = you may contact me by phone, $q$ = you may contact me by email,
$r$ = I am available on October 8th in Minneapolis, $s$ = I am available on October 8th in Hong Kong: the first "or" is
**inclusive** (you could both email and call), the second is **exclusive** (nobody can be in Minneapolis and Hong Kong
on the same day), so the letter says

$$(p \lor q) \land (r \oplus s).$$

*In words:* at least one of the two contact routes is open, and the candidate is in exactly one of the two cities.

````python
# ── 3.2 Inclusive vs exclusive or, and the cover-letter formula (p ∨ q) ∧ (r ⊕ s) ──
print(truth_table({"p ∨ q": lambda p, q: p or q, "p ⊕ q": lambda p, q: xor(p, q)}, ["p", "q"]).to_string(index=False))
letter = truth_table({"(p ∨ q) ∧ (r ⊕ s)": lambda p, q, r, s: (p or q) and xor(r, s)}, ["p", "q", "r", "s"])
true_rows = letter[letter.iloc[:, -1] == "T"]
print(f"\ncover letter: {len(letter)} rows, the formula is true in {len(true_rows)} of them:")
print(true_rows.to_string(index=False))
````

🔍 **What just happened?** The two "or" columns differ only in the first row: for T, T the inclusive $p \lor q$ is
true and the exclusive $p \oplus q$ is false. The cover-letter formula has four variables, so $2^4 = 16$ rows, and it
is true in exactly 6 of them: 3 ways for $p \lor q$ to hold (TT, TF, FT) times 2 ways for $r \oplus s$ (TF, FT). In
every true row exactly one of $r, s$ is T — the formula has correctly ruled out being in two cities at once.

### 3.3 "If and only if" — the biconditional

**Mutual implication** [⇔, "if and only if", "iff"]. $p \Leftrightarrow q$ is true when $p$ and $q$ have the **same
truth value** (both true or both false) and false otherwise. It is called the **biconditional** because an
implication ("if $p$ then $q$", Section 4) is also called a *conditional*, and $p \Leftrightarrow q$ is two
conditionals at once: "if $p$ then $q$" **and** "if $q$ then $p$" (we prove this in Section 7). In science "iff"
states a **necessary and sufficient** condition, and mathematicians use it in definitions.

The *lac* operon rule of the biological question is a biconditional. With $z$ = "the *lac* genes are transcribed",
$\ell$ = "lactose is present", $g$ = "glucose is present":

$$z \;\Leftrightarrow\; (\ell \land \lnot g).$$

*In words:* the genes are on exactly when lactose is present and glucose is absent — on in that situation (sufficiency)
and off in every other situation (necessity). The simplification ignores the low "leaky" expression without lactose and
the weak induction when both sugars are present.

````python
# ── 3.3 The lac-operon rule as a biconditional: which (lactose, glucose, genes-on) combinations does it allow? ──
lac = truth_table({"ℓ ∧ ¬g": lambda l, g, z: l and not g,
                   "z ⇔ (ℓ ∧ ¬g)": lambda l, g, z: iff(z, l and not g)}, ["l", "g", "z"])
lac.columns = ["ℓ lactose", "g glucose", "z genes on", "ℓ ∧ ¬g", "z ⇔ (ℓ ∧ ¬g)"]
print(lac.to_string(index=False))
allowed = lac[lac["z ⇔ (ℓ ∧ ¬g)"] == "T"]
print(f"\nthe rule allows {len(allowed)} of the 8 combinations; an observation in any other row would refute it")
````

🔍 **What just happened?** Treating $z$ as a third variable, the biconditional is true in exactly the 4 rows where the
column $z$ equals the column $\ell \land \lnot g$: genes on with lactose and no glucose (row 3), genes off in the other
three situations (rows 2, 6, 8). A biconditional therefore works like a **constraint**: it lists which combinations
of inputs and outputs the model permits. Observing, say, transcription without lactose (a row with $\ell$ = F and
$z$ = T) would falsify the rule — which is how constitutive *lacI* mutants, which transcribe without inducer, were
recognised as breaking it.

### 3.4 "If" versus "if and only if" in English

English often says "if" when it means "if and only if". Think of a number between 10 and 1,000,000, and let
$p$ = "your number is prime", $q$ = "your number is even", $r$ = "your number is evenly divisible by a positive integer
other than 1 and itself". Translate:

1. "If the number you're thinking of is even, then it isn't prime."
2. "The number you're thinking of isn't prime if it's evenly divisible by an integer other than 1 and itself."

**Solution.** The "if" in (1) is an implication, $q \Rightarrow \lnot p$. The "if" in (2) is really "if and only if",
because the sentence is a *definition*: not being prime **is** having such a divisor, so $\lnot p \Leftrightarrow r$.

🧮 **Math refresher — divisibility and primes.** An integer $n$ is *evenly divisible* by $d$ (written $d \mid n$, "$d$
divides $n$") if $n/d$ is a whole number — in Python `n % d == 0` (the remainder is 0). A **prime** is an integer greater
than 1 whose only positive divisors are 1 and itself: 2, 3, 5, 7, 11, … The number 2 is the only even prime, which is
why the exercise starts at 10.

````python
# ── 3.4 "If" versus "if and only if": check both translations on every number from 10 to 3,000 ──
def is_prime(n: int) -> bool:
    """True if n > 1 has no divisor d with 2 ≤ d ≤ √n."""
    return n > 1 and all(n % d != 0 for d in range(2, int(n ** 0.5) + 1))


def has_other_divisor(n: int) -> bool:
    """True if n is evenly divisible by a positive integer other than 1 and n itself (literal translation)."""
    return any(n % d == 0 for d in range(2, n))


numbers = range(10, 3001)
print("(1) q ⇒ ¬p  holds for every n in 10..3000:", all(implies(n % 2 == 0, not is_prime(n)) for n in numbers))
print("(2) ¬p ⇔ r  holds for every n in 10..3000:", all(iff(not is_prime(n), has_other_divisor(n)) for n in numbers))
print("    the converse of (1), ¬p ⇒ q, fails first at n =",
      next(n for n in numbers if not implies(not is_prime(n), n % 2 == 0)), "(odd, yet not prime)")
print("outside the range, n = 2: q ⇒ ¬p is", implies(2 % 2 == 0, not is_prime(2)), "(2 is even AND prime)")
````

🔍 **What just happened?** Both translations hold for all 2,991 numbers we checked. Implication (1) is one-way: its
reverse, "if the number isn't prime then it is even", already fails at 15 = 3 × 5. Biconditional (2) is two-way — the
definition of "not prime" — so no number can separate its two sides. And the range matters: at $n = 2$ the antecedent
"2 is even" is true and the consequent "2 is not prime" is false, so (1) fails there. A logical statement is always a
statement *about some collection of objects*; Section 8 makes that collection explicit with quantifiers.

### 📘 Lecture notes — propositions and logical connectives

**Why logic.** Logic is the building block of mathematical reasoning: it formalises statements, lets us analyse
arguments, and underlies sets, probability ("and/or/if–then" events) and statistics (what a test lets us conclude).
**Propositional logic** treats statements as indivisible units whose truth values are unknown until assigned;
**predicate logic** adds variables over a set and quantifiers, so the statement can be fully evaluated — it extends
propositional logic with more detail.

**Definitions.** A *proposition* is a statement that is either true or false ("2 + 2 = 4"; "a bear is CMU's mascot").
An *atomic* proposition is conceptually indivisible (a *Boolean variable*, written $p, q, r$); a *compound* proposition
combines simpler ones with logical connectives; a compound proposition over $p_1, \dots, p_k$ is a *Boolean formula*.
A *truth table* lists all $2^k$ truth assignments (one per row, in the order TT, TF, FT, FF) and the value of the whole
proposition in the last column.

| name | symbol/use | description | truth table (rows TT, TF, FT, FF) |
|---|---|---|---|
| negation | $\lnot p$ | "not $p$" | T → F, F → T |
| conjunction | $p \land q$ | "$p$ and $q$" | T F F F |
| disjunction | $p \lor q$ | "$p$ or $q$" (inclusive) | T T T F |
| exclusive or | $p \oplus q$ | "$p$ or $q$ but not both" | F T T F |
| implication | $p \Rightarrow q$ | "if $p$, then $q$" | T F T T |
| mutual implication | $p \Leftrightarrow q$ | "$p$ if and only if $q$" ("iff") | T F F T |

¬ is a **unary** operator (it builds a compound proposition from one simpler proposition); the other five are
**binary** (they combine two).

**Worked translations.** Alice: $p \land q$. John: $r \lor s$. Kevin: $t \land \lnot u$. Jessica:
$v \lor (w \land x)$. Cover letter: $(p \lor q) \land (r \oplus s)$ — "email or phone" is inclusive, "Minneapolis or
Hong Kong on the same day" exclusive. Numbers between 10 and 1,000,000: "if even then not prime" is $q \Rightarrow \lnot p$;
"isn't prime if it has a divisor other than 1 and itself" is the definition $\lnot p \Leftrightarrow r$.

**Common confusions.** (1) "Or" in logic is inclusive unless written ⊕. (2) English "if" sometimes means "iff",
especially in definitions. (3) "But" is logically "and". (4) Atomic vs compound is about structure, not truth value.

## 4 · Implication in depth

> 📍 **Where we are.** Of all the connectives, implication is the one that confuses students most — and the one
> mathematics uses most: every theorem is an "if … then …". We look at it from four angles: its truth table, the
> promise analogy, vacuous truth, and a picture with sets.

**Definition (implication).** $p \Rightarrow q$ means "**if** $p$ is true, **then** $q$ is true". It is **false only
when $p$ is true and $q$ is false** — in all other cases it is true. In $p \Rightarrow q$, $p$ is called the
**antecedent** (or hypothesis) and $q$ the **consequent** (or conclusion).

**Implication is not causation.** "If $p$ then $q$" sounds as if $p$ causes $q$ — but $p \Rightarrow q$ only says that
in every situation where $p$ is true, $q$ is true as well: knowing $p$ lets us conclude $q$. It is a statement about
truth values, not about mechanisms.

**The promise analogy.** Read $p \Rightarrow q$ as a promise: "If it rains, I will bring an umbrella."

| situation | $p$: it rains | $q$: I bring an umbrella | promise | $p \Rightarrow q$ |
|---|---|---|---|---|
| it rains, I bring one | T | T | kept | T |
| it rains, I don't | T | F | **broken** | **F** |
| no rain, I bring one anyway | F | T | not applicable — not broken | T |
| no rain, no umbrella | F | F | not applicable — not broken | T |

The same holds for "if you finish your homework, you get dessert": if you don't finish, whatever happens to the dessert
cannot break the promise. And for "if it is raining, the ground is wet": rain and a wet ground keeps the statement
(true); no rain leaves the ground free to be wet (a sprinkler!) or dry (true either way); only **rain with a dry ground**
contradicts it.

**Vacuous truth: "false implies anything", "anything implies true".**

* If the antecedent $p$ is **false**, $p \Rightarrow q$ is true whatever $q$ is: "If 2 + 3 = 4, then you will eat
  tofu for dinner" is true, no matter what you eat; "if 2 + 3 = 4 then 1 = 1" and "if 2 + 3 = 4 then 1 = 0" are both
  true; "if pigs fly, I'll give you a million dollars" costs nothing.
* If the consequent $q$ is **true**, $p \Rightarrow q$ is true whatever $p$ is: "if 1 = 0 then 2 + 2 = 4" and "if
  2 + 2 = 4 then 2 + 2 = 4" are both true.
* "If the moon is a secret Sumerian surveillance device, then 2 + 3 = 5" is true twice over: its hypothesis is false
  *and* its conclusion is true.

A statement that is true only because its antecedent is false is called **vacuously true**: it cannot be disproven,
because the situation it talks about never arises. The truth of $p \Rightarrow q$ depends only on whether $p$ is true
and $q$ false **at the same time**.

````python
# ── 4.1 Implication: false only when a true antecedent meets a false consequent ──
print(truth_table({"p ⇒ q": lambda p, q: implies(p, q),
                   "¬p ∨ q": lambda p, q: (not p) or q}, ["p", "q"]).to_string(index=False))
examples = [("1 + 1 = 2", 1 + 1 == 2, "2 + 3 = 5", 2 + 3 == 5),
            ("2 + 3 = 4", 2 + 3 == 4, "2 + 2 = 4", 2 + 2 == 4),
            ("2 + 3 = 4", 2 + 3 == 4, "2 + 3 = 6", 2 + 3 == 6),
            ("2 + 2 = 4", 2 + 2 == 4, "2 + 1 = 5", 2 + 1 == 5)]
print()
for a_text, a, c_text, c in examples:
    print(f"({a_text}) ⇒ ({c_text}):   {'T' if a else 'F'} ⇒ {'T' if c else 'F'}   is {implies(a, c)}")
````

🔍 **What just happened?** The $p \Rightarrow q$ column reads T, F, T, T: a single F, in the row $p$ = T, $q$ = F. The
column $\lnot p \lor q$ is identical — that is literally how the toolkit implements `implies` (`(not p) or q`), and it
is a useful way to remember the definition: "either the hypothesis fails, or the conclusion holds". Python then
evaluated the four classic arithmetic examples: "1 + 1 = 2 implies 2 + 3 = 5" (true implies true) is true;
"2 + 3 = 4 implies 2 + 2 = 4" (false implies true) and "2 + 3 = 4 implies 2 + 3 = 6" (false implies false) are true;
only "2 + 2 = 4 implies 2 + 1 = 5" (true implies false) is false.

**Why define it like this?** The best reason comes from mathematics itself. We want the statement "for every whole
number $n$: if 4 divides $n$, then 2 divides $n$" to be **true** — it is a theorem. But for $n = 3$ both parts are false,
and for $n = 2$ the hypothesis is false and the conclusion true. If "false implies …" were counted as false, these
numbers would refute a true theorem. The next cell checks this.

````python
# ── 4.2 Why "false implies anything" is the right definition ──
def tf(b: bool) -> str:
    """Print a truth value as T or F."""
    return "T" if b else "F"


rows = [(n, tf(n % 4 == 0), tf(n % 2 == 0), tf(implies(n % 4 == 0, n % 2 == 0))) for n in range(1, 11)]
print(pd.DataFrame(rows, columns=["n", "4 divides n", "2 divides n", "(4 | n) ⇒ (2 | n)"]).to_string(index=False))
print("\ntrue for every n in 1..1000?", all(implies(n % 4 == 0, n % 2 == 0) for n in range(1, 1001)))


def strict_if(p: bool, q: bool) -> bool:
    """A WRONG 'if' that is false whenever the hypothesis is false (it is really 'and')."""
    return p and q


print("with the wrong 'if', the theorem fails already at n =",
      next(n for n in range(1, 1001) if not strict_if(n % 4 == 0, n % 2 == 0)))
````

🔍 **What just happened?** For $n = 1, \dots, 10$ the implication column is T in every row, and it stays true for all
1,000 numbers. Rows such as $n = 1$ (F ⇒ F) and $n = 2$ (F ⇒ T) are *vacuously* true: they say nothing about numbers
divisible by 4, so they cannot contradict the theorem. With the "wrong if", which treats a false hypothesis as a failure,
the theorem collapses at $n = 1$. Material implication is the only definition under which universal "if–then" theorems
can be true.

### 4.3 The set picture: $p \Rightarrow q$ means $P \subseteq Q$

Think of $p \Rightarrow q$ as "every time $p$ is true, $q$ must also be true". Let $P$ be the set of cases (objects,
situations) in which $p$ is true and $Q$ the set of cases in which $q$ is true. Then

$$\bigl(p \Rightarrow q \text{ in every case}\bigr) \;\Longleftrightarrow\; P \subseteq Q .$$

*In words:* the implication holds in all cases exactly when the region of $p$-cases lies entirely inside the region of
$q$-cases ($\subseteq$ reads "is a subset of", Module 03). A **counterexample** is a case in $P$ but outside $Q$; the
implication holds when that region is empty.

````python
# ── 4.3 p ⇒ q as P ⊆ Q: two examples on the numbers 1–12, with the counterexample region shaded ──
U = set(range(1, 13))                               # the universe of cases: n = 1, …, 12
multiples_of_4 = {n for n in U if n % 4 == 0}
multiples_of_3 = {n for n in U if n % 3 == 0}
evens = {n for n in U if n % 2 == 0}
fig, axes = plt.subplots(1, 3, figsize=(12, 3.8))
ax = axes[0]                                        # schematic: P drawn inside Q
ax.add_patch(plt.Rectangle((-2.1, -1.9), 4.2, 3.7, fill=False, color="0.4"))
ax.add_patch(plt.Circle((0.2, -0.1), 1.55, color="tab:blue", alpha=0.12))
ax.add_patch(plt.Circle((0.2, -0.1), 1.55, fill=False, color="tab:blue", lw=2))
ax.add_patch(plt.Circle((-0.35, 0.0), 0.7, color="#ffd54f", alpha=0.9))
ax.text(-0.35, 0.0, "P", ha="center", va="center", fontsize=13, fontweight="bold")
ax.text(1.05, 0.95, "Q", fontsize=13, fontweight="bold", color="tab:blue")
ax.set_xlim(-2.2, 2.2)
ax.set_ylim(-2.0, 1.9)
ax.set_aspect("equal")
ax.axis("off")
ax.set_title("p ⇒ q  ⇔  P ⊆ Q")
counterexample = lambda in_p, in_q: in_p and not in_q          # shade P \ Q, where p ⇒ q fails
draw_venn({"P": multiples_of_4, "Q": evens}, universe=U, ax=axes[1], highlight=counterexample,
          title="p: 4 | n,  q: 2 | n   (holds)")
draw_venn({"P": multiples_of_3, "Q": evens}, universe=U, ax=axes[2], highlight=counterexample,
          title="p: 3 | n,  q: 2 | n   (fails)")
plt.tight_layout()
plt.show()
print("counterexamples to (4 | n) ⇒ (2 | n):", sorted(multiples_of_4 - evens))
print("counterexamples to (3 | n) ⇒ (2 | n):", sorted(multiples_of_3 - evens))
````

🔍 **What just happened?** Left: the schematic — $P$ sits inside $Q$, so there is nowhere for a counterexample to live.
Middle: on the numbers 1–12, the multiples of 4 {4, 8, 12} all lie inside the evens; the region "in $P$ but not in
$Q$" is empty (∅), so the implication holds for every $n$. Right: the multiples of 3 {3, 6, 9, 12} are *not* inside the
evens; the shaded region holds the counterexamples 3 and 9, each of which makes "if 3 divides $n$ then $n$ is even"
false. The numbers listed in the corner ("U: …", outside both circles) are vacuously true cases: the hypothesis is false
for them (and so is the conclusion). The numbers that are in $Q$ but not in $P$ are vacuously true too — "$p$ is false,
$q$ is true". Whenever you meet "if …
then …" about a collection of objects, picture this diagram.

**Sufficient and necessary.** When $p \Rightarrow q$ holds, $p$ is a **sufficient** condition for $q$ (being in $P$ is
enough to be in $Q$) and $q$ is a **necessary** condition for $p$ (you cannot be in $P$ without being in $Q$). English
has many ways to say $p \Rightarrow q$: "if $p$ then $q$", "$q$ if $p$", "$p$ only if $q$", "$q$ whenever $p$",
"$p$ is sufficient for $q$", "$q$ is necessary for $p$".

## 5 · Converse, inverse and contrapositive

> 📍 **Where we are.** An implication has three relatives, obtained by swapping and/or negating its two parts. Only one
> of them says the same thing. Confusing them is the most common error in scientific reasoning — and the logic of
> knockout experiments rests on getting it right.

For the implication $p \Rightarrow q$:

| name | formula | example for "if it rains, the ground is wet" |
|---|---|---|
| implication | $p \Rightarrow q$ | if it rains, the ground is wet |
| **converse** | $q \Rightarrow p$ | if the ground is wet, it rains |
| **inverse** | $\lnot p \Rightarrow \lnot q$ | if it does not rain, the ground is not wet |
| **contrapositive** | $\lnot q \Rightarrow \lnot p$ | if the ground is not wet, it does not rain |

The contrapositive says the same as the original; the converse and the inverse say something different (a sprinkler
makes the ground wet without rain). The next cell proves it with a truth table.

````python
# ── 5.1 Which relatives of p ⇒ q say the same thing? ──
family = truth_table({"p ⇒ q": lambda p, q: implies(p, q),
                      "q ⇒ p (converse)": lambda p, q: implies(q, p),
                      "¬p ⇒ ¬q (inverse)": lambda p, q: implies(not p, not q),
                      "¬q ⇒ ¬p (contrapositive)": lambda p, q: implies(not q, not p)}, ["p", "q"])
print(family.to_string(index=False))
cols = family.columns
print("\nimplication ≡ contrapositive:", (family[cols[2]] == family[cols[5]]).all())
print("converse    ≡ inverse       :", (family[cols[3]] == family[cols[4]]).all())
print("implication ≡ converse      :", (family[cols[2]] == family[cols[3]]).all())
````

🔍 **What just happened?** The implication and its contrapositive have the same column (T, F, T, T): they are
**logically equivalent**, $p \Rightarrow q \equiv \lnot q \Rightarrow \lnot p$. The converse and the inverse share a
different column (T, T, F, T) — they are equivalent to each other (the inverse is the contrapositive of the converse)
but not to the original. The disagreement is in rows 2 and 3: when exactly one of $p$, $q$ is true. So "if it rains the
ground is wet" does **not** give you "if the ground is wet it rained" — that inference has a name, *affirming the
consequent*, and it is a fallacy.

⚠️ **Common confusion.** From "$p \Rightarrow q$" and "$q$ is true" you may **not** conclude $p$ (affirming the
consequent); from "$p \Rightarrow q$" and "$p$ is false" you may **not** conclude $\lnot q$ (denying the antecedent).
The two valid moves are: from $p$ conclude $q$ (*modus ponens*) and from $\lnot q$ conclude $\lnot p$ (*modus
tollens* — the contrapositive in action).

### 5.2 Necessary, sufficient, and what a knockout tests

Biologists say "gene X is **necessary** for phenotype Y". In logic that is $Y \Rightarrow X$: every organism showing
the phenotype has the gene. Its contrapositive, $\lnot X \Rightarrow \lnot Y$, is the same claim seen from the other
side — "remove the gene and the phenotype disappears" — and that is exactly what a **knockout** experiment checks.
"X is **sufficient** for Y" is the different claim $X \Rightarrow Y$, tested by adding X (overexpression, rescue) in a
context that lacks Y. A real example: mice and humans with loss-of-function mutations in the transcription factor gene
*Foxn1* ("nude" mice) have no functional thymus, so *Foxn1* is **necessary** for thymus development. It is not
**sufficient**: thymus development needs many other genes too. The table below is typed in and illustrative.

````python
# ── 5.2 Necessary vs sufficient: an illustrative knockout table (typed in, not real data) ──
mice = pd.DataFrame({
    "mouse":             ["m1", "m2", "m3", "m4", "m5", "m6", "m7", "m8"],
    "Foxn1 functional":  [True, True, True, False, False, True, False, True],
    "other gene":        [True, True, False, True, False, True, True, False],   # another gene the thymus needs
})
mice["thymus"] = mice["Foxn1 functional"] & mice["other gene"]          # the (simplified) truth: both are needed
X, Y = mice["Foxn1 functional"], mice["thymus"]
mice["Y ⇒ X"] = [implies(y, x) for x, y in zip(X, Y)]                # necessary
mice["¬X ⇒ ¬Y"] = [implies(not x, not y) for x, y in zip(X, Y)]      # what a knockout tests (contrapositive)
mice["X ⇒ Y"] = [implies(x, y) for x, y in zip(X, Y)]                # sufficient
print(mice.replace({True: "T", False: "F"}).to_string(index=False))
print("\nFoxn1 necessary   (∀ mice: Y ⇒ X)  :", bool(mice["Y ⇒ X"].all()))
print("knockout version  (∀ mice: ¬X ⇒ ¬Y):", bool(mice["¬X ⇒ ¬Y"].all()))
print("Foxn1 sufficient  (∀ mice: X ⇒ Y)  :", bool(mice["X ⇒ Y"].all()),
      "| counterexamples:", list(mice.loc[~mice["X ⇒ Y"], "mouse"]))
````

🔍 **What just happened?** In the simulated world, a thymus needs *Foxn1* **and** another gene. The necessity claim
$Y \Rightarrow X$ holds for all eight mice, and so — row by row — does its contrapositive $\lnot X \Rightarrow \lnot Y$:
every *Foxn1*-deficient mouse (m4, m5, m7) lacks a thymus. The sufficiency claim $X \Rightarrow Y$ fails for m3 and m8,
which have a working *Foxn1* but lack the other gene. Notice how a universal claim ("for all mice") is refuted by a
single row — the topic of Section 8.

### 🌉 Bridge — necessity, sufficiency and the logic of experiments

Loss-of-function experiments (knockouts, CRISPR screens, RNA interference, drug inhibition) test **necessity** through
the contrapositive: if removing X never removes Y, X is not necessary for Y. Gain-of-function experiments
(overexpression, rescue, transplantation) test **sufficiency**. Many papers need both, which is why "necessary and
sufficient" — a biconditional — is the gold standard of a mechanistic claim. Material implication is not causation:
even a perfectly true $p \Rightarrow q$ in a data set may be produced by a hidden common cause, and separating the two
needs interventions and a causal model (BiologyToML
[41 · perturbation and causal inference](../../../../ml/tutorial/BiologyToML/notebooks/41_perturbation_causal_inference.ipynb)). The probabilistic version of
"$p$ makes $q$ more likely" is conditional probability, $P(q \mid p)$ (Module 17).

### 📘 Lecture notes — implication, converse, inverse and contrapositive

**Definition.** $p \Rightarrow q$ ("if $p$ then $q$") is false only when $p$ is true and $q$ is false; in every other
case it is true. $p$ is the *antecedent* (hypothesis), $q$ the *consequent* (conclusion). Equivalently
$p \Rightarrow q \equiv \lnot p \lor q$.

**How to read it.** (1) A truth relationship, **not causation**: in every situation where $p$ is true, $q$ is also true.
(2) A **promise**: it is broken only if $p$ happens and $q$ does not; if $p$ never happens the promise is not broken.
(3) **Vacuous truth**: "false implies anything" (a false $p$ makes $p \Rightarrow q$ true) and "anything implies true"
(a true $q$ makes it true). (4) **Sets**: with $P$ = cases where $p$ holds and $Q$ = cases where $q$ holds,
"$p \Rightarrow q$ in every case" $\iff P \subseteq Q$.

**Worked examples.** "1 + 1 = 2 ⇒ 2 + 3 = 5" (T ⇒ T) true; "2 + 3 = 4 ⇒ 2 + 2 = 4" (F ⇒ T) true;
"2 + 3 = 4 ⇒ 2 + 3 = 6" (F ⇒ F) true; "2 + 2 = 4 ⇒ 2 + 1 = 5" (T ⇒ F) **false**. "If it is raining, the ground is
wet" is falsified only by rain on a dry ground.

**Relatives.** Converse $q \Rightarrow p$, inverse $\lnot p \Rightarrow \lnot q$, contrapositive
$\lnot q \Rightarrow \lnot p$. Only the contrapositive is equivalent to the original:

$$p \Rightarrow q \;\equiv\; \lnot q \Rightarrow \lnot p, \qquad q \Rightarrow p \;\equiv\; \lnot p \Rightarrow \lnot q .$$

*In words:* an implication says the same as its contrapositive, and the converse says the same as the inverse — but the
two pairs say different things. If $p \Rightarrow q$, then $p$ is sufficient for $q$ and $q$ is necessary for $p$.

**Common confusions.** Implication ≠ causation; implication ≠ its converse (affirming the consequent is a fallacy);
⇒ is not commutative ($p \Rightarrow q \not\equiv q \Rightarrow p$, Section 7).

> 🎓 **Exam tip.** To decide whether an implication is true, look for the one bad row: is there a case with $p$ true and
> $q$ false? If none exists, the implication is true — including all the cases where $p$ is false.

## 6 · Order of operations: precedence and parentheses

> 📍 **Where we are.** Longer formulas need rules for how to read them, exactly as $2 + 3 \times 4$ needs the rule
> "multiply before you add" (it is $2 + 12 = 14$, not $5 \times 4 = 20$). Logic has the same kind of rules.

**Unary and binary operators.** The ¬ connective is a **unary** operator — it builds a compound proposition from a
single simpler one. The other five connectives (∧, ∨, ⊕, ⇒, ⇔) are **binary** operators, which build a compound
proposition from two simpler ones.

**Precedence** (Latin *prae-* "before" + *cedere* "go": *what goes first*) decides the order in which operators are
evaluated when there are no parentheses. A full description of any programming language includes a precedence table,
arranged from "binds the tightest" (highest precedence) to "binds the loosest" (lowest); the table tells us when we
**must** add parentheses to make an expression mean what we want, and when they are optional. The convention used in
02-680:

| precedence | operators |
|---|---|
| 1 · highest (binds tightest) | negation ¬ |
| 2 · medium | conjunction ∧, disjunction ∨, exclusive or ⊕ |
| 3 · lower | implication ⇒ |
| 4 · lowest (binds loosest) | biconditional ⇔ |

Examples. $p \land q \Rightarrow r$ means $(p \land q) \Rightarrow r$, because ∧ binds more tightly than ⇒; if ⇒ had
gone first we would have $p \land (q \Rightarrow r)$ — a **different** proposition. $\lnot p \land q$ means
$(\lnot p) \land q$ ("¬ binds more tightly than ∧"), not $\lnot(p \land q)$. $p \Rightarrow q \Leftrightarrow r$ means
$(p \Rightarrow q) \Leftrightarrow r$. And

$$p \lor q \Rightarrow \lnot r \land \ell \quad\text{means}\quad (p \lor q) \Rightarrow (\lnot r \land \ell).$$

*In words:* both versions are technically the same statement — "if $p$ or $q$, then not $r$ and $\ell$" — but the
parenthesised one is much clearer. **Best practice: use parentheses.**

⚠️ **Common confusion — ∧ versus ∨.** The table above gives ∧, ∨ and ⊕ the *same* precedence, but conventions differ:
many logic textbooks and most programming languages (Python included) let ∧ bind more tightly than ∨, so that
$p \land q \lor r$ means $(p \land q) \lor r$. Because different readers resolve $p \land q \lor r$ differently, never
rely on precedence among ∧, ∨ and ⊕ — always write the parentheses.

````python
# ── 6.1 Two readings of p ∧ q ⇒ r and of ¬p ∧ q: where do they disagree? ──
readings = truth_table({"(p ∧ q) ⇒ r": lambda p, q, r: implies(p and q, r),
                        "p ∧ (q ⇒ r)": lambda p, q, r: p and implies(q, r)}, ["p", "q", "r"])
readings["agree?"] = np.where(readings["(p ∧ q) ⇒ r"] == readings["p ∧ (q ⇒ r)"], "", "← differ")
print(readings.to_string(index=False))
negation_first = truth_table({"(¬p) ∧ q": lambda p, q: (not p) and q,
                              "¬(p ∧ q)": lambda p, q: not (p and q)}, ["p", "q"])
negation_first["agree?"] = np.where(negation_first["(¬p) ∧ q"] == negation_first["¬(p ∧ q)"], "", "← differ")
print()
print(negation_first.to_string(index=False))
````

🔍 **What just happened?** The correct reading $(p \land q) \Rightarrow r$ and the wrong one $p \land (q \Rightarrow r)$
agree in the four rows with $p$ = T and disagree in all four rows with $p$ = F: when $p$ is false, $p \land q$ is
false, so the correct reading is vacuously **true**, while the wrong reading starts with "$p$ ∧ …" and is **false**.
The assignment $p$ = F, $q$ = T, $r$ = F is one such row: $(F \land T) \Rightarrow F = F \Rightarrow F = T$, but
$F \land (T \Rightarrow F) = F \land F = F$. Likewise $(\lnot p) \land q$ and $\lnot(p \land q)$ disagree in two rows
(TF and FF). Precedence is not pedantry: it decides truth values.

````python
# ── 6.2 Draw the two parse trees of p ∧ q ⇒ r for p = F, q = T, r = F (green = T, red = F) ──
OPS = {"¬": lambda a: not a, "∧": lambda a, b: a and b, "∨": lambda a, b: a or b,
       "⊕": xor, "⇒": implies, "⇔": iff}


def evaluate(tree, values: dict) -> bool:
    """Truth value of a formula tree such as ("⇒", ("∧", "p", "q"), "r") under the given variable values."""
    if isinstance(tree, str):
        return values[tree]
    op, *children = tree
    return OPS[op](*(evaluate(child, values) for child in children))


def draw_parse_tree(tree, values: dict, ax, title: str):
    """Draw a formula tree top-down (connectives inside, variables at the leaves), coloured by truth value."""
    next_leaf = [0.0]

    def place(node, depth):
        if isinstance(node, str):
            x = next_leaf[0]
            next_leaf[0] += 1.0
            label = node
        else:
            child_x = [place(child, depth + 1) for child in node[1:]]
            x = float(np.mean(child_x))
            for cx in child_x:
                ax.plot([x, cx], [-depth, -depth - 1], color="0.5", lw=1.2, zorder=1)
            label = node[0]
        value = evaluate(node, values)
        ax.text(x, -depth, f"{label}\n{'T' if value else 'F'}", ha="center", va="center", fontsize=10, zorder=2,
                bbox=dict(boxstyle="round,pad=0.35", fc="#c8e6c9" if value else "#ffcdd2", ec="0.3"))
        return x

    place(tree, 0)
    ax.set_xlim(-0.7, next_leaf[0] - 0.3)
    ax.set_ylim(-2.6, 0.6)
    ax.axis("off")
    ax.set_title(title)


assignment = {"p": False, "q": True, "r": False}
fig, axes = plt.subplots(1, 2, figsize=(9, 3.4))
draw_parse_tree(("⇒", ("∧", "p", "q"), "r"), assignment, axes[0], "correct: (p ∧ q) ⇒ r   →  T")
draw_parse_tree(("∧", "p", ("⇒", "q", "r")), assignment, axes[1], "wrong: p ∧ (q ⇒ r)   →  F")
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** A formula is really a **tree**: the operator applied *last* sits at the root, the variables
are the leaves, and every node's truth value is computed from its children's. Precedence decides the shape of the tree.
In the correct tree (left) ∧ is evaluated first, $F \land T = F$, and the root $F \Rightarrow F$ is T. In the wrong tree
(right) the ⇒ node is evaluated first, $T \Rightarrow F = F$, and the root $F \land F$ is F. Same letters, same
connectives, same truth assignment — different trees, different answers.

````python
# ── 6.3 Python's own precedence: not > and > or  (and Python has no ⇒ operator — we call implies) ──
p, q = False, False
print("not p and q  →", not p and q, "  same as (not p) and q:", (not p) and q, "| not (p and q):", not (p and q))
p, q, r = True, True, False
print("p or q and r →", p or q and r, "   same as p or (q and r):", p or (q and r), " | (p or q) and r:", (p or q) and r)
long_formula = truth_table({"(p ∨ q) ⇒ (¬r ∧ ℓ)": lambda p, q, r, l: implies(p or q, (not r) and l)},
                           ["p", "q", "r", "l"])
print("\n(p ∨ q) ⇒ (¬r ∧ ℓ) is true in", int((long_formula.iloc[:, -1] == "T").sum()), "of", len(long_formula), "rows")
````

🔍 **What just happened?** Python follows "not before and before or". With $p$ = $q$ = F, `not p and q` is
`(not p) and q` = T ∧ F = False, whereas `not (p and q)` = ¬F = True — the two readings of $\lnot p \land q$ from
Section 6.1. With $p$ = $q$ = T and $r$ = F, `p or q and r` is `p or (q and r)` = True, whereas `(p or q) and r` = False. So in Python `and` binds more tightly than `or` — one of the conventions mentioned in
the ⚠️ box, and a reason to write parentheses. Python has no ⇒ symbol, so we write `implies(a, b)` with explicit
arguments, which forces us to decide the grouping. The long formula $(p \lor q) \Rightarrow (\lnot r \land \ell)$ is
true in 7 of 16 rows: the 4 rows with $p$ = $q$ = F (vacuously), plus the 3 rows with $p \lor q$ true and
$r$ = F, $\ell$ = T.

## 7 · Logical equivalence and the laws of logic

> 📍 **Where we are.** We have met pairs of formulas with identical truth tables (implication and contrapositive;
> $p \Rightarrow q$ and $\lnot p \lor q$). Such pairs are "the same statement in different clothes". This section names
> the idea, lists the standard laws — the algebra of logic — and proves every one of them by brute force.

**Definitions.**

* Two propositions are **logically equivalent**, written $A \equiv B$, if they have the same truth value under **every**
  truth assignment — identical truth-table columns. Like the algebraic identity $(x + y)^2 = x^2 + 2xy + y^2$, which
  holds whatever $x$ and $y$ are, an equivalence holds whatever $p, q, r$ are.
* A **tautology** is a proposition that is true in every row (e.g. $p \lor \lnot p$). A **contradiction** is false in
  every row (e.g. $p \land \lnot p$). Anything else is **contingent**.
* $A \equiv B$ exactly when $A \Leftrightarrow B$ is a tautology.

Equivalences let us **simplify** formulas, **prove** statements by rewriting them into simpler known-true ones,
**transform** a statement into a more useful form (De Morgan's laws move a negation inside a formula), and see
relationships between connectives (an implication can be written with ¬ and ∨ only).

````python
# ── 7.1 Two tools: compare two formulas on every assignment; classify a single formula ──
def equivalent(f, g, variables: list):
    """(True, None) if f and g agree on every truth assignment, else (False, the first disagreeing assignment)."""
    for values in itertools.product([True, False], repeat=len(variables)):
        env = dict(zip(variables, values))
        if bool(f(**env)) != bool(g(**env)):
            return False, env
    return True, None


def classify(f, variables: list) -> str:
    """'tautology' (true in every row), 'contradiction' (false in every row) or 'contingent'."""
    column = [bool(f(**dict(zip(variables, v)))) for v in itertools.product([True, False], repeat=len(variables))]
    return "tautology" if all(column) else "contradiction" if not any(column) else "contingent"


for text, f in [("p ∨ ¬p", lambda p: p or not p), ("p ∧ ¬p", lambda p: p and not p),
                ("p ⇒ p", lambda p: implies(p, p)), ("p ⇒ ¬p", lambda p: implies(p, not p))]:
    print(f"{text:7s} is a {classify(f, ['p'])}")
ok, _ = equivalent(lambda p, q: iff(p, q), lambda p, q: implies(p, q) and implies(q, p), ["p", "q"])
print("\np ⇔ q ≡ (p ⇒ q) ∧ (q ⇒ p) ?", ok)
print("so (p ⇔ q) ⇔ ((p ⇒ q) ∧ (q ⇒ p)) is a",
      classify(lambda p, q: iff(iff(p, q), implies(p, q) and implies(q, p)), ["p", "q"]))
````

🔍 **What just happened?** `classify` confirmed that $p \lor \lnot p$ (the law of the excluded middle) and
$p \Rightarrow p$ are tautologies, that $p \land \lnot p$ is a contradiction, and that $p \Rightarrow \lnot p$ is
merely contingent (it is true exactly when $p$ is false — it is equivalent to $\lnot p$). `equivalent` then proved the
**mutual-implication law** $p \Leftrightarrow q \equiv (p \Rightarrow q) \land (q \Rightarrow p)$ — the reason ⇔ is
called the *bi*conditional — and the last line shows the link between the two ideas: an equivalence $A \equiv B$ is the
same as the tautology $A \Leftrightarrow B$.

### 7.2 The laws of logic

| law | statement |
|---|---|
| commutativity | $p \lor q \equiv q \lor p$, $\;p \land q \equiv q \land p$, $\;p \oplus q \equiv q \oplus p$, $\;p \Leftrightarrow q \equiv q \Leftrightarrow p$ |
| associativity | $p \lor (q \lor r) \equiv (p \lor q) \lor r$, and likewise for ∧, ⊕ and ⇔ |
| distribution of ∧ over ∨ | $p \land (q \lor r) \equiv (p \land q) \lor (p \land r)$ |
| distribution of ∨ over ∧ | $p \lor (q \land r) \equiv (p \lor q) \land (p \lor r)$ |
| mutual implication | $(p \Rightarrow q) \land (q \Rightarrow p) \equiv p \Leftrightarrow q$ |
| De Morgan's laws | $\lnot(p \land q) \equiv \lnot p \lor \lnot q$, $\;\lnot(p \lor q) \equiv \lnot p \land \lnot q$ |
| identity | $p \land \text{True} \equiv p$, $\;p \lor \text{False} \equiv p$ |
| domination | $p \lor \text{True} \equiv \text{True}$, $\;p \land \text{False} \equiv \text{False}$ |
| idempotence | $p \lor p \equiv p$, $\;p \land p \equiv p$ |
| double negation | $\lnot(\lnot p) \equiv p$ |
| implication as disjunction; contrapositive | $p \Rightarrow q \equiv \lnot p \lor q \equiv \lnot q \Rightarrow \lnot p$ |

Just as in addition, where $(a + b) + c = a + (b + c)$ and $a + b = b + a$, the operators ∧, ∨, ⊕ and ⇔ are
**associative** (grouping does not matter) and **commutative** (order does not matter): "it is raining and it is cold"
says the same as "it is cold and it is raining", and "I'll study logic or math, or biology" the same as "logic, or math
or biology". **Implication is neither** — and **negation**, being unary, has only one operand, so the two properties do
not even apply to it. This is a classic "thought experiment"; the next cell finds the counterexamples.

````python
# ── 7.2 Verify the laws of logic by brute force — and catch the impostors ──
V1, V2, V3 = ["p"], ["p", "q"], ["p", "q", "r"]
LAWS = [
    ("commutativity of ∨", "p ∨ q ≡ q ∨ p", lambda p, q: p or q, lambda p, q: q or p, V2),
    ("commutativity of ∧", "p ∧ q ≡ q ∧ p", lambda p, q: p and q, lambda p, q: q and p, V2),
    ("commutativity of ⊕", "p ⊕ q ≡ q ⊕ p", lambda p, q: xor(p, q), lambda p, q: xor(q, p), V2),
    ("commutativity of ⇔", "p ⇔ q ≡ q ⇔ p", lambda p, q: iff(p, q), lambda p, q: iff(q, p), V2),
    ("associativity of ∨", "p ∨ (q ∨ r) ≡ (p ∨ q) ∨ r", lambda p, q, r: p or (q or r), lambda p, q, r: (p or q) or r, V3),
    ("associativity of ∧", "p ∧ (q ∧ r) ≡ (p ∧ q) ∧ r", lambda p, q, r: p and (q and r), lambda p, q, r: (p and q) and r, V3),
    ("associativity of ⊕", "p ⊕ (q ⊕ r) ≡ (p ⊕ q) ⊕ r", lambda p, q, r: xor(p, xor(q, r)), lambda p, q, r: xor(xor(p, q), r), V3),
    ("associativity of ⇔", "p ⇔ (q ⇔ r) ≡ (p ⇔ q) ⇔ r", lambda p, q, r: iff(p, iff(q, r)), lambda p, q, r: iff(iff(p, q), r), V3),
    ("distribution ∧ over ∨", "p ∧ (q ∨ r) ≡ (p ∧ q) ∨ (p ∧ r)", lambda p, q, r: p and (q or r),
     lambda p, q, r: (p and q) or (p and r), V3),
    ("distribution ∨ over ∧", "p ∨ (q ∧ r) ≡ (p ∨ q) ∧ (p ∨ r)", lambda p, q, r: p or (q and r),
     lambda p, q, r: (p or q) and (p or r), V3),
    ("mutual implication", "(p ⇒ q) ∧ (q ⇒ p) ≡ p ⇔ q", lambda p, q: implies(p, q) and implies(q, p),
     lambda p, q: iff(p, q), V2),
    ("De Morgan (∧)", "¬(p ∧ q) ≡ ¬p ∨ ¬q", lambda p, q: not (p and q), lambda p, q: (not p) or (not q), V2),
    ("De Morgan (∨)", "¬(p ∨ q) ≡ ¬p ∧ ¬q", lambda p, q: not (p or q), lambda p, q: (not p) and (not q), V2),
    ("identity", "p ∧ True ≡ p", lambda p: p and True, lambda p: p, V1),
    ("identity", "p ∨ False ≡ p", lambda p: p or False, lambda p: p, V1),
    ("domination", "p ∨ True ≡ True", lambda p: p or True, lambda p: True, V1),
    ("domination", "p ∧ False ≡ False", lambda p: p and False, lambda p: False, V1),
    ("idempotence", "p ∨ p ≡ p", lambda p: p or p, lambda p: p, V1),
    ("idempotence", "p ∧ p ≡ p", lambda p: p and p, lambda p: p, V1),
    ("double negation", "¬(¬p) ≡ p", lambda p: not (not p), lambda p: p, V1),
    ("implication as or", "p ⇒ q ≡ ¬p ∨ q", lambda p, q: implies(p, q), lambda p, q: (not p) or q, V2),
    ("contrapositive", "p ⇒ q ≡ ¬q ⇒ ¬p", lambda p, q: implies(p, q), lambda p, q: implies(not q, not p), V2),
    ("✘ ⇒ commutative?", "p ⇒ q ≡ q ⇒ p", lambda p, q: implies(p, q), lambda p, q: implies(q, p), V2),
    ("✘ ⇒ associative?", "(p ⇒ q) ⇒ r ≡ p ⇒ (q ⇒ r)", lambda p, q, r: implies(implies(p, q), r),
     lambda p, q, r: implies(p, implies(q, r)), V3),
    ("✘ naive negation", "¬(p ∧ q) ≡ ¬p ∧ ¬q", lambda p, q: not (p and q), lambda p, q: (not p) and (not q), V2),
]
rows = []
for name, text, f, g, variables in LAWS:
    ok, counter = equivalent(f, g, variables)
    witness = "" if ok else ", ".join(f"{k}={tf(v)}" for k, v in counter.items())
    rows.append({"law": name, "statement": text, "holds": "yes" if ok else "NO", "counterexample": witness})
print(pd.DataFrame(rows).to_string(index=False))
````

🔍 **What just happened?** All 22 genuine laws hold on every row of their truth tables — each "yes" is a complete proof,
because a truth table checks *every* possible case. The three impostors fail, and `equivalent` hands us the
counterexample: ⇒ is **not commutative** ($p$ = T, $q$ = F gives T ⇒ F = F but F ⇒ T = T); ⇒ is **not associative**
($p$ = F, $q$ = T, $r$ = F gives $(F \Rightarrow T) \Rightarrow F = T \Rightarrow F = F$ but
$F \Rightarrow (T \Rightarrow F) = F \Rightarrow F = T$) — so a chain like $p \Rightarrow q \Rightarrow r$ always needs
parentheses; and the tempting "distribute the ¬" rule $\lnot(p \land q) \equiv \lnot p \land \lnot q$ fails at
$p$ = T, $q$ = F. De Morgan's laws say what really happens: **negating flips ∧ into ∨ and ∨ into ∧**.

### 7.3 Truth tables for compound propositions

For a longer proposition, build the table **one sub-formula per column**, from the inside out. To find the truth table
of $(p \Rightarrow q) \land (q \lor p)$, compute the columns of $p \Rightarrow q$ and of $q \lor p$, and put a T in the
final column for precisely those rows in which both had a T.

````python
# ── 7.3 Compound propositions, one sub-formula per column ──
step = truth_table({"p ⇒ q": lambda p, q: implies(p, q), "q ∨ p": lambda p, q: q or p,
                    "(p ⇒ q) ∧ (q ∨ p)": lambda p, q: implies(p, q) and (q or p)}, ["p", "q"])
print(step.to_string(index=False))
second = truth_table({"p ∧ q": lambda p, q: p and q, "¬q": lambda p, q: not q,
                      "p ∧ q ⇒ ¬q": lambda p, q: implies(p and q, not q)}, ["p", "q"])
print()
print(second.to_string(index=False))
print("\n(p ⇒ q) ∧ (q ∨ p) ≡ q ?      ", equivalent(lambda p, q: implies(p, q) and (q or p), lambda p, q: q, ["p", "q"])[0])
print("p ∧ q ⇒ ¬q ≡ ¬p ∨ ¬q ?       ", equivalent(lambda p, q: implies(p and q, not q),
                                                    lambda p, q: (not p) or (not q), ["p", "q"])[0])
````

🔍 **What just happened?** The first table has a T in the last column exactly where both helper columns have one —
rows TT and FT — so $(p \Rightarrow q) \land (q \lor p)$ is true exactly when $q$ is true: it is a long way of writing
$q$. In the second table (read with precedence: $(p \land q) \Rightarrow \lnot q$) the only F is in the first row: the
proposition is true **precisely when at least one of $p$ and $q$ is false**, i.e. it is equivalent to
$\lnot p \lor \lnot q$ — which, by De Morgan, is $\lnot(p \land q)$.

### 7.4 De Morgan's laws

Negating a compound statement requires a little thought. **De Morgan's laws**:

$$\lnot(p \lor q) \equiv \lnot p \land \lnot q, \qquad \lnot(p \land q) \equiv \lnot p \lor \lnot q .$$

*In words:* "not ($p$ or $q$)" — neither is true — means "not $p$ **and** not $q$"; and "not both $p$ and $q$" means
"either not $p$ **or** not $q$ (or both)". "It is not true that I will eat pizza or pasta" means I eat neither; "it is not
true that it is raining and cold" means it is not raining, or it is not cold. The proof is a truth table with ten columns.

````python
# ── 7.4 De Morgan's laws: the ten-column truth-table proof ──
demorgan = truth_table({
    "p ∨ q": lambda p, q: p or q, "¬(p ∨ q)": lambda p, q: not (p or q),
    "¬p": lambda p, q: not p, "¬q": lambda p, q: not q, "¬p ∧ ¬q": lambda p, q: (not p) and (not q),
    "p ∧ q": lambda p, q: p and q, "¬(p ∧ q)": lambda p, q: not (p and q),
    "¬p ∨ ¬q": lambda p, q: (not p) or (not q)}, ["p", "q"])
print(demorgan.to_string(index=False))
print("\ncolumn ¬(p ∨ q) = column ¬p ∧ ¬q :", (demorgan["¬(p ∨ q)"] == demorgan["¬p ∧ ¬q"]).all())
print("column ¬(p ∧ q) = column ¬p ∨ ¬q :", (demorgan["¬(p ∧ q)"] == demorgan["¬p ∨ ¬q"]).all())
````

🔍 **What just happened?** Compare columns 4 and 7: $\lnot(p \lor q)$ and $\lnot p \land \lnot q$ are both F, F, F, T
— true only when $p$ and $q$ are both false. Columns 9 and 10, $\lnot(p \land q)$ and $\lnot p \lor \lnot q$, are both
F, T, T, T — false only when both are true. Identical columns on all four rows: both laws are proved.

The same laws hold for **sets**, with "not in" (complement), "in both" (∩) and "in either" (∪) in place of ¬, ∧, ∨ —
Module 03 builds on this. The picture: shade everything outside the union of $A$ and $B$, and you get exactly the
region that is outside $A$ and outside $B$.

````python
# ── 7.5 De Morgan as pictures: shade the region where a formula about membership in A and B is true ──
def shade_two_sets(region, ax, title: str):
    """Two-set Venn diagram inside a universe rectangle, shading every point where region(in_A, in_B) is true —
    including the area outside both circles (which the toolkit's draw_venn leaves unshaded)."""
    gx, gy = np.meshgrid(np.linspace(-2.1, 2.1, 211), np.linspace(-1.5, 1.5, 151))
    in_a = (gx + 0.55) ** 2 + gy ** 2 <= 1.0
    in_b = (gx - 0.55) ** 2 + gy ** 2 <= 1.0
    mask = np.vectorize(lambda a, b: bool(region(a, b)))(in_a, in_b)
    ax.contourf(gx, gy, mask.astype(float), levels=[0.5, 1.5], colors=["#ffd54f"])
    ax.add_patch(plt.Rectangle((-2.1, -1.5), 4.2, 3.0, fill=False, color="0.4"))
    for cx, label, colour in [(-0.55, "A", "tab:blue"), (0.55, "B", "tab:orange")]:
        ax.add_patch(plt.Circle((cx, 0), 1.0, fill=False, lw=2, color=colour))
        ax.text(cx + (-0.75 if cx < 0 else 0.75), 0.95, label, fontsize=11, fontweight="bold", color=colour)
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-1.6, 1.6)
    ax.set_aspect("equal")
    ax.axis("off")
    ax.set_title(title, fontsize=9)


panels = [("¬(A ∨ B)", lambda a, b: not (a or b)), ("¬A ∧ ¬B", lambda a, b: (not a) and (not b)),
          ("¬(A ∧ B)", lambda a, b: not (a and b)), ("¬A ∨ ¬B", lambda a, b: (not a) or (not b))]
fig, axes = plt.subplots(1, 4, figsize=(12, 2.8))
for ax, (title, region) in zip(axes, panels):
    shade_two_sets(region, ax, f"shaded: x ∈ {title}")
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** Each panel shades the points $x$ for which the formula about "$x$ is in A" and "$x$ is in B"
is true. Panels 1 and 2 are identical — only the area outside both circles is yellow — which is De Morgan's first law
drawn as a picture. Panels 3 and 4 are identical too: everything except the lens-shaped overlap, De Morgan's second
law. A drawing is not a proof (the truth table was), but it is a good way to *remember* which way the connective flips.

````python
# ── 7.6 De Morgan on the SIMULATED cells: "neither a T-lineage nor a B-lineage marker" ──
neither = ~(cd3 | cd19)                         # ¬(CD3⁺ ∨ CD19⁺)
both_negative = ~cd3 & ~cd19                    # ¬CD3⁺ ∧ ¬CD19⁺
wrong = ~cd3 | ~cd19                            # ¬CD3⁺ ∨ ¬CD19⁺ — a different set of cells
print("¬(CD3⁺ ∨ CD19⁺):", int(neither.sum()), "cells | ¬CD3⁺ ∧ ¬CD19⁺:", int(both_negative.sum()),
      "cells | identical cell by cell:", bool((neither == both_negative).all()))
print("¬CD3⁺ ∨ ¬CD19⁺ (the naive version):", int(wrong.sum()), "cells")
print("\n🧪 who is in the double-negative gate?")
print(cells.loc[neither, "population"].value_counts().to_string())
````

### 📘 Lecture notes — precedence, logical equivalence and the laws of logic

**Operators.** ¬ is unary; ∧, ∨, ⊕, ⇒, ⇔ are binary. **Precedence** (from "binds tightest" to "binds loosest"):
¬, then ∧ / ∨ / ⊕, then ⇒, then ⇔. So $p \land q \Rightarrow r$ is $(p \land q) \Rightarrow r$ (different from
$p \land (q \Rightarrow r)$, e.g. at $p$ = F, $q$ = T, $r$ = F), $\lnot p \land q$ is $(\lnot p) \land q$, and
$p \lor q \Rightarrow \lnot r \land \ell$ is $(p \lor q) \Rightarrow (\lnot r \land \ell)$. Conventions for ∧ vs ∨
differ between books and languages — use parentheses.

**Equivalence.** $A \equiv B$ if $A$ and $B$ have the same truth value under every assignment (identical truth-table
columns); then $A \Leftrightarrow B$ is a tautology. Tautology: always true ($p \lor \lnot p$); contradiction: always
false ($p \land \lnot p$).

**Laws.** Commutativity and associativity of ∧, ∨, ⊕, ⇔; distribution of ∧ over ∨ and of ∨ over ∧; mutual implication
$(p \Rightarrow q) \land (q \Rightarrow p) \equiv p \Leftrightarrow q$; De Morgan
$\lnot(p \land q) \equiv \lnot p \lor \lnot q$ and $\lnot(p \lor q) \equiv \lnot p \land \lnot q$; identity, domination,
idempotence, double negation; $p \Rightarrow q \equiv \lnot p \lor q \equiv \lnot q \Rightarrow \lnot p$. Implication is
neither commutative (T, F) nor associative (F, T, F); for the unary ¬ the properties do not apply.

**Worked examples.** $(p \Rightarrow q) \land (q \lor p)$ has column T, F, T, F ($\equiv q$). $p \land q \Rightarrow \lnot q$
has column F, T, T, T: true precisely when at least one of $p, q$ is false. De Morgan's ten-column table: columns
$\lnot(p \lor q)$ and $\lnot p \land \lnot q$ are F, F, F, T; columns $\lnot(p \land q)$ and $\lnot p \lor \lnot q$ are
F, T, T, T.

> 🎓 **Exam tip.** To prove an equivalence, write one column per sub-formula and compare the two final columns on all
> $2^k$ rows; to disprove one, a single row suffices — say which row.

## 8 · Predicate logic: free variables, predicates and quantifiers

> 📍 **Where we are.** So far every letter stood for a whole statement that is simply true or false. Science, however,
> constantly talks about *collections*: all genes on a chip, some patients in a trial, every cell in a sample.
> Propositional logic has no words for that. Predicate logic adds them.

**Bound and free variables.** Up to now we always knew whether $p$ and $q$ were true or false. Often a statement looks
more like this:

$$r \leftarrow \text{“}2 + x = 4\text{”}.$$

*In words:* $r$ is the name of the sentence "2 plus $x$ equals 4" — and without a definition of $x$ we cannot say
whether it is true. It is true for $x = 2$ and false for $x = 5$. The variable $x$ is **free** (unbound): nothing in the
sentence tells us its value. A sentence with a free variable is not yet a proposition; it is a **predicate**, waiting
for a value or a quantifier to **bind** the variable.

⚠️ **Common confusion — "tautology".** Propositions without free variables (like "2 + 2 = 4") have a definite truth
value, but that does not make them *tautologies*. A tautology is a formula that is true under **every** assignment of
its variables, such as $p \lor \lnot p$; "2 + 2 = 4" is simply a true proposition.

````python
# ── 8.1 A free variable: the truth of "2 + x = 4" depends on x ──
def two_plus_x_is_four(x) -> bool:
    """The predicate 2 + x = 4; x is free until we choose a value for it."""
    return 2 + x == 4


for x in [-1, 0, 1, 2, 3, 4.5]:
    print(f"x = {x!s:4}:  2 + x = 4  is  {two_plus_x_is_four(x)}")
print("values of x in -10…10 that make it true:", [x for x in range(-10, 11) if two_plus_x_is_four(x)])
````

🔍 **What just happened?** As a Python function, a predicate is just a Boolean formula with an input: until we call it
with a value, it has no truth value. For $x = 2$ it is `True`, for every other value we tried it is `False`. Choosing
a value for $x$ is the simplest way to *bind* the variable; the more powerful way is to quantify it.

**Why predicate logic.** Simple propositional logic has a significant limitation: it cannot deal with **properties of
individual objects within a collection**. "All students in this class passed" would be one indivisible letter $P$,
with no way to talk about a particular student or about "some" students. Predicate logic breaks statements into a
**predicate** — a property $P(x)$ that becomes true or false once $x$ is chosen — and **quantifiers** that say for how
many $x$ it must hold.

🧮 **Math refresher — sets of numbers, ∈ and |·|.** $\mathbb Z$ is the set of all integers $\{\dots, -2, -1, 0, 1, 2,
\dots\}$; $x \in S$ reads "$x$ is an element of $S$" or simply "$x$ in $S$" (Module 03 treats sets properly). In a
quantified statement the colon ":" reads "such that". $|a|$ is the absolute value — the distance of $a$ from 0:
$|{-4}| = |4| = 4$.

**Universal quantifier** [*for all*, ∀]. Let $P$ be a predicate over a set $S$. The proposition $\forall x \in S: P(x)$
is true if $P(x)$ is true for **every** $x \in S$. Example — "for any integer $x$, the value of $x \cdot 0$ is 0":

$$\forall x \in \mathbb Z:\; x \cdot 0 = 0 .$$

*In words:* "for all $x$ in the integers, $x$ times 0 equals 0". Notice that "$x \cdot 0 = 0$" on its own is a
predicate with a free $x$; the ∀ binds it, and the whole line is a proposition (a true one). The LaTeX command for ∀ is
`\forall`.

**Existential quantifier** [*there exists*, ∃]. The proposition $\exists x \in S: P(x)$ is true if $P(x)$ is true for
**at least one** $x \in S$. Sometimes we do not want to talk about all items but about one (or more) in particular:

$$\exists y \in \mathbb Z:\; y^2 = |y + y| .$$

*In words:* "there exists an integer $y$ such that $y$ squared equals the absolute value of $y + y$". It is true: $y = 0$
works ($0 = |0|$), and so do $y = 2$ ($4 = |4|$) and $y = -2$ ($4 = |-4|$). One witness would have been enough. The
LaTeX command for ∃ is `\exists`. (The standard name is *quantifier*; some texts write "qualifier", which is not the
technical term.)

**Quantifiers are loops.** If you looped through all the items of $S$, ∀ asks "did *every* item satisfy $P$?" — the
loop can stop at the first **counterexample** — and ∃ asks "did *some* item satisfy $P$?" — the loop can stop at the
first **witness**. Python's `all` and `any` are exactly these loops.

````python
# ── 8.2 Quantifiers are loops: ∀ = "no counterexample found", ∃ = "a witness found" ──
def for_all(predicate, domain) -> bool:
    """∀x ∈ domain: predicate(x) — stops at the first counterexample."""
    for x in domain:
        if not predicate(x):
            return False
    return True


def there_exists(predicate, domain) -> bool:
    """∃x ∈ domain: predicate(x) — stops at the first witness."""
    for x in domain:
        if predicate(x):
            return True
    return False


window = range(-1000, 1001)            # a finite window of ℤ; ℤ itself is infinite
print("∀x: x·0 = 0        :", for_all(lambda x: x * 0 == 0, window), "| all():", all(x * 0 == 0 for x in window))
print("∃y: y² = |y + y|   :", there_exists(lambda y: y ** 2 == abs(y + y), window),
      "| any():", any(y ** 2 == abs(y + y) for y in window))
print("   all witnesses   :", [y for y in window if y ** 2 == abs(y + y)])
print("∃x: x > 100        :", there_exists(lambda x: x > 100, window), "| first witness:",
      next(x for x in window if x > 100))
print("∀x: x² > 0         :", for_all(lambda x: x ** 2 > 0, window), "| first counterexample:",
      next(x for x in window if not x ** 2 > 0))
print("\nempty domain: all([]) =", all([]), "| any([]) =", any([]))
````

🔍 **What just happened?** Our hand-written loops and Python's `all`/`any` agree on every statement. On the window
−1000…1000, $\forall x: x \cdot 0 = 0$ is `True` and $\exists y: y^2 = |y+y|$ is `True` with exactly three witnesses,
−2, 0 and 2. That list is complete for all of $\mathbb Z$, not just the window: $|y + y| = 2|y|$ and $y^2 = |y|^2$, so
the equation says $|y|^2 = 2|y|$, i.e. $|y|\,(|y| - 2) = 0$, so $|y| = 0$ or $|y| = 2$. The claim $\forall x: x^2 > 0$
fails, and the loop reports its first counterexample, $x = 0$.

Two cautions. (1) $\mathbb Z$ is infinite, so a computer can **refute** a ∀-claim (one counterexample) or **confirm** an
∃-claim (one witness), but it can never *prove* a ∀-claim over an infinite set by checking — that needs an argument
(here: $x \cdot 0 = 0$ is a rule of arithmetic). (2) Over an **empty** domain, `all([])` is `True` and `any([])` is
`False`: "every element of ∅ has property $P$" is vacuously true — there is no counterexample — exactly like
"false implies anything".

### 8.3 One counterexample is enough

Biology is full of universal claims ("all cancers have a *TP53* mutation", "every B cell is CD3-negative"). In logic
such a claim is a ∀-statement, and it is refuted by a **single** counterexample. *TP53* is the most frequently mutated
gene in human cancer — yet large sequencing surveys find it mutated in well under all tumours: roughly a third overall, and far more often in some cancer types than in others. Below, a SIMULATED
cohort with a mutation probability of 0.45 (an illustrative value), and universal claims about our simulated cells.

````python
# ── 8.3 Universal claims meet data: a SIMULATED tumour cohort and the SIMULATED cells ──
rng_tumours = np.random.default_rng(SEED + 8)
tp53_mutated = rng_tumours.random(500) < 0.45          # SIMULATED: 500 tumours, each mutated with probability 0.45
print("∀ tumours: TP53 mutated ?", bool(tp53_mutated.all()), "| first counterexample: tumour index", int(np.argmin(tp53_mutated)))
print("∃ tumour : TP53 mutated ?", bool(tp53_mutated.any()), "| fraction mutated:", round(float(tp53_mutated.mean()), 3))

is_b = cells["population"] == "B cell"                 # 🧪 hidden truth (only known because we simulated)
is_t = cells["population"].isin(["CD4 T cell", "CD8 T cell"])
claims = {"∀ cells: B cell ⇒ CD3⁻": ("∀", ~is_b | ~cd3),      # p ⇒ q is ¬p ∨ q, cell by cell
          "∀ cells: T cell ⇒ CD3⁺": ("∀", ~is_t | cd3),
          "∃ cell : CD4⁺ ∧ CD8⁺": ("∃", cd4 & cd8)}
print()
for text, (kind, column) in claims.items():
    if kind == "∀":
        print(f"{text:24s} → {bool(column.all())!s:5}  ({int((~column).sum())} counterexamples)")
    else:
        print(f"{text:24s} → {bool(column.any())!s:5}  ({int(column.sum())} witness)")
print("\nthe counterexample B cells:")
print(cells[is_b & cd3].to_string())
````

🔍 **What just happened?** Tumour index 1 (the second tumour) is already unmutated, so "all tumours carry a *TP53*
mutation" is false; the existential claim is true, with 45.2 % of the cohort mutated. Among the cells, "every B cell is
CD3⁻" fails because of 2 counterexamples — B cells whose CD3 signal (2.05 and 2.10) crept over the threshold by noise —
and "every T cell is CD3⁺" fails for 2 cells as well. The existential claim "some cell is CD4⁺ and CD8⁺" is true with a
single witness. Biology and logic disagree about the word "all": a biologist reading "all B cells are CD3⁻" means "up
to measurement error and rare exceptions", a logician means *no exceptions at all*. Writing claims with explicit
quantifiers forces you to say which one you mean — and this is one reason why statistics (Modules 21–26) replaces "all"
with "a fraction" and "a probability".

### 📘 Lecture notes — predicates, free variables and quantifiers

**Free vs bound.** In $r \leftarrow$ "$2 + x = 4$" the variable $x$ is *free* (unbound): without a definition of $x$ we
cannot say whether the statement is true. A quantifier (or a chosen value) *binds* it.

**Predicates.** Propositional logic cannot deal with properties of individual objects within a collection; predicate
logic can. A predicate $P$ over a set $S$ assigns to every $x \in S$ a proposition $P(x)$.

**Quantifiers.**

$$\forall x \in S: P(x) \;\text{ is true iff } P(x) \text{ holds for every } x \in S; \qquad \exists x \in S: P(x) \;\text{ is true iff } P(x) \text{ holds for at least one } x \in S .$$

*In words:* "for all $x$ in $S$, $P(x)$" needs no exceptions; "there exists an $x$ in $S$ such that $P(x)$" needs one
example. The colon means "such that"; ∀ is `\forall` and ∃ is `\exists` in LaTeX.

**Worked examples.** $\forall x \in \mathbb Z: x \cdot 0 = 0$ (true). $\exists y \in \mathbb Z: y^2 = |y + y|$ (true;
witnesses $-2, 0, 2$ — more than one). $\exists x \in \mathbb Z: x > 100$ (true, e.g. 101).

**Quantifiers as loops.** ∀ = `all(...)`: loop over the items and stop at the first counterexample; ∃ = `any(...)`:
stop at the first witness. Over an empty set ∀ is (vacuously) true and ∃ is false.

**Common confusions.** One counterexample refutes a ∀-statement; one witness proves an ∃-statement; checking finitely
many cases never proves a ∀-statement about an infinite set. "Tautology" means true under every assignment, not
"has a known truth value".

> 🎓 **Exam tip.** Before deciding whether a quantified statement is true, write down the set the variable ranges over:
> "$\exists x: x^2 = 2$" is false over $\mathbb Z$ but true over $\mathbb R$.

### 🌉 Bridge — from logic to sets, probability and statistics

Every idea of this module has a twin in the next ones. A predicate defines a **set**, $\{x \in S : P(x)\}$; then ∧, ∨,
¬ become intersection, union and complement, and $p \Rightarrow q$ becomes $P \subseteq Q$ ([Module 03 · 🧺 Sets and functions: collections and mappings](03_sets_and_functions.ipynb)). In probability an
**event** is such a set of outcomes, "A and B" is $A \cap B$, and De Morgan's laws compute
$P(\text{neither } A \text{ nor } B) = 1 - P(A \cup B)$ ([Module 16 · 🎲 Probability: sample spaces, events and axioms](16_probability_spaces.ipynb)); BiologyToML
[02 · probability for biologists](../../../../ml/tutorial/BiologyToML/notebooks/02_probability_for_biologists.ipynb) uses exactly these rules. Quantifiers
reappear in definitions: "$f \in O(g)$ iff there **exist** $c$ and $x_0$ such that **for all** $x \ge x_0$, …"
([Module 05 · ⏱️ Asymptotics: how fast functions grow](05_asymptotics_growth_of_functions.ipynb)), and in hypothesis testing, where "we fail to reject $H_0$" is a carefully weaker statement than
"$H_0$ is true" ([Module 25 · 🚦 Hypothesis testing: errors, power, z- and t-tests](25_hypothesis_testing.ipynb)).

## 9 · Negating quantifiers and nested quantifiers

> 📍 **Where we are.** We can now write quantified statements. Two skills remain, and both are needed constantly in
> proofs: saying precisely what the **opposite** of a quantified statement is, and reading statements with **several**
> quantifiers, where order changes the meaning.

**Negation — De Morgan again.** Negating a quantified statement is not as simple as putting "not" in front; De Morgan's
laws come to the rescue once more:

$$\lnot\,\bigl[\forall x: P(x)\bigr] \;\Longleftrightarrow\; \bigl[\exists x: \lnot P(x)\bigr], \qquad
\lnot\,\bigl[\exists x: P(x)\bigr] \;\Longleftrightarrow\; \bigl[\forall x: \lnot P(x)\bigr].$$

*In words:* "it is false that everybody is $P$" is equivalent to "somebody isn't $P$"; and "it is false that somebody
is $P$" is equivalent to "everybody isn't $P$". The negation moves inside, and on its way it **flips the quantifier**.
"Not all students passed" means "some student failed" (not "no student passed"); "there is no unicorn in this
classroom" means "every thing in this classroom is not a unicorn". Notice that what is left to do, $\lnot P(x)$, is just
the negation of a proposition — and if $P(x)$ is compound, the propositional De Morgan's laws finish the job.

Why these are De Morgan's laws: over a finite set $S = \{a, b, c\}$, "$\forall x: P(x)$" is the long conjunction
$P(a) \land P(b) \land P(c)$ and "$\exists x: P(x)$" the long disjunction $P(a) \lor P(b) \lor P(c)$; negating flips
∧ into ∨ and ∨ into ∧.

````python
# ── 9.1 Check De Morgan for quantifiers on 2,000 random finite "worlds" ──
rng_worlds = np.random.default_rng(SEED + 9)
agree_forall, agree_exists = 0, 0
for _ in range(2000):
    size = int(rng_worlds.integers(0, 8))                     # a random finite set of 0–7 objects …
    P_values = rng_worlds.random(size) < rng_worlds.random()  # … and a random truth value P(x) for each object
    agree_forall += (not all(P_values)) == any(not v for v in P_values)
    agree_exists += (not any(P_values)) == all(not v for v in P_values)
print("¬∀x P(x) ⇔ ∃x ¬P(x) held in", agree_forall, "of 2000 worlds")
print("¬∃x P(x) ⇔ ∀x ¬P(x) held in", agree_exists, "of 2000 worlds")

# a compound predicate on the SIMULATED T cells: "every T cell is CD3⁺ and CD19⁻"
claim = (cd3 & ~cd19)[is_t].all()                  # ∀ T cells: CD3⁺ ∧ ¬CD19⁺
negation = (~cd3 | cd19)[is_t].any()               # ∃ T cell : ¬CD3⁺ ∨ CD19⁺  (quantifier flipped, then De Morgan)
print("\n∀ T cells: CD3⁺ ∧ CD19⁻ :", bool(claim), "| its negation ∃ T cell: CD3⁻ ∨ CD19⁺ :", bool(negation))
````

🔍 **What just happened?** In all 2,000 random worlds — including empty ones, where "for all" is vacuously true and
"there exists" false — the two negation rules held. On the simulated cells, the claim "every T cell is CD3⁺ and CD19⁻"
is false and its correctly formed negation, "some T cell is CD3⁻ **or** CD19⁺", is true (the two noisy CD3⁻ T cells of
Section 8 are its witnesses). Note the double flip: ∀ became ∃ (quantifier De Morgan) and ∧ became ∨ (propositional De
Morgan). Exactly one of a statement and its negation is always true.

### 9.2 Nested quantifiers: order matters

In predicate logic we can combine several quantifiers in one expression — this is called **nesting** — and changing
their order can completely change the meaning. Read nested quantifiers from the **outside in**: fix a value for the
outer variable, then deal with the inner one.

$$\forall x \in \mathbb Z:\; \bigl[\exists y \in \mathbb Z:\; x = -1 \cdot y\bigr].$$

*In words:* "for every integer $x$ there exists an integer $y$ such that $x = -1$ times $y$". Outer quantifier: pick
any integer $x$. Inner quantifier: for **that** $x$, find a $y$. Predicate: $x = -y$. It is **true**: choose $y = -x$
(for $x = 5$ take $y = -5$, for $x = -10$ take $y = 10$, for $x = 0$ take $y = 0$). The witness $y$ is allowed to
**depend on** $x$.

$$\exists y \in \mathbb Z:\; \bigl[\forall x \in \mathbb Z:\; x = -1 \cdot y\bigr].$$

*In words:* "there exists an integer $y$ such that for every integer $x$, $x = -y$" — one single $y$ that works for all
$x$ at once. It is **false**: whatever $y$ you fix, $-y$ is one particular number $k$, and "every integer equals $k$" is
false. No single $y$ can produce all possible $x$ values. Swapping the quantifiers turned a true statement into a false
one. (The same holds for $\exists x\, \forall y: x = -y$: no single $x$ equals $-y$ for every $y$.)

⚠️ **Common confusion.** When you swap ∀ and ∃, keep each variable attached to its own quantifier — write
$\exists y\,\forall x$ (or $\exists x\,\forall y$), not a version where the letters silently change roles. In
$\forall x\,\exists y$ the inner $y$ may depend on $x$; in $\exists y\,\forall x$ it may not. An everyday example: "for
every person there is a car they own" (each person may own a different car) versus "there is a car that every person
owns" (one car shared by everybody).

````python
# ── 9.2 Order matters: ∀x ∃y (x = −y) versus ∃y ∀x (x = −y), on the window −5…5 ──
W = range(-5, 6)
forall_exists = all(any(x == -1 * y for y in W) for x in W)
exists_forall = any(all(x == -1 * y for x in W) for y in W)
print("∀x ∃y: x = −y :", forall_exists, "| a witness y for each x:", {x: next(y for y in W if x == -1 * y) for x in W})
print("∃y ∀x: x = −y :", exists_forall)
for y in [-1, 0, 2]:
    print(f"   fixing y = {y:2d} fails already at x = {next(x for x in W if x != -1 * y)}")
````

🔍 **What just happened?** The ∀∃ statement is `True`, and the dictionary shows its strategy: each $x$ gets its own
witness, $y = -x$ (−5 ↦ 5, 0 ↦ 0, 5 ↦ −5). The ∃∀ statement is `False`: fixing $y = -1$, $0$ or $2$ fails at the very
first $x$ tried ($-5$), because a single $y$ can only match one $x$. In code the difference is the nesting of the
loops: `all(any(...) for x)` lets the inner search restart for every $x$; `any(all(...) for y)` demands one $y$ that
survives the whole inner loop.

### 9.3 Negating nested quantifiers

Negating nested quantifiers is a **recursive** procedure: start with the outermost quantifier, apply the negation one
layer at a time moving inward, and flip each quantifier on the way (¬∀ → ∃¬, ¬∃ → ∀¬):

$$\lnot\bigl[\forall x\,\exists y: P(x,y)\bigr] \;\Leftrightarrow\; \exists x\,\lnot\bigl[\exists y: P(x,y)\bigr] \;\Leftrightarrow\; \exists x\,\forall y: \lnot P(x,y).$$

*In words:* "it is false that every $x$ has a partner $y$" means "some $x$ has no partner at all". The same steps give

$$\lnot\bigl[\exists x\,\forall y: P(x,y)\bigr] \;\Leftrightarrow\; \forall x\,\lnot\bigl[\forall y: P(x,y)\bigr] \;\Leftrightarrow\; \forall x\,\exists y: \lnot P(x,y).$$

*In words:* "there is no $x$ that works with every $y$" means "for every $x$ there is some $y$ that it does not work
with". (Some sources write this example with the letter $x$ for both quantifiers, $\lnot[\exists x: [\forall x: P(x)]]$;
the logic is the same, but two different letters keep the two variables apart and avoid confusion.)

A table of truth values $P(x, y)$ — rows $x$, columns $y$ — makes every nested statement a NumPy one-liner:
$\exists y: P(x,y)$ for each row is `P.any(axis=1)`, $\forall x$ of that is `.all()`.

````python
# ── 9.3 Negating nested quantifiers, checked on 1,000 random tables P[x, y] ──
rng_tables = np.random.default_rng(SEED + 10)
checks = {"¬[∀x ∃y P] ⇔ ∃x ∀y ¬P": 0, "¬[∃x ∀y P] ⇔ ∀x ∃y ¬P": 0,
          "∃x ∃y P ⇔ ∃y ∃x P": 0, "∀x ∃y P ⇔ ∃y ∀x P   (✘)": 0}
for _ in range(1000):
    P = rng_tables.random((4, 5)) < rng_tables.random()          # rows = x (4 values), columns = y (5 values)
    checks["¬[∀x ∃y P] ⇔ ∃x ∀y ¬P"] += (not P.any(axis=1).all()) == (~P).all(axis=1).any()
    checks["¬[∃x ∀y P] ⇔ ∀x ∃y ¬P"] += (not P.all(axis=1).any()) == (~P).any(axis=1).all()
    checks["∃x ∃y P ⇔ ∃y ∃x P"] += P.any(axis=1).any() == P.any(axis=0).any()
    checks["∀x ∃y P ⇔ ∃y ∀x P   (✘)"] += P.any(axis=1).all() == P.all(axis=0).any()
for statement, count in checks.items():
    print(f"{statement:28s} held in {count:4d} of 1000 random tables")
````

🔍 **What just happened?** Both recursive negation rules held in all 1,000 random tables. Quantifiers **of the same
kind** commute — $\exists x\,\exists y$ equals $\exists y\,\exists x$ in every table (and likewise $\forall\forall$). But
the mixed swap "$\forall x\,\exists y \Leftrightarrow \exists y\,\forall x$" held in only 713 tables and failed in
the other 287 — whenever every row has a T but no column is entirely T. (One direction always holds: if a single $y$ works for every
$x$, then certainly every $x$ has some working $y$.)

### 9.4 The grid: row-major or column-major, "no cell is filled ≡ every cell is empty"

To decide whether any cell of a grid (a matrix $A$ with $n$ rows and $m$ columns) is filled, you can scan it in
**row-major** order (row by row) or **column-major** order (column by column):

```text
row-major                          column-major
for i = 1 to n:                    for j = 1 to m:
    for j = 1 to m:                    for i = 1 to n:
        if A[i, j] then                    if A[i, j] then
            return True                        return True
return False                       return False
```

If every cell is inspected, the result is identical, whichever order is used: $\exists r\,\exists c: P(r, c)$ and
$\exists c\,\exists r: P(r, c)$ are the same statement. And what does it mean for "some cell is filled" to be **false**?
Two ways of saying it — (1) there does not exist a filled cell, (2) every cell is unfilled:

$$\lnot\,\bigl[\exists r\;\exists c: P(r, c)\bigr] \;\equiv\; \forall r\;\forall c: \lnot P(r, c).$$

*In words:* "no cell is filled" is the same as "every cell is empty". Our biological grid is a 96-well screening plate
(rows A–H, columns 1–12), where "filled" means "the compound in this well was a **hit**".

````python
# ── 9.4 A SIMULATED 96-well screening plate: rows A–H × columns 1–12, True = hit ──
rng_plate = np.random.default_rng(SEED + 1)
plate = rng_plate.random((8, 12)) < 0.15             # SIMULATED: each well is a hit with probability 0.15
ROWS = list("ABCDEFGH")
print(pd.DataFrame(np.where(plate, "●", "·"), index=ROWS, columns=range(1, 13)).to_string())
statements = {
    "∃r ∃c: hit      (some well is a hit)": plate.any(),
    "∀r ∀c: ¬hit     (no well is a hit)": (~plate).all(),
    "∀r ∃c: hit      (every row has a hit)": plate.any(axis=1).all(),
    "∃c ∀r: hit      (some column is all hits)": plate.all(axis=0).any(),
    "∀c ∃r: hit      (every column has a hit)": plate.any(axis=0).all(),
}
print()
for text, value in statements.items():
    print(f"{text:44s} {bool(value)}")


def first_hit(grid, row_major: bool = True):
    """Scan the plate like the pseudo-code above; return (found?, wells inspected, first hit well)."""
    n_rows, n_cols = grid.shape
    order = ([(i, j) for i in range(n_rows) for j in range(n_cols)] if row_major
             else [(i, j) for j in range(n_cols) for i in range(n_rows)])
    for step, (i, j) in enumerate(order, start=1):
        if grid[i, j]:
            return True, step, f"{ROWS[i]}{j + 1}"
    return False, len(order), None


print("\nrow-major scan   :", first_hit(plate, row_major=True))
print("column-major scan:", first_hit(plate, row_major=False))
````

🔍 **What just happened?** The simulated plate has 12 hits (●). "Some well is a hit" is `True`, so its negation "no
well is a hit" is `False` — exactly one of a statement and its negation holds. "Every row has a hit" ($\forall r\,\exists c$)
is `True` — each of the rows A–H contains a ●; but "some column is all hits" ($\exists c\,\forall r$) is `False`, and so
is "every column has a hit": the order and the kind of quantifiers change the claim. Both scans answer "yes, there is a
hit", as they must, but they meet a different first hit after a different number of steps: the row-major scan stops
at A3 after inspecting 3 wells, the column-major scan at D1 after 4 wells. The answer to ∃∃ does not depend on the
order; the work does.

````python
# ── 9.5 Picture the two scanning orders on the SIMULATED plate ──
fig, axes = plt.subplots(1, 2, figsize=(11, 3.6))
for ax, row_major in zip(axes, [True, False]):
    hit_r, hit_c = np.nonzero(plate)
    ax.scatter(np.tile(np.arange(12), 8), np.repeat(np.arange(8), 12), s=60, facecolors="none", edgecolors="0.75")
    ax.scatter(hit_c, hit_r, s=60, color="tab:red", zorder=3, label="hit")
    if row_major:
        for i in range(8):
            ax.annotate("", xy=(11.3, i), xytext=(-0.3, i), arrowprops=dict(arrowstyle="-|>", color="0.3", lw=0.8))
    else:
        for j in range(12):
            ax.annotate("", xy=(j, 7.3), xytext=(j, -0.3), arrowprops=dict(arrowstyle="-|>", color="0.3", lw=0.8))
    found, steps, well = first_hit(plate, row_major)
    ax.set_xticks(range(12))
    ax.set_xticklabels(range(1, 13))
    ax.set_yticks(range(8))
    ax.set_yticklabels(ROWS)
    ax.invert_yaxis()
    ax.set_xlabel("column")
    ax.set_ylabel("row")
    ax.set_title(f"{'row-major' if row_major else 'column-major'} scan: first hit {well} after {steps} wells")
axes[0].legend(loc="lower right", fontsize=7)
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** The arrows show the two orders in which the nested loops visit the 96 wells: along the rows
(left) or down the columns (right). Both routes cover every well, so both settle "∃ a hit" — and if the plate had no
hit, both would inspect all 96 wells before answering "every well is empty". The titles report where each scan
stopped. In NumPy you rarely write these loops: `plate.any()`, `plate.any(axis=1).all()` and

### 📘 Lecture notes — negation and nested quantifiers

**De Morgan for quantifiers.**
$\lnot[\forall x \in S: P(x)] \Leftrightarrow [\exists x \in S: \lnot P(x)]$ and
$\lnot[\exists x \in S: P(x)] \Leftrightarrow [\forall x \in S: \lnot P(x)]$ — "everybody is P is false" ≡ "somebody
isn't P"; "somebody is P is false" ≡ "everybody isn't P". After flipping the quantifier, negate the predicate with the
propositional rules.

**Nested quantifiers** combine several quantifiers; the order can completely change the meaning. Read them from the
outside in. $\forall x \in \mathbb Z\,[\exists y \in \mathbb Z: x = -1 \cdot y]$ is **true** (take $y = -x$);
$\exists y \in \mathbb Z\,[\forall x \in \mathbb Z: x = -1 \cdot y]$ is **false** (no single $y$ produces all $x$).
Quantifiers of the same kind commute ($\exists r\,\exists c \equiv \exists c\,\exists r$); mixed ones do not.

**Negating nested quantifiers** is recursive: start outside, apply the negation one layer at a time moving inward,
flipping each quantifier: $\lnot[\exists x\,\forall y: P(x,y)] \Leftrightarrow \forall x\,\lnot[\forall y: P(x,y)]
\Leftrightarrow \forall x\,\exists y: \lnot P(x,y)$ (use two different letters for the two variables).

**The grid.** Row-major and column-major scans inspect the same cells and give the same answer. "No cell is filled"
≡ "every cell is empty": $\lnot[\exists r\,\exists c: P(r,c)] \equiv \forall r\,\forall c: \lnot P(r,c)$.

> 🎓 **Exam tip.** To negate, push the ¬ inward one symbol at a time and write every intermediate line; at the end, no ¬
> should stand in front of a quantifier or a parenthesis. Then translate your answer back into English as a check.

## 10 · Logic in action: Boolean networks of gene regulation

> 📍 **Where we are.** Everything is in place to build a real model. In a **Boolean network** each gene is either ON (1)
> or OFF (0), and each gene's next state is a Boolean formula of the current states of its regulators — exactly like the
> *lac* rule of Section 1. Iterating the formulas shows how the whole circuit behaves over time.

**Definitions.** For $n$ genes the **state** of the cell is a tuple of $n$ truth values, so there are $2^n$ possible
states (8 for 3 genes). In a *synchronous* Boolean network all genes update together at each time step:
$x_i(t+1) = f_i\bigl(x_1(t), \dots, x_n(t)\bigr)$, where each $f_i$ is a Boolean formula. Because the number of states
is finite and the update is deterministic, every trajectory must eventually revisit a state and then repeat forever.
The states it ends up cycling through form an **attractor**: a single state that maps to itself (a **fixed point** —
a stable expression programme, such as a cell type) or a **cycle** of several states (an **oscillation**).

**The repressilator.** In 2000 Michael Elowitz and Stanislas Leibler built a synthetic circuit in *E. coli* from three
repressor genes arranged in a loop: LacI represses *tetR*, TetR represses the λ phage repressor gene *cI*, and cI
represses *lacI*. As a Boolean network ("a gene is ON at the next step if and only if its repressor is OFF now"):

$$\text{lacI}(t+1) = \lnot\,\text{cI}(t), \qquad \text{tetR}(t+1) = \lnot\,\text{lacI}(t), \qquad \text{cI}(t+1) = \lnot\,\text{tetR}(t).$$

*In words:* each gene is switched on in the next step exactly when the gene that represses it is currently off — three
negations chained in a loop.

````python
# ── 10.1 The repressilator as a Boolean network: the complete state-transition table ──
GENES = ("lacI", "tetR", "cI")


def repressilator_step(state: tuple) -> tuple:
    """Synchronous update of (lacI, tetR, cI): cI ⊣ lacI, LacI ⊣ tetR, TetR ⊣ cI."""
    lacI, tetR, cI = state
    return (not cI, not lacI, not tetR)


def as_bits(state: tuple) -> str:
    """(True, False, True) → '101' (order lacI, tetR, cI)."""
    return "".join("1" if s else "0" for s in state)


all_states = list(itertools.product([False, True], repeat=3))
transitions = {as_bits(s): as_bits(repressilator_step(s)) for s in all_states}
print("state (lacI tetR cI) → next state")
for state, nxt in transitions.items():
    print(f"        {state}        →    {nxt}")
````

🔍 **What just happened?** `itertools.product` listed all $2^3 = 8$ states and the update rule gave each one a unique
successor — the network is a **function** from the set of states to itself (functions are the topic of Module 03). Read
one line: from 100 (only *lacI* on), LacI keeps *tetR* off, *cI* is on because TetR is absent, and *lacI* stays on
because cI was absent: next state 101. Notice that every state appears exactly once on the right-hand side as well.

````python
# ── 10.2 Follow every state until it repeats: the cycles reached are the attractors ──
def find_attractors(step_map: dict) -> list:
    """Attractor cycles of a deterministic map given as a dict state → next state."""
    attractors = []
    for start in step_map:
        path, state = [], start
        while state not in path:                     # walk until a state repeats (it must: finitely many states)
            path.append(state)
            state = step_map[state]
        cycle = path[path.index(state):]
        if not any(set(cycle) == set(a) for a in attractors):
            attractors.append(cycle)
    return attractors


attractors = find_attractors(transitions)
for cycle in attractors:
    print(f"attractor with {len(cycle)} state(s): {' → '.join(cycle)} → {cycle[0]}")
````

🔍 **What just happened?** The network has two attractors and no fixed point. The **6-cycle**
001 → 011 → 010 → 110 → 100 → 101 → 001 is the oscillation the repressilator was built for: the "on" state is handed
around the loop, each gene switching on as its repressor switches off — the Boolean shadow of the oscillating
fluorescence Elowitz and Leibler observed in living bacteria. The **2-cycle** 000 ↔ 111 (all genes flipping together)
is an artefact of updating all genes at exactly the same instant; real molecules never stay perfectly synchronised,
which is one reason modellers also use *asynchronous* updates. A loop with an **odd** number of repressions has no
fixed point at all — try to find a state that maps to itself.

````python
# ── 10.3 Picture the dynamics: the state-transition graph and one trajectory over time ──
six_cycle, two_cycle = attractors[1], attractors[0]
pos = {s: (np.cos(np.pi / 2 - 2 * np.pi * k / 6), np.sin(np.pi / 2 - 2 * np.pi * k / 6)) for k, s in enumerate(six_cycle)}
pos.update({two_cycle[0]: (2.4, 0.6), two_cycle[1]: (2.4, -0.6)})
node_colours = {s: "#c8e6c9" for s in six_cycle} | {s: "#ffe0b2" for s in two_cycle}
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), gridspec_kw={"width_ratios": [1, 1.3]})
draw_graph(list(transitions.items()), nodes=list(pos), directed=True, pos=pos, ax=axes[0],
           node_colors=node_colours, title="state-transition graph (lacI tetR cI)")

state, trajectory = (True, False, False), []
for t in range(13):
    trajectory.append([int(s) for s in state])
    state = repressilator_step(state)
axes[1].imshow(np.array(trajectory).T, cmap="Greens", aspect="auto", vmin=-0.2, vmax=1.2)
axes[1].set_yticks(range(3))
axes[1].set_yticklabels(GENES)
axes[1].set_xticks(range(13))
axes[1].set_xlabel("time step t")
axes[1].set_title("trajectory from state 100: dark = ON (period 6)")
plt.tight_layout()
plt.show()
````

🔍 **What just happened?** Left: each of the 8 states is a node and each arrow is one application of the update rule —
a directed graph (Module 06). The six green states form one directed cycle (the oscillation), the two orange states the
synchronous-update artefact; every state has exactly one outgoing arrow, because the update rule is a function. Right:
starting from 100, each gene is ON for three steps and OFF for three, *lacI*, *cI* and *tetR* switching on in turn, and
the pattern repeats with period 6 — a three-gene logical circuit that keeps time.

### 🌉 Bridge — Boolean network models of gene regulation

Stuart Kauffman (1969) proposed that the attractors of large random Boolean networks behave like **cell types**:
stable patterns of gene activity to which the network returns after a disturbance. René Thomas (1973) formalised
Boolean models of genetic control circuits and formulated the rules — later proved as theorems — that a positive
feedback loop is needed for **multiple stable states** (memory, differentiation) and a negative loop for sustained
**oscillations** (as above). Today Boolean models are used for
signalling networks, cell-fate decisions and the prediction of drug-combination effects, because they need only the
wiring and the logic — no rate constants. They are the deterministic, two-valued cousins of the probabilistic models
of the course: the state-transition graph is a graph (Module 06), letting each transition happen with a probability turns
it into a Markov chain ([Module 20 · ⛓️ Graphical models: Bayesian networks and Markov chains](20_bayesian_networks_markov_chains.ipynb)), and replacing "ON/OFF" by probabilities of being ON leads to Bayesian networks (BiologyToML
[15 · Bayesian networks](../../../../ml/tutorial/BiologyToML/notebooks/15_bayesian_networks_representation.ipynb)). Synthetic biology uses the same logic in the
other direction: Gardner, Cantor and Collins (2000) built a genetic **toggle switch** — two mutually repressing genes —
whose two fixed points store one bit of memory.

## 📌 Key takeaways

- A **proposition** is a statement that is true or false; atomic propositions are Boolean variables, compound ones are
  built with the connectives ¬, ∧, ∨, ⊕, ⇒, ⇔, whose **truth tables** ($2^k$ rows) are their definitions.
- Translating English needs care: "or" is inclusive unless it is ⊕; "but" is "and"; "if" sometimes means "iff";
  parentheses are part of the meaning ($v \lor (w \land x) \ne (v \lor w) \land x$).
- **Implication** $p \Rightarrow q$ is false only for T ⇒ F; it is not causation; a false antecedent makes it
  (vacuously) true; over a collection of cases it means $P \subseteq Q$. It equals its **contrapositive**
  $\lnot q \Rightarrow \lnot p$, but not its **converse** — knockouts test necessity through the contrapositive.
- **Precedence**: ¬, then ∧/∨/⊕, then ⇒, then ⇔ — but conventions for ∧ vs ∨ differ, so write parentheses. A formula
  is a tree; precedence decides its shape and therefore its truth value.
- **Logical equivalence** = identical truth tables; the laws of logic (commutativity, associativity, distributivity,
  mutual implication, De Morgan) are proved by checking every row. ⇒ is neither commutative nor associative.
- **De Morgan**: negation flips ∧ ↔ ∨ and ∀ ↔ ∃. "Not all" means "some not"; "not some" means "all not".
- **Predicates** with free variables become propositions when the variables are bound by **quantifiers**; ∀ = `all`
  (one counterexample refutes it), ∃ = `any` (one witness proves it).
- With **nested quantifiers**, order matters: $\forall x\,\exists y$ lets $y$ depend on $x$, $\exists y\,\forall x$
  does not. Negate them one layer at a time.
- **Boolean networks** turn gene-regulation rules into dynamics; their attractors (fixed points, cycles) model stable
  cell states and oscillations.
