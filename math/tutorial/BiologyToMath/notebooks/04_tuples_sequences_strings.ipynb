## 🧫 The biological question

**Why is a 20-letter guide RNA enough to find one single place in a 3-billion-letter genome?** CRISPR–Cas9 genome
editing works because the Cas9 protein is steered by a guide RNA whose 20-nucleotide "spacer" pairs with the target
DNA (next to a short PAM motif, NGG for the commonly used *Streptococcus pyogenes* Cas9). A geneticist designing a
guide must ask: *how many different 20-letter DNA words exist, and how likely is it that my word occurs somewhere
else in the genome by chance?*

To answer, we must treat DNA not as a *set* of bases — the set of bases in almost every gene is just {A, C, G, T} —
but as an **ordered sequence**: ACG and GCA contain the same letters but encode different amino acids (threonine and
alanine). The mathematics of ordered collections — **tuples**, **Cartesian products** and **strings** — answers
the question in one line, and it reappears every time a biologist counts something ordered:

````python
# ── Course toolkit (identical to ../course_toolkit.py v1.0; copied so this notebook runs alone) ──
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np


def circle_layout(nodes: list) -> dict:
    """Place nodes evenly on a circle (first node at the top, clockwise)."""
    n = len(nodes)
    angles = np.pi / 2 - 2 * np.pi * np.arange(n) / max(n, 1)
    return {v: (float(np.cos(a)), float(np.sin(a))) for v, a in zip(nodes, angles)}


def tree_layout(edges: list, root) -> dict:
    """Positions for drawing a rooted tree top-down: root at the top, leaves evenly spaced.

    ``edges`` are (parent, child) pairs or undirected pairs; the tree is explored from ``root``.
    """
    nbrs = {}
    for u, v in edges:
        nbrs.setdefault(u, []).append(v)
        nbrs.setdefault(v, []).append(u)
    children, order, seen, stack = {}, [], {root}, [root]
    while stack:
        u = stack.pop()
        order.append(u)
        children[u] = [w for w in nbrs.get(u, []) if w not in seen]
        seen.update(children[u])
        stack.extend(reversed(children[u]))
    depth = {root: 0}
    for u in order:
        for w in children[u]:
            depth[w] = depth[u] + 1
    x_pos, next_leaf = {}, [0]

    def place(u):
        if not children[u]:
            x_pos[u] = float(next_leaf[0])
            next_leaf[0] += 1
        else:
            for w in children[u]:
                place(w)
            x_pos[u] = float(np.mean([x_pos[w] for w in children[u]]))

    place(root)
    return {u: (x_pos[u], -float(depth[u])) for u in order}


def draw_graph(edges: list, nodes=None, directed: bool = False, pos=None, ax=None, node_colors=None,
               edge_labels=None, title: str | None = None, font_size: int = 9):
    """Draw a small graph with matplotlib only (no extra packages needed).

    ``edges`` is a list of (u, v) pairs; ``nodes`` (optional) adds isolated nodes and fixes the
    order; ``pos`` maps node → (x, y) (default: ``circle_layout``).  Directed edges get arrows,
    self-loops are drawn as small rings above the node, ``edge_labels`` maps (u, v) → text (e.g. a
    weight) and ``node_colors`` maps node → colour.  Node circles grow with the length of their
    label.  Returns the Axes.
    """
    from matplotlib.transforms import offset_copy

    if nodes is None:
        nodes = []
        for u, v in edges:
            for w in (u, v):
                if w not in nodes:
                    nodes.append(w)
    pos = circle_layout(list(nodes)) if pos is None else pos
    if ax is None:
        _, ax = plt.subplots(figsize=(4.2, 3.6))
    radius = {v: max(0.62 * font_size * len(str(v)) / 2 + 4, 0.9 * font_size) for v in nodes}  # points
    for u, v in edges:
        if u == v:
            ring = offset_copy(ax.transData, fig=ax.figure, x=0, y=1.25 * radius[u], units="points")
            ax.scatter([pos[u][0]], [pos[u][1]], s=(1.3 * radius[u]) ** 2, facecolors="none",
                       edgecolors="0.35", linewidths=1.3, transform=ring, zorder=1)
            continue
        style = "-|>" if directed else "-"
        ax.annotate("", xy=pos[v], xytext=pos[u],
                    arrowprops=dict(arrowstyle=style, color="0.35", lw=1.3, mutation_scale=12,
                                    shrinkA=radius[u], shrinkB=radius[v],
                                    connectionstyle="arc3,rad=0.08" if directed else "arc3,rad=0"))
        if edge_labels and (u, v) in edge_labels:
            (x1, y1), (x2, y2) = pos[u], pos[v]
            ax.text((x1 + x2) / 2, (y1 + y2) / 2, str(edge_labels[(u, v)]), fontsize=font_size - 1,
                    ha="center", va="center", color="darkred", zorder=4,
                    bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85))
    xs = [pos[v][0] for v in nodes]
    ys = [pos[v][1] for v in nodes]
    colours = [(node_colors or {}).get(v, "#cfe2f3") for v in nodes]
    ax.scatter(xs, ys, s=[(2 * radius[v]) ** 2 for v in nodes], c=colours, edgecolors="0.2",
               linewidths=1.2, zorder=2, clip_on=False)
    for v in nodes:
        ax.text(pos[v][0], pos[v][1], str(v), ha="center", va="center", fontsize=font_size, zorder=3,
                clip_on=False)
    ax.margins(0.22)
    ax.set_aspect("equal", adjustable="datalim")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax


NUCLEOTIDES = ("A", "C", "G", "T")


COMPLEMENT = {"A": "T", "C": "G", "G": "C", "T": "A"}


# First-order Markov-chain transition probabilities P(next | current) for DNA, rows = current base,
# columns = next base, both in the order A, C, G, T.  Rounded to two decimals from the classic
# CpG-island ("+") and background ("−") models of Durbin et al. (1998), Table 3.1, which were
# estimated from human sequence.  In CpG islands a C is often followed by a G; elsewhere CG is rare
# because methylated C in CpG mutates to T.
CPG_ISLAND_TRANSITIONS = np.array([
    [0.18, 0.27, 0.43, 0.12],
    [0.17, 0.37, 0.27, 0.19],
    [0.16, 0.34, 0.37, 0.13],
    [0.08, 0.36, 0.38, 0.18],
])


BACKGROUND_TRANSITIONS = np.array([
    [0.30, 0.20, 0.29, 0.21],
    [0.32, 0.30, 0.08, 0.30],
    [0.25, 0.25, 0.29, 0.21],
    [0.18, 0.24, 0.29, 0.29],
])


def reverse_complement(seq: str) -> str:
    """Reverse complement of a DNA string (A↔T, C↔G, read backwards)."""
    return "".join(COMPLEMENT[b] for b in reversed(seq))


def simulate_dna(length: int, transitions=None, initial=None, seed: int = 0) -> str:
    """Random DNA string.  Without ``transitions`` the bases are i.i.d. uniform; otherwise they
    follow a first-order Markov chain with the given 4 × 4 transition matrix (order A, C, G, T)
    and initial distribution ``initial`` (default uniform)."""
    rng = np.random.default_rng(seed)
    initial = np.full(4, 0.25) if initial is None else np.asarray(initial, dtype=float)
    if transitions is None:
        return "".join(rng.choice(list(NUCLEOTIDES), size=length, p=initial))
    P = np.asarray(transitions, dtype=float)
    states = [rng.choice(4, p=initial)]
    for _ in range(length - 1):
        states.append(rng.choice(4, p=P[states[-1]]))
    return "".join(NUCLEOTIDES[s] for s in states)


def kmer_counts(seq: str, k: int) -> dict:
    """Counts of every overlapping k-mer (substring of length k) of ``seq``, sorted alphabetically."""
    counts = Counter(seq[i:i + k] for i in range(len(seq) - k + 1))
    return dict(sorted(counts.items()))

print("🧰 toolkit loaded:", ", ".join(sorted(['BACKGROUND_TRANSITIONS', 'COMPLEMENT', 'CPG_ISLAND_TRANSITIONS', 'NUCLEOTIDES', 'circle_layout', 'draw_graph', 'kmer_counts', 'reverse_complement', 'simulate_dna', 'tree_layout'])))
````

## 1 · Why order matters: sets versus ordered collections

> 📍 **Where we are.** Module 03 built everything from **sets**: collections in which only membership counts. A set
> is a well-defined collection of *distinct* objects, so $\{1, 2, 3\} = \{3, 2, 1\}$ — no duplicates and no order.
> That is exactly right for "*which* genes are up-regulated" or "*which* cell types are in this tissue". But many
> biological objects are **arrangements**, and a set cannot see arrangements.

**A real-world example: coordinates.** A field biologist records where a sample was collected as latitude and
longitude, say 31.2° N and 121.4° E (Shanghai). The *set* {latitude, longitude} = {31.2, 121.4} cannot distinguish
$(31.2,\ 121.4)$ from $(121.4,\ 31.2)$ — yet the second pair, read as (latitude, longitude), is not even a place on
Earth, because latitudes only run from −90° to +90°.

The same problem appears everywhere:

| Situation | As a set | As a tuple | Why the tuple is better |
|---|---|---|---|
| chessboard position | {3, 5} | (3, 5) | row vs column matters |
| point in the plane | {x, y} | (x, y) | (3, 5) ≠ (5, 3) |
| RGB colour | {255, 0, 0} | (255, 0, 0) | (255, 0, 0) is red, (0, 0, 255) is blue |
| codon | {A, C, G} | (A, C, G) | ACG codes for threonine, GCA for alanine |
| sampling site | {31.2, 121.4} | (31.2, 121.4) | latitude vs longitude |

````python
# ── 1.1 Sets forget order (and duplicates); tuples keep both ──
site_as_set_1, site_as_set_2 = {31.2, 121.4}, {121.4, 31.2}
site_1, site_2 = (31.2, 121.4), (121.4, 31.2)          # ordered pairs ⟨latitude, longitude⟩
print("as sets:   equal?", site_as_set_1 == site_as_set_2)
print("as tuples: equal?", site_1 == site_2)


def is_on_earth(point: tuple) -> bool:
    """True if point = (latitude, longitude) is a real place: latitude in [-90, 90], longitude in [-180, 180]."""
    latitude, longitude = point
    return -90 <= latitude <= 90 and -180 <= longitude <= 180


for p in (site_1, site_2):
    print(f"{p} read as (latitude, longitude) is a place on Earth? {is_on_earth(p)}")

print("\n{1, 2, 3} == {3, 2, 1}:", {1, 2, 3} == {3, 2, 1}, "  |  (1, 2, 3) == (3, 2, 1):", (1, 2, 3) == (3, 2, 1))
print("set('GATTACA')   =", sorted(set("GATTACA")), "→", len(set("GATTACA")), "elements")
print("tuple('GATTACA') =", tuple("GATTACA"), "→", len(tuple("GATTACA")), "components")
````

🔍 **What just happened?** The two *sets* are equal (`True`) because a set only records membership; the two *tuples*
are different (`False`) because position 1 holds the latitude and position 2 the longitude. Reading the swapped pair
as (latitude, longitude) gives a latitude of 121.4°, which `is_on_earth` rejects. The last two lines show the second
thing a set throws away: duplicates. The 7-letter word GATTACA becomes the 4-element set {A, C, G, T}, while its
tuple keeps all **7** letters in order, with their repeats (three A's and two T's).

### 1.2 Order changes the biology

The genetic code reads DNA (or RNA) three letters at a time. A **codon** is therefore an ordered triple of bases,
not a set of three bases.

````python
# ── 1.2 Same letters, different codons: ACG, GCA and CAG ──
mini_code = {"ACG": "Thr (threonine)", "GCA": "Ala (alanine)", "CAG": "Gln (glutamine)"}
for codon in ["ACG", "GCA", "CAG"]:
    print(f"codon {codon}: as a set {sorted(set(codon))}, as a tuple {tuple(codon)} → {mini_code[codon]}")
print("\nsame SET of bases?  ", set("ACG") == set("GCA") == set("CAG"))
print("same TUPLE of bases?", tuple("ACG") == tuple("GCA"), tuple("GCA") == tuple("CAG"))
````

🔍 **What just happened?** All three codons are built from the same set {A, C, G}, yet the ribosome reads them as three
*different* amino acids — threonine, alanine and glutamine. A mutation that only *rearranges* bases (an inversion of a
short segment, for instance) can therefore change a protein even though the "set of bases" is untouched. Whenever
position carries information, we need an ordered collection.

## 2 · Tuples: components, names and notation

> 📍 **Where we are.** We know *why* we need ordered collections. Now we define them precisely, fix the notation that
> 02-680 uses, and say exactly when two tuples are equal.

A **sequence** — also called a **list** or a **tuple** — is an *ordered* collection of objects, which are called its
**components** or **entries**. The order of the components matters, and **duplicates are allowed**. When the number
of components is small we use special names:

| Number of components | Name | Example |
|---|---|---|
| 2 | (ordered) pair | ⟨0, 1⟩ |
| 3 | triple | ⟨Northfield, Minnesota, USA⟩ |
| 4 | quadruple | ⟨HBB, 6, E, V⟩ |
| $n$ | $n$-tuple (or $n$-tuplet) | $\langle a_1, a_2, \dots, a_n\rangle$ |

**Notation.** When being precise, mathematicians (and the 02-680 textbook by Liben-Nowell) use **angle brackets**:
$\langle 128, 128, 0\rangle$. Very often we are lazy and write parentheses, $(128, 128, 0)$. What we never do is use
curly brackets, because $\{\ \}$ is reserved for **sets**. So $\langle 4, 9\rangle$, $(4, 9)$ and $\{4, 9\}$ are three
ways of writing — respectively — a pair, the same pair, and a *set* that forgets the order.

**Components by position.** The $i$-th component of $x = \langle x_1, \dots, x_n\rangle$ is written $x_i$ (or
sometimes $x[i]$). Mathematics counts from 1; Python counts from 0, so $x_1$ is `x[0]` in code.

**Equality of tuples.** Two tuples are equal when they have the same length *and* agree position by position:

$$\langle x_1, \dots, x_n\rangle = \langle y_1, \dots, y_m\rangle \iff n = m \;\wedge\; \forall i \in \{1, \dots, n\}: x_i = y_i .$$

*In words:* tuples are equal exactly when they have the same number of components and the first components agree,
the second components agree, and so on — the order of the components is part of the tuple's identity. Compare sets,
where $\{4, 9\} = \{9, 4\}$ and $\{4, 4\} = \{4\}$.

````python
# ── 2.1 Equality of tuples, implemented straight from the definition ──
def tuples_equal(x: tuple, y: tuple) -> bool:
    """⟨x1,…,xn⟩ = ⟨y1,…,ym⟩  ⇔  n = m and x_i = y_i for every position i (the definition as a loop)."""
    if len(x) != len(y):                     # n = m ?
        return False
    for i in range(len(x)):                  # ∀ i: x_i = y_i ?
        if x[i] != y[i]:
            return False
    return True


tests = [((4, 9), (4, 9)), ((4, 9), (9, 4)), ((4, 4), (4,)), ((1, 2, 3), (1, 2)), ((), ())]
for x, y in tests:
    print(f"{str(x):10s} vs {str(y):8s}: by the definition {str(tuples_equal(x, y)):5s} | Python's == {x == y}")

print("\nduplicates: the pair (4, 4) has", len((4, 4)), "components; the set {4, 4} has", len({4, 4}), "element")
place = ("Northfield", "Minnesota", "USA")
print("x_1 (maths) = x[0] (Python) =", place[0], "| x_3 = x[2] =", place[2], "| x[-1] (last) =", place[-1])
````

🔍 **What just happened?** Our loop agrees with Python's built-in `==` on every test: ⟨4, 9⟩ equals itself but not
⟨9, 4⟩; a pair is never equal to a 1-tuple, even if the components look alike; and the two **empty tuples** `()` are
equal (both have length 0, and "for every position" is vacuously true — the vacuous truth of Module 02). The pair
(4, 4) has two components, while the set {4, 4} has one element. Finally, the first component $x_1$ of
⟨Northfield, Minnesota, USA⟩ is `place[0]` in Python: keep the 1-based (maths) and 0-based (code) indexing apart.

### 2.2 A tuple you look at every day: RGB colours

When we tell a computer how to display a colour, we often give its **RGB value**: the tuple
⟨red, green, blue⟩ of three intensities between 0 and 255. The order is important — the next cell reads the same seven

````python
# ── 2.2 RGB tuples (typed in): the order of the components decides the colour ──
RAINBOW_RGB = {"Violet": (128, 0, 128), "Indigo": (74, 0, 130), "Blue": (0, 0, 255), "Green": (0, 255, 0),
               "Yellow": (255, 255, 0), "Orange": (255, 128, 0), "Red": (255, 0, 0)}

fig, axes = plt.subplots(1, 2, figsize=(9, 3.4))
for ax, backwards in [(axes[0], False), (axes[1], True)]:
    for row, (name, rgb) in enumerate(RAINBOW_RGB.items()):
        shown = rgb[::-1] if backwards else rgb                   # reverse the order of the components
        ax.add_patch(plt.Rectangle((0, -row), 1.0, 0.85, color=[c / 255 for c in shown]))
        ax.text(1.15, -row + 0.42, f"{name:7s} read as {shown}", va="center", family="monospace", fontsize=8)
    ax.set_xlim(0, 4.2); ax.set_ylim(-6.3, 1.0); ax.axis("off")
axes[0].set_title("tuples read as ⟨red, green, blue⟩")
axes[1].set_title("the same tuples read backwards, as ⟨blue, green, red⟩")
plt.tight_layout(); plt.show()

unchanged = [name for name, rgb in RAINBOW_RGB.items() if rgb == rgb[::-1]]
print("colours that survive reversal:", unchanged, "— their tuples read the same in both directions")
````

### 2.3 Biological tuples: variants and genotypes

A protein variant is naturally a quadruple ⟨gene, position, reference amino acid, variant amino acid⟩. The classic
example is the sickle-cell variant of β-globin, traditionally written HBB E6V (glutamate → valine at residue 6; residue
7 if the initiator methionine is counted). Genotypes raise a subtler question: a diploid cell carries two copies of
each chromosome, so the genotype at a single nucleotide position is a *pair* of alleles. Is it ordered?

````python
# ── 2.3 A variant is a quadruple; a genotype is a pair — ordered (phased) or not (unphased) ──
sickle = ("HBB", 6, "E", "V")            # ⟨gene, residue, reference amino acid, variant amino acid⟩
print("variant tuple:", sickle, "→", len(sickle), "components (a quadruple)")
print("swapping reference and variant gives a different variant:", sickle != ("HBB", 6, "V", "E"))

alleles = ["A", "T"]
phased = list(itertools.product(alleles, repeat=2))                 # ⟨allele from mother, allele from father⟩
print("\nphased genotypes (ordered pairs):  ", phased, "→", len(phased))
unphased = sorted({tuple(sorted(pair)) for pair in phased})         # forget the order, keep both copies
print("unphased genotypes (sorted pairs): ", unphased, "→", len(unphased))
print("as Python SETS, AA and AT would be", sorted({"A", "A"}), "and", sorted({"A", "T"}),
      "— the set {A, A} = {A} forgets the second copy")
````

🔍 **What just happened?** The sickle-cell variant is a 4-tuple, and order matters inside it: ⟨HBB, 6, E, V⟩ ≠
⟨HBB, 6, V, E⟩ (the second would be the reverse mutation). For genotypes, a **phased** genotype records which allele
came from the mother and which from the father, so it is an ordered pair — 4 possibilities ⟨A,A⟩, ⟨A,T⟩, ⟨T,A⟩,
⟨T,T⟩. Most genotyping assays cannot tell ⟨A,T⟩ from ⟨T,A⟩: the **unphased** genotype is a pair whose order is
forgotten but whose copies are kept, giving the 3 genotypes AA, AT, TT that Hardy–Weinberg proportions describe
(Module 18). A *set* is the wrong model for both: the set {A, A} is {A}, which cannot tell a homozygote AA from a
"genotype" with one allele. The fix used in the code — sort the pair, keep it as a tuple — is the standard way to
represent an unordered collection *with* repeats (a **multiset**).

### 📘 Lecture notes — sets versus tuples

**Sets** contain a collection of distinct elements where order does not matter: $\{1, 2, 3\} = \{3, 2, 1\}$, no
duplicates, no order — suitable for representing *which* elements are in a group. **But in many contexts order
matters**: sets cannot distinguish between different arrangements of the same elements (the set {latitude, longitude}
cannot tell $(31.2, 121.4)$ from $(121.4, 31.2)$).

| Situation | Set | Tuple | Why the tuple is better |
|---|---|---|---|
| Chessboard position | {3, 5} | (3, 5) | row vs column matters |
| 2D point | {x, y} | (x, y) | (3, 5) ≠ (5, 3) |
| RGB colour | {255, 0, 0} | (255, 0, 0) | red ≠ blue |

**Definition.** A **sequence** (also known as a **list** or **tuple**) is an *ordered* collection of objects, called
its **components** or **entries**. Unlike sets, order matters and duplicates are allowed.

**Names.** With 2, 3, 4 or $n$ components: (ordered) **pair**, **triple**, **quadruple**, **$n$-tuple** (or
$n$-tuplet). Examples: ⟨0, 1⟩, ⟨Northfield, Minnesota, USA⟩, (128, 128, 0).

**Notation.** Precisely, angle brackets $\langle a_1, a_2, \dots, a_n\rangle$; lazily, parentheses
$(a_1, a_2, \dots, a_n)$; **never** curly brackets, which denote sets. Component $i$ of $x$ is $x_i$ (or $x[i]$).

**Equality.** $\langle x_1..x_n\rangle = \langle y_1..y_m\rangle \iff n = m \wedge \forall i: x_i = y_i$. Hence
$\langle 4, 9\rangle \ne \langle 9, 4\rangle$ although $\{4, 9\} = \{9, 4\}$, and $\langle 4, 4\rangle$ is a genuine
pair although $\{4, 4\} = \{4\}$.

**Example (RGB).** Colours are triples ⟨red, green, blue⟩: Violet ⟨128, 0, 128⟩, Indigo ⟨74, 0, 130⟩, Blue
⟨0, 0, 255⟩, Green ⟨0, 255, 0⟩, Yellow ⟨255, 255, 0⟩, Orange ⟨255, 128, 0⟩, Red ⟨255, 0, 0⟩.

**Common confusions.** (i) "Same elements" is not "same tuple": ACG, GCA and CAG have equal sets of letters but are
different codons. (ii) Python counts positions from 0, mathematics from 1. (iii) A collection that is unordered but has
repeats (an unphased genotype AA) is neither a set nor a tuple — represent it as a sorted tuple (a multiset).

> 🎓 **Exam tip.** When asked to model something, first ask *does position carry meaning?* and *can an element
> repeat?* — if either answer is yes, use a tuple. Be able to say whether two given tuples are equal and *why* (length?
> which position differs?), and what the set { } of the same symbols loses.

## 3 · The Cartesian product: building sets of tuples

> 📍 **Where we are.** We can write single tuples. Very often we need the set of *all* tuples whose components come
> from given sets — all possible codons, all conditions of an experiment, all points of the plane. One operator builds
> them all: the **Cartesian product**, named after René Descartes, who described points of the plane by pairs of
> numbers.

$$S \times T = \{\langle x, y\rangle \mid x \in S \wedge y \in T\}.$$

*In words:* $S\times T$ ("$S$ cross $T$") is the set of all ordered pairs whose **first** component comes from $S$
and whose **second** component comes from $T$ — every element of $S$ paired with every element of $T$.

**Example.** Take the first two example sets of Module 03: $A = \{\text{"Welcome"}, \text{"to"}, \text{"02-680"}\}$ and
$B = \{x^2 \mid x = 2 \vee x = 3\} = \{4, 9\}$. Then

$$A\times B = \{\langle\text{"Welcome"}, 4\rangle, \langle\text{"to"}, 4\rangle, \langle\text{"02-680"}, 4\rangle,
\langle\text{"Welcome"}, 9\rangle, \langle\text{"to"}, 9\rangle, \langle\text{"02-680"}, 9\rangle\}.$$

*In words:* each of the 3 words is paired with each of the 2 numbers, giving 6 ordered pairs, word first.

The two sets need not be different: the product of a set with itself is so common that it has its own notation,
$B\times B = B^2 = \{\langle 4,4\rangle, \langle 4,9\rangle, \langle 9,4\rangle, \langle 9,9\rangle\}$. Another small
example: $\{0, 1\}\times\{2, 3\} = \{\langle 0,2\rangle, \langle 0,3\rangle, \langle 1,2\rangle, \langle 1,3\rangle\}$.

````python
# ── 3.1 The Cartesian product as a set comprehension (set-builder notation in Python) ──
def cartesian(S, T) -> set:
    """S × T = {⟨x, y⟩ | x ∈ S ∧ y ∈ T}: every element of S paired with every element of T, S first."""
    return {(x, y) for x in S for y in T}


A = {"Welcome", "to", "02-680"}
B = {x ** 2 for x in range(10) if x == 2 or x == 3}            # B = {x² | x = 2 ∨ x = 3}
print("B =", sorted(B))
A_x_B = cartesian(A, B)
print("A × B =", sorted(A_x_B, key=str), "→", len(A_x_B), "pairs")
print("B × A =", sorted(cartesian(B, A), key=str))
print("A × B = B × A ?", A_x_B == cartesian(B, A))
print("B² = B × B =", sorted(cartesian(B, B)))
print("{0, 1} × {2, 3} =", sorted(cartesian({0, 1}, {2, 3})))
print("A × ∅ =", cartesian(A, set()), "| agrees with itertools.product?", A_x_B == set(itertools.product(A, B)))
````

🔍 **What just happened?** The comprehension `{(x, y) for x in S for y in T}` is a word-for-word translation of
$\{\langle x, y\rangle \mid x\in S \wedge y \in T\}$, and it agrees with Python's `itertools.product`. $A\times B$ has
the six pairs listed above (Python sorts strings by character code, so "02-680" comes before "Welcome" and "to" —
the order of *printing* is irrelevant, since $A\times B$ is a set). $B\times A$ contains ⟨4, 'Welcome'⟩ instead of
⟨'Welcome', 4⟩, so $A\times B \neq B\times A$: **the Cartesian product is not commutative**. $B^2$ has four pairs
including ⟨4, 4⟩ and ⟨9, 9⟩, and a product with the empty set is empty — there is no way to choose the second
component.

### 3.2 How many tuples? The multiplication principle

**Claim.** For finite sets, $|A\times B| = |A|\cdot|B|$.

**Proof.** List the pairs by their first component. For each of the $|A|$ possible first components $a$ there are
exactly $|B|$ pairs $\langle a, b\rangle$, one for every $b\in B$, and pairs with different first components are
different tuples. So the pairs fall into $|A|$ disjoint groups of size $|B|$:

$$|A\times B| = \underbrace{|B| + |B| + \dots + |B|}_{|A| \text{ times}} = |A|\cdot|B| .$$

*In words:* if you can choose the first component in $|A|$ ways and, independently, the second in $|B|$ ways, you can
choose the pair in $|A|\cdot|B|$ ways. This **multiplication principle** is the root of almost every counting argument
in biology and in probability (Module 16). The picture of $\Sigma\times\Sigma$ for the DNA alphabet makes it visible:
a 4 × 4 grid of the 16 **dinucleotides**.

````python
# ── 3.2 Picture of a Cartesian product: Σ × Σ for Σ = {A, C, G, T}, the 16 dinucleotides ──
SIGMA = list(NUCLEOTIDES)
dinucleotides = list(itertools.product(SIGMA, SIGMA))
print("|Σ × Σ| =", len(dinucleotides), "= |Σ| · |Σ| =", len(SIGMA), "·", len(SIGMA))

fig, ax = plt.subplots(figsize=(4.8, 4.4))
for i, first in enumerate(SIGMA):
    for j, second in enumerate(SIGMA):
        colour = {("C", "G"): "tab:red", ("G", "C"): "tab:blue"}.get((first, second), "0.88")
        ax.scatter(j, -i, s=1100, color=colour, zorder=1)
        ax.text(j, -i, f"⟨{first},{second}⟩", ha="center", va="center", fontsize=8, zorder=2,
                color="white" if colour != "0.88" else "black")
ax.set_xticks(range(4)); ax.set_xticklabels([f"y = {b}" for b in SIGMA])
ax.set_yticks([0, -1, -2, -3]); ax.set_yticklabels([f"x = {b}" for b in SIGMA])
ax.set_xlim(-0.6, 3.6); ax.set_ylim(-3.6, 0.6)
ax.set_xlabel("second component y ∈ Σ"); ax.set_ylabel("first component x ∈ Σ")
ax.set_title("Σ × Σ = 4 rows × 4 columns = 16 ordered pairs\n⟨C,G⟩ (red) and ⟨G,C⟩ (blue) are different elements")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** Every row of the grid is one choice of the first base and every column one choice of the
second, so the grid has $4\times 4 = 16$ cells — the multiplication principle as a picture. The two coloured cells are
the dinucleotides ⟨C, G⟩ ("CpG") and ⟨G, C⟩ ("GpC"). As *sets* they are both {C, G}; as tuples they are different
elements of $\Sigma\times\Sigma$, and in Section 8 we will see that real genomes treat them very differently.

### 3.3 Products of more than two sets

The Cartesian product of the sets $A_1, A_2, \dots, A_n$ is the set of ordered $n$-tuples whose $i$-th component comes
from $A_i$:

$$A_1\times A_2\times\dots\times A_n = \{\langle a_1, a_2, \dots, a_n\rangle \mid a_i \in A_i \text{ for } i = 1, 2, \dots, n\},
\qquad |A_1\times\dots\times A_n| = |A_1|\cdot|A_2|\cdots|A_n| .$$

*In words:* choose the first component from $A_1$, the second from $A_2$, …, the last from $A_n$; every such choice is
one element, and the number of elements is the product of the sizes (the multiplication principle applied $n-1$
times).

**Question.** What is $A\times B\times C$ for $A = \{0, 1\}$, $B = \{1, 2\}$ and $C = \{0, 1, 2\}$? Try it on paper
before running the cell — list the triples in "dictionary" (lexicographic) order: first by $a$, then by $b$, then by
$c$.

````python
# ── 3.3 A × B × C for A = {0, 1}, B = {1, 2}, C = {0, 1, 2} ──
A3, B3, C3 = [0, 1], [1, 2], [0, 1, 2]
triples = list(itertools.product(A3, B3, C3))           # lexicographic order: a, then b, then c
print("A × B × C has", len(triples), "triples = |A|·|B|·|C| =", len(A3), "·", len(B3), "·", len(C3))
print("first row (a = 0): ", triples[:6])
print("second row (a = 1):", triples[6:])
````

🔍 **What just happened?** There are $2\cdot 2\cdot 3 = 12$ triples, and `itertools.product` lists them in exactly the
lexicographic order you (hopefully) used: $(0,1,0), (0,1,1), (0,1,2), (0,2,0), (0,2,1), (0,2,2)$, then the same six
with first component 1. The rightmost component changes fastest — like the digits of an odometer.

### 3.4 An annoying pedantic point: $(A\times B)\times C$ versus $A\times B\times C$

The 02-680 textbook (Liben-Nowell, §2.4) admits that we are being sloppy: the Cartesian product was defined for **two**
sets, so $S\times S\times S$ "must" mean either $S\times(S\times S)$ or $(S\times S)\times S$. The elements of
$(\{0,1\}\times\{0,1\})\times\{0,1\}$ are pairs whose first component is itself a pair, such as
$\langle\langle 0, 1\rangle, 1\rangle$; the elements of $\{0,1\}\times(\{0,1\}\times\{0,1\})$ look like
$\langle 0, \langle 1, 1\rangle\rangle$. Strictly, neither is the triple $\langle 0, 1, 1\rangle$, so strictly
$(A\times B)\times C \neq A\times B\times C$. We ignore this and simply write
$\langle 0, 1, 1\rangle \in \{0,1\}\times\{0,1\}\times\{0,1\}$. Why is that harmless?

````python
# ── 3.4 Nested pairs vs flat triples: different objects, but a perfect one-to-one correspondence ──
flat, left_nested, right_nested = (0, 1, 1), ((0, 1), 1), (0, (1, 1))
print("⟨0,1,1⟩ = ⟨⟨0,1⟩,1⟩ ?", flat == left_nested, "| ⟨0,1,1⟩ = ⟨0,⟨1,1⟩⟩ ?", flat == right_nested)
print("lengths:", len(flat), len(left_nested), len(right_nested), "(a triple vs two pairs)")

bits = [0, 1]
left_product = set(itertools.product(itertools.product(bits, bits), bits))     # ({0,1} × {0,1}) × {0,1}
flat_product = set(itertools.product(bits, bits, bits))                         # {0,1} × {0,1} × {0,1}


def flatten_left(t: tuple) -> tuple:
    """⟨⟨a, b⟩, c⟩ ↦ ⟨a, b, c⟩."""
    (a, b), c = t
    return (a, b, c)


print("\n|(B×B)×B| =", len(left_product), "| |B×B×B| =", len(flat_product), "| sets equal?", left_product == flat_product)
print("flattening maps one onto the other exactly:", {flatten_left(t) for t in left_product} == flat_product)
````

🔍 **What just happened?** As Python objects the flat triple and the two nested pairs are all different (`False`,
`False`), and they do not even have the same length — so the sets $(B\times B)\times B$ and $B\times B\times B$ are
not equal. But they have the same size (8) and the function `flatten_left` turns every nested pair
$\langle\langle a,b\rangle,c\rangle$ into the triple $\langle a,b,c\rangle$, hitting every triple exactly once. Because
this correspondence is perfect (a *bijection*, Module 03), nothing we care about — counting, listing, reasoning —
depends on the bracketing, and mathematicians drop the inner brackets. Notice that order matters at every level:
$\langle 4, 9\rangle \neq \langle 9, 4\rangle$.

⚠️ **Common confusion.** $A\times B$ and $B\times A$ have the same *size* but are different sets unless $A = B$ (or
one of them is empty). And $A\times\varnothing = \varnothing$, not $A$: with nothing to choose for the second
component, there are no pairs at all.

### 🌉 Bridge — experimental designs are Cartesian products

A **full factorial design** tests every combination of the levels of every factor: it *is* a Cartesian product. An
immunologist who studies a knockout mouse might cross genotype {wild type, *Tlr4* knockout} (TLR4 is the receptor for
bacterial lipopolysaccharide, LPS) × treatment {vehicle, LPS, LPS + drug} × time {0, 6, 24 h} × biological replicate
{1, 2, 3}. The multiplication principle tells you the bench work and the sequencing budget *before* you start. The
infection experiment that runs through this course (`simulate_tissue_expression`, Modules 03, 07, 08, …) is the
product {spleen, kidney, liver} × {control, infected} × {replicate 1, …, 4}.

````python
# ── 3.5 A factorial design (illustrative) is a Cartesian product: list it and count it ──
factors = {"genotype": ["wild type", "Tlr4 knockout"],
           "treatment": ["vehicle", "LPS", "LPS + drug"],
           "time_h": [0, 6, 24],
           "replicate": [1, 2, 3]}
design = pd.DataFrame(list(itertools.product(*factors.values())), columns=list(factors))
print("samples:", len(design), "=", " × ".join(str(len(v)) for v in factors.values()))
print("distinct conditions (ignoring replicates):", len(design.drop(columns="replicate").drop_duplicates()))
print(design.head(7).to_string(index=False))

infection = list(itertools.product(["spleen", "kidney", "liver"], ["control", "infected"], [1, 2, 3, 4]))
print("\ncourse infection experiment:", len(infection), "samples, e.g.", infection[0], "…", infection[-1])
````

🔍 **What just happened?** The design table has $2\times 3\times 3\times 3 = 54$ rows — 54 mice to house and 54 RNA
libraries to sequence — of which 18 distinct conditions (54 / 3 replicates). The printed rows show the odometer order
again: replicate changes fastest, genotype slowest. The course's infection experiment is the product
$3\times 2\times 4 = 24$ samples, from ('spleen', 'control', 1) to ('liver', 'infected', 4) — exactly the 24 rows of the
expression matrix you met in Module 03 and will use again in Modules 07 and 08. Real studies rarely afford *every* combination of many factors
($2^{10} = 1024$ conditions for ten on/off factors!), which is why statisticians invented fractional designs; but
the full product is always the starting point.

### 📘 Lecture notes — the Cartesian product

**Definition (two sets).** $S\times T = \{\langle x, y\rangle \mid x\in S \wedge y\in T\}$ — equivalently
$A\times B = \{\langle a, b\rangle : a\in A \text{ and } b\in B\}$: all ordered pairs whose first component comes from
$A$ and whose second component comes from $B$. It is a very useful way to **construct sets of tuples**: in essence it
creates all possible pairs of elements from two sets.

**Worked example.** $A = \{\text{"Welcome"}, \text{"to"}, \text{"02-680"}\}$, $B = \{x^2 \mid x = 2\vee x = 3\} = \{4, 9\}$:
$A\times B = \{\langle\text{"Welcome"},4\rangle, \langle\text{"to"},4\rangle, \langle\text{"02-680"},4\rangle,
\langle\text{"Welcome"},9\rangle, \langle\text{"to"},9\rangle, \langle\text{"02-680"},9\rangle\}$.
A set with itself: $B\times B = B^2 = \{\langle 4,4\rangle, \langle 4,9\rangle, \langle 9,4\rangle, \langle 9,9\rangle\}$.
Also $\{0,1\}\times\{2,3\} = \{\langle 0,2\rangle, \langle 0,3\rangle, \langle 1,2\rangle, \langle 1,3\rangle\}$.

**$n$ sets.** $A_1\times\dots\times A_n = \{(a_1, \dots, a_n) \mid a_i\in A_i \text{ for } i = 1, \dots, n\}$, the set of
ordered $n$-tuples. Example: $\{0,1\}\times\{1,2\}\times\{0,1,2\} = \{(0,1,0), (0,1,1), (0,1,2), (0,2,0), (0,2,1),
(0,2,2), (1,1,0), (1,1,1), (1,1,2), (1,2,0), (1,2,1), (1,2,2)\}$ — 12 triples.

**Size.** $|A\times B| = |A|\cdot|B|$ and $|A_1\times\dots\times A_n| = \prod_i |A_i|$ (multiplication principle).

**The pedantic point.** Strictly, $(A\times B)\times C$ (elements $\langle\langle a,b\rangle,c\rangle$) is not the same
set as $A\times B\times C$ (elements $\langle a,b,c\rangle$); we ignore the difference and write
$\langle 0,1,1\rangle\in\{0,1\}\times\{0,1\}\times\{0,1\}$ instead of
$\langle\langle 0,1\rangle,1\rangle\in(\{0,1\}\times\{0,1\})\times\{0,1\}$, because flattening is a bijection.

**Common confusions.** $A\times B\ne B\times A$ in general (order matters: $\langle 4,9\rangle\ne\langle 9,4\rangle$);
$A\times\varnothing = \varnothing$.

> 🎓 **Exam tip.** Expect to list a small product in lexicographic order, to state its size *before* listing it, and to
> explain why $(A\times B)\times C$ and $A\times B\times C$ are "the same" only up to re-bracketing.

## 4 · Relations: subsets of a Cartesian product

> 📍 **Where we are.** A Cartesian product contains *all* pairs. Interesting structure appears when we keep only the
> pairs that satisfy some condition. Such a subset is called a **relation** — and two objects of this course are
> relations in disguise: functions (Module 03) and directed graphs (Module 06).

A **relation** from $S$ to $T$ is any subset $R\subseteq S\times T$; when $T = S$ we speak of a relation *on* $S$. We
write $a\,R\,b$, or $(a,b)\in R$, when the pair satisfies the relation.

**Question.** What are the ordered pairs of the "less than or equal to" relation on $\{0, 1, 2, 3\}$, which contains
$(a, b)$ if $a\le b$?

**Solution.** $(a,b)$ belongs to $R$ if and only if both $a$ and $b$ belong to $\{0,1,2,3\}$ and $a\le b$:

$$R_{\le} = \{(a, b)\in\{0,1,2,3\}^2 \mid a \le b\} = \{(0,0), (0,1), (0,2), (0,3), (1,1), (1,2), (1,3), (2,2), (2,3), (3,3)\}.$$

*In words:* out of the 16 pairs of $\{0,1,2,3\}\times\{0,1,2,3\}$, keep the 10 whose first number is not larger than the
second.

````python
# ── 4.1 The relation ≤ on S = {0, 1, 2, 3}: a subset of S × S, drawn as a grid ──
S = [0, 1, 2, 3]
R_le = {(a, b) for a in S for b in S if a <= b}                 # {(a, b) ∈ S × S | a ≤ b}
R_lt = {(a, b) for a in S for b in S if a < b}
R_eq = {(a, b) for a in S for b in S if a == b}
print("R≤ =", sorted(R_le))
print("|R≤| =", len(R_le), "| |R<| =", len(R_lt), "| |R=| =", len(R_eq), "| |S × S| =", len(cartesian(S, S)))
print("R≤ ⊆ S × S ?", R_le <= cartesian(S, S), "| R≤ = R< ∪ R= ?", R_le == R_lt | R_eq)

fig, ax = plt.subplots(figsize=(4.0, 3.8))
for a in S:
    for b in S:
        inside = (a, b) in R_le
        ax.scatter(b, a, s=420, color="tab:green" if inside else "white", edgecolors="0.3", zorder=2)
ax.set_xticks(S); ax.set_yticks(S); ax.invert_yaxis()
ax.set_xlabel("second component b"); ax.set_ylabel("first component a")
ax.set_title("the relation a ≤ b: 10 of the 16 pairs\n(filled = in R)")
ax.set_xlim(-0.6, 3.6); ax.set_ylim(3.6, -0.6); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The comprehension keeps exactly the 10 pairs of the solution. They split into the 4 pairs
on the diagonal ($a = b$) and the 6 above it ($a < b$); there are $\binom{4}{2} = 6$ ways to choose two different
numbers, and only one order of each pair satisfies $<$. In the grid, a relation on a finite set is a pattern of
filled cells in the $|S|\times|S|$ table of all pairs. Keep this picture in mind: in Module 06 the same table, filled
with 1s and 0s, becomes the **adjacency matrix** of a graph.

### 4.2 Functions are relations with exactly one partner

Module 03 defined a function $f: S\to T$ as a rule that gives every input exactly one output. As a set, a function is
the relation $\{(s, f(s)) \mid s\in S\}\subseteq S\times T$ in which **every $s\in S$ appears as a first component
exactly once**. The genetic code is the perfect example: it is a relation between codons and amino acids, and we can
check from the pairs alone that it is a function. We type the standard code in the compact form used by sequence
databases (one character per codon, codons ordered by the Cartesian product $\{T, C, A, G\}^3$; `*` marks a stop).

````python
# ── 4.2 The standard genetic code (typed in) as a relation ⊆ Codons × AminoAcids ──
CODE_STRING = "FFLLSSSSYY**CC*WLLLLPPPPHHQQRRRRIIIMTTTTNNKKSSRRVVVVAAAADDEEGGGG"   # standard code, order TCAG
DNA_CODONS = ["".join(c) for c in itertools.product("TCAG", repeat=3)]        # {T, C, A, G}³ in table order
GENETIC_CODE = dict(zip(DNA_CODONS, CODE_STRING))                             # codon → amino acid ('*' = stop)
code_relation = set(GENETIC_CODE.items())                                     # the set of pairs (codon, aa)

print("codons:", len(DNA_CODONS), "| pairs in the relation:", len(code_relation))
first_components = Counter(codon for codon, _ in code_relation)
print("every codon has exactly one partner → the relation is a function:", set(first_components.values()) == {1})

codons_per_aa = Counter(aa for _, aa in code_relation)
by_multiplicity = {}
for aa, n in sorted(codons_per_aa.items()):
    by_multiplicity.setdefault(n, []).append(aa)
print("amino acids grouped by number of codons:", dict(sorted(by_multiplicity.items(), reverse=True)))
print("the reverse relation (aa → codon) is a function?", set(codons_per_aa.values()) == {1})
print("ATG →", GENETIC_CODE["ATG"], "| stop codons:", [c for c, aa in GENETIC_CODE.items() if aa == "*"])
````

🔍 **What just happened?** The relation contains 64 pairs, one per codon, and every codon occurs as a first component
exactly once — so it is a function from $\{T,C,A,G\}^3$ to the 20 amino acids plus "stop". Read in the other direction
it is *not* a function: leucine (L), arginine (R) and serine (S) have six codons each; alanine, glycine, proline,
threonine and valine four; isoleucine and stop three; nine amino acids two; and only methionine (M, the start codon
ATG) and tryptophan (W) one. This redundancy is called the **degeneracy** of the genetic code, and it is why many
mutations at third codon positions are "silent". (We used DNA letters; in mRNA every T reads U.)

**Definition.** A relation from $S$ to $T$ is a subset $R\subseteq S\times T$; a relation on $S$ is a subset of

## 5 · Sequences over one set: $S^n$

> 📍 **Where we are.** Many products multiply a set with itself: codons are triples of bases, colours are triples of
> intensities, genotypes at $J$ SNPs are $J$-tuples of allele counts. The notation $S^n$ handles them all, and it is the
> step from tuples to the **vectors** of Module 07.

For a set $S$ and a positive integer $n$,

$$S^n = \underbrace{S\times S\times\dots\times S}_{n \text{ times}}, \qquad |S^n| = |S|^n .$$

*In words:* $S^n$ is the set of all sequences of length $n$ in which every component is an element of $S$; by the
multiplication principle there are $|S|$ choices for each of the $n$ positions, hence $|S|^n$ sequences.

**Examples.**
* **Binary triples.** $\{0,1\}^3 = \{\langle 0,0,0\rangle, \langle 0,0,1\rangle, \langle 0,1,0\rangle, \langle 0,1,1\rangle,
  \langle 1,0,0\rangle, \langle 1,0,1\rangle, \langle 1,1,0\rangle, \langle 1,1,1\rangle\}$ — all 8 combinations of three
  binary digits.
* **RGB colours.** The RGB values of digital images are elements of $\{0, 1, \dots, 255\}^3$ (also written
  $\{x\in\mathbb Z : x\in[0, 255]\}^3$): all ordered triples of red, green and blue intensities.
* **The plane.** $\mathbb R\times\mathbb R$ is written $\mathbb R^2$; each point $(x, y)\in\mathbb R^2$ is a 2-tuple of
  real numbers.
* **Biology.** Codons are elements of $\{A, C, G, U\}^3$; the genotypes of one person at $J$ SNPs (each coded 0, 1 or 2
  copies of the minor allele) form an element of $\{0, 1, 2\}^J$.

````python
# ── 5.1 Counting S^n: binary triples, colours, codons and genotype tuples ──
binary_triples = list(itertools.product([0, 1], repeat=3))
print("{0,1}³ =", binary_triples, "→", len(binary_triples), "= 2³")
print("|{0,…,255}³| =", f"{len(range(256)) ** 3:,}", "colours = 256³")
codons = list(itertools.product("ACGU", repeat=3))
print("|{A,C,G,U}³| =", len(codons), "codons = 4³, e.g.", "".join(codons[0]), "".join(codons[27]), "".join(codons[-1]))

for J in [1, 2, 10, 100, 1_000_000]:
    n_digits = math.floor(J * math.log10(3)) + 1                  # number of decimal digits of 3^J
    size = f"= {3 ** J:,}" if J <= 10 else f"≈ 10^{J * math.log10(3):,.1f}"
    print(f"genotype tuples at J = {J:>9,} SNPs: |{{0,1,2}}^J| = 3^J {size}  ({n_digits:,} digit{'s' if n_digits != 1 else ''})")
````

🔍 **What just happened?** `itertools.product(…, repeat=n)` is exactly $S^n$. Binary triples: $2^3 = 8$. Colours:
$256^3 = 16{,}777{,}216$ — the "16.7 million colours" of a computer screen. Codons: $4^3 = 64$, listed in
lexicographic order from AAA to UUU; the middle one printed, `codons[27]`, is CGU — check it by counting in base 4
with A = 0, C = 1, G = 2, U = 3: $27 = 1\cdot 16 + 2\cdot 4 + 3$ gives C, G, U. Genotype tuples explode:
$3^{10} = 59{,}049$ at ten SNPs, about $10^{47.7}$ at 100 SNPs, and a number with 477,122 digits for a million-SNP
genotyping array. Only about $10^{11}$ humans have ever lived (a common demographic estimate), so every person's
genome-wide genotype is, for all practical purposes, unique (identical twins aside) — the same multiplication
principle, applied to a few dozen highly variable markers, underlies forensic DNA identification.

### 5.3 The rational numbers as pairs

Module 03 described a rational number as a ratio of integers. With tuples we can say exactly what that means:

$$\mathbb Q \;\leftrightarrow\; \mathbb Z\times\{x \mid x\in\mathbb Z,\ x > 0\}, \qquad \frac{n}{d} \;\leftrightarrow\; \langle n, d\rangle .$$

*In words:* a rational number is represented as a pair ⟨numerator, denominator⟩ whose first component is any integer
and whose second component is a positive integer. The fractions 1/2 and 202/808 would be represented as ⟨1, 2⟩ and
⟨202, 808⟩.

But there is a catch: different pairs can name the same number. We must reduce fractions to lowest terms to establish
the equivalence of pairs like ⟨2, 4⟩ and ⟨1, 2⟩. Two pairs name the same rational exactly when
$\langle n, d\rangle \sim \langle n', d'\rangle \iff n\cdot d' = n'\cdot d$ (cross-multiplication).

````python
# ── 5.3 Rationals as pairs ⟨n, d⟩ ∈ ℤ × ℤ⁺: equivalent pairs and lowest terms ──
def lowest_terms(pair: tuple) -> tuple:
    """Reduce ⟨n, d⟩ (d > 0) by the greatest common divisor: the canonical pair of the rational n/d."""
    n, d = pair
    g = math.gcd(n, d)
    return (n // g, d // g)


def same_rational(p: tuple, q: tuple) -> bool:
    """⟨n, d⟩ ~ ⟨n', d'⟩  ⇔  n·d' = n'·d."""
    return p[0] * q[1] == q[0] * p[1]


for pair in [(1, 2), (2, 4), (202, 808), (-3, 6), (0, 5)]:
    print(f"pair {str(pair):11s} → lowest terms {str(lowest_terms(pair)):8s} | Fraction: {Fraction(*pair)}")
print("\n⟨2,4⟩ ~ ⟨1,2⟩ ?", same_rational((2, 4), (1, 2)), "| ⟨2,4⟩ ~ ⟨1,4⟩ ?", same_rational((2, 4), (1, 4)),
      "| ⟨202,808⟩ ~ ⟨1,4⟩ ?", same_rational((202, 808), (1, 4)))

pairs = list(itertools.product(range(-3, 4), range(1, 5)))        # n ∈ {−3,…,3}, d ∈ {1,…,4}
print("pairs in {−3,…,3} × {1,…,4}:", len(pairs), "| distinct rationals they name:", len({lowest_terms(p) for p in pairs}))
````

🔍 **What just happened?** ⟨2, 4⟩ reduces to ⟨1, 2⟩, and ⟨202, 808⟩ to ⟨1, 4⟩ (202 = 2·101 and 808 = 8·101, so the
greatest common divisor is 202); Python's `Fraction` performs the same reduction. Cross-multiplication confirms
⟨2, 4⟩ ~ ⟨1, 2⟩ ($2\cdot 2 = 1\cdot 4$) but **not** ⟨2, 4⟩ ~ ⟨1, 4⟩ ($2\cdot 4 \ne 1\cdot 4$). The last line shows why
the pair picture needs care: the 28 pairs of $\{-3, \dots, 3\}\times\{1, \dots, 4\}$ name only 19 different rationals —
for example ⟨0, 1⟩, ⟨0, 2⟩, ⟨0, 3⟩, ⟨0, 4⟩ all name 0.

⚠️ **Common confusion.** The set $\mathbb Z\times\mathbb Z^{+}$ is *bigger* than $\mathbb Q$ in the sense that many
pairs name the same number. To get exactly one pair per rational, keep only the pairs in lowest terms
($\gcd(n, d) = 1$) — or, equivalently, treat a rational as the whole *class* of equivalent pairs. The same issue
arises whenever a biological object has several equivalent descriptions (the same DNA molecule read from either
strand, Section 9).

### 📘 Lecture notes — sequences over one set, $S^n$

**Definition.** For a set $S$ and a positive integer $n$, $S^n = S\times S\times\dots\times S$ ($n$ times): the set of
all sequences of length $n$ whose components are elements of $S$. $|S^n| = |S|^n$.

**Examples.** A 3-tuple can be viewed as an element of a product of three sets. RGB values are elements of
$\{0, 1, \dots, 255\}^3$ ($256^3 = 16{,}777{,}216$ ordered triples of red, green and blue intensities). The Boolean
triples $\{0,1\}^3 = \{\langle 0,0,0\rangle, \langle 0,0,1\rangle, \langle 0,1,0\rangle, \langle 0,1,1\rangle,
\langle 1,0,0\rangle, \langle 1,0,1\rangle, \langle 1,1,0\rangle, \langle 1,1,1\rangle\}$ — all 8 combinations of 3
binary digits. The plane $\mathbb R\times\mathbb R = \mathbb R^2$; each point $(x, y)\in\mathbb R^2$ is a 2-tuple of real
numbers, and $(2, 3)\ne(3, 2)$.

**Rationals as pairs.** $\mathbb Q$ can be modelled by $\mathbb Z\times\{x\mid x\in\mathbb Z, x>0\}$: $\frac nd\leftrightarrow\langle n, d\rangle$,
e.g. $\frac12\leftrightarrow\langle 1,2\rangle$ and $\frac{202}{808}\leftrightarrow\langle 202, 808\rangle$. Equivalent pairs
($n d' = n' d$, e.g. ⟨2, 4⟩ and ⟨1, 2⟩) name the same number; reduce to lowest terms to get one pair per rational.

**Common confusion.** $S^n$ (tuples of length $n$) is not the power set $\mathcal P(S)$ (all subsets) — compare
$|\{0,1\}^3| = 8$ with $|\mathcal P(\{0,1,2\})| = 8$: same size, completely different elements (the binary encoding of
Module 03 is a bijection between them).

> 🎓 **Exam tip.** For "how many …?" questions, first identify the set $S$ and the length $n$, then answer $|S|^n$; for
> rationals, always check whether two pairs are equivalent by cross-multiplication.

## 6 · Tuples of varying length and the Kleene star

> 📍 **Where we are.** Every $S^n$ contains tuples of *one* fixed length. DNA reads, proteins and words have many
> different lengths, so we need sets that mix lengths. Unions (Module 03) of the sets $S^n$ do it — and taking the union
> over *all* lengths gives the Kleene star, the set of all finite sequences.

**Question: how do we represent a set of tuples of different lengths?** Remember that the elements of a set need not
be objects of the same type. With $B = \{4, 9\}$, let $B^2$ be all 2-tuples over $B$ and $B^3$ all 3-tuples; then

$$B^2\cup B^3 = \left\{\begin{matrix}\langle 4,4\rangle, \langle 4,9\rangle, \langle 9,4\rangle, \langle 9,9\rangle,\\
\langle 4,4,4\rangle, \langle 4,4,9\rangle, \langle 4,9,4\rangle, \langle 4,9,9\rangle,\\
\langle 9,4,4\rangle, \langle 9,4,9\rangle, \langle 9,9,4\rangle, \langle 9,9,9\rangle\end{matrix}\right\}.$$

*In words:* the union simply collects the 4 pairs and the 8 triples into one set of 12 tuples. No pair equals a
triple (different lengths), so the two sets are disjoint and the sizes add: $|B^2\cup B^3| = 4 + 8 = 12$.

**Binary numbers without leading zeros.** Suppose we want to enumerate all binary numbers with at most 8 digits. We
omit leading zeros because 00010101 is usually written 10101. So we want the strings that are at most 8 bits long
and do not start with 0 — a 1 followed by anything of length 0 to 7:

$$\{1\}\times\bigcup_{i=0}^{7}\{0,1\}^i .$$

*In words:* $\{1\}$ forces the first digit to be 1, and $\{0,1\}^i$ supplies the remaining $i$ digits, for every
length $i = 0, 1, \dots, 7$ of the suffix — so the whole number has $1 + i$ digits, from 1 to 8.

🧮 **Math refresher — indexed unions and a geometric sum.** $\bigcup_{i=0}^{7} X_i$ means
$X_0\cup X_1\cup\dots\cup X_7$, just as $\sum_{i=0}^{7} x_i$ means $x_0 + x_1 + \dots + x_7$. Counting the binary numbers
needs $\sum_{i=0}^{7} 2^i$. Call it $T = 1 + 2 + 4 + \dots + 2^7$. Then $2T = 2 + 4 + \dots + 2^7 + 2^8$, and
subtracting, everything cancels except $2T - T = 2^8 - 1$. So $T = 2^8 - 1 = 255$; in general
$\sum_{i=0}^{n} r^i = \frac{r^{n+1} - 1}{r - 1}$ for $r\neq 1$.

````python
# ── 6.1 B² ∪ B³ and the binary numbers {1} × ⋃_{i=0}^{7} {0,1}^i ──
B = [4, 9]
B2, B3 = set(itertools.product(B, repeat=2)), set(itertools.product(B, repeat=3))
print("|B²| =", len(B2), "| |B³| =", len(B3), "| disjoint?", B2.isdisjoint(B3), "| |B² ∪ B³| =", len(B2 | B3))


def binary_numbers(max_digits: int = 8, first_suffix_length: int = 0) -> set:
    """{1} × ⋃_{i} {0,1}^i for i = first_suffix_length, …, max_digits − 1; each tuple glued into a string."""
    numbers = set()
    for i in range(first_suffix_length, max_digits):
        for suffix in itertools.product("01", repeat=i):
            numbers.add("1" + "".join(suffix))
    return numbers


numbers = binary_numbers()
in_order = sorted(numbers, key=lambda s: int(s, 2))
print("\nhow many:", len(numbers), "| first six:", in_order[:6], "| last:", in_order[-1])
print("as integers they are exactly {1, …, 255}?", {int(s, 2) for s in numbers} == set(range(1, 256)))
print("Σ_{i=0}^{7} 2^i =", sum(2 ** i for i in range(8)), "= 2^8 − 1 =", 2 ** 8 - 1)
from_one = binary_numbers(first_suffix_length=1)                 # the union started at i = 1 by mistake
print("starting the union at i = 1 instead gives", len(from_one), "numbers; missing:", numbers - from_one)
````

🔍 **What just happened?** $B^2$ and $B^3$ are disjoint, so their union has $4 + 8 = 12$ elements. The binary-number
formula produces **255** strings — 1, 10, 11, 100, 101, 110, …, 11111111 — and converting them to integers gives
exactly the numbers 1 to 255 ($= 2^8 - 1$, the geometric sum). Each number appears once because "no leading zeros"
makes the binary representation unique.

⚠️ **Common confusion.** The index of the union must start at $i = 0$. The suffix of length 0 is the empty tuple, and
$\{1\}\times\{0,1\}^0$ contributes the one-digit number **1**. Writing $\bigcup_{i=1}^{7}$ silently drops it: the cell
shows that this version yields only 254 numbers and that the missing one is `'1'`.

### 6.2 The Kleene star: tuples of any length

That last formula leads to the notation already used for strings in Module 03, $\Sigma^*$. Sometimes we want the
set of all tuples of *any* finite length. Then we use the **Kleene star** (or Kleene operator), written $^*$, which
generates all finite-length tuples formed from elements of a set:

$$S^* = \bigcup_{i=0}^{\infty} S^i = S^0\cup S^1\cup S^2\cup S^3\cup\cdots, \qquad S^0 = \{\langle\rangle\}.$$

*In words:* $S^*$ contains the empty tuple, all 1-tuples, all pairs, all triples, … — every tuple of every finite
length whose components come from $S$. The set $S^*$ is infinite (if $S$ is not empty), but each of its elements is a
finite tuple.

The definition $S^0 = \{\langle\rangle\}$ — the set containing only the **empty tuple** — holds for every $S$. For
$B = \{4, 9\}$:
$B^0 = \{\langle\rangle\}$, $B^1 = \{\langle 4\rangle, \langle 9\rangle\}$, $B^2 = \{\langle 4,4\rangle, \langle 4,9\rangle,
\langle 9,4\rangle, \langle 9,9\rangle\}$, $B^3 = \{\langle 4,4,4\rangle, \dots, \langle 9,9,9\rangle\}$, so
$B^* = \{\langle\rangle, \langle 4\rangle, \langle 9\rangle, \langle 4,4\rangle, \langle 4,9\rangle, \langle 9,4\rangle,
\dots, \langle 4,4,4\rangle, \dots\}$.

````python
# ── 6.2 The Kleene star, one length at a time; the empty tuple and the empty string ──
def kleene_star_up_to(S, max_length: int) -> list:
    """The finite part S⁰ ∪ S¹ ∪ … ∪ S^max_length of S*, listed by length (shortest first)."""
    tuples = []
    for i in range(max_length + 1):
        tuples.extend(itertools.product(sorted(S), repeat=i))
    return tuples


B_star = kleene_star_up_to({4, 9}, 3)
print("B* up to length 3:", len(B_star), "tuples = 1 + 2 + 4 + 8")
print("  ", B_star[:7], "…", B_star[-1])
print("B⁰ =", set(itertools.product([4, 9], repeat=0)), "→ one element, the empty tuple; |B⁰| =", len(list(itertools.product([4, 9], repeat=0))))
print("∅* up to length 5:", kleene_star_up_to(set(), 5), "→ even the empty set has a Kleene star with one element")
print('the empty string ε = "": length', len(""), "| {ε} has", len({""}), "element | ∅ has", len(set()), "elements")
for n in [5, 10, 20]:
    print(f"DNA strings of length ≤ {n:2d}: Σ_(i=0..{n}) 4^i = {sum(4 ** i for i in range(n + 1)):,}")
````

🔍 **What just happened?** Up to length 3, $B^*$ has $1 + 2 + 4 + 8 = 15$ tuples, starting with the empty tuple `()`
and ending with ⟨9, 9, 9⟩. $B^0$ is the set `{()}` with **one** element — the empty tuple — not the empty set. Even
$\varnothing^* = \{\langle\rangle\}$: there are no components to choose, but the tuple with no components still exists.
For strings, the empty tuple is the **empty string** ε (Python's `""`): $\Sigma^0 = \{\varepsilon\}$, ε has length 0,
and ε is unique — there is only one string with no letters. Finally, the number of DNA strings of length at most $n$
is the geometric sum $\frac{4^{n+1}-1}{3}$: 1,365 for $n = 5$, about 1.4 million for $n = 10$ and
1.47 × 10¹² for $n = 20$.

⚠️ **Common confusion.** Three different things look "empty": the empty set $\varnothing$ (no elements), the empty
tuple ⟨⟩ or empty string ε (a sequence with no components), and the set $\{\langle\rangle\} = S^0$ (a set with *one*
element, the empty tuple). Writing "$S^0 = \langle\rangle$" mixes up a set with its element — it is $S^0 = \{\langle\rangle\}$.

### 6.3 A picture of $\{0,1\}^*$: the prefix tree

The elements of $S^*$ can be arranged as a tree: the empty tuple at the top, and every tuple pointing to the tuples
obtained by appending one more component. (Trees are the subject of Module 06; here the picture just organises the
counting.)

````python
# ── 6.3 The binary strings of length ≤ 3 as a prefix tree, and the sizes |{0,1}^i| ──
strings_up_to_3 = ["".join(t) for t in kleene_star_up_to({"0", "1"}, 3)]
label = {s: (s if s else "ε") for s in strings_up_to_3}
tree_edges = [(label[s[:-1]], label[s]) for s in strings_up_to_3 if s]        # parent = string minus its last symbol
pos = tree_layout(tree_edges, "ε")
level_colour = {0: "#f4cccc", 1: "#fce5cd", 2: "#d9ead3", 3: "#cfe2f3"}
node_colours = {label[s]: level_colour[len(s)] for s in strings_up_to_3}

fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), gridspec_kw={"width_ratios": [1.6, 1]})
draw_graph(tree_edges, pos=pos, ax=axes[0], node_colors=node_colours, font_size=8,
           title="{0,1}⁰ ∪ {0,1}¹ ∪ {0,1}² ∪ {0,1}³: 1 + 2 + 4 + 8 = 15 strings")
lengths = np.arange(0, 9)
axes[1].bar(lengths, 2 ** lengths, color="tab:blue", label="|{0,1}^i| = 2^i")
axes[1].plot(lengths, 2 ** (lengths + 1) - 1, "o-", color="tab:orange", label="strings of length ≤ i: 2^(i+1) − 1")
axes[1].set_xlabel("length i"); axes[1].set_ylabel("number of binary strings"); axes[1].legend()
axes[1].set_title("each extra position doubles the count")
plt.tight_layout(); plt.show()
print("edges in the prefix tree:", len(tree_edges), "| nodes:", len(strings_up_to_3))
````

🔍 **What just happened?** The left panel organises the 15 binary strings of length ≤ 3 by length: ε at the top
(level 0), then 0 and 1, then the four strings of length 2 and the eight of length 3. Every string except ε has
exactly one parent (itself minus its last symbol), so there are 14 edges for 15 nodes — one fewer edge than nodes, the
signature of a tree that Module 06 will prove. The right panel shows the doubling: $|\{0,1\}^i| = 2^i$, and the
cumulative count $2^{i+1} - 1$ is always one less than the next power of two — the geometric sum again. At $i = 7$ it
is 255, the number of binary numbers of Section 6.1: putting a "1" in front of each string of length ≤ 7 turns it into
exactly one binary number with at most 8 digits.

### 📘 Lecture notes — tuples of varying length and the Kleene star

**Mixed lengths.** Sets need not contain objects of one type, so a set may contain tuples of different lengths:
with $B = \{4, 9\}$, $B^2\cup B^3$ contains the 4 pairs and the 8 triples (12 tuples).

**Binary numbers without leading zeros** (up to 8 digits): $\{1\}\times\bigcup_{i=0}^{7}\{0,1\}^i$ — the $\{1\}$
ensures the first digit is 1, $\{0,1\}^i$ the remaining digits for every suffix length $i = 0, \dots, 7$. It has
$\sum_{i=0}^{7}2^i = 2^8 - 1 = 255$ elements, the numbers 1 to 255. (The union must start at $i = 0$; starting at
$i = 1$ loses the number 1.)

**Kleene star.** $S^* = \bigcup_{i=0}^{\infty} S^i$: all finite-length tuples formed from elements of $S$, e.g.
$B^* = B^0\cup B^1\cup B^2\cup B^3\cup\cdots = \{\langle\rangle, \langle 4\rangle, \langle 9\rangle, \langle 4,4\rangle,
\langle 4,9\rangle, \langle 9,4\rangle, \dots, \langle 4,4,4\rangle, \dots\}$.

**Empty tuple and empty string.** $S^0 = \{\langle\rangle\}$ for every $S$ (the set containing only the empty tuple).
For an alphabet, $\Sigma^0 = \{\varepsilon\}$: the empty string ε contains zero elements and is unique.

**Common confusion.** $\varnothing$ (no elements) ≠ ⟨⟩ = ε (a sequence of length 0) ≠ $\{\langle\rangle\}$ (a set with
one element).

> 🎓 **Exam tip.** To count a union of $S^i$ over a range of $i$, check that the pieces are disjoint (different lengths
> always are) and add the sizes with the geometric-sum formula; double-check the endpoints of the index range.

## 7 · Strings over an alphabet: DNA, RNA and proteins

> 📍 **Where we are.** We now have all the pieces to say precisely what a DNA sequence is. In certain contexts,
> sequences of elements from the same set are called **strings**, and the set is called an **alphabet**.

For a set $\Sigma$, called an **alphabet**, a **string over $\Sigma$** is an element of $\Sigma^n$ for some non-negative
integer $n$; in other words, a string is any element of

$$\Sigma^* = \bigcup_{n\in\mathbb Z^{\ge 0}} \Sigma^n, \qquad \text{and the length of } x\in\Sigma^n \text{ is } |x| = n .$$

*In words:* a string is a finite sequence (tuple) of symbols from a given alphabet, of any length $0, 1, 2, \dots$; its
length is the number of symbols. For example, the set of all 5-letter English words is a subset of $\{A, B, \dots, Z\}^5$:
each word is a 5-tuple of letters from $\Sigma = \{A, B, \dots, Z\}$.

**Notation conventions for strings.** When writing strings we usually omit the angle brackets and commas (unlike
regular tuples). These strings are technically tuples, but we write them in a simplified form:

| String | Belongs to | Explanation |
|---|---|---|
| ABRACADABRA | $\{A, B, \dots, Z\}^{11}$ | a string of 11 capital letters |
| 11010011 | $\{0, 1\}^8$ | a binary string of length 8 |
| ATGGCC | $\{A, C, G, T\}^6$ | a DNA string of length 6 (two codons) |
| AUGGCC | $\{A, C, G, U\}^6$ | the same message as RNA |
| MKVLAAGIVG | $\Sigma_{\text{protein}}^{10}$ | a peptide over the 20 one-letter amino-acid codes |

The biological alphabets are $\Sigma_{\text{DNA}} = \{A, C, G, T\}$, $\Sigma_{\text{RNA}} = \{A, C, G, U\}$ and the
20 amino acids $\Sigma_{\text{protein}} = \{A, C, D, E, F, G, H, I, K, L, M, N, P, Q, R, S, T, V, W, Y\}$.

````python
# ── 7.1 Membership in Σ*: "every symbol belongs to the alphabet" ──
ALPHABETS = {"DNA": set("ACGT"), "RNA": set("ACGU"), "protein": set("ACDEFGHIKLMNPQRSTVWY"),
             "capitals": set("ABCDEFGHIJKLMNOPQRSTUVWXYZ"), "binary": set("01")}


def is_string_over(s: str, alphabet: set) -> bool:
    """s ∈ Σ*  ⇔  ∀ positions i: s_i ∈ Σ  (the empty string is over every alphabet)."""
    return all(symbol in alphabet for symbol in s)


examples = [("ABRACADABRA", "capitals"), ("11010011", "binary"), ("ATGGCC", "DNA"), ("AUGGCC", "RNA"),
            ("AUGGCC", "DNA"), ("MKVLAAGIVG", "protein"), ("ATGNNC", "DNA"), ("", "DNA")]
for s, name in examples:
    verdict = is_string_over(s, ALPHABETS[name])
    where = f"∈ ({name})^{len(s)}" if verdict else "∉ " + f"({name})*"
    print(f"{s!r:14s} {where:20s} length {len(s)}")
print("\n5-letter strings over {A,…,Z}:", f"{26 ** 5:,}", "— only a few thousand of them are English words")
````

🔍 **What just happened?** `is_string_over` is the quantifier statement $\forall i: s_i\in\Sigma$ written with `all`.
ABRACADABRA lies in $\{A..Z\}^{11}$, 11010011 in $\{0,1\}^8$, ATGGCC is DNA and AUGGCC is RNA — but AUGGCC is *not* a
DNA string (U is not in $\Sigma_{\text{DNA}}$), and ATGNNC is not either: the letter N, which real sequencing files use
for "unknown base", lies outside $\{A, C, G, T\}$, so software must either extend the alphabet (the IUPAC code adds N,
R, Y, …) or reject such reads. The empty string passes for every alphabet (vacuous truth again) and lies in
$\Sigma^0$. There are $26^5 = 11{,}881{,}376$ five-letter strings, of which only a tiny subset are words — and,
likewise, only a tiny subset of all DNA strings are functional genes.

### 7.2 Operations on strings: concatenation, reversal, complement

* **Concatenation** $x\cdot y$ (often just $xy$) writes $y$ after $x$; its length is $|xy| = |x| + |y|$, and ε is the
  identity: $\varepsilon x = x\varepsilon = x$. Concatenation is not commutative: in general $xy\neq yx$.
* **Reverse complement** (Module 03) is a function $\Sigma_{\text{DNA}}^*\to\Sigma_{\text{DNA}}^*$: complement every
  base (A ↔ T, C ↔ G) and read backwards — the sequence of the other DNA strand, read in its own 5′→3′ direction.
* **Transcription** is a function $\Sigma_{\text{DNA}}^*\to\Sigma_{\text{RNA}}^*$ that replaces T by U (for the coding
  strand).

````python
# ── 7.2 Concatenation, the empty string, reverse complement and transcription ──
x, y, eps = "GAATTC", "TTAA", ""
print("x·y =", x + y, "| |x·y| =", len(x + y), "= |x| + |y| =", len(x) + len(y))
print("y·x =", y + x, "| x·y = y·x ?", x + y == y + x)
print("ε·x = x = x·ε ?", eps + x == x == x + eps)
print("reverse complement of", x, "=", reverse_complement(x), "→ the EcoRI site is its own reverse complement")
print("reverse complement of ATGGCC =", reverse_complement("ATGGCC"))


def transcribe(dna: str) -> str:
    """Coding-strand DNA → mRNA: the function Σ_DNA* → Σ_RNA* that replaces every T by U."""
    return dna.replace("T", "U")


print("transcribe(ATGGCCTAA) =", transcribe("ATGGCCTAA"))
print(f"protein strings of length 100: 20^100 ≈ 10^{100 * math.log10(20):.1f}")
````

🔍 **What just happened?** Concatenating GAATTC and TTAA gives GAATTCTTAA of length 6 + 4 = 10, while the other order
gives TTAAGAATTC — a different string, so concatenation is not commutative (just as Cartesian products are not). The
empty string changes nothing. GAATTC, the recognition site of the restriction enzyme EcoRI, equals its own reverse
complement: such **biological palindromes** are typical of restriction sites, because the enzyme (a dimer) binds both
strands symmetrically. Transcription turned ATGGCCTAA into AUGGCCUAA. Finally, there are about $10^{130}$ protein
strings of length 100 — vastly more than could ever have been made in the history of life, which is why protein
design (BiologyToML Module 39) is a search through an astronomically large set of tuples.

## 8 · k-mers: counting tuples in a genome

> 📍 **Where we are.** We can now answer the opening question. A **k-mer** is a string of length $k$, i.e. an element of
> $\Sigma^k$; the k-mers *of a sequence* are its substrings of length $k$. Genome assemblers, read aligners,
> metagenomic classifiers and CRISPR guide designers all reason about k-mers, and all of them rely on one comparison:
> the number of possible k-mers versus the length of the genome.

A DNA string of length $L$ contains $L - k + 1$ overlapping k-mers (start positions $1, \dots, L-k+1$), while there are
$|\Sigma^k| = 4^k$ possible k-mers. Hence

$$\#\{\text{distinct k-mers in the sequence}\} \;\le\; \min\left(4^k,\; L - k + 1\right).$$

*In words:* a sequence cannot contain more different k-mers than exist ($4^k$), nor more than it has positions
($L - k + 1$). For small $k$, every possible k-mer occurs (many times); once $4^k$ is much larger than $L$, almost every
k-mer that occurs, occurs only once — it becomes a unique **address** of its position.

````python
# ── 8.1 4^k versus genome length (genome sizes rounded, illustrative) and the CRISPR 20-mer ──
genomes = {"E. coli (≈ 4.6 Mb)": 4.6e6, "baker's yeast (≈ 12 Mb)": 1.2e7, "human (≈ 3.1 Gb, haploid)": 3.1e9}
for name, size in genomes.items():
    k = 1
    while 4 ** k <= size:                       # smallest k with more possible k-mers than positions
        k += 1
    print(f"{name:27s}: 4^{k - 1} = {4 ** (k - 1):>13,} ≤ L < 4^{k} = {4 ** k:>14,}  → k = {k}")

positions = 2 * 3.1e9                           # both strands of the human genome
print(f"\n4^20 = {4 ** 20:,} possible 20-mers vs ≈ {positions:.1e} positions (both strands)")
print(f"expected chance occurrences of one given 20-mer in a random genome of that size: {positions / 4 ** 20:.4f}")
````

🔍 **What just happened?** For a random genome, k-mers start to be "rare" once $4^k$ exceeds the genome length: at
$k = 12$ for *E. coli* (4.6 Mb lies between $4^{11}\approx 4.2$ million and $4^{12}\approx 16.8$ million) and for
yeast, and at $k = 16$ for the human genome (3.1 Gb lies between $4^{15}\approx 1.07\times 10^9$ and
$4^{16}\approx 4.29\times 10^9$). For a CRISPR guide, $4^{20}\approx 1.1\times 10^{12}$ possible 20-mers face about
$6.2\times 10^9$ positions on the two strands: a given 20-mer would be expected to occur by chance only about
**0.0056** times. That is the answer to the opening question — a 20-letter address is long enough to be unique.
(Real genomes are far from random — repeats and duplicated genes make some 20-mers occur many times — so guide-design
software checks every candidate for near-matches elsewhere, the "off-targets".)

### 8.2 Counting k-mers in the simulated genome

`kmer_counts(seq, k)` from the toolkit returns a dictionary k-mer → number of occurrences. We count the k-mers of our
50-kb random genome for $k = 1, \dots, 12$.

````python
# ── 8.2 Distinct and unique k-mers in the SIMULATED 50-kb genome, k = 1 … 12 ──
L = len(GENOME)
rows = []
for k in range(1, 13):
    counts = kmer_counts(GENOME, k)
    rows.append({"k": k, "possible 4^k": 4 ** k, "positions L−k+1": L - k + 1, "distinct seen": len(counts),
                 "upper bound": min(4 ** k, L - k + 1),
                 "fraction seen once": sum(1 for c in counts.values() if c == 1) / len(counts)})
kmer_table = pd.DataFrame(rows)
print(kmer_table.to_string(index=False, formatters={"fraction seen once": "{:.3f}".format}))

fig, axes = plt.subplots(1, 2, figsize=(10, 3.6))
axes[0].semilogy(kmer_table["k"], kmer_table["possible 4^k"], "s--", label="possible k-mers 4^k")
axes[0].semilogy(kmer_table["k"], kmer_table["positions L−k+1"], "^--", label="positions L − k + 1")
axes[0].semilogy(kmer_table["k"], kmer_table["distinct seen"], "o-", color="k", label="distinct k-mers seen")
axes[0].set_xlabel("k"); axes[0].set_ylabel("count (log scale)"); axes[0].legend()
axes[0].set_title("SIMULATED 50-kb genome: distinct k-mers ≤ min(4^k, L−k+1)")
axes[1].plot(kmer_table["k"], kmer_table["fraction seen once"], "o-", color="tab:red")
axes[1].axvline(math.log(L, 4), color="0.5", ls=":", label=f"4^k = L  (k = log₄ L ≈ {math.log(L, 4):.1f})")
axes[1].set_xlabel("k"); axes[1].set_ylabel("fraction of distinct k-mers seen exactly once"); axes[1].legend()
axes[1].set_title("k-mers become unique addresses once 4^k ≫ L")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** For $k\le 6$ every one of the $4^k$ possible k-mers occurs (4,096 distinct 6-mers among
49,995 positions) and none occurs only once. Between $k = 7$ and $k = 8$ the two limits cross — the dotted line marks
$4^k = L$ at $k = \log_4 50{,}000\approx 7.8$ — and the curve of distinct k-mers bends from the $4^k$ line onto the
$L - k + 1$ line: at $k = 8$, 35,002 of the 65,536 possible 8-mers occur. From $k = 10$ on, almost all positions carry
a different k-mer (48,898 distinct 10-mers among 49,991 positions) and 97.8 % of the k-mers seen occur exactly once; at
$k = 12$ the fraction is 99.9 %. This is why read aligners look up reads through k-mer "seeds" of length ≈ 16–31,
and why k-mer–based assemblers and classifiers use k-mers of similar or larger size: long enough to be unique in a
real (repetitive) genome, short enough that a read with a few sequencing errors still contains many error-free
k-mers.

### 8.3 Order matters in real DNA: CG is not GC

In a uniformly random sequence the dinucleotides ⟨C, G⟩ and ⟨G, C⟩ are equally common. Real vertebrate genomes are
different: a C followed by a G (a "CpG") is often methylated, and methylated C mutates to T unusually often, so CpG is
**depleted** outside special regions called **CpG islands** (often found at gene promoters). The toolkit's Markov chains
reproduce this with transition probabilities estimated from human DNA. Biologists measure it with the
observed-to-expected ratio $\text{CpG}_{o/e} = \dfrac{\#\text{CG}\cdot L}{\#\text{C}\cdot\#\text{G}}$, which is about 1
for random sequence.

````python
# ── 8.3 Dinucleotides ⟨C,G⟩ vs ⟨G,C⟩ in three SIMULATED sequences ──
sequences = {"uniform random": GENOME[:20_000],
             "background (Markov)": simulate_dna(20_000, transitions=BACKGROUND_TRANSITIONS, seed=5),
             "CpG island (Markov)": simulate_dna(20_000, transitions=CPG_ISLAND_TRANSITIONS, seed=6)}
summary = []
for name, seq in sequences.items():
    di = kmer_counts(seq, 2)
    summary.append({"sequence": name, "CG": di.get("CG", 0), "GC": di.get("GC", 0),
                    "CpG o/e": di.get("CG", 0) * len(seq) / (seq.count("C") * seq.count("G")),
                    "GC content": (seq.count("C") + seq.count("G")) / len(seq)})
cpg = pd.DataFrame(summary)
print(cpg.to_string(index=False, formatters={"CpG o/e": "{:.2f}".format, "GC content": "{:.3f}".format}))

fig, ax = plt.subplots(figsize=(6.4, 3.4))
xpos = np.arange(len(cpg))
ax.bar(xpos - 0.2, cpg["CG"], width=0.4, color="tab:red", label="⟨C,G⟩ = CG")
ax.bar(xpos + 0.2, cpg["GC"], width=0.4, color="tab:blue", label="⟨G,C⟩ = GC")
ax.set_xticks(xpos); ax.set_xticklabels(cpg["sequence"]); ax.set_ylabel("count in 20,000 bases")
ax.set_title("same set {C, G}, different tuples: CpG is depleted outside CpG islands"); ax.legend()
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** In the uniform sequence CG and GC are equally frequent (1,248 vs 1,234; o/e ratio 1.01). In
the background model CG drops to 411 while GC stays at 1,137 — an o/e ratio of 0.36, the CpG depletion of bulk human
DNA. In the CpG-island model both are frequent and the o/e ratio is 0.77, with a GC content of about 0.69. The classic
definition of a CpG island (Gardiner-Garden & Frommer 1987) asks for GC content above 50 % and CpG o/e above 0.6 over at
least 200 bases — a criterion stated entirely in terms of 1-mers and *ordered* 2-mers. Module 20 returns to these two
Markov chains to decide whether a stretch of DNA is an island.

## 9 · Codons and reading frames

> 📍 **Where we are.** The genetic code (Section 4) is a function on $\Sigma^3$. To translate a gene we apply it to
> consecutive, non-overlapping triples of a string — but *where* we start counting matters, and so does *which strand*
> we read. Tuples, Cartesian products, strings and functions now work together.

A DNA string can be cut into codons starting at position 1, 2 or 3; these are the three **reading frames** of the
strand (starting at position 4 just repeats frame 1 minus one codon). The other strand, read 5′→3′, is the reverse
complement, with three frames of its own. A double-stranded DNA molecule therefore has **six reading frames**,
written +1, +2, +3 and −1, −2, −3. A protein-coding gene is an **open reading frame (ORF)**: in one frame, a start
codon ATG followed by many codons and finally one of the three stop codons TAA, TAG, TGA.

Because 3 of the 64 codons are stops, a random codon is a stop with probability $3/64$. In random DNA, the number of
non-stop codons before the first stop therefore has mean

$$\frac{1 - 3/64}{3/64} = \frac{61}{3}\approx 20.3 \text{ codons}.$$

*In words:* in random sequence you hit a stop codon roughly every 21 codons, so a stretch of hundreds of codons without
a stop is very unlikely by chance — a long ORF is evidence of a real gene. (The formula is the mean of a geometric
distribution, Module 18; here we just check it by counting.)

````python
# ── 9.1 Translate one string in all six reading frames (a gene planted in SIMULATED DNA) ──
def translate(dna: str, frame: int = 0) -> str:
    """Apply the genetic code to the consecutive codons of dna, starting at offset frame (0, 1 or 2)."""
    return "".join(GENETIC_CODE[dna[i:i + 3]] for i in range(frame, len(dna) - 2, 3))


def six_frames(dna: str) -> dict:
    """Translations of the three forward frames (+1, +2, +3) and of the reverse complement (−1, −2, −3)."""
    rc = reverse_complement(dna)
    frames = {f"+{f + 1}": translate(dna, f) for f in range(3)}
    frames.update({f"−{f + 1}": translate(rc, f) for f in range(3)})
    return frames


rng_gene = np.random.default_rng(SEED + 9)
sense_codons = [c for c, aa in GENETIC_CODE.items() if aa != "*"]
planted_gene = "ATG" + "".join(rng_gene.choice(sense_codons, size=40)) + "TAA"     # 1 + 40 codons + stop
dna = simulate_dna(22, seed=91) + planted_gene + simulate_dna(17, seed=92)
print("DNA length:", len(dna), f"| gene planted at position 23 (offset 22 → frame +{22 % 3 + 1})")
for name, protein in six_frames(dna).items():
    print(f"frame {name}: {protein}")
````

🔍 **What just happened?** The 165-letter string contains a planted gene of 1 + 40 + 1 = 42 codons, starting at
position 23. Because $22 = 3\cdot 7 + 1$, the gene lies in frame **+2**: there, after seven codons of flanking DNA
(`FFSIRYP`), the translation shows the planted protein `MRDGPSTKVK…GPIV` — an M followed by 40 amino acids — ending at
the single `*` of the planted TAA. The other five frames are translations too, but they mean nothing: in only 54 or 55 codons
each of them contains between one and four stop codons (`*`), as expected for sequence that does not code in that
frame. Gene-finding programs begin with exactly this six-frame scan.

### 9.2 How long are open reading frames by chance?

We now scan all six frames of the 50-kb random genome and measure every ORF: a run that starts at an ATG and ends at the
next stop codon in the same frame.

````python
# ── 9.2 Stop-free runs and ORFs in the six frames of the SIMULATED random genome ──
def orf_lengths(protein: str) -> list:
    """Lengths (in codons, stop excluded) of the ORFs of a translated frame: first 'M' of a segment → next '*'."""
    lengths = []
    for segment in protein.split("*")[:-1]:          # each piece before a '*' ends at a stop codon
        start = segment.find("M")
        if start >= 0:
            lengths.append(len(segment) - start)
    return lengths


frames = six_frames(GENOME)
runs = [len(seg) for protein in frames.values() for seg in protein.split("*")[:-1]]   # codons between stops
orfs = [n for protein in frames.values() for n in orf_lengths(protein)]
print(f"stop-free runs: {len(runs):,} | mean length {np.mean(runs):.1f} codons (theory 61/3 = {61 / 3:.1f})")
print(f"ORFs (ATG → stop): {len(orfs):,} | median {np.median(orfs):.0f} codons | longest {max(orfs)} codons")
print(f"ORFs of ≥ 100 codons: {sum(n >= 100 for n in orfs)} | of ≥ 300 codons: {sum(n >= 300 for n in orfs)}")

fig, ax = plt.subplots(figsize=(6.4, 3.4))
bins = np.arange(0, max(runs) + 10, 5)
ax.hist(runs, bins=bins, color="tab:gray", alpha=0.8, label="stop-free runs (SIMULATED genome)")
p_stop = 3 / 64                                     # P(a random codon is a stop)
expected = len(runs) * ((1 - p_stop) ** bins[:-1] - (1 - p_stop) ** bins[1:])   # P(run length in each bin) × runs
ax.plot(bins[:-1] + 2.5, expected, "r-", label="random-codon expectation")
ax.set_xlabel("codons between consecutive stops"); ax.set_ylabel("number of runs"); ax.legend()
ax.set_title("in random DNA a stop codon comes every ≈ 21 codons")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** Across the six frames (about 100,000 codons) the stop codons cut the random genome into
4,728 stop-free runs with a mean length of 20.1 codons — very close to the predicted 61/3 ≈ 20.3 — and the histogram
follows the red random-codon expectation, which falls by half roughly every 15 codons because
$(61/64)^{15}\approx 0.49$. Proper ORFs (from an ATG to the next stop) are shorter still: 1,134 of them, median 15
codons. The longest is 165 codons and 15 exceed 100 codons, but none reaches 300 — while a typical bacterial protein is
about 300 amino acids long. A long ORF is therefore a strong statistical hint of a real gene, the basic idea behind
prokaryotic gene finders; the 15 chance ORFs above 100 codons also show why real gene finders add further evidence
(codon usage, Markov models of coding DNA — Module 20) instead of using length alone.

### 9.3 From strings to vectors: one-hot encoding

Machine-learning models need numbers. The standard encoding of a DNA string $x\in\Sigma^L$ is a $4\times L$ table of 0s
and 1s with a single 1 in each column, in the row of the base at that position: a tuple of $L$ "one-hot" columns.

````python
# ── 9.3 One-hot encoding of a DNA string: a tuple of L columns, each with a single 1 ──
def one_hot(dna: str) -> np.ndarray:
    """4 × L array; entry (b, i) is 1 if position i holds base b (rows in the order A, C, G, T)."""
    encoding = np.zeros((4, len(dna)), dtype=int)
    for i, base in enumerate(dna):
        encoding[NUCLEOTIDES.index(base), i] = 1
    return encoding


site = "GAATTCAGG"
X_onehot = one_hot(site)
print(pd.DataFrame(X_onehot, index=list(NUCLEOTIDES), columns=[f"{i + 1}:{b}" for i, b in enumerate(site)]))
print("column sums:", X_onehot.sum(axis=0), "| shape:", X_onehot.shape)
print("decoding recovers the string:", "".join(NUCLEOTIDES[r] for r in X_onehot.argmax(axis=0)) == site)
````

🔍 **What just happened?** The 9-letter string became a 4 × 9 array: every column sums to 1 and the row of the 1
names the base, so the encoding is a bijection between $\Sigma^9$ and these 0/1 tables — nothing is lost (decoding
gives back GAATTCAGG). Position and order are kept because columns are ordered: this is a tuple of tuples, i.e. a
**matrix** (Module 08). Convolutional networks that predict transcription-factor binding read exactly such matrices.

## 📌 Key takeaways

- A **set** records only membership; a **tuple** (sequence, list) is an *ordered* collection in which duplicates are
  allowed. Write tuples with ⟨ ⟩ or ( ), never with { }. $\langle 4, 9\rangle\ne\langle 9, 4\rangle$, while
  $\{4, 9\} = \{9, 4\}$.
- Two tuples are equal iff they have the same length and agree position by position.
- The **Cartesian product** $S\times T = \{\langle x, y\rangle\mid x\in S\wedge y\in T\}$ builds all pairs;
  $|A_1\times\dots\times A_n| = \prod|A_i|$ (multiplication principle). It is not commutative, and
  $(A\times B)\times C$ equals $A\times B\times C$ only up to re-bracketing.
- A **relation** is a subset of a product (≤ on {0, 1, 2, 3} has 10 pairs); a function is a relation in which every
  input appears exactly once (the genetic code).
- $S^n$ is the set of length-$n$ sequences over $S$, with $|S|^n$ elements: 64 codons, $256^3$ colours, $3^J$
  genotype tuples, the plane $\mathbb R^2$. Rationals can be modelled as pairs ⟨n, d⟩ once equivalent pairs are identified.
- Unions of the $S^i$ mix lengths; the **Kleene star** $S^* = \bigcup_{i\ge0}S^i$ contains all finite sequences, with
  $S^0 = \{\langle\rangle\}$. The binary numbers with at most 8 digits are $\{1\}\times\bigcup_{i=0}^{7}\{0,1\}^i$ (255 of them).
- A **string** over an alphabet Σ is an element of $\Sigma^*$; DNA, RNA and proteins are strings over alphabets of
  4, 4 and 20 letters. ε is the unique string of length 0.
- $4^k$ vs genome length explains why k-mers of length ≈ 16–31 (and 20-nt CRISPR guides) act as unique addresses, and
  why a long open reading frame is unlikely to arise by chance.
