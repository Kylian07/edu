## 🧫 The biological question

When a pathogen spreads through the body, every organ answers it — but not in the same way. Some genes are switched on
almost everywhere: inflammatory **cytokines** (IL-1β, TNF, IL-6, CXCL8) that recruit immune cells, and
**interferon-stimulated genes** (STAT1, IRF1, MX1, ISG15) that put cells into a defensive state, and the **acute-phase**
programme (serum amyloid A, the iron regulator hepcidin, complement C3), made mostly by the liver but induced in other
organs too. And every organ also changes genes of its own. Our experiment (SIMULATED, the course's recurring infection study) measured 400 genes by log2
expression in **spleen, kidney and liver** of 4 control and 4 infected animals each. Four questions:

* ***Which genes respond to infection, and in which tissues?***
* ***Which part of the response is shared by all three tissues, and which is tissue-specific?***
* ***How sure are we*** — how many reported genes are false, how many responders do we miss, and would more animals help?
* ***What sits upstream of the response, and what does one snapshot say about a process that unfolds over days?***

None of them can be answered by one formula. They need the whole course, in the order of the **research cycle**:
question → design → data → explore → model → test → interpret → report.

| Biological question | Mathematical tool | Where |
|---|---|---|
| What exactly counts as "responds", "shared", "tissue-specific"? | predicates, quantifiers, sets, hypotheses | Section 1 |
| What can this design answer, and is the model identifiable? | Cartesian products, design matrices, rank | Section 2 |
| What varies most — tissue or infection? Do the controls behave? | centring, projections, SVD, matrix products | Section 3 |
| Does SAA1 respond in the liver, and by how much? | regression = maximum likelihood, $t$-test, CI | Section 4 |
| Which of the 400 genes respond, with few false discoveries? How many animals are enough? | 1,200 tests, Benjamini–Hochberg, simulation of power | Section 5 |
| Are the largest fold changes exaggerated? | empirical Bayes, normal prior, shrinkage | Section 6 |
| Which regulators sit upstream of the responding genes? | induced subgraphs, adjacency powers, reachability | Section 7 |
| Is "infected vs control" a snapshot of a dynamic process? | ODEs, equilibria, Jacobian, eigenvalues | Section 8 |
| What can we claim, and can anyone reproduce it? | tables, figures, limitations, fingerprints | Section 9 |

````python
# ── Course toolkit (identical to ../course_toolkit.py v1.0; copied so this notebook runs alone) ──
import itertools

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


def jaccard(a, b) -> float:
    """Jaccard index |A ∩ B| / |A ∪ B| of two collections (1.0 for two empty sets)."""
    a, b = set(a), set(b)
    union = a | b
    return len(a & b) / len(union) if union else 1.0


def _venn_order(x):
    """Sort key for Venn-diagram labels: numbers in numeric order first, then everything else as text."""
    if isinstance(x, (int, float, np.integer, np.floating)) and not isinstance(x, bool):
        return (0, float(x), "")
    return (1, 0.0, str(x))


def draw_venn(sets: dict, universe=None, ax=None, show_elements: bool = True, max_items: int = 6,
              title: str | None = None, highlight=None):
    """Draw a Venn diagram of two or three sets with matplotlib only.

    ``sets`` maps a label to a Python set, e.g. ``{"S": {1, 2, 3}, "T": {3, 4}}``.  Each region
    shows its elements (or ``n = …`` if it has more than ``max_items``).  If ``universe`` is given,
    a rectangle is drawn and the elements outside every set are listed in a corner.
    ``highlight`` is a function of the membership tuple, e.g. ``lambda s, t: s and not t`` for
    S \\ T; the regions where it is true are shaded.  Returns the Axes.
    """
    labels = list(sets)
    if len(labels) not in (2, 3):
        raise ValueError("draw_venn draws two or three sets")
    if ax is None:
        _, ax = plt.subplots(figsize=(4.6, 3.8))
    if len(labels) == 2:
        centres = [(-0.55, 0.0), (0.55, 0.0)]
    else:
        centres = [(-0.55, 0.32), (0.55, 0.32), (0.0, -0.6)]
    radius = 1.0
    colours = ["tab:blue", "tab:orange", "tab:green"]
    # classify a fine grid of points to find each region (used for shading and label positions)
    gx, gy = np.meshgrid(np.linspace(-2.2, 2.2, 441), np.linspace(-2.0, 1.9, 391))
    inside = [((gx - cx) ** 2 + (gy - cy) ** 2) <= radius ** 2 for cx, cy in centres]
    if highlight is not None:
        mask = np.zeros_like(gx, dtype=bool)
        for member in itertools.product([True, False], repeat=len(labels)):
            if any(member) and highlight(*member):
                region = np.ones_like(gx, dtype=bool)
                for k, m in enumerate(member):
                    region &= inside[k] if m else ~inside[k]
                mask |= region
        ax.contourf(gx, gy, mask.astype(float), levels=[0.5, 1.5], colors=["#ffd54f"], alpha=0.8)
    for (cx, cy), lab, col in zip(centres, labels, colours):
        ax.add_patch(plt.Circle((cx, cy), radius, fill=False, lw=2, color=col))
        lx = cx + (-1.0 if cx < 0 else 1.0 if cx > 0 else 0.0) * 0.95
        ly = cy + (0.95 if cy >= 0 else -1.15)
        ax.text(lx, ly, lab, ha="center", va="center", fontsize=11, fontweight="bold", color=col)
    elements = set().union(*sets.values())
    for member in itertools.product([True, False], repeat=len(labels)):
        if not any(member):
            continue
        region_items = [x for x in elements if all((x in sets[l]) == m for l, m in zip(labels, member))]
        region = np.ones_like(gx, dtype=bool)
        for k, m in enumerate(member):
            region &= inside[k] if m else ~inside[k]
        if not region.any():
            continue
        px, py = gx[region].mean(), gy[region].mean()
        listing = ", ".join(str(x) for x in sorted(region_items, key=_venn_order))
        if show_elements and len(region_items) <= max_items and len(listing) <= 16:
            text = listing if region_items else "∅"
        else:
            text = f"n = {len(region_items)}"
        ax.text(px, py, text, ha="center", va="center", fontsize=8, wrap=True)
    if universe is not None:
        ax.add_patch(plt.Rectangle((-2.1, -1.9), 4.2, 3.7, fill=False, lw=1, color="0.4"))
        outside = sorted(set(universe) - elements, key=_venn_order)
        text = ", ".join(map(str, outside)) if len(outside) <= max_items else f"n = {len(outside)}"
        ax.text(-2.0, -1.8, "U: " + (text if outside else "∅"), fontsize=8, va="bottom", color="0.3")
    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.0, 1.9)
    ax.set_aspect("equal")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax


def circle_layout(nodes: list) -> dict:
    """Place nodes evenly on a circle (first node at the top, clockwise)."""
    n = len(nodes)
    angles = np.pi / 2 - 2 * np.pi * np.arange(n) / max(n, 1)
    return {v: (float(np.cos(a)), float(np.sin(a))) for v, a in zip(nodes, angles)}


def adjacency_matrix(nodes: list, edges: list, directed: bool = False) -> np.ndarray:
    """Adjacency matrix A with A[i, j] = number of edges from nodes[i] to nodes[j].

    For an undirected graph every edge is counted in both directions (A is symmetric);
    a self-loop {v, v} adds 1 to the diagonal entry A[v, v].  List each undirected edge once:
    giving both (u, v) and (v, u) with ``directed=False`` counts that edge twice (a multigraph).
    """
    index = {v: i for i, v in enumerate(nodes)}
    A = np.zeros((len(nodes), len(nodes)), dtype=int)
    for u, v in edges:
        A[index[u], index[v]] += 1
        if not directed and u != v:
            A[index[v], index[u]] += 1
    return A


def draw_graph(edges: list, nodes=None, directed: bool = False, pos=None, ax=None, node_colors=None,
               edge_labels=None, title: str | None = None, font_size: int = 9):
    """Draw a small graph with matplotlib only (no extra packages needed).

    ``edges`` is a list of (u, v) pairs; ``nodes`` (optional) adds isolated nodes and fixes the
    order; ``pos`` maps node → (x, y) (default: ``circle_layout``).  Directed edges get arrows,
    self-loops are drawn as small rings above the node, ``edge_labels`` maps (u, v) → text (e.g. a
    weight) and ``node_colors`` maps node → colour.  Node circles grow with the length of their
    label.  Returns the Axes.
    """
    from matplotlib.transforms import offset_copy

    if nodes is None:
        nodes = []
        for u, v in edges:
            for w in (u, v):
                if w not in nodes:
                    nodes.append(w)
    pos = circle_layout(list(nodes)) if pos is None else pos
    if ax is None:
        _, ax = plt.subplots(figsize=(4.2, 3.6))
    radius = {v: max(0.62 * font_size * len(str(v)) / 2 + 4, 0.9 * font_size) for v in nodes}  # points
    for u, v in edges:
        if u == v:
            ring = offset_copy(ax.transData, fig=ax.figure, x=0, y=1.25 * radius[u], units="points")
            ax.scatter([pos[u][0]], [pos[u][1]], s=(1.3 * radius[u]) ** 2, facecolors="none",
                       edgecolors="0.35", linewidths=1.3, transform=ring, zorder=1)
            continue
        style = "-|>" if directed else "-"
        ax.annotate("", xy=pos[v], xytext=pos[u],
                    arrowprops=dict(arrowstyle=style, color="0.35", lw=1.3, mutation_scale=12,
                                    shrinkA=radius[u], shrinkB=radius[v],
                                    connectionstyle="arc3,rad=0.08" if directed else "arc3,rad=0"))
        if edge_labels and (u, v) in edge_labels:
            (x1, y1), (x2, y2) = pos[u], pos[v]
            ax.text((x1 + x2) / 2, (y1 + y2) / 2, str(edge_labels[(u, v)]), fontsize=font_size - 1,
                    ha="center", va="center", color="darkred", zorder=4,
                    bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85))
    xs = [pos[v][0] for v in nodes]
    ys = [pos[v][1] for v in nodes]
    colours = [(node_colors or {}).get(v, "#cfe2f3") for v in nodes]
    ax.scatter(xs, ys, s=[(2 * radius[v]) ** 2 for v in nodes], c=colours, edgecolors="0.2",
               linewidths=1.2, zorder=2, clip_on=False)
    for v in nodes:
        ax.text(pos[v][0], pos[v][1], str(v), ha="center", va="center", fontsize=font_size, zorder=3,
                clip_on=False)
    ax.margins(0.22)
    ax.set_aspect("equal", adjustable="datalim")
    ax.axis("off")
    if title:
        ax.set_title(title)
    return ax


NUCLEOTIDES = ("A", "C", "G", "T")


COMPLEMENT = {"A": "T", "C": "G", "G": "C", "T": "A"}


# First-order Markov-chain transition probabilities P(next | current) for DNA, rows = current base,
# columns = next base, both in the order A, C, G, T.  Rounded to two decimals from the classic
# CpG-island ("+") and background ("−") models of Durbin et al. (1998), Table 3.1, which were
# estimated from human sequence.  In CpG islands a C is often followed by a G; elsewhere CG is rare
# because methylated C in CpG mutates to T.
CPG_ISLAND_TRANSITIONS = np.array([
    [0.18, 0.27, 0.43, 0.12],
    [0.17, 0.37, 0.27, 0.19],
    [0.16, 0.34, 0.37, 0.13],
    [0.08, 0.36, 0.38, 0.18],
])


BACKGROUND_TRANSITIONS = np.array([
    [0.30, 0.20, 0.29, 0.21],
    [0.32, 0.30, 0.08, 0.30],
    [0.25, 0.25, 0.29, 0.21],
    [0.18, 0.24, 0.29, 0.29],
])


def reverse_complement(seq: str) -> str:
    """Reverse complement of a DNA string (A↔T, C↔G, read backwards)."""
    return "".join(COMPLEMENT[b] for b in reversed(seq))


def simulate_dna(length: int, transitions=None, initial=None, seed: int = 0) -> str:
    """Random DNA string.  Without ``transitions`` the bases are i.i.d. uniform; otherwise they
    follow a first-order Markov chain with the given 4 × 4 transition matrix (order A, C, G, T)
    and initial distribution ``initial`` (default uniform)."""
    rng = np.random.default_rng(seed)
    initial = np.full(4, 0.25) if initial is None else np.asarray(initial, dtype=float)
    if transitions is None:
        return "".join(rng.choice(list(NUCLEOTIDES), size=length, p=initial))
    P = np.asarray(transitions, dtype=float)
    states = [rng.choice(4, p=initial)]
    for _ in range(length - 1):
        states.append(rng.choice(4, p=P[states[-1]]))
    return "".join(NUCLEOTIDES[s] for s in states)


def markov_log_probability(seq: str, transitions, initial=None) -> float:
    """Natural-log probability of a DNA string under a first-order Markov chain:
    log P(x1) + Σ log P(x_t | x_{t-1})."""
    idx = {b: i for i, b in enumerate(NUCLEOTIDES)}
    initial = np.full(4, 0.25) if initial is None else np.asarray(initial, dtype=float)
    P = np.asarray(transitions, dtype=float)
    logp = np.log(initial[idx[seq[0]]])
    for a, b in zip(seq[:-1], seq[1:]):
        logp += np.log(P[idx[a], idx[b]])
    return float(logp)


# Named genes of the simulated infection experiment.  The infection-response genes are real genes
# known to be induced during infection (cytokines, interferon-stimulated genes, acute-phase
# proteins; lysozyme, LYZ, is an antimicrobial enzyme rather than a classic acute-phase protein and is
# grouped with them only for simplicity); the reference genes are classic "housekeeping" genes used
# to normalise qPCR.
INFECTION_PATHWAYS = {
    "cytokine signalling": ["IL1B", "TNF", "IL6", "CXCL8"],
    "interferon response": ["IRF1", "STAT1", "MX1", "ISG15"],
    "acute phase": ["SAA1", "HAMP", "C3", "LYZ"],
}


HOUSEKEEPING_GENES = ["ACTB", "GAPDH", "EEF1A1", "RPL13A", "TBP", "HPRT1",
                      "PPIA", "UBC", "YWHAZ", "SDHA", "TUBA1A", "B2M"]


# A simplified, illustrative cartoon of innate-immune signalling among the named genes
# (regulator → target).  NF-κB (NFKB1) induces cytokines; TNF and IL-1β activate NF-κB (a feedback
# loop); IL-6 acts through STAT3 to induce acute-phase genes; STAT1 induces interferon-stimulated genes.
INFLAMMATION_NETWORK_EDGES = [
    ("NFKB1", "IL1B"), ("NFKB1", "TNF"), ("NFKB1", "IL6"), ("NFKB1", "CXCL8"),
    ("TNF", "NFKB1"), ("IL1B", "NFKB1"), ("IL6", "STAT3"), ("STAT3", "SAA1"),
    ("STAT3", "HAMP"), ("STAT3", "C3"), ("STAT1", "IRF1"), ("STAT1", "MX1"),
    ("STAT1", "ISG15"), ("IRF1", "ISG15"), ("NFKB1", "LYZ"),
]


def simulate_tissue_expression(n_genes: int = 400, tissues=("spleen", "kidney", "liver"), n_reps: int = 4,
                               seed: int = 2680) -> dict:
    """SIMULATED log2 gene expression from an infection experiment in several tissues.

    Every tissue has ``n_reps`` control and ``n_reps`` infected samples.  The 12 genes of
    ``INFECTION_PATHWAYS`` respond to infection in every tissue (up by 1.5–3 log2 units); further
    "g…" genes respond in one tissue only or in two tissues, so the sets of truly changed genes
    overlap like the differentially-expressed-gene sets of a real multi-tissue study.  The 12
    ``HOUSEKEEPING_GENES`` are stable (tiny noise, no response).  About a third of all genes differ in
    baseline level between tissues.

    Returns a dict with ``X`` (samples × genes, log2 expression), ``samples`` (DataFrame with
    columns sample, tissue, condition), ``genes`` (list), ``true_lfc`` (DataFrame tissues × genes,
    true infected − control difference), ``true_de`` (dict tissue → set of truly changed genes),
    ``pathways`` (dict pathway → genes) and ``source``.
    """
    rng = np.random.default_rng(seed)
    named = [g for genes in INFECTION_PATHWAYS.values() for g in genes]
    n_other = n_genes - len(named) - len(HOUSEKEEPING_GENES)
    if n_other < 60:
        raise ValueError("n_genes must be at least 84")
    genes = named + HOUSEKEEPING_GENES + [f"g{i:03d}" for i in range(1, n_other + 1)]
    J, T = len(genes), len(tissues)
    base = rng.normal(7.0, 1.8, J)
    base[len(named):len(named) + len(HOUSEKEEPING_GENES)] = rng.uniform(9.0, 12.0, len(HOUSEKEEPING_GENES))
    tissue_shift = rng.normal(0.0, 1.0, (T, J)) * (rng.random(J) < 0.35)
    tissue_shift[:, len(named):len(named) + len(HOUSEKEEPING_GENES)] = 0.0
    lfc = np.zeros((T, J))
    lfc[:, :len(named)] = rng.uniform(1.5, 3.0, (T, len(named)))
    other = list(range(len(named) + len(HOUSEKEEPING_GENES), J))
    rng.shuffle(other)
    sizes_single = [24, 18, 14, 12, 10][:T]
    pairs = list(itertools.combinations(range(T), 2))
    sizes_pair = [10, 5, 4, 3, 3, 2, 2, 2, 2, 2][:len(pairs)]
    k = 0

    def effect(m):
        return rng.uniform(1.0, 2.5, m) * np.where(rng.random(m) < 0.7, 1.0, -1.0)

    for t, m in enumerate(sizes_single):
        idx = other[k:k + m]
        lfc[t, idx] = effect(m)
        k += m
    for (t1, t2), m in zip(pairs, sizes_pair):
        idx = other[k:k + m]
        e = effect(m)
        lfc[t1, idx] = e
        lfc[t2, idx] = e * rng.uniform(0.8, 1.2, m)
        k += m
    noise_sd = rng.uniform(0.25, 0.55, J)
    noise_sd[len(named):len(named) + len(HOUSEKEEPING_GENES)] = 0.08
    rows, X = [], []
    for t, tissue in enumerate(tissues):
        for condition in ("control", "infected"):
            for r in range(1, n_reps + 1):
                mean = base + tissue_shift[t] + (lfc[t] if condition == "infected" else 0.0)
                X.append(mean + rng.normal(0.0, noise_sd))
                rows.append((f"{tissue}_{condition[:3]}{r}", tissue, condition))
    samples = pd.DataFrame(rows, columns=["sample", "tissue", "condition"])
    true_lfc = pd.DataFrame(lfc, index=list(tissues), columns=genes)
    true_de = {tissue: {g for g, v in zip(genes, lfc[t]) if v != 0} for t, tissue in enumerate(tissues)}
    return {"X": np.array(X), "samples": samples, "genes": genes, "true_lfc": true_lfc,
            "true_de": true_de, "pathways": {k_: list(v) for k_, v in INFECTION_PATHWAYS.items()},
            "source": "SIMULATED log2 expression (BiologyToMath simulate_tissue_expression)"}


CELL_TYPE_MARKERS = {
    "T cell": ["CD3E", "CD3D", "IL7R"],
    "B cell": ["CD79A", "MS4A1", "CD19"],
    "macrophage": ["CD68", "LYZ", "CSF1R"],
    "fibroblast": ["COL1A1", "COL1A2", "DCN"],
}


def simulate_cell_counts(n_cells=(160, 120, 90, 70), cell_types=("T cell", "B cell", "macrophage", "fibroblast"),
                         n_background: int = 28, overdispersion: float = 0.0, seed: int = 680) -> dict:
    """SIMULATED single-cell RNA counts (molecules of each gene detected in each cell).

    Each cell type has a true mean count λ for every gene: its three ``CELL_TYPE_MARKERS`` genes are
    high (λ ≈ 6–15) in that type and low (≈ 0.1–0.5) elsewhere; ``n_background`` other genes share
    similar λ (≈ 0.2–4) in all types.  Counts are Poisson(λ); with ``overdispersion`` φ > 0 each
    cell's λ is first multiplied by a Gamma(1/φ, φ) factor (mean 1), which makes counts negative-
    binomial with variance λ + φλ² — closer to real data.

    Returns a dict with ``counts`` (cells × genes, int), ``labels`` (cell type of each cell),
    ``genes``, ``rates`` (DataFrame cell types × genes of true λ) and ``source``.
    """
    rng = np.random.default_rng(seed)
    markers = [g for t in cell_types for g in CELL_TYPE_MARKERS[t]]
    genes = markers + [f"bg{i:02d}" for i in range(1, n_background + 1)]
    rates = np.zeros((len(cell_types), len(genes)))
    bg = rng.uniform(0.2, 4.0, n_background)
    for t, ctype in enumerate(cell_types):
        for j, g in enumerate(markers):
            rates[t, j] = rng.uniform(6.0, 15.0) if g in CELL_TYPE_MARKERS[ctype] else rng.uniform(0.1, 0.5)
        rates[t, len(markers):] = bg * rng.uniform(0.8, 1.25, n_background)
    labels, counts = [], []
    for t, (ctype, n) in enumerate(zip(cell_types, n_cells)):
        lam = np.tile(rates[t], (n, 1))
        if overdispersion > 0:
            lam = lam * rng.gamma(1.0 / overdispersion, overdispersion, size=lam.shape)
        counts.append(rng.poisson(lam))
        labels += [ctype] * n
    return {"counts": np.vstack(counts), "labels": np.array(labels), "genes": genes,
            "rates": pd.DataFrame(rates, index=list(cell_types), columns=genes),
            "source": "SIMULATED single-cell counts (BiologyToMath simulate_cell_counts)"}


def simulate_cholesterol_trial(n: int = 12, effect: float = -18.0, seed: int = 2026) -> pd.DataFrame:
    """SIMULATED paired drug trial: LDL cholesterol (mg/dL) of ``n`` patients before and after
    treatment.  Patients differ a lot from each other (between-patient SD 30 mg/dL around 160), the
    measurement/day-to-day noise is smaller (SD 10), and the drug changes LDL by ``effect`` on
    average (the hidden truth).  Columns: patient, before, after."""
    rng = np.random.default_rng(seed)
    level = rng.normal(160.0, 30.0, n)
    before = level + rng.normal(0.0, 10.0, n)
    after = level + effect + rng.normal(0.0, 10.0, n)
    return pd.DataFrame({"patient": [f"P{i + 1:02d}" for i in range(n)],
                         "before": before.round(1), "after": after.round(1)})

print("🧰 toolkit loaded:", ", ".join(sorted(['BACKGROUND_TRANSITIONS', 'CELL_TYPE_MARKERS', 'COMPLEMENT', 'CPG_ISLAND_TRANSITIONS', 'HOUSEKEEPING_GENES', 'INFECTION_PATHWAYS', 'INFLAMMATION_NETWORK_EDGES', 'NUCLEOTIDES', '_venn_order', 'adjacency_matrix', 'circle_layout', 'draw_graph', 'draw_venn', 'jaccard', 'markov_log_probability', 'reverse_complement', 'simulate_cell_counts', 'simulate_cholesterol_trial', 'simulate_dna', 'simulate_tissue_expression'])))
````

## 1 · State the questions precisely (logic, sets, hypotheses)

> 📍 **Where we are.** Before touching the data we write down exactly what we want to know. [Module 02 · 🔣 Logic: the grammar of precise reasoning](02_logic.ipynb) gave us predicates and
> quantifiers, [Module 03 · 🧺 Sets and functions: collections and mappings](03_sets_and_functions.ipynb) sets built from predicates, [Module 25 · 🚦 Hypothesis testing: errors, power, z- and t-tests](25_hypothesis_testing.ipynb) null and alternative hypotheses. A vague question ("what does
> infection do?") can never be wrong — and therefore can never be tested.

**The objects.** $G$ is the set of the 400 measured genes and $T = \{\text{spleen}, \text{kidney}, \text{liver}\}$. For a gene
$g$ in tissue $t$, let $\mu^{\text{inf}}_{g,t}$ and $\mu^{\text{ctl}}_{g,t}$ be the *true* mean log2 expression in infected and
control animals — population quantities, not averages of our 8 samples — and

$$\delta_{g,t} = \mu^{\text{inf}}_{g,t} - \mu^{\text{ctl}}_{g,t}$$

*In words:* $\delta_{g,t}$ is the true log2 fold change of gene $g$ in tissue $t$; $\delta = 1$ means infection doubles the
expression, $\delta = -1$ halves it.

**The predicate.**

$$R(g, t)\ :\Longleftrightarrow\ \delta_{g,t} \neq 0$$

*In words:* "gene $g$ responds to infection in tissue $t$" is *defined* to mean that its true mean differs between infected
and control animals of that tissue. $R$ is a predicate with two free variables (Module 02); it becomes a proposition —
true or false — once $g$ and $t$ are fixed.

**Sets of responders.**

$$D_t = \{g\in G : R(g,t)\},\qquad S = \bigcap_{t\in T} D_t = \{g\in G : \forall t\in T\ \ R(g,t)\},\qquad A = \bigcup_{t\in T} D_t = \{g\in G : \exists t\in T\ \ R(g,t)\}$$

*In words:* $D_t$ collects the responders of one tissue; the **shared response** $S$ contains the genes that respond in
*every* tissue (an intersection is a ∀), and $A$ the genes that respond in *at least one* tissue (a union is an ∃).

**Tissue-specific responders**, e.g. for the spleen:

$$D_{\text{spleen}}\setminus\big(D_{\text{kidney}}\cup D_{\text{liver}}\big) = \{g\in G : R(g,\text{spleen})\ \land\ \lnot R(g,\text{kidney})\ \land\ \lnot R(g,\text{liver})\}$$

*In words:* set difference is "and not": responds in the spleen and in neither of the other two tissues.

**The hypotheses.** For every pair $(g, t)$,

$$H_0^{(g,t)}:\ \delta_{g,t} = 0 \qquad\text{vs}\qquad H_1^{(g,t)}:\ \delta_{g,t}\neq 0$$

*In words:* the null hypothesis says that the gene does not respond in that tissue; the alternative is two-sided because a
gene can go up or down. With 400 genes in 3 tissues these are **1,200 hypotheses**, so we must control an error rate for
the whole family (Module 26): we will keep the **false discovery rate** at $q = 0.05$ within each tissue.

⚠️ **Common confusion — a test estimates $D_t$; it does not reveal it.** Our list of discoveries $\hat D_t$ is a *random*
set: it misses responders (Type II errors) and may contain non-responders (Type I errors), and retaining $H_0$ never shows
that $\delta = 0$ (Module 25). This matters most for the shared response. If each tissue's list catches a truly shared gene
with probability $r$, and the three tissues behave independently, the intersection $\hat S = \bigcap_t\hat D_t$ catches it
with probability $r^3$ — with $r = 0.8$ only $0.8^3 = 0.51$. **Intersecting discovery lists under-estimates the shared
response.** We will check this prediction against the answer key in Section 5.

First, a tiny illustrative table to make sure that the set operations and the quantifiers really say the same thing.

````python
# ── 1.1 Sets of responders are predicates in disguise (a toy table of R(g, t), ILLUSTRATIVE) ──
toy = pd.DataFrame({"spleen": [1, 1, 1, 0, 0, 0], "kidney": [1, 0, 1, 1, 0, 0], "liver": [1, 0, 0, 1, 0, 1]},
                   index=["IL6", "gA", "gB", "gC", "ACTB", "gD"]).astype(bool)
G_toy, T_toy = list(toy.index), list(toy.columns)


def responds(g, t):
    """The predicate R(g, t) of the toy table."""
    return bool(toy.loc[g, t])


D_toy = {t: {g for g in G_toy if responds(g, t)} for t in T_toy}          # set-builder notation = a comprehension
S_sets = D_toy["spleen"] & D_toy["kidney"] & D_toy["liver"]
S_logic = {g for g in G_toy if all(responds(g, t) for t in T_toy)}        # ∀ t  ↔  all(...)
A_sets = D_toy["spleen"] | D_toy["kidney"] | D_toy["liver"]
A_logic = {g for g in G_toy if any(responds(g, t) for t in T_toy)}        # ∃ t  ↔  any(...)
spleen_only = D_toy["spleen"] - (D_toy["kidney"] | D_toy["liver"])
nowhere = set(G_toy) - A_sets
print(toy.astype(int).to_string())
print("\nD_t:", D_toy)
print("shared S (∩):", S_sets, "| ∀-version identical:", S_sets == S_logic)
print("responds somewhere A (∪):", sorted(A_sets), "| ∃-version identical:", A_sets == A_logic)
print("spleen-specific (\\):", spleen_only)
print("responds nowhere:", nowhere, "| De Morgan, = ∩ of the complements:",
      nowhere == set.intersection(*[set(G_toy) - D_toy[t] for t in T_toy]))
print("S ⊆ D_spleen, i.e. 'shared ⇒ responds in spleen':", S_sets <= D_toy["spleen"])
````

🔍 **What just happened?** On the toy table (1 = responds) the intersection and the ∀-comprehension give the same shared
set, $\{$IL6$\}$; the union and the ∃-comprehension give the same five genes; the only spleen-specific gene is gA; and the
genes that respond nowhere, $\{$ACTB$\}$, are both the complement of the union and the intersection of the three
complements — De Morgan's law for sets (Module 03) is De Morgan's law for quantifiers (Module 02):
$\lnot\exists t\,R(g,t) \equiv \forall t\,\lnot R(g,t)$. Finally, $S\subseteq D_{\text{spleen}}$ is the set form of the
implication "shared ⇒ responds in spleen" (Module 02: $p\Rightarrow q$ over a collection of cases means $P\subseteq Q$).

A good study also writes its **analysis plan** before the data are analysed — which test, which error rate, which controls,
and what would make us distrust the result. Writing it down first protects against the most human of errors: choosing the
analysis after seeing which one looks best.

### 📘 Toolbox recap — questions as mathematics

- A **predicate** $R(g,t)$ turns "responds" into a statement that is true or false once $g$ and $t$ are fixed ([Module 02 · 🔣 Logic: the grammar of precise reasoning](02_logic.ipynb)).
- **Sets from predicates** ([Module 03 · 🧺 Sets and functions: collections and mappings](03_sets_and_functions.ipynb)): $D_t = \{g : R(g,t)\}$; $\bigcap$ ↔ ∀ ↔ `all`, $\bigcup$ ↔ ∃ ↔ `any`, $\setminus$ ↔
  "and not"; De Morgan holds for both; $S\subseteq D_t$ ↔ "shared ⇒ responds in $t$".
- **Hypotheses** ([Module 25 · 🚦 Hypothesis testing: errors, power, z- and t-tests](25_hypothesis_testing.ipynb)): $H_0: \delta = 0$ vs $H_1: \delta\neq 0$ for every gene and tissue; many hypotheses ⇒ an error rate
  for the family, here the FDR ([Module 26 · 📊 Categorical data and multiple testing](26_categorical_tests_multiple_testing.ipynb)).
- A **discovery list estimates** a set; it is not the set. Intersections of lists lose genes at rate $\approx r^k$.

## 2 · The design as mathematics (tuples, Cartesian products, matrices, rank)

> 📍 **Where we are.** The design of an experiment decides which questions its data can answer at all. Here it becomes
> mathematics: samples are tuples ([Module 04 · 🔤 Tuples, sequences and strings: when order matters](04_tuples_sequences_strings.ipynb)), the design is a Cartesian product, a model is a matrix equation ([Module 08 · 🔢 Matrices: tables you can multiply](08_matrices.ipynb), [Module 27 · 📉 Linear regression: the whole course in one model](27_linear_regression_review.ipynb)), and
> whether the model has a unique answer is a question about rank ([Module 11 · 🧱 Linear independence, span, basis and rank](11_independence_basis_rank.ipynb)).

Each sample is a triple $s = \langle t, c, r\rangle$ of a tissue $t\in T$, a condition $c\in C = \{\text{control}, \text{infected}\}$
and a replicate number $r\in\text{Rep} = \{1, 2, 3, 4\}$. The full design is the set of samples

$$\mathcal E = T\times C\times\text{Rep},\qquad |\mathcal E| = |T|\cdot|C|\cdot|\text{Rep}| = 3\cdot 2\cdot 4 = 24$$

*In words:* every tissue is measured under every condition, four times; the Cartesian product lists all combinations and
the multiplication principle counts them (Module 04).

````python
# ── 2.1 The design is a Cartesian product T × C × Rep ──
design_tuples = set(itertools.product(TISSUES, ["control", "infected"], [1, 2, 3, 4]))
observed_tuples = {(t, c, int(name[-1])) for name, t, c in samples[["sample", "tissue", "condition"]].to_numpy()}
print(f"|T × C × Rep| = {len(design_tuples)};  samples in the data: {len(observed_tuples)};  "
      f"same set of tuples: {design_tuples == observed_tuples}")
print("three of the 24 tuples:", sorted(design_tuples)[:3])
print("balanced design — every (tissue, condition) group has size:", samples.groupby(["tissue", "condition"]).size().unique())
````

🔍 **What just happened?** The 24 sample names, parsed back into tuples, are exactly the 24 elements of
$T\times C\times\text{Rep}$: nothing is missing and nothing is duplicated, and every group has 4 animals. A **balanced** design is
worth protecting — it makes many formulas simple (we use one below) and keeps every comparison equally precise.

**A model for one gene.** Write $x_s = \mathbb 1[c_s = \text{infected}]$, $k_s = \mathbb 1[t_s = \text{kidney}]$ and
$l_s = \mathbb 1[t_s = \text{liver}]$ — indicator functions (Modules 03, 06), called **dummy variables** in statistics. The
*additive* model is

$$y_s = \beta_0 + \beta_{\text{inf}}\,x_s + \beta_{\text{kid}}\,k_s + \beta_{\text{liv}}\,l_s + \varepsilon_s$$

*In words:* the log expression of sample $s$ is a baseline (control spleen), plus an infection effect if the animal is
infected, plus a shift if the tissue is kidney or liver, plus noise. Stacking the 24 equations gives

$$\mathbf y = D\boldsymbol\beta + \boldsymbol\varepsilon,\qquad D\in\mathbb R^{24\times 4},\qquad \hat{\boldsymbol\beta} = (D^\top D)^{-1}D^\top\mathbf y$$

*In words:* one row of the **design matrix** $D$ per sample, one column per coefficient; least squares (Module 27) has a
unique answer exactly when $D^\top D$ is invertible, i.e. when the columns of $D$ are linearly independent (rank 4).

**The dummy-variable trap.** Why not one column per tissue? Because every sample is in exactly one tissue,
$\mathbb 1[\text{spleen}] + \mathbb 1[\text{kidney}] + \mathbb 1[\text{liver}] = 1$, which is the intercept column. So

$$D_{\text{trap}}\,\mathbf v = \mathbf 0\qquad\text{for}\qquad \mathbf v = (1, 0, -1, -1, -1)^\top$$

*In words:* the intercept column minus the three tissue columns is the zero vector — the five columns are linearly
dependent, the rank is 4 < 5, $D_{\text{trap}}^\top D_{\text{trap}}$ is singular (it has the eigenvalue 0), and adding any
multiple of $\mathbf v$ to a solution $\boldsymbol\beta$ changes no fitted value: the coefficients are not identifiable.

````python
# ── 2.2 Two design matrices: the dummy-variable trap and the full-rank fix ──
ones = np.ones(len(samples))
is_t = {t: (tissue == t).astype(float) for t in TISSUES}
D_trap = np.column_stack([ones, infected, is_t["spleen"], is_t["kidney"], is_t["liver"]])
D_add = np.column_stack([ones, infected, is_t["kidney"], is_t["liver"]])          # spleen = reference tissue
v_null = np.array([1, 0, -1, -1, -1])
for name, Dm in [("D_trap (5 columns)", D_trap), ("D_add  (4 columns)", D_add)]:
    print(f"{name}: rank {np.linalg.matrix_rank(Dm)};  eigenvalues of DᵀD = {np.linalg.eigvalsh(Dm.T @ Dm).round(3)}")
print("D_trap · (1, 0, −1, −1, −1) =", D_trap @ v_null)

fig, axes = plt.subplots(1, 2, figsize=(8.5, 4.2), gridspec_kw={"width_ratios": [5, 4]})
for ax, Dm, cols, title in [(axes[0], D_trap, ["1", "infected", "spleen", "kidney", "liver"], "D_trap: rank 4 < 5 columns"),
                            (axes[1], D_add, ["1", "infected", "kidney", "liver"], "D_add: rank 4 = 4 columns")]:
    ax.imshow(Dm, cmap="Greys", vmin=0, vmax=1.3, aspect="auto")
    ax.set_xticks(range(len(cols))); ax.set_xticklabels(cols, rotation=30)
    ax.set_yticks(range(0, 24, 4)); ax.set_yticklabels(samples["sample"][::4], fontsize=7)
    ax.set_title(title)
axes[0].set_ylabel("sample (row of the design matrix)")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The trap matrix has 5 columns but rank 4, and one eigenvalue of $D_{\text{trap}}^\top D_{\text{trap}}$
is 0 — the matrix is singular (Modules 09 and 12: invertible ⇔ 0 is not an eigenvalue), so the normal equations have
infinitely many solutions. The vector $(1, 0, -1, -1, -1)$ is sent to the zero vector: it is a basis of the null space
(rank–nullity: $5 - 4 = 1$, Module 11). Dropping one tissue column — the spleen becomes the *reference* — gives a matrix of
full column rank whose $D^\top D$ has only positive eigenvalues. In the pictures each black cell is a 1: the trap's three
tissue columns visibly add up to the intercept column. The rule: **with an intercept, a factor with $k$ levels needs $k-1$
dummy columns.**

**Additive or interaction?** The additive model forces the infection effect to be the same in every tissue — exactly what
our question doubts. The **interaction** model adds the products $x_s k_s$ and $x_s l_s$:

$$y_s = \beta_0 + \beta_{\text{kid}}k_s + \beta_{\text{liv}}l_s + \beta_{\text{inf}}x_s + \gamma_{\text{kid}}\,x_sk_s + \gamma_{\text{liv}}\,x_sl_s + \varepsilon_s$$

*In words:* the infection effect is $\beta_{\text{inf}}$ in the spleen, $\beta_{\text{inf}}+\gamma_{\text{kid}}$ in the kidney and
$\beta_{\text{inf}}+\gamma_{\text{liv}}$ in the liver — each tissue gets its own response. Because $\hat{\boldsymbol\beta} =
(D^\top D)^{-1}D^\top\mathbf y$ is linear in $\mathbf y$, we can fit all 400 genes at once: put the genes as the columns of
$Y = X$ and compute $\hat B = (D^\top D)^{-1}D^\top X$, one column of coefficients per gene.

````python
# ── 2.3 400 regressions in one matrix product: the additive vs the interaction model ──
D_int = np.column_stack([D_add, infected * is_t["kidney"], infected * is_t["liver"]])
print("interaction design: rank", np.linalg.matrix_rank(D_int), "of", D_int.shape[1], "columns")
B_add = np.linalg.solve(D_add.T @ D_add, D_add.T @ X)            # 4 × 400: one column of coefficients per gene
B_int = np.linalg.solve(D_int.T @ D_int, D_int.T @ X)            # 6 × 400
effects = pd.DataFrame({"spleen": B_int[1], "kidney": B_int[1] + B_int[4], "liver": B_int[1] + B_int[5],
                        "additive β_inf": B_add[1]}, index=genes)
unnamed = effects.index.str.startswith("g")
spread = effects[TISSUES].max(axis=1) - effects[TISSUES].min(axis=1)
example_gene = spread[unnamed].idxmax()                           # the unnamed gene whose response differs most
print(effects.loc[["SAA1", "MX1", "ACTB", example_gene]].round(3).to_string())
print("balanced design ⇒ additive β_inf = average of the three tissue effects:",
      np.allclose(effects["additive β_inf"], effects[TISSUES].mean(axis=1)))
print("residual degrees of freedom: additive", 24 - 4, "| interaction", 24 - 6, "| one tissue alone", 8 - 2)
````

🔍 **What just happened?** Two matrix products fitted 2 × 400 regressions. *SAA1* rises by 3.32, 2.95 and 2.39 log2 units in
spleen, kidney and liver, *MX1* by 2.76, 1.51 and 2.86, and the housekeeping gene *ACTB* by almost nothing (−0.03 to 0.05) —
positive and negative controls behave as they should. The unnamed gene with the most tissue-dependent response, *g188*,
barely moves in spleen (0.06) and kidney (0.22) but falls by 3.04 log2 units in the liver; the additive model reports a
single compromise of −0.92 that describes no tissue at all — for a gene that responds in one tissue only, "the infection
effect" is not a meaningful single number. The check confirms a consequence of balance: the additive
$\hat\beta_{\text{inf}}$ is *exactly* the average of the three tissue-specific effects ($(3.32+2.95+2.39)/3 = 2.89$ for
*SAA1*).

Our question is about tissue-specific responses, so from now on we use the interaction model — equivalently, we analyse
each tissue separately. Both give the same fold-change estimates; they differ only in how the noise variance is estimated
(from all 24 samples with 18 degrees of freedom, or from the 8 samples of one tissue with 6). The tissue-by-tissue
analysis assumes less (each tissue may have its own noise level), which is why the plan chose it.

### 📘 Toolbox recap — the design

- Samples are **tuples** $\langle t, c, r\rangle$; a full design is a **Cartesian product**, $|T\times C\times\text{Rep}| = \prod$ of the
  sizes ([Module 04 · 🔤 Tuples, sequences and strings: when order matters](04_tuples_sequences_strings.ipynb)); balanced designs make estimates simple and equally precise.
- **Dummy variables** are indicator functions; the model is $\mathbf y = D\boldsymbol\beta + \boldsymbol\varepsilon$ and
  $\hat{\boldsymbol\beta} = (D^\top D)^{-1}D^\top\mathbf y$ ([Module 08 · 🔢 Matrices: tables you can multiply](08_matrices.ipynb), [Module 27 · 📉 Linear regression: the whole course in one model](27_linear_regression_review.ipynb)); many genes at once: $\hat B = (D^\top D)^{-1}D^\top Y$.
- **Identifiability = full column rank** ([Module 11 · 🧱 Linear independence, span, basis and rank](11_independence_basis_rank.ipynb)): dependent columns ⇒ a null-space vector ⇒ $D^\top D$ singular, with an
  eigenvalue 0 ([Module 09 · 🧮 Linear systems and matrix inverses](09_linear_systems_inverses.ipynb), [Module 12 · 🌀 Eigenvalues and eigenvectors](12_eigenvalues_eigenvectors.ipynb)). A factor with $k$ levels gets $k-1$ dummies next to an intercept.
- **Interactions** let effects differ between groups; an additive model averages them away.

## 3 · Explore before you test (vectors, matrices, SVD)

> 📍 **Where we are.** Before any test, look at the whole matrix: what varies most, are there outliers, do the controls
> behave? [Module 13 · 🧩 Diagonalization and the singular value decomposition](13_diagonalization_svd.ipynb) found that the SVD of this matrix separates tissues on its first two components and infection on the third.
> Here we make that statement quantitative with the design matrices of Section 2, and then use a **projection** ([Module 11 · 🧱 Linear independence, span, basis and rank](11_independence_basis_rank.ipynb),
> [Module 27 · 📉 Linear regression: the whole course in one model](27_linear_regression_review.ipynb)) to remove the tissue effect altogether.

**Centring and the SVD.** With $\bar{\mathbf x}$ the vector of gene means,

$$X_c = X - \mathbf 1\bar{\mathbf x}^\top = U\Sigma V^\top,\qquad Z = U\Sigma,\qquad \text{share of variance of component }k = \frac{\sigma_k^2}{\sum_j\sigma_j^2}$$

*In words:* subtract each gene's mean, factor the centred matrix by the SVD; the columns of $Z$ are the samples'
principal-component **scores**, the rows of $V^\top$ the genes' **loadings**, and $\sigma_k^2$ is the variance carried by
component $k$ (Module 13: SVD of the centred matrix = PCA).

**Which factor does a component reflect?** Regress its scores $\mathbf z_k$ on the columns of a design matrix $F$ and compute

$$R^2 = 1 - \frac{\|\mathbf z_k - F\hat{\mathbf b}\|^2}{\|\mathbf z_k - \bar z_k\mathbf 1\|^2}$$

*In words:* the fraction of the component's variance that the factor (tissue, or infection) explains — 1 if the component is
entirely that factor, 0 if it has nothing to do with it (Module 27).

````python
# ── 3.1 SVD of the centred matrix: which design factor does each component reflect? ──
Xc = X - X.mean(axis=0)                                   # centre every gene (column)
U, sing, Vt = np.linalg.svd(Xc, full_matrices=False)
var_share = sing ** 2 / np.sum(sing ** 2)
Z = U * sing                                              # principal-component scores, one row per sample


def r_squared(z, F):
    """Fraction of the variance of z explained by least squares on the columns of F (F contains the intercept)."""
    b = np.linalg.lstsq(F, z, rcond=None)[0]
    resid = z - F @ b
    return 1 - resid @ resid / np.sum((z - z.mean()) ** 2)


F_tissue = np.column_stack([ones, is_t["kidney"], is_t["liver"]])
F_infect = np.column_stack([ones, infected])
pc_table = pd.DataFrame({"% variance": 100 * var_share[:5],
                         "R² tissue": [r_squared(Z[:, k], F_tissue) for k in range(5)],
                         "R² infection": [r_squared(Z[:, k], F_infect) for k in range(5)]},
                        index=[f"PC{k + 1}" for k in range(5)])
print(pc_table.round(3).to_string())

COL_T = {"spleen": "tab:purple", "kidney": "tab:green", "liver": "tab:brown"}
fig, axes = plt.subplots(1, 3, figsize=(12, 3.6))
axes[0].bar(range(1, 11), 100 * var_share[:10], color="tab:blue")
axes[0].set_xlabel("component k"); axes[0].set_ylabel("% of variance"); axes[0].set_title("scree plot")
for t in TISSUES:
    for cond, mk in [(0, "o"), (1, "^")]:
        m = (tissue == t) & (infected == cond)
        axes[1].scatter(Z[m, 0], Z[m, 1], color=COL_T[t], marker=mk, s=36,
                        label=f"{t}, {'infected' if cond else 'control'}")
axes[1].set_xlabel("PC1 score"); axes[1].set_ylabel("PC2 score"); axes[1].set_title("PC1–PC2: the tissues")
axes[1].legend(fontsize=6.5)
order = np.lexsort((tissue, infected))
axes[2].bar(range(24), Z[order, 2], color=["tab:red" if v else "tab:blue" for v in infected[order]])
axes[2].axhline(0, color="k", lw=0.8)
axes[2].set_xticks([5.5, 17.5]); axes[2].set_xticklabels(["12 control samples", "12 infected samples"])
axes[2].set_ylabel("PC3 score"); axes[2].set_title("PC3: infection")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The first two components carry 28.3 % and 27.2 % of the variance, and tissue explains 96 % and
98 % of their scores ($R^2$ 0.956 and 0.981) while infection explains less than 1 %: **PC1 and PC2 are the tissues** (three
groups need two dimensions to be told apart). PC3 (13.2 % of the variance) is the opposite — infection explains 95 % of it,
tissue 2 %: **PC3 is the infection response**, with every control sample on one side of zero and every infected sample on
the other. PC4 and PC5 are explained by neither factor; with 4–5 % each they are the start of the noise floor.

So the largest source of variation in this experiment is *where* a sample comes from, not *what happened to it*. When the
nuisance factor is known, we can remove it exactly. The projection onto the column space of $F = F_{\text{tissue}}$ is
$H_T = F(F^\top F)^{-1}F^\top$ (the hat matrix of Module 27), and

$$X_w = (I - H_T)\,X$$

*In words:* replace every gene's values by the residuals of a regression on tissue — for this $F$ that is the same as
subtracting each tissue's mean from its own samples ("centring within tissue"). What is left is variation *within* tissues,
where infection is the only design factor.

````python
# ── 3.2 Remove the tissue effect by a projection, then look again ──
H_T = F_tissue @ np.linalg.solve(F_tissue.T @ F_tissue, F_tissue.T)       # 24 × 24 hat matrix of the tissue design
X_w = (np.eye(24) - H_T) @ X
X_within = X.copy()
for t in TISSUES:
    X_within[tissue == t] -= X[tissue == t].mean(axis=0)
print("projection residuals = centring within tissue? max |difference| =", f"{np.abs(X_w - X_within).max():.1e}")
print("H_T is a projection: max |H² − H| =", f"{np.abs(H_T @ H_T - H_T).max():.1e}", "| trace(H_T) =", round(np.trace(H_T), 6))
print(f"total variance ‖X_c‖² = {np.sum(Xc ** 2):.1f};  after removing tissue ‖X_w‖² = {np.sum(X_w ** 2):.1f} "
      f"({np.sum(X_w ** 2) / np.sum(Xc ** 2):.1%} remains)")
U_w, sing_w, Vt_w = np.linalg.svd(X_w, full_matrices=False)
Z_w = U_w * sing_w
share_w = sing_w ** 2 / np.sum(sing_w ** 2)
print("within-tissue PCs: % variance", (100 * share_w[:3]).round(1),
      "| R² infection", [round(r_squared(Z_w[:, k], F_infect), 3) for k in range(3)])
top_load = np.argsort(-np.abs(Vt_w[0]))[:14]
print("14 genes with the largest |loading| on within-tissue PC1:", list(genes[top_load]))

fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.6), gridspec_kw={"width_ratios": [1, 1.4]})
for t in TISSUES:
    for cond, col in [(0, "tab:blue"), (1, "tab:red")]:
        m = (tissue == t) & (infected == cond)
        axes[0].scatter(Z_w[m, 0], Z_w[m, 1], color=col, marker={"spleen": "o", "kidney": "s", "liver": "^"}[t], s=36,
                        label=f"{t}, {'infected' if cond else 'control'}")
axes[0].set_xlabel("within-tissue PC1"); axes[0].set_ylabel("within-tissue PC2")
axes[0].set_title("after removing tissue: PC1 = infection"); axes[0].legend(fontsize=6.5)
named_set = set(genes[:12])
axes[1].bar(range(14), Vt_w[0, top_load], color=["tab:red" if g in named_set else "0.6" for g in genes[top_load]])
axes[1].set_xticks(range(14)); axes[1].set_xticklabels(genes[top_load], rotation=60, fontsize=7)
axes[1].axhline(0, color="k", lw=0.8)
axes[1].set_ylabel("loading on PC1"); axes[1].set_title("top loadings (red = candidate infection genes)")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The projection residuals agree with centring within tissue to rounding error ($\sim10^{-15}$),
$H_T$ is idempotent ($H^2 = H$) with trace 3 = the number of columns of $F$ — the facts of Module 27 about hat matrices.
Removing the tissue effect removes more than half of all variation: of $\|X_c\|^2 = 5053$ only 2318 (45.9 %) is left. In
what remains, **PC1 is the infection response** (29.4 % of the within-tissue variance, $R^2 = 0.95$ for infection): controls
and infected samples of all three tissues now separate along one axis. Eleven of its 14 largest loadings belong to
candidate infection genes, all with the same sign — the shared programme; the other three are unnamed genes, *g280* (with
the opposite sign: a gene switched *off* by infection), *g219* and *g354* — candidates for responses the literature did not
tell us about. PC2 (11.2 %) is something else again: it contrasts the spleen with the other two tissues, in *opposite* directions for
control and infected samples — a tissue × infection **interaction**, the part of the response that differs between tissues,
which removing the tissue *baselines* leaves untouched.

Exploration can also use a matrix of *our own* knowledge. Let $W\in\mathbb R^{J\times 4}$ have $w_{g,p} = 1/|P_p|$ if gene $g$
belongs to gene set $P_p$ and 0 otherwise. Then

$$\text{scores} = X_c\,W\in\mathbb R^{24\times 4}$$

*In words:* each sample's score for a gene set is the average of its centred expression over the genes of that set — a
matrix product chains "samples × genes" with "genes × gene sets" (Module 08). We add the housekeeping genes as a fourth,
**negative-control** set: their score should be flat.

````python
# ── 3.3 Pathway scores by one matrix product, and the housekeeping negative control ──
gene_sets = dict(INFECTION_PATHWAYS)
gene_sets["housekeeping (control)"] = HOUSEKEEPING_GENES
W = np.zeros((len(genes), len(gene_sets)))
for p, members in enumerate(gene_sets.values()):
    W[np.isin(genes, members), p] = 1 / len(members)
scores = pd.DataFrame(Xc @ W, columns=list(gene_sets))
scores["tissue"], scores["condition"] = tissue, samples["condition"].to_numpy()
print(scores.groupby(["tissue", "condition"]).mean().loc[TISSUES].round(2).to_string())
gene_sd = X_w.std(axis=0, ddof=3)                        # within-tissue SD of every gene (3 tissue means removed)
is_hk = np.isin(genes, HOUSEKEEPING_GENES)
print(f"\nwithin-tissue SD: housekeeping genes ≤ {gene_sd[is_hk].max():.3f};  median of the other 388 genes "
      f"{np.median(gene_sd[~is_hk]):.3f}")

fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.6), gridspec_kw={"width_ratios": [1.3, 1]})
im = axes[0].imshow(scores[list(gene_sets)].to_numpy().T, cmap="RdBu_r", vmin=-3, vmax=3, aspect="auto")
axes[0].set_yticks(range(4)); axes[0].set_yticklabels(list(gene_sets), fontsize=7.5)
axes[0].set_xticks(range(0, 24, 4)); axes[0].set_xticklabels(samples["sample"][::4], rotation=45, fontsize=7)
axes[0].set_title("gene-set scores = X_c W (centred log2)")
plt.colorbar(im, ax=axes[0], fraction=0.03)
axes[1].hist([gene_sd[~is_hk], gene_sd[is_hk]], bins=30, stacked=True, color=["0.6", "tab:green"],
             label=["other genes", "housekeeping genes"])
axes[1].set_xlabel("within-tissue SD of a gene (log2)"); axes[1].set_ylabel("number of genes")
axes[1].set_title("negative control: housekeeping genes are stable"); axes[1].legend(fontsize=7)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** One $24\times400$ by $400\times4$ product gave every sample a score for each gene set. All three
infection programmes are clearly higher in infected than in control samples of every tissue (for example the
interferon score is −1.43 vs 0.32 in the spleen and −1.09 vs 1.23 in the liver), while the **housekeeping score stays
within ±0.02 of zero** everywhere. Gene by gene, the housekeeping genes have within-tissue standard deviations of at most
0.10 log2 units, against a median of 0.44 for the other genes: the negative controls behave, so the data pass the
first "warning sign" of the plan. Notice what the heat map also shows: the control *spleen* has the lowest cytokine score —
baseline levels differ between tissues, which is exactly why every test compares infected with control *within* a tissue.

## 4 · Model one gene (probability, maximum likelihood, regression)

> 📍 **Where we are.** Before 1,200 tests, one done slowly. We take *SAA1* (serum amyloid A, an acute-phase protein made by
> the liver) in the liver and ask: does it respond to infection, by how much, and how sure are we? Everything here was
> derived in [Module 23 · 🎯 Parameter estimation and maximum likelihood](23_maximum_likelihood.ipynb), [Module 25 · 🚦 Hypothesis testing: errors, power, z- and t-tests](25_hypothesis_testing.ipynb) and [Module 27 · 📉 Linear regression: the whole course in one model](27_linear_regression_review.ipynb); the point is to see the pieces fit.

**The model.** For the $n = 8$ liver samples, with $x_i\in\{0,1\}$ the infection indicator,

$$y_i = \beta_0 + \beta_1x_i + \varepsilon_i,\qquad \varepsilon_i\overset{\text{iid}}{\sim}\mathcal N(0,\sigma^2)
\quad\Longleftrightarrow\quad y_i\mid x_i\sim\mathcal N(\beta_0+\beta_1x_i,\ \sigma^2)$$

*In words:* the control mean is $\beta_0$, infection shifts it by $\beta_1$ (the log2 fold change $\delta$ of Section 1 — called
the slope $\beta_s$ in Module 27), and each animal scatters around its group mean with Gaussian noise of variance $\sigma^2$.

**Maximum likelihood = least squares.** The log-likelihood (Modules 18, 23) is

$$\ell(\beta_0,\beta_1,\sigma^2) = -\frac n2\log(2\pi\sigma^2) - \frac{1}{2\sigma^2}\sum_{i=1}^n\big(y_i-\beta_0-\beta_1x_i\big)^2$$

*In words:* for any fixed $\sigma^2$ the only part that depends on $\boldsymbol\beta$ is minus a positive constant times the
residual sum of squares, so maximising the likelihood over $\boldsymbol\beta$ **is** minimising the RSS. With $D = [\mathbf 1, \mathbf x]$,
$n_1$ infected and $n_0$ control animals, the normal equations $D^\top D\boldsymbol\beta = D^\top\mathbf y$ read

$$n\beta_0 + n_1\beta_1 = \sum_i y_i,\qquad n_1\beta_0 + n_1\beta_1 = \sum_{i:\,x_i=1}y_i
\quad\Longrightarrow\quad \hat\beta_0 = \bar y_0,\qquad \hat\beta_1 = \bar y_1-\bar y_0$$

*In words:* subtracting the second equation from the first leaves $n_0\beta_0 = \sum_{x_i=0}y_i$, so the intercept is the
control mean; then the slope is the difference of the group means. Setting $\partial\ell/\partial\sigma^2 = 0$ gives
$\hat\sigma^2_{\text{MLE}} = \text{RSS}/n$; the unbiased version is $s^2 = \text{RSS}/(n-2)$ (Module 27).

**Uncertainty.** From $\operatorname{Cov}(\hat{\boldsymbol\beta}) = \sigma^2(D^\top D)^{-1}$,

$$\operatorname{SE}(\hat\beta_1) = s\sqrt{\tfrac1{n_0}+\tfrac1{n_1}},\qquad t = \frac{\hat\beta_1}{\operatorname{SE}(\hat\beta_1)}\sim t_{n-2}\ \text{under }H_0,\qquad \hat\beta_1\pm t_{n-2,\,0.975}\operatorname{SE}(\hat\beta_1)$$

*In words:* the standard error of a difference of two means; the $t$ statistic measures the difference in standard errors and
follows Student's $t$ with $n-2 = 6$ degrees of freedom if $H_0:\beta_1 = 0$ is true; the 95 % confidence interval is the
estimate plus or minus about 2.45 standard errors. This $t$ is exactly **Student's two-sample $t$ statistic** with pooled
variance $s^2$ (Module 25) — regression on a 0/1 predictor *is* the two-sample $t$-test.

````python
# ── 4.1 SAA1 in the liver: least squares = maximum likelihood, by hand and with SciPy ──
in_liver = tissue == "liver"
y_saa1 = X[in_liver, np.flatnonzero(genes == "SAA1")[0]]
x_liv = infected[in_liver]
D1 = np.column_stack([np.ones(x_liv.size), x_liv])
beta_hat = np.linalg.solve(D1.T @ D1, D1.T @ y_saa1)             # normal equations
rss = np.sum((y_saa1 - D1 @ beta_hat) ** 2)
n_liv, p_liv = D1.shape
s2 = rss / (n_liv - p_liv)
se_slope = np.sqrt(s2 * np.linalg.inv(D1.T @ D1)[1, 1])
t_slope = beta_hat[1] / se_slope
p_slope = 2 * stats.t.sf(abs(t_slope), n_liv - p_liv)
t975 = stats.t.ppf(0.975, n_liv - p_liv)
ci_slope = beta_hat[1] + np.array([-1, 1]) * t975 * se_slope
print("control  :", y_saa1[x_liv == 0].round(3), " mean", y_saa1[x_liv == 0].mean().round(4))
print("infected :", y_saa1[x_liv == 1].round(3), " mean", y_saa1[x_liv == 1].mean().round(4))
print(f"β̂0 = {beta_hat[0]:.4f} (control mean), β̂1 = {beta_hat[1]:.4f} (difference of means = log2 fold change)")
print(f"RSS = {rss:.4f}:  σ̂²_MLE = RSS/n = {rss / n_liv:.4f},  s² = RSS/(n − 2) = {s2:.4f}")
print(f"SE(β̂1) = {se_slope:.4f},  t = {t_slope:.3f} on {n_liv - p_liv} df,  p = {p_slope:.2e}")
print(f"95 % CI for β1: [{ci_slope[0]:.3f}, {ci_slope[1]:.3f}]  (t quantile {t975:.3f});  "
      f"fold change 2^β̂1 = {2 ** beta_hat[1]:.2f}")
student = stats.ttest_ind(y_saa1[x_liv == 1], y_saa1[x_liv == 0])
welch = stats.ttest_ind(y_saa1[x_liv == 1], y_saa1[x_liv == 0], equal_var=False)
print(f"scipy Student: t = {student.statistic:.3f}, p = {student.pvalue:.2e};   "
      f"Welch: t = {welch.statistic:.3f}, p = {welch.pvalue:.2e} (df = {welch.df:.2f})")
````

🔍 **What just happened?** The four control livers have *SAA1* around 8.5–8.9 and the four infected livers around
10.6–11.4 log2 units. Least squares returns the control mean, $\hat\beta_0 = 8.654$, and the difference of means,
$\hat\beta_1 = 2.394$ — a $2^{2.394}\approx 5.3$-fold induction. The MLE of the noise variance, $\text{RSS}/n = 0.065$, is
smaller than the unbiased $s^2 = 0.087$ by the factor $6/8$. With $\operatorname{SE} = 0.209$ the statistic is $t = 11.48$ on
6 df, $p = 2.6\times10^{-5}$, and the 95 % CI is [1.884, 2.904] log2 units: the response is real and between about 3.7- and
7.5-fold. SciPy's Student test reproduces $t$ and $p$ exactly. **Welch's** test gives the *same* $t$ — with equal group sizes
the two standard errors coincide — but estimates only 4.87 degrees of freedom and so a larger $p$ ($1.0\times10^{-4}$):
the price of not assuming equal variances.

Let us also confirm "MLE = least squares" by brute force — maximise the likelihood numerically — and look at the
likelihood as a function of $\beta_1$ alone. For each $\beta_1$, the best $\beta_0$ and $\sigma^2$ can be plugged in, giving the
**profile log-likelihood** $\ell_p(\beta_1) = -\frac n2\log\!\big(2\pi\,\text{RSS}(\beta_1)/n\big) - \frac n2$.

````python
# ── 4.2 Numerical maximum likelihood, and the profile log-likelihood of β1 ──
def neg_log_lik(theta, y, Dm):
    """−log-likelihood of y = Dβ + ε with ε ~ N(0, σ²); theta = (β0, β1, log σ)."""
    beta, log_sigma = theta[:-1], theta[-1]
    resid = y - Dm @ beta
    sigma2 = np.exp(2 * log_sigma)
    return 0.5 * len(y) * np.log(2 * np.pi * sigma2) + resid @ resid / (2 * sigma2)


mle = optimize.minimize(neg_log_lik, x0=np.array([8.0, 0.0, 0.0]), args=(y_saa1, D1), method="BFGS")
print(f"numerical MLE: β0 = {mle.x[0]:.4f}, β1 = {mle.x[1]:.4f}, σ² = {np.exp(2 * mle.x[2]):.4f};  "
      f"closed form: {beta_hat[0]:.4f}, {beta_hat[1]:.4f}, {rss / n_liv:.4f}")
b1_grid = np.linspace(1.2, 3.6, 400)
profile = []
for b1 in b1_grid:
    r_ = y_saa1 - b1 * x_liv
    rss_b1 = np.sum((r_ - r_.mean()) ** 2)                         # best β0 for this β1 is the mean of y − β1·x
    profile.append(-0.5 * n_liv * np.log(2 * np.pi * rss_b1 / n_liv) - n_liv / 2)
profile = np.array(profile)

fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.5))
jit = np.random.default_rng(SEED + 42).uniform(-0.06, 0.06, n_liv)
axes[0].scatter(x_liv + jit, y_saa1, color=["tab:red" if v else "tab:blue" for v in x_liv], zorder=3)
axes[0].plot([-0.2, 0.2], [beta_hat[0]] * 2, "k-"); axes[0].plot([0.8, 1.2], [beta_hat.sum()] * 2, "k-")
axes[0].errorbar(1.32, beta_hat.sum(), yerr=t975 * se_slope, fmt="none", ecolor="k", capsize=4)
axes[0].set_xticks([0, 1]); axes[0].set_xticklabels(["control", "infected"]); axes[0].set_xlim(-0.5, 1.6)
axes[0].set_ylabel("SAA1, log2 expression"); axes[0].set_title("SAA1 in liver: group means, ±95 % CI of the difference")
axes[1].plot(b1_grid, profile - profile.max(), color="tab:purple")
axes[1].axvline(beta_hat[1], color="k", ls="--", lw=1, label=f"β̂1 = {beta_hat[1]:.3f}")
axes[1].axvspan(*ci_slope, color="tab:orange", alpha=0.25, label="95 % t-interval")
axes[1].set_xlabel("β1 (log2 fold change)"); axes[1].set_ylabel("profile log-likelihood − max")
axes[1].set_title("the likelihood peaks at the difference of means"); axes[1].legend(fontsize=7)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** BFGS, knowing nothing about least squares, climbed the likelihood to $\beta_0 = 8.6540$,
$\beta_1 = 2.3940$ and $\sigma^2 = 0.0653$ — the closed-form values to four decimals, including $\hat\sigma^2 = \text{RSS}/n$, not
$\text{RSS}/(n-2)$. The profile log-likelihood (right) is a smooth hill whose top sits exactly at $\hat\beta_1$; the $t$-interval
covers the part of the hill within about three log-likelihood units of the top. A narrow hill means precise information
about $\beta_1$ (Module 23: the curvature of the log-likelihood measures information).

## 5 · Test every gene (hypothesis tests, FDR, power)

> 📍 **Where we are.** Section 4 did one test; the plan asks for 1,200. Because the design matrix is the same for every gene,
> all regressions of a tissue are **one** matrix computation (Section 2), and [Module 26 · 📊 Categorical data and multiple testing](26_categorical_tests_multiple_testing.ipynb) gave us the Benjamini–Hochberg (BH)
> procedure to keep the false discovery rate at $q$.

For tissue $t$ put the 8 samples × 400 genes into $Y_t$ and use $D = [\mathbf 1, \mathbf x]$:

$$\hat B = (D^\top D)^{-1}D^\top Y_t\in\mathbb R^{2\times400},\qquad s_g^2 = \frac{\|\mathbf y_g - D\hat{\boldsymbol\beta}_g\|^2}{n-2},\qquad t_g = \frac{\hat\beta_{1,g}}{s_g\sqrt{[(D^\top D)^{-1}]_{22}}}$$

*In words:* the second row of $\hat B$ holds the 400 log2 fold changes; each gene gets its own residual variance and its own
$t$ statistic, and a $p$-value from $t_{6}$.

**BH at level $q$.** Sort the $m = 400$ $p$-values, $p_{(1)}\le\dots\le p_{(m)}$, find the largest $k$ with

$$p_{(k)}\le\frac{k}{m}\,q$$

*In words:* compare the $k$-th smallest $p$-value with a threshold that grows linearly with its rank, and reject the $k$
hypotheses with the smallest $p$-values; for independent tests the expected fraction of false discoveries is at most
$\frac{m_0}{m}q\le q$ (Module 26).

````python
# ── 5.1 A t-test for every gene in every tissue: 3 × 400 regressions as matrix products ──
def regress_all_genes(Y, x):
    """Least squares y = β0 + β1·x for every column of Y at once.
    Returns arrays (one entry per gene): slope β̂1, its standard error, t statistic and two-sided p-value (n − 2 df)."""
    Dm = np.column_stack([np.ones(x.size), x])
    inv = np.linalg.inv(Dm.T @ Dm)
    B = inv @ Dm.T @ Y                                     # 2 × J coefficients
    resid = Y - Dm @ B
    df = x.size - 2
    se = np.sqrt(np.sum(resid ** 2, axis=0) / df * inv[1, 1])
    t = B[1] / se
    return B[1], se, t, 2 * stats.t.sf(np.abs(t), df)


def bh_reject(pvals, q=0.05):
    """Benjamini–Hochberg step-up procedure: True for every hypothesis rejected at FDR level q."""
    pvals = np.asarray(pvals)
    m = pvals.size
    order = np.argsort(pvals)
    passes = pvals[order] <= np.arange(1, m + 1) / m * q
    reject = np.zeros(m, dtype=bool)
    if passes.any():
        reject[order[:np.max(np.nonzero(passes)[0]) + 1]] = True
    return reject


screen = {}
for t in TISSUES:
    m_t = tissue == t
    lfc, se, tstat, pval = regress_all_genes(X[m_t], infected[m_t])
    screen[t] = pd.DataFrame({"log2FC": lfc, "SE": se, "t": tstat, "p": pval, "BH": bh_reject(pval)}, index=genes)
    scipy_p = stats.ttest_ind(X[m_t & (infected == 1)], X[m_t & (infected == 0)], axis=0).pvalue
    n_up = int((screen[t]["BH"] & (lfc > 0)).sum())
    print(f"{t:6s}: {int(screen[t]['BH'].sum()):2d} discoveries at FDR 0.05 ({n_up} up, "
          f"{int(screen[t]['BH'].sum()) - n_up} down);  max |p − SciPy Student p| = {np.abs(pval - scipy_p).max():.1e}")
discovered = {t: set(screen[t].index[screen[t]["BH"]]) for t in TISSUES}
print("housekeeping genes discovered anywhere:", sorted(set().union(*discovered.values()) & set(HOUSEKEEPING_GENES)))
print("positive controls (12 candidate genes) discovered:", {t: len(discovered[t] & set(genes[:12])) for t in TISSUES})
print(f"SAA1 in liver: t = {screen['liver'].loc['SAA1', 't']:.3f}, p = {screen['liver'].loc['SAA1', 'p']:.2e} (Section 4)")
````

````python
# ── 5.2 Volcano plots and p-value histograms (the plan's second warning sign) ──
fig, axes = plt.subplots(2, 3, figsize=(11.5, 6.4), gridspec_kw={"height_ratios": [1.4, 1]})
for k, t in enumerate(TISSUES):
    d = screen[t]
    ax = axes[0, k]
    ax.scatter(d.loc[~d.BH, "log2FC"], -np.log10(d.loc[~d.BH, "p"]), s=8, color="0.7", label="not discovered")
    ax.scatter(d.loc[d.BH, "log2FC"], -np.log10(d.loc[d.BH, "p"]), s=10, color="tab:red", label="BH discovery")
    hk_rows = d.index.isin(HOUSEKEEPING_GENES)
    ax.scatter(d.loc[hk_rows, "log2FC"], -np.log10(d.loc[hk_rows, "p"]), s=16, color="tab:green", marker="s",
               label="housekeeping")
    p_cut = d.loc[d.BH, "p"].max()
    ax.axhline(-np.log10(p_cut), color="tab:red", ls="--", lw=1)
    ax.set_title(f"{t}: {int(d.BH.sum())} discoveries (BH cut-off p ≤ {p_cut:.1e})", fontsize=9)
    ax.set_xlabel("log2 fold change (infected − control)")
    axes[1, k].hist(d["p"], bins=20, range=(0, 1), color="tab:blue", edgecolor="white")
    axes[1, k].axhline(400 / 20, color="k", ls=":", lw=1)
    axes[1, k].set_xlabel("p-value")
axes[0, 0].set_ylabel("−log10 p"); axes[0, 0].legend(fontsize=7, loc="upper left")
axes[1, 0].set_ylabel("number of genes")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** Our matrix version reproduces SciPy's Student $t$-test for all 1,200 genes (differences
$\sim10^{-14}$). At FDR 0.05, BH reports **47** genes in the spleen, **41** in the kidney and **28** in the liver, mostly
up-regulated. (These are Student's pooled-variance $t$-tests, the regression of Section 2; Module 26 screened the same
data with Welch's test, which has the same $t$ for equal group sizes but fewer degrees of freedom, hence larger $p$-values
and fewer discoveries: 39, 34 and 25.) Both warning signs of the plan stay quiet: **no housekeeping gene** is discovered (green squares sit at the
bottom of the volcanoes), and the $p$-value histograms have the healthy shape "flat floor + spike near 0" — the floor is

behave: 10 of the 12 candidate genes are discovered in the spleen and 11 in each of the other tissues. The dashed lines

### 5.3 Opening the answer key

Now — and only now — we compare with the truth. For a discovered set $\hat D$ and the true set $D$:

$$\text{precision} = \frac{|\hat D\cap D|}{|\hat D|},\qquad \text{recall} = \frac{|\hat D\cap D|}{|D|},\qquad \text{FDP} = 1-\text{precision},\qquad J(\hat D, D) = \frac{|\hat D\cap D|}{|\hat D\cup D|}$$

*In words:* precision is the fraction of our list that is right, recall (the power of the screen) the fraction of true
responders we found, the false discovery proportion what BH promises to keep small on average, and the Jaccard index
(Module 03) a single overlap score that punishes both kinds of error.

````python
# ── 5.3 Grading the screen against the hidden truth (the answer key is opened here) ──
true_de = answer_key["true_de"]
grade_rows = []
for t in TISSUES:
    d_hat, d_true = discovered[t], true_de[t]
    tp = len(d_hat & d_true)
    grade_rows.append({"tissue": t, "discovered": len(d_hat), "true responders": len(d_true), "true pos.": tp,
                       "false pos.": len(d_hat - d_true), "precision": tp / len(d_hat), "recall": tp / len(d_true),
                       "Jaccard": jaccard(d_hat, d_true)})
grade = pd.DataFrame(grade_rows).set_index("tissue")
print(grade.round(3).to_string())
S_hat, S_true = set.intersection(*discovered.values()), set.intersection(*true_de.values())
print(f"\nshared response: |Ŝ| = {len(S_hat)} found, |S| = {len(S_true)} true, {len(S_hat & S_true)} correct; "
      f"missed: {sorted(S_true - S_hat)}")
r_shared = np.array([len(discovered[t] & S_true) / len(S_true) for t in TISSUES])
print("per-tissue recall of the 12 shared genes:", r_shared.round(3), f"→ product {np.prod(r_shared):.3f} → "
      f"expected ≈ {12 * np.prod(r_shared):.1f} of 12 in all three lists")
print("missed true responders (all tissues): median |true log2FC| =",
      round(float(np.median([abs(answer_key['true_lfc'].loc[t, g]) for t in TISSUES for g in true_de[t] - discovered[t]])), 2),
      "vs found:", round(float(np.median([abs(answer_key['true_lfc'].loc[t, g]) for t in TISSUES for g in true_de[t] & discovered[t]])), 2))
fig, axes = plt.subplots(1, 2, figsize=(9.5, 4.0))
draw_venn({t: discovered[t] for t in TISSUES}, ax=axes[0], title="discovered (BH, q = 0.05)")
draw_venn({t: true_de[t] for t in TISSUES}, ax=axes[1], title="true responders (answer key)")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The screen is good but not perfect. Precision is 0.94 (spleen), 0.90 (kidney) and 0.96 (liver):
8 false discoveries among 116, a false discovery proportion of 7 % overall — close to the promised 5 % (BH controls the
*average* FDP; a single experiment can be above it). Recall is 0.86, 0.84 and 0.77: about one true responder in six (22 of 130) is
missed, and the misses are the small effects (median true $|\delta|$ of the missed genes 1.35 against 1.91 for the found ones).
The Jaccard indices, 0.75–0.82, combine both errors.

The **shared response** shows the predicted bias. The true $S$ has 12 genes; $\hat S$ has 8, all correct, but four (C3,
IL1B, ISG15, TNF) are missing because each was missed in one tissue. The per-tissue recall of these genes, about 0.83–0.92,
gives $0.83\cdot0.92\cdot0.92\approx0.70$ and an expected $12\times0.70\approx8.4$ genes found in all three lists — close to the 8 we
found. The Venn diagrams show the same story region by region: the central region (shared) shrinks from 12 to 8, and
the genes it loses reappear in the two-tissue regions — while the spleen-only region *grows* from 24 true to 28 discovered
genes. **A missed detection in one tissue masquerades as tissue specificity**: "found in the spleen but not in the liver" is
weaker evidence for a spleen-specific response than it looks.

**Would more animals help?** The simulator can run the experiment again with 2–8 replicates per group. Seed 2680 is our
data set; seeds 2681–2699 are new experiments (each with its own hidden truth, but the same numbers of responders).

````python
# ── 5.4 How many animals? The whole screen repeated with 2–8 replicates (SIMULATED, 20 experiments each) ──
power_rows = []
for n_reps in [2, 3, 4, 6, 8]:
    for seed in range(2680, 2700):
        sim = simulate_tissue_expression(n_reps=n_reps, seed=seed)
        tis_s = sim["samples"]["tissue"].to_numpy()
        inf_s = (sim["samples"]["condition"] == "infected").to_numpy().astype(float)
        genes_s = np.array(sim["genes"])
        disc_s = {}
        for t in TISSUES:
            m_t = tis_s == t
            disc_s[t] = set(genes_s[bh_reject(regress_all_genes(sim["X"][m_t], inf_s[m_t])[3])])
            tp = len(disc_s[t] & sim["true_de"][t])
            power_rows.append({"replicates": n_reps, "seed": seed, "tissue": t, "discoveries": len(disc_s[t]),
                               "recall": tp / len(sim["true_de"][t]), "FDP": (len(disc_s[t]) - tp) / max(len(disc_s[t]), 1)})
        shared_true = set.intersection(*sim["true_de"].values())
        power_rows[-1]["shared found"] = len(set.intersection(*disc_s.values()) & shared_true) / len(shared_true)
power = pd.DataFrame(power_rows)
summary = power.groupby("replicates").agg(discoveries=("discoveries", "mean"), recall=("recall", "mean"),
                                          FDR=("FDP", "mean"), FDR_se=("FDP", lambda v: v.std(ddof=1) / np.sqrt(len(v))),
                                          shared_found=("shared found", "mean"))
print(summary.round(3).to_string())

fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.5))
for t, col in COL_T.items():
    rec = power[power.tissue == t].groupby("replicates")["recall"].mean()
    axes[0].plot(rec.index, rec.values, "o-", color=col, label=f"recall, {t}")
axes[0].plot(summary.index, summary["shared_found"], "k^--", label="shared genes found by ∩")
axes[0].set_xlabel("replicate animals per group"); axes[0].set_ylabel("fraction"); axes[0].set_ylim(0, 1.03)
axes[0].set_title("power of the screen (BH, q = 0.05)"); axes[0].legend(fontsize=7)
axes[1].errorbar(summary.index, summary["FDR"], yerr=1.96 * summary["FDR_se"], fmt="o-", color="tab:red", capsize=3)
axes[1].axhline(0.05, color="k", ls=":", label="q = 0.05")
axes[1].set_xlabel("replicate animals per group"); axes[1].set_ylabel("mean false discovery proportion")
axes[1].set_title("FDR stays near q (±95 % CI over 60 screens)"); axes[1].legend(fontsize=7)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** Each row of the table averages 60 screens (20 experiments × 3 tissues). With **2 animals per
group**, the screen finds almost nothing (0.35 discoveries per tissue, recall 0.005): each test has only 2 degrees of
freedom, $t_2$ has very heavy tails, and almost no $p$-value is small enough to survive BH. With **3** animals recall jumps to
0.51, with **4** (our design) to 0.83, with **6** to 0.96 and with **8** to 0.99. The shared response suffers most, as the
$r^3$ argument predicts: the intersection recovers none of the 12 shared genes with 2 animals, 26 % with 3, 79 % with 4 and
99–100 % with 6 or 8. The false discovery rate stays near $q = 0.05$ for every design (0.036–0.051 from 3 animals on; the
0.072 at 2 animals rests on a handful of discoveries and its 95 % interval, 0.01–0.13, covers 0.05): BH controls false
discoveries whatever the sample size — what more animals buy is **power**. This is the experimental-design lesson of
Module 25, now for a whole screen: decide the number of animals *before* the experiment, by simulation if no formula fits.

- **All genes at once:** $\hat B = (D^\top D)^{-1}D^\top Y$; per-gene $s_g^2$, $t_g$, $p_g$ ([Module 08 · 🔢 Matrices: tables you can multiply](08_matrices.ipynb), [Module 27 · 📉 Linear regression: the whole course in one model](27_linear_regression_review.ipynb)); check against a library ([Module 25 · 🚦 Hypothesis testing: errors, power, z- and t-tests](25_hypothesis_testing.ipynb)).

## 6 · Borrow strength across genes (Bayes, MAP, empirical Bayes)

> 📍 **Where we are.** Each fold change of Section 5 used only that gene's 8 samples. But we have 400 genes, and most of them
> barely change — that is information about *every* gene. [Module 24 · 🧭 Bayesian estimation and maximum a posteriori](24_bayesian_estimation_map.ipynb) showed that a normal prior turns the MAP estimate of a mean
> into a precision-weighted average that shrinks noisy estimates. Here the prior is learned from the 400 genes themselves:
> **empirical Bayes**, the idea behind limma and DESeq2.

**A two-level model.** For gene $g$ in one tissue,

$$\hat\beta_g\mid\delta_g\ \sim\ \mathcal N(\delta_g,\ \mathrm{se}_g^2),\qquad \delta_g\ \sim\ \mathcal N(0,\ \tau^2)$$

*In words:* the measured fold change $\hat\beta_g$ scatters around the true one $\delta_g$ with the standard error of Section 5
(the likelihood), and the true fold changes of all genes scatter around 0 with spread $\tau$ (the prior): most genes change
little, a few a lot.

**The posterior** (Module 24's normal–normal formula with $\bar x\to\hat\beta_g$, $\sigma^2/n\to\mathrm{se}_g^2$, $\mu_0 = 0$,
$\sigma_0^2\to\tau^2$):

$$\tilde\delta_g = E[\delta_g\mid\hat\beta_g] = \underbrace{\frac{\tau^2}{\tau^2+\mathrm{se}_g^2}}_{w_g}\,\hat\beta_g,\qquad
\operatorname{Var}(\delta_g\mid\hat\beta_g) = \frac{\tau^2\,\mathrm{se}_g^2}{\tau^2+\mathrm{se}_g^2}$$

*In words:* the shrunken estimate is the raw one multiplied by a weight $w_g$ between 0 and 1 — close to 1 for a precisely
measured gene (small $\mathrm{se}_g$), small for a noisy one. Because the posterior is normal, this is both the posterior
mean and the MAP estimate.

**Estimating the prior from the data.** Averaging over genes, $\hat\beta_g\sim\mathcal N(0,\ \tau^2+\mathrm{se}_g^2)$, so
$E[\hat\beta_g^2] = \tau^2 + \mathrm{se}_g^2$ and

$$\hat\tau^2 = \frac1J\sum_g\hat\beta_g^2 - \frac1J\sum_g\mathrm{se}_g^2$$

*In words:* the spread of the observed fold changes is the spread of the true ones plus the measurement noise; subtract the
noise to estimate the prior variance (a method-of-moments estimate, truncated at 0).

````python
# ── 6.1 Empirical-Bayes shrinkage of the log2 fold changes, graded against the true fold changes ──
true_lfc = answer_key["true_lfc"]
shrink_rows = []
for t in TISSUES:
    d = screen[t]
    tau2 = max(0.0, np.mean(d["log2FC"] ** 2) - np.mean(d["SE"] ** 2))
    w = tau2 / (tau2 + d["SE"] ** 2)
    d["shrunk"] = w * d["log2FC"]
    d["post_sd"] = np.sqrt(tau2 * d["SE"] ** 2 / (tau2 + d["SE"] ** 2))
    d["true"] = true_lfc.loc[t, d.index].to_numpy()
    is_null = d["true"] == 0
    err_raw, err_shr = (d["log2FC"] - d["true"]) ** 2, (d["shrunk"] - d["true"]) ** 2
    shrink_rows.append({"tissue": t, "τ̂": np.sqrt(tau2), "mean weight w": w.mean(),
                        "MSE raw": err_raw.mean(), "MSE shrunk": err_shr.mean(),
                        "MSE raw, true δ = 0": err_raw[is_null].mean(), "MSE shrunk, true δ = 0": err_shr[is_null].mean(),
                        "MSE raw, true δ ≠ 0": err_raw[~is_null].mean(), "MSE shrunk, true δ ≠ 0": err_shr[~is_null].mean()})
print(pd.DataFrame(shrink_rows).set_index("tissue").round(4).T.to_string())

d = screen["spleen"]
fig, axes = plt.subplots(1, 3, figsize=(12, 3.6))
for ax, col, name in [(axes[0], "log2FC", "raw β̂ (least squares)"), (axes[1], "shrunk", "shrunken δ̃ (empirical Bayes)")]:
    ax.scatter(d["true"], d[col], s=8, alpha=0.6)
    ax.plot([-3, 3.5], [-3, 3.5], "k--", lw=1)
    ax.set_xlabel("true log2 fold change δ"); ax.set_ylabel("estimate")
    ax.set_title(f"spleen: {name}\nMSE = {np.mean((d[col] - d['true']) ** 2):.4f}", fontsize=9)
axes[2].scatter(d["SE"], d["shrunk"] / d["log2FC"], s=8, color="tab:purple")
axes[2].set_xlabel("standard error se_g of the raw estimate"); axes[2].set_ylabel("weight w_g = δ̃ / β̂")
axes[2].set_title("noisier genes are shrunk more")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The fitted prior SD $\hat\tau$ is 0.71 (spleen), 0.61 (kidney) and 0.57 (liver); with standard errors
mostly between 0.1 and 0.4 the weights average 0.87, 0.83 and 0.82, so a typical estimate is pulled 13–18 % towards 0. Overall the
shrunken estimates are **closer to the truth** in every tissue — the mean squared error falls from 0.097 to 0.078 in the
spleen, from 0.087 to 0.070 in the kidney and from 0.082 to 0.063 in the liver (about −20 %). The split shows *where* the gain
comes from: for the hundreds of genes whose true change is 0, shrinkage removes noise (spleen MSE 0.096 → 0.068); for the
true responders it **adds bias** (0.106 → 0.142), because a normal prior centred at 0 pulls real fold changes of 2 or 3
towards 0. The middle panel shows it: the cloud at $\delta = 0$ is tighter, the responders sit below the diagonal. The right
panel shows the rule behind it: the weight falls as the standard error grows.

Is one data set enough to judge? Let us repeat the experiment.

````python
# ── 6.2 Does shrinkage win in general? 20 repeated experiments, and the "winner's curse" ──
wins, mse_pairs = 0, []
for seed in range(2680, 2700):
    sim = simulate_tissue_expression(seed=seed)
    tis_s = sim["samples"]["tissue"].to_numpy()
    inf_s = (sim["samples"]["condition"] == "infected").to_numpy().astype(float)
    for t in TISSUES:
        m_t = tis_s == t
        lfc_s, se_s, _, _ = regress_all_genes(sim["X"][m_t], inf_s[m_t])
        tau2_s = max(0.0, np.mean(lfc_s ** 2) - np.mean(se_s ** 2))
        shr_s = tau2_s / (tau2_s + se_s ** 2) * lfc_s
        truth_s = sim["true_lfc"].loc[t].to_numpy()
        mse_pairs.append((np.mean((lfc_s - truth_s) ** 2), np.mean((shr_s - truth_s) ** 2)))
mse_pairs = np.array(mse_pairs)
print(f"60 screens: mean MSE raw {mse_pairs[:, 0].mean():.4f}, shrunk {mse_pairs[:, 1].mean():.4f}; "
      f"shrinkage better in {np.mean(mse_pairs[:, 1] < mse_pairs[:, 0]):.0%} of them")
for t in TISSUES:
    d = screen[t]
    top = d["log2FC"].abs().sort_values(ascending=False).index[:20]
    print(f"{t:6s}: 20 largest |raw| estimates average {d.loc[top, 'log2FC'].abs().mean():.3f}; their true |δ| "
          f"{d.loc[top, 'true'].abs().mean():.3f}; shrunken {d.loc[top, 'shrunk'].abs().mean():.3f}")
````

🔍 **What just happened?** Over 60 independent screens, shrinkage lowered the mean squared error (0.082 → 0.069 on average)
in about 98 % of them — not a lucky accident. The **winner's curse** is visible too: the 20 genes with the largest raw fold
changes overstate their true effects (spleen: 2.50 measured vs 2.28 true), because being at the top of a list selects genes
whose noise happened to push them up (Module 24 showed the same with simulated fold changes). Here, though, the shrunken
values *undershoot* (2.17): a single normal prior is the wrong shape for these data — the truth is a **spike** of exact zeros
plus a **slab** of large effects, and one bell curve cannot fit both. Better priors (a mixture of a spike at 0 and a wide
normal, or the flexible priors of the "adaptive shrinkage" method of Stephens, 2017) shrink the nulls hard and the large
effects little. The lesson is general: a prior is a modelling choice, and its errors show up exactly where the prior is
wrong.

### 📘 Toolbox recap — borrowing strength

- **Two levels:** $\hat\beta_g\mid\delta_g\sim\mathcal N(\delta_g,\mathrm{se}_g^2)$ (likelihood), $\delta_g\sim\mathcal N(0,\tau^2)$ (prior) ([Module 24 · 🧭 Bayesian estimation and maximum a posteriori](24_bayesian_estimation_map.ipynb)).
- **Posterior mean = MAP** $= w_g\hat\beta_g$ with $w_g = \tau^2/(\tau^2+\mathrm{se}_g^2)$ — a precision-weighted average with 0;
  posterior variance $\tau^2\mathrm{se}_g^2/(\tau^2+\mathrm{se}_g^2)$.
- **Empirical Bayes:** estimate $\tau^2$ from all genes, $\hat\tau^2 = \overline{\hat\beta^2} - \overline{\mathrm{se}^2}$ ([Module 21 · ⚖️ Expectation, variance and moments](21_expectation_variance_moments.ipynb): $E[X^2] = \operatorname{Var}+\text{mean}^2$).
- **Trade-off:** less variance, more bias; lower total error here (−20 %), but true large effects are under-estimated when the
  prior's shape is wrong. Raw top hits suffer from the winner's curse.

## 7 · Network view (graphs and adjacency matrices)

> 📍 **Where we are.** A list of genes is not yet biology. Which regulators could drive the response? [Module 06 · 🕸️ Graphs and trees: networks and phylogenies](06_graphs_and_trees.ipynb) represented gene
> regulation as a directed graph and [Module 08 · 🔢 Matrices: tables you can multiply](08_matrices.ipynb) turned it into an adjacency matrix whose powers count paths. Here we put the
> discoveries of Section 5 onto the course's (illustrative) inflammation network.

**Induced subgraph.** For a directed graph $G = \langle V, E\rangle$ and a set of nodes $V'\subseteq V$,

$$G[V'] = \big\langle V',\ \{(u,v)\in E : u\in V' \land v\in V'\}\big\rangle$$

*In words:* keep the chosen nodes and *every* edge between two of them. With $V'$ = the discovered genes, $G[V']$ shows how
the discoveries are wired among themselves — and, by omission, which regulators are missing.

**Paths from matrix powers.** With $A_{ij} = 1$ if $i\to j$,

$$(A^k)_{ij} = \text{number of directed walks of length } k \text{ from } i \text{ to } j,\qquad j \text{ reachable from } i\iff \Big(\sum_{k=1}^{|V|-1}A^k\Big)_{ij}>0$$

*In words:* the $(i,j)$ entry of the $k$-th power counts the ways to go from $i$ to $j$ in exactly $k$ steps (Module 08); summing
the powers up to $|V|-1$ marks every node that can be reached at all — the genes downstream of a regulator.

````python
# ── 7.1 The discoveries on the inflammation network (an illustrative cartoon): induced subgraph and degrees ──
NET_EDGES = list(INFLAMMATION_NETWORK_EDGES)
NET_NODES = sorted({g for edge in NET_EDGES for g in edge})
NET_POS = {"NFKB1": (0.0, 0.0), "TNF": (-1.3, 1.3), "IL1B": (1.3, 1.3), "CXCL8": (-1.7, -0.5), "LYZ": (-0.8, -1.5),
           "IL6": (1.4, -0.7), "STAT3": (2.8, -0.7), "SAA1": (4.1, 0.4), "HAMP": (4.3, -0.7), "C3": (4.1, -1.8),
           "STAT1": (5.7, 0.0), "IRF1": (6.9, 1.2), "MX1": (6.9, -1.2), "ISG15": (8.0, 0.0)}       # as in Module 06
discovered_any = set().union(*discovered.values())
measured = set(genes)
V_disc = [v for v in NET_NODES if v in discovered_any]
E_disc = [(u, v) for u, v in NET_EDGES if u in V_disc and v in V_disc]
A_net = adjacency_matrix(NET_NODES, NET_EDGES, directed=True)
deg = pd.DataFrame({"in (full)": A_net.sum(axis=0), "out (full)": A_net.sum(axis=1)}, index=NET_NODES)
A_sub = adjacency_matrix(V_disc, E_disc, directed=True)
in_sub, out_sub = dict(zip(V_disc, A_sub.sum(axis=0))), dict(zip(V_disc, A_sub.sum(axis=1)))
deg["in (induced)"] = [str(in_sub[v]) if v in in_sub else "–" for v in NET_NODES]
deg["out (induced)"] = [str(out_sub[v]) if v in out_sub else "–" for v in NET_NODES]
deg["status"] = ["discovered" if v in discovered_any else ("not measured" if v not in measured else "not discovered")
                 for v in NET_NODES]
print(f"network: {len(NET_NODES)} genes, {len(NET_EDGES)} edges;  induced subgraph on the {len(V_disc)} discovered "
      f"network genes: {len(E_disc)} edges {E_disc}")
print(deg.to_string())

status_colour = {"discovered": "#f4a6a6", "not measured": "0.85", "not discovered": "#cfe2f3"}
fig, axes = plt.subplots(1, 2, figsize=(12, 3.9), gridspec_kw={"width_ratios": [1.25, 1]})
draw_graph(NET_EDGES, nodes=NET_NODES, directed=True, pos=NET_POS, ax=axes[0], font_size=7,
           node_colors={v: status_colour[s] for v, s in deg["status"].items()},
           title="full network: red = discovered, grey = not measured")
draw_graph(E_disc, nodes=V_disc, directed=True, pos={v: NET_POS[v] for v in V_disc}, ax=axes[1], font_size=7,
           node_colors={v: "#f4a6a6" for v in V_disc}, title=f"induced subgraph G[V′]: only {len(E_disc)} edges survive")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** All 12 measured network genes respond in at least one tissue and are discovered (red), but the
induced subgraph keeps only **4 of the 15 edges** — the interferon module STAT1 → IRF1, STAT1 → MX1, STAT1 → ISG15 and
IRF1 → ISG15. The cytokine and acute-phase genes become isolated nodes, because the two hubs that connect them, **NFKB1**
(out-degree 5) and **STAT3** (out-degree 3), are *not measured* in our panel (grey). Lesson of Module 06: an induced
subgraph depends critically on the node list, and the most important regulators of a response — transcription factors
that are activated by phosphorylation rather than by more transcription — are often missing from it.

Which regulators could explain every discovery? Matrix powers answer it.

````python
# ── 7.2 Upstream regulators from adjacency powers; is co-expression along edges evidence of regulation? ──
idx = {v: i for i, v in enumerate(NET_NODES)}
A2, A3 = A_net @ A_net, A_net @ A_net @ A_net
reach = sum(np.linalg.matrix_power(A_net, k) for k in range(1, len(NET_NODES))) > 0
print("closed 2-step walks NFKB1 → … → NFKB1 (feedback loops):", A2[idx["NFKB1"], idx["NFKB1"]],
      "| 3-step walks NFKB1 → … → SAA1:", A3[idx["NFKB1"], idx["SAA1"]])
for src in ["NFKB1", "STAT1"]:
    print(f"downstream of {src}:", [v for v in NET_NODES if reach[idx[src], idx[v]] and v != src])
explained = all(any(v == src or reach[idx[src], idx[v]] for src in ["NFKB1", "STAT1"]) for v in V_disc)
print("∀ discovered network gene v ∃ src ∈ {NFKB1, STAT1}: v = src or v is reachable from src  →", explained)
print("upstream regulators (ancestors) of SAA1:", [v for v in NET_NODES if reach[idx[v], idx["SAA1"]]])

X_group = X.copy()                                         # remove tissue AND infection: centre within the 6 groups
for t in TISSUES:
    for cond in (0, 1):
        m = (tissue == t) & (infected == cond)
        X_group[m] -= X[m].mean(axis=0)
print("\nedge between measured genes   r within tissue   r within tissue × condition")
for u, v in E_disc:
    i, j = np.flatnonzero(genes == u)[0], np.flatnonzero(genes == v)[0]
    print(f"{u:>6s} → {v:<6s} {np.corrcoef(X_w[:, i], X_w[:, j])[0, 1]:18.3f} {np.corrcoef(X_group[:, i], X_group[:, j])[0, 1]:24.3f}")
````

🔍 **What just happened?** $(A^2)_{\text{NFKB1},\text{NFKB1}} = 2$ counts the two feedback loops NFKB1 → TNF → NFKB1 and
NFKB1 → IL1B → NFKB1, and $(A^3)_{\text{NFKB1},\text{SAA1}} = 1$ is the path NFKB1 → IL6 → STAT3 → SAA1. The reachability matrix
splits the cartoon into two programmes: NFKB1 reaches the cytokines and, through IL6 and STAT3, the acute-phase genes;
STAT1 reaches the interferon genes. The ∀∃ statement evaluates to **True**: every discovered network gene is NFKB1, STAT1 or
downstream of one of them — two upstream regulators suffice to explain the whole response *in this network*. (SAA1's
ancestors include NFKB1, TNF, IL1B, IL6 and STAT3 — every one of them a candidate drug target or knockout experiment.)

The last table is a warning. Along the four surviving edges, expression is strongly correlated across samples within
tissues ($r$ = 0.78–0.89) — tempting evidence for regulation. But once infection is also removed (centring within the 6
tissue × condition groups) the correlations fall to between −0.28 and −0.11, i.e. noise: the simulator contains **no
regulation at all**; the genes are correlated only because infection moves them together. This is the confounding lesson
of Module 27 (*STAT1* and *MX1*) on a network scale: co-expression is evidence of a shared cause, not of an edge.
Regulation needs interventions (knockdowns, perturbation screens) or time courses.

## 8 · Dynamics: an infection is a process, not a snapshot (calculus, eigenvalues)

> 📍 **Where we are.** Our data compare infected and control animals at **one** time point. But infection is a process:
> the pathogen grows, the immune response rises, and both change for days. [Module 14 · 📈 Calculus: derivatives, integrals and rates of change](14_calculus_derivatives_integrals.ipynb) introduced differential equations and
> [Module 15 · ⛰️ Gradients, Jacobians, optimization and dynamics](15_multivariable_calculus_dynamics.ipynb) equilibria, Jacobians and stability. Here a deliberately simple **toy model** — illustrative, not a validated model
> of any infection — shows what a snapshot can and cannot tell us.

**The model.** Let $P(t)$ be the pathogen load and $E(t)$ the level of an immune **effector** (think of the product of an
interferon-stimulated gene), both in arbitrary units, $t$ in days:

$$\frac{dP}{dt} = rP - kPE,\qquad \frac{dE}{dt} = s + aP - dE$$

*In words:* the pathogen replicates at rate $r$ and is killed at a rate proportional to how much effector there is ($k$); the
effector is made at a baseline rate $s$, induced in proportion to the pathogen ($a$), and decays at rate $d$. The rates we use,
$r = 1$, $k = 0.5$, $s = 0.125$, $a = 0.5$, $d = 0.25$ per day, are **illustrative**.

**Equilibria** ($\dot P = \dot E = 0$). From the first equation $P(r - kE) = 0$, so $P = 0$ or $E = r/k$.

* **Pathogen-free (the control animal):** $P_0 = 0$, $E_0 = s/d$.
* **Infected:** $E^\ast = r/k$, and from the second equation $s + aP^\ast - dr/k = 0$, so $P^\ast = \dfrac{dr/k - s}{a} = \dfrac{s}{a}(\mathcal R - 1)$ with

$$\mathcal R = \frac{r\,d}{k\,s}$$

*In words:* $\mathcal R$ compares the pathogen's growth rate $r$ with the killing rate it meets in an uninfected animal,
$kE_0 = ks/d$. A positive infected equilibrium exists only if $\mathcal R > 1$ — the pathogen outgrows baseline immunity (an
analogue of the $R_0$ of the SIR model of Module 15).

**Stability** (Module 15: linearise with the Jacobian, then look at the eigenvalues):

$$J(P,E) = \begin{pmatrix} r - kE & -kP\\ a & -d\end{pmatrix},\qquad
J(0, E_0) = \begin{pmatrix} r - ks/d & 0\\ a & -d\end{pmatrix},\qquad
J(P^\ast, E^\ast) = \begin{pmatrix} 0 & -kP^\ast\\ a & -d\end{pmatrix}$$

*In words:* at the pathogen-free state the Jacobian is triangular, so its eigenvalues are its diagonal entries (Module 12):
$r - ks/d$ and $-d$ — the state is **unstable** exactly when $\mathcal R > 1$, i.e. a tiny inoculum grows. At the infected
equilibrium the trace is $-d < 0$ and the determinant $akP^\ast > 0$, so both eigenvalues have negative real parts: it is
**stable**. A striking consequence for the snapshot: the effector's fold change at the infected equilibrium is
$E^\ast/E_0 = \dfrac{r/k}{s/d} = \mathcal R$, so its **log2 fold change equals $\log_2\mathcal R$**.

````python
# ── 8.1 Equilibria, Jacobian and eigenvalues — symbolically with SymPy, then with the illustrative rates ──
P_, E_ = sp.symbols("P E")
r_, k_, s_, a_, d_ = sp.symbols("r k s a d", positive=True)
field = sp.Matrix([r_ * P_ - k_ * P_ * E_, s_ + a_ * P_ - d_ * E_])
J_sym = field.jacobian([P_, E_])
equilibria = sp.solve(list(field), [P_, E_], dict=True)
print("Jacobian:", J_sym.tolist())
RATES = {r_: 1.0, k_: 0.5, s_: 0.125, a_: 0.5, d_: 0.25}            # per day — ILLUSTRATIVE
for eq in equilibria:
    J_eq = sp.simplify(J_sym.subs(eq))
    J_num = np.array(J_eq.subs(RATES).tolist(), dtype=float)
    lam = np.linalg.eigvals(J_num)
    print(f"\nequilibrium P = {sp.simplify(eq[P_])}, E = {sp.simplify(eq[E_])}  →  (P, E) = "
          f"({float(eq[P_].subs(RATES)):.2f}, {float(eq[E_].subs(RATES)):.2f})")
    print("  J =", J_eq.tolist(), "→ eigenvalues", np.round(lam, 4),
          "→", "stable" if np.all(lam.real < 0) else "unstable")
R_inv = float((r_ * d_ / (k_ * s_)).subs(RATES))
E0_val, Es_val = float((s_ / d_).subs(RATES)), float((r_ / k_).subs(RATES))
Ps_val = float(((d_ * r_ / k_ - s_) / a_).subs(RATES))
print(f"\n𝓡 = rd/(ks) = {R_inv:.2f};  E* / E0 = {Es_val:.2f} / {E0_val:.2f} = {Es_val / E0_val:.2f} → log2 fold change {np.log2(Es_val / E0_val):.2f}")
````

🔍 **What just happened?** SymPy found the two equilibria, $(P, E) = (0, s/d)$ and $\big((dr - ks)/(ak),\ r/k\big)$ — the formulas
above — and with the illustrative rates they are $(0, 0.5)$ and $(0.75, 2)$. At the pathogen-free state the eigenvalues are
$0.75$ and $-0.25$: one positive, so a small inoculum grows (a saddle). At the infected state they are
$-0.125\pm0.415i$: negative real parts, so the system returns to it after a disturbance — and the imaginary parts mean it
does so in **damped oscillations** with period $2\pi/0.415\approx15$ days. Since $\mathcal R = 4$, the effector settles at four
times its baseline: a log2 fold change of exactly 2, in the range of the candidate genes of our experiment. Now the time
course.

````python
# ── 8.2 A simulated infection: time course, phase portrait, and what a snapshot measures ──
r_v, k_v, s_v, a_v, d_v = (RATES[sym] for sym in (r_, k_, s_, a_, d_))


def pathogen_effector(t, state):
    """Right-hand side of the toy model: dP/dt = rP − kPE, dE/dt = s + aP − dE."""
    P, E = state
    return [r_v * P - k_v * P * E, s_v + a_v * P - d_v * E]


course = integrate.solve_ivp(pathogen_effector, (0, 90), [0.001, E0_val], dense_output=True, rtol=1e-9, atol=1e-12,
                             max_step=0.1)
days = np.linspace(0, 90, 9001)
P_t, E_t = course.sol(days)
lfc_t = np.log2(E_t / E0_val)                              # what an infected-vs-control snapshot would measure
for day in [3, 7, 10, 14, 21, 28, 60]:
    P_d, E_d = course.sol(day)
    print(f"day {day:2d}: pathogen {P_d:6.3f}, effector {E_d:5.3f} → log2 fold change {np.log2(E_d / E0_val):5.2f}")
lam_star = np.linalg.eigvals(np.array([[0.0, -k_v * Ps_val], [a_v, -d_v]]))[0]   # eigenvalue at the infected state
peaks = np.flatnonzero((E_t[1:-1] > E_t[:-2]) & (E_t[1:-1] > E_t[2:])) + 1
print("effector peaks on days", days[peaks].round(2), "| successive deviations from E* shrink by",
      (np.abs(E_t[peaks][1:] - Es_val) / np.abs(E_t[peaks][:-1] - Es_val)).round(3),
      "| linear theory e^(Re λ · period) =", round(float(np.exp(lam_star.real * 2 * np.pi / abs(lam_star.imag))), 3))

fig, axes = plt.subplots(1, 3, figsize=(12.5, 3.6))
axes[0].plot(days, P_t, color="tab:red", label="pathogen P")
axes[0].plot(days, E_t, color="tab:blue", label="effector E")
axes[0].axhline(Es_val, color="tab:blue", ls=":", lw=1); axes[0].axhline(Ps_val, color="tab:red", ls=":", lw=1)
axes[0].set_xlim(0, 60); axes[0].set_xlabel("day after infection"); axes[0].set_ylabel("level (arbitrary units)")
axes[0].set_title("time course (dotted: infected equilibrium)"); axes[0].legend(fontsize=7)
P_grid, E_grid = np.meshgrid(np.linspace(0.001, 1.8, 18), np.linspace(0.2, 2.9, 18))
dP, dE = pathogen_effector(0, [P_grid, E_grid])
speed = np.hypot(dP, dE)                                   # draw directions only (unit arrows)
axes[1].quiver(P_grid, E_grid, dP / speed, dE / speed, angles="xy", color="0.75", width=0.004, scale=28)
axes[1].plot(P_t, E_t, color="k", lw=1.2, label="trajectory")
axes[1].axhline(r_v / k_v, color="tab:red", ls="--", lw=1, label="dP/dt = 0 (E = r/k)")
P_line = np.linspace(0, 1.8, 50)
axes[1].plot(P_line, (s_v + a_v * P_line) / d_v, color="tab:blue", ls="--", lw=1, label="dE/dt = 0")
axes[1].plot([0, Ps_val], [E0_val, Es_val], "o", color="k", ms=5)
axes[1].set_xlim(0, 1.8); axes[1].set_ylim(0.2, 2.9)
axes[1].set_xlabel("pathogen P"); axes[1].set_ylabel("effector E"); axes[1].set_title("phase plane: a damped spiral")
axes[1].legend(fontsize=6.5, loc="lower right")
axes[2].plot(days, lfc_t, color="tab:purple")
axes[2].axhline(np.log2(R_inv), color="k", ls=":", lw=1, label=f"log2 R = {np.log2(R_inv):.0f}")
for day in [7, 10, 14, 21]:
    axes[2].plot(day, np.log2(course.sol(day)[1] / E0_val), "o", color="tab:orange")
    axes[2].annotate(f"day {day}", (day, np.log2(course.sol(day)[1] / E0_val)), textcoords="offset points",
                     xytext=(4, -12), fontsize=7)
axes[2].set_xlim(0, 60); axes[2].set_xlabel("day of sampling"); axes[2].set_ylabel("log2 fold change of E")
axes[2].set_title("the snapshot depends on when you look"); axes[2].legend(fontsize=7)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** From an inoculum of 0.001 the pathogen grows almost exponentially at rate $r - kE_0 = 0.75$ per
day (the unstable eigenvalue) and peaks around day 12; the effector follows with a delay, overshoots to its maximum on day
14.4, and the system spirals into the infected equilibrium $(0.75, 2)$ — the phase portrait shows the trajectory leaving the
pathogen-free saddle at $(0, 0.5)$ and winding around the intersection of the two nullclines. The simulation confirms the
linear analysis quantitatively: effector peaks are 15.2–15.4 days apart (linear theory: $2\pi/0.415 = 15.2$), and each deviation from $E^\ast$ is 0.15
times the previous one — exactly $e^{\operatorname{Re}\lambda\cdot\text{period}} = e^{-0.125\times15.2} = 0.150$.

The right panel is the lesson for the experiment. The *same* gene, in the *same* animals, would show a log2 fold change of
**0.24 on day 7, 1.23 on day 10, 2.34 on day 14** (overshoot) and 1.87 on day 21, before settling at $\log_2\mathcal R = 2$.
A snapshot measures one point of this curve. If the three tissues were sampled at different times — or the infection
reached them at different times — "tissue-specific responses" could partly be **timing**. Only a time course can separate
the two.

## (a) A CpG-island detector

**Question.** In vertebrate genomes the dinucleotide CG is rare — methylated C in CG mutates readily to T — except in
**CpG islands**, CG-rich stretches of a few hundred to a few thousand bases, often at promoters. Where are the islands in
14 kb of DNA? **Data (SIMULATED).** Background DNA from the background Markov chain with two embedded islands (300 bp and
1,200 bp) from the CpG-island chain of Module 20. **Method.** Score every pair of neighbouring bases by the log-odds
$\log\big(P_{\text{island}}(b\mid a)/P_{\text{bg}}(b\mid a)\big)$, sum the scores over a window centred on each base, and call the
base "island" when the window score is positive:

$$S_w(i) = \sum_{\text{pairs }(a,b)\text{ in the window of width }w\text{ around }i}\log\frac{P_{\text{island}}(b\mid a)}{P_{\text{bg}}(b\mid a)}$$

*In words:* the log of the likelihood ratio "island vs background" for the window (Module 20) — positive when the island
chain explains the window better. **Ground truth:** the simulated island positions, per base. **The trade-off to study:**
the window width $w$.

````python
# ── (a) Starter: a CpG-island detector on 14 kb of SIMULATED DNA with two hidden islands ──
segments = [("background", 5000, 101), ("island", 300, 102), ("background", 4100, 103), ("island", 1200, 104),
            ("background", 3400, 105)]
dna = "".join(simulate_dna(length, CPG_ISLAND_TRANSITIONS if kind == "island" else BACKGROUND_TRANSITIONS, seed=sd)
              for kind, length, sd in segments)
is_island = np.concatenate([np.full(length, kind == "island") for kind, length, _ in segments])   # the hidden truth
code_of = {b: i for i, b in enumerate(NUCLEOTIDES)}
codes = np.array([code_of[b] for b in dna])
pair_score = np.log(CPG_ISLAND_TRANSITIONS / BACKGROUND_TRANSITIONS)[codes[:-1], codes[1:]]   # pair (i, i+1)
cum_score = np.concatenate([[0.0], np.cumsum(pair_score)])


def window_scores(width):
    """Log-odds score of the window of `width` bases centred on every base (windows are clipped at the two ends)."""
    n_pairs = width - 1
    start = np.clip(np.arange(len(dna)) - width // 2, 0, len(pair_score) - n_pairs)
    return cum_score[start + n_pairs] - cum_score[start]


w0 = dna[5050:5250]                                      # one 200-bp window inside the first island
print(f"window 5050–5249: cumulative sums give {cum_score[5249] - cum_score[5050]:.4f}; markov_log_probability gives "
      f"{markov_log_probability(w0, CPG_ISLAND_TRANSITIONS) - markov_log_probability(w0, BACKGROUND_TRANSITIONS):.4f}")
cpg_rows = []
for width in [25, 50, 100, 200, 400, 800]:
    called = window_scores(width) > 0
    tp = int(np.sum(called & is_island))
    cpg_rows.append({"window w (bp)": width, "precision": tp / max(called.sum(), 1), "recall": tp / is_island.sum(),
                     "called segments": int(np.sum(np.diff(called.astype(int)) == 1) + called[0])})
print(pd.DataFrame(cpg_rows).round(3).to_string(index=False))

fig, axes = plt.subplots(3, 1, figsize=(10, 5.6), sharex=True)
for ax, width in zip(axes, [50, 200, 800]):
    sc = window_scores(width)
    ax.fill_between(np.arange(len(dna)), sc.min(), sc.max(), where=is_island, color="tab:orange", alpha=0.25,
                    label="true island")
    ax.plot(sc, lw=0.7, color="tab:blue"); ax.axhline(0, color="k", lw=0.8, ls="--")
    ax.set_ylabel(f"S, w = {width}")
axes[0].legend(loc="upper left", fontsize=7); axes[-1].set_xlabel("position (bp)")
axes[0].set_title("window log-odds scores: narrow windows are noisy, wide windows blur and miss short islands")
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The cumulative-sum trick gives the same window score as the toolkit's
`markov_log_probability` (the difference of the two chains' log-probabilities; the first-base terms cancel), so every window
of every width costs one subtraction. The table is the **window-size trade-off**. Narrow windows (25–50 bp) catch almost all
island bases (recall 0.89–0.96) but fire all over the background (precision 0.47–0.69; 302 and 87 called segments). At 100–200
bp precision and recall are both near or above 0.9 and the calls form a handful of segments (4 at 200 bp). Wider windows blur:
at 400 bp the calls spill over the island edges (precision 0.87) and fragment near the short island; at 800 bp every window
that contains the 300-bp island also contains at least 500 bp of background, its score never becomes positive, and the
island disappears (recall 0.80, a single called segment). A window must be shorter than the features you want to find and
long enough to average out the noise.

## (b) Population viability with a Leslie matrix

**Question.** A long-lived species (think of a turtle or a seabird) is monitored in five age classes: 1, 2, 3 and 4 years
and adults (5+). Is the population declining, what will its age structure be, and which management action — protecting
juveniles, protecting adults or increasing reproduction — helps most? **Data (ILLUSTRATIVE numbers).** Fecundities
$F = (0, 0, 0.25, 0.6, 1.0)$ female recruits per female per year and survival probabilities $0.35, 0.6, 0.75, 0.8$ between
classes, with adult survival $0.8$ per year. **Model.** $\mathbf n_{t+1} = L\mathbf n_t$ with a Leslie matrix whose last class
keeps its survivors (the diagonal entry 0.8 — the standard "plus class" for long-lived adults). **Tools.** The dominant
eigenvalue $\lambda$ is the long-run growth factor and its right eigenvector $\mathbf w$ the stable age structure (Modules 12,
13); the **left** eigenvector $\mathbf v$ ($\mathbf v^\top L = \lambda\mathbf v^\top$) gives the reproductive values, and the
sensitivity of $\lambda$ to each entry is

$$\frac{\partial\lambda}{\partial L_{ij}} = \frac{v_i\,w_j}{\mathbf v^\top\mathbf w},\qquad e_{ij} = \frac{L_{ij}}{\lambda}\,\frac{\partial\lambda}{\partial L_{ij}}$$

*In words:* how much the growth rate changes per unit change of one entry (a derivative, Module 14) is the product of the
value of the class an individual moves *into* and the abundance of the class it comes *from*; the **elasticity** $e_{ij}$ is
the proportional version (% change of $\lambda$ per % change of the entry), and the elasticities sum to 1. **Ground truth:**
the formula can be checked against finite differences, and every management scenario can be projected.

````python
# ── (b) Starter: growth rate, stable structure, sensitivities and management of an ILLUSTRATIVE population ──
fecundity = np.array([0.0, 0.0, 0.25, 0.6, 1.0])          # female recruits per female per year, classes 1–4 and 5+
survival = np.array([0.35, 0.6, 0.75, 0.8])               # survival 1→2, 2→3, 3→4, 4→5+
adult_survival = 0.8                                      # 5+ → 5+


def population_matrix(F, S, s_adult):
    """Leslie matrix with fecundities F in row 1, survivals S on the sub-diagonal and an adult 'plus' class."""
    L = np.zeros((5, 5))
    L[0] = F
    L[np.arange(1, 5), np.arange(4)] = S
    L[4, 4] = s_adult
    return L


def dominant_eig(M):
    """Dominant eigenvalue (real and positive here) and its eigenvector, normalised to sum to 1."""
    vals, vecs = np.linalg.eig(M)
    k = np.argmax(np.abs(vals))
    vec = np.abs(vecs[:, k].real)
    return vals[k].real, vec / vec.sum()


L_pop = population_matrix(fecundity, survival, adult_survival)
lam_pop, w_pop = dominant_eig(L_pop)                      # growth factor and stable age structure
v_pop = dominant_eig(L_pop.T)[1]
v_pop = v_pop / v_pop[0]                                  # reproductive values, class 1 = 1
sens = np.outer(v_pop, w_pop) / (v_pop @ w_pop)
elas = L_pop / lam_pop * sens
h = 1e-6
fd_check = max(abs((dominant_eig(L_pop + h * np.eye(5)[i][:, None] * np.eye(5)[j][None, :])[0] - lam_pop) / h - sens[i, j])
               for i, j in zip(*np.nonzero(L_pop)))
print(f"λ = {lam_pop:.4f} ({100 * (lam_pop - 1):+.1f} % per year) | stable age structure w = {w_pop.round(3)}")
print("reproductive values v =", v_pop.round(2), "| max |finite difference − formula| =", f"{fd_check:.1e}")
print("elasticities of the non-zero entries (sum = %.4f):" % elas.sum())
for i, j in zip(*np.nonzero(L_pop)):
    print(f"   L[{i + 1},{j + 1}] = {L_pop[i, j]:.2f}   sensitivity {sens[i, j]:.3f}   elasticity {elas[i, j]:.3f}")
scenarios = {"no action": L_pop,
             "juvenile survival 1→2 + 0.05": population_matrix(fecundity, survival + [0.05, 0, 0, 0], adult_survival),
             "adult survival + 0.05": population_matrix(fecundity, survival, adult_survival + 0.05),
             "fecundity × 1.10": population_matrix(1.1 * fecundity, survival, adult_survival)}
fig, axes = plt.subplots(1, 2, figsize=(10.5, 3.6))
for name, M in scenarios.items():
    lam_s = dominant_eig(M)[0]
    print(f"{name:30s} λ = {lam_s:.4f} ({100 * (lam_s - 1):+.1f} % per year)")
    n_t, totals = 100 * w_pop, [100.0]
    for _ in range(40):
        n_t = M @ n_t
        totals.append(n_t.sum())
    axes[0].plot(totals, label=f"{name}: λ = {lam_s:.4f}")
axes[0].set_xlabel("year"); axes[0].set_ylabel("population (start = 100)"); axes[0].legend(fontsize=7)
axes[0].set_title("40-year projections under four scenarios")
im = axes[1].imshow(np.where(L_pop > 0, elas, np.nan), cmap="viridis")
axes[1].set_xticks(range(5)); axes[1].set_xticklabels(["1", "2", "3", "4", "5+"]); axes[1].set_xlabel("from class j")
axes[1].set_yticks(range(5)); axes[1].set_yticklabels(["1", "2", "3", "4", "5+"]); axes[1].set_ylabel("to class i")
axes[1].set_title("elasticities e_ij"); plt.colorbar(im, ax=axes[1], fraction=0.04)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** The dominant eigenvalue is $\lambda = 0.970$: the population shrinks by about 3 % per year and, at
the stable age structure, 39 % of individuals are one-year-olds and 32 % adults. The sensitivity formula agrees with finite
differences to within $5\times10^{-7}$ — a numerical derivative of an eigenvalue confirms a theorem about eigenvectors. The elasticities
sum to 1, and **adult survival** carries by far the largest one (0.45); the four survivals between age classes have about
0.1 each, the adult fecundity 0.096, and the fecundities of the young classes almost nothing (0.007 and 0.012). The
projections agree: adding 0.05 to adult survival raises $\lambda$ to 0.9986 (almost stable), the same absolute gain in
first-year survival only to 0.9855, and 10 % more recruits to 0.9810. For a long-lived
species, **keeping adults alive matters most** — the conclusion of the classic loggerhead-turtle analysis of Crouse, Crowder
& Caswell (1987), which shifted conservation effort from protecting nests to reducing the deaths of juveniles and adults in fishing nets
(turtle excluder devices).

## (c) Design and analyse a cholesterol trial

**Question.** Does a drug lower LDL cholesterol, by how much, and how many patients does a trial need? **Data (SIMULATED).**
The course's paired trial, `simulate_cholesterol_trial`: LDL before and after treatment in each patient; patients differ a
lot from each other (SD 30 mg/dL), the measurement noise is smaller (SD 10 mg/dL) and the hidden true effect is
**−18 mg/dL**. **Tools.** The paired $t$-test on the differences $\delta_i = \text{after}_i-\text{before}_i$ (Module 25), the
$t$ confidence interval (Module 22), power by simulation and by the non-central $t$ distribution, and a rank-based test for
robustness. With the simulator's numbers each difference has SD $\sigma_\delta = \sqrt{10^2+10^2}\approx14.1$, so

$$\text{power}(n) = P\Big(|T|>t_{n-1,0.975}\Big),\qquad T\sim t_{n-1}\big(\text{non-centrality } \sqrt n\,|\Delta|/\sigma_\delta\big),\quad \Delta = -18$$

*In words:* the paired test rejects when $|T|$ exceeds the critical value; under the alternative, $T$ follows a $t$
distribution shifted by the effect measured in standard errors (Module 25). **Ground truth:** the true effect $-18$ mg/dL.

````python
# ── (c) Starter: paired vs unpaired analysis, power by simulation, and one outlier (SIMULATED trial) ──
trial = simulate_cholesterol_trial()                     # 12 patients; true effect −18 mg/dL (hidden)
diff = (trial["after"] - trial["before"]).to_numpy()     # after − before: negative = LDL fell
n_tr = diff.size
ci_tr = diff.mean() + np.array([-1, 1]) * stats.t.ppf(0.975, n_tr - 1) * diff.std(ddof=1) / np.sqrt(n_tr)
print(f"mean change {diff.mean():.2f} mg/dL, 95 % CI [{ci_tr[0]:.2f}, {ci_tr[1]:.2f}]  (truth −18)")
print(f"paired t-test   p = {stats.ttest_rel(trial['after'], trial['before']).pvalue:.4f}")
print(f"unpaired t-test p = {stats.ttest_ind(trial['after'], trial['before']).pvalue:.4f}  (ignores the pairing)")


def simulated_power(n, n_sim=300, alpha=0.05):
    """Fraction of simulated trials of size n in which the paired / unpaired t-test rejects, and the CI coverage."""
    rej_p = rej_u = cover = 0
    for rep in range(n_sim):
        tr = simulate_cholesterol_trial(n=n, seed=10_000 + rep)
        d_ = (tr["after"] - tr["before"]).to_numpy()
        se_ = d_.std(ddof=1) / np.sqrt(n)
        q_ = stats.t.ppf(1 - alpha / 2, n - 1)
        rej_p += abs(d_.mean() / se_) > q_
        cover += d_.mean() - q_ * se_ <= -18 <= d_.mean() + q_ * se_
        a_, b_ = tr["after"].to_numpy(), tr["before"].to_numpy()
        sp2 = (a_.var(ddof=1) + b_.var(ddof=1)) / 2
        rej_u += abs((a_.mean() - b_.mean()) / np.sqrt(2 * sp2 / n)) > stats.t.ppf(1 - alpha / 2, 2 * n - 2)
    return rej_p / n_sim, rej_u / n_sim, cover / n_sim


sigma_d = np.sqrt(10 ** 2 + 10 ** 2)
power_tab = []
for n in [4, 6, 8, 10, 12, 16, 20]:
    nc = np.sqrt(n) * 18 / sigma_d
    q_ = stats.t.ppf(0.975, n - 1)
    theory = stats.nct.sf(q_, n - 1, nc) + stats.nct.cdf(-q_, n - 1, nc)
    sim_p, sim_u, cov = simulated_power(n)
    power_tab.append({"n patients": n, "paired: simulation": sim_p, "paired: theory": theory,
                      "unpaired: simulation": sim_u, "CI coverage": cov})
power_tab = pd.DataFrame(power_tab)
print(power_tab.round(3).to_string(index=False))

typo = trial.copy()
typo.loc[3, "after"] += 100                             # a data-entry error: 297.4 instead of 197.4 mg/dL
print(f"\nwith one typo in patient {typo.loc[3, 'patient']}: paired t-test p = "
      f"{stats.ttest_rel(typo['after'], typo['before']).pvalue:.3f};  Wilcoxon signed-rank p = "
      f"{stats.wilcoxon(typo['after'], typo['before']).pvalue:.3f}  (without the typo: "
      f"{stats.wilcoxon(trial['after'], trial['before']).pvalue:.3f})")

fig, ax = plt.subplots(figsize=(6.4, 3.4))
ax.plot(power_tab["n patients"], power_tab["paired: simulation"], "o", color="tab:blue", label="paired, simulation")
ax.plot(power_tab["n patients"], power_tab["paired: theory"], "-", color="tab:blue", label="paired, non-central t")
ax.plot(power_tab["n patients"], power_tab["unpaired: simulation"], "s--", color="tab:red", label="unpaired, simulation")
ax.axhline(0.8, color="k", ls=":", lw=1, label="80 % power")
ax.set_xlabel("patients in the trial"); ax.set_ylabel("power (α = 0.05)"); ax.set_ylim(0, 1.03)
ax.set_title("pairing removes between-patient variation"); ax.legend(fontsize=7)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** In the 12-patient trial LDL fell by 12.9 mg/dL on average, 95 % CI [−21.00, −4.75] — the interval
covers the true −18. The **paired** test gives $p = 0.005$; the **unpaired** test on the same numbers gives $p = 0.25$,
because it compares two clouds whose spread is dominated by differences *between* patients (SD 30), which pairing removes.
The power table makes this a design rule: the paired analysis reaches 80 % power with about 8 patients (simulation 0.87,
non-central $t$ 0.87) and about 98 % with 12, while the unpaired analysis of the same trials stays below 35 % even with 20.
Simulated and theoretical power agree within simulation error, and the 95 % CIs cover the truth in 94–95 % of the
simulated trials — as promised. One data-entry error (+100 mg/dL in patient P04) destroys the $t$-test ($p = 0.67$): the mean
and SD of the differences are both dragged by one value. The rank-based Wilcoxon signed-rank test is far less affected
($p = 0.077$ vs 0.009 without the typo) — robustness has a price in power, but not a catastrophic one.

## (d) A single-cell classifier: MLE vs MAP, and a model violation

**Question.** Can we tell T cells, B cells, macrophages and fibroblasts apart from their RNA counts, using only a few
labelled reference cells per type? **Data (SIMULATED).** `simulate_cell_counts`: 440 cells × 40 genes (12 markers, 28
background genes), Poisson counts with known rates — or, with `overdispersion` $\varphi>0$, negative-binomial counts (Module 21).
**Method.** A **Poisson naive Bayes** classifier (Module 17): estimate each type's rate $\lambda_{c,g}$ from the reference
cells and assign a new cell to the type with the largest

$$\log P(\mathbf x\mid c) = \sum_{g}\big(x_g\log\lambda_{c,g} - \lambda_{c,g} - \log x_g!\big)$$

*In words:* the log-probability of the cell's counts if it were of type $c$, assuming independent Poisson genes; with equal
priors the largest log-likelihood wins. The rates come either from the **MLE** $\hat\lambda = \bar x$ (Module 23) or from the
**MAP** under a Gamma(shape 1.5, scale 1) prior, $\hat\lambda = (\sum x + 0.5)/(n+1)$ (Module 24). **Ground truth:** the true cell
types and the true rates.

````python
# ── (d) Starter: Poisson naive Bayes with MLE vs MAP rates, few reference cells, and over-dispersion (SIMULATED) ──
def fit_rates(counts, labels, types, pseudo=0.0, extra=0.0):
    """Per-type Poisson rates (sum of counts + pseudo) / (number of cells + extra): the MLE for pseudo = extra = 0,
    the MAP under a Gamma(shape 1.5, scale 1) prior for pseudo = 0.5, extra = 1."""
    return np.array([(counts[labels == c].sum(axis=0) + pseudo) / ((labels == c).sum() + extra) for c in types])


def class_log_lik(counts, rates):
    """log P(counts | type) for every cell (rows) and type (columns); xlogy makes 0·log 0 = 0 and x·log 0 = −∞."""
    return np.stack([(xlogy(counts, lam) - lam - gammaln(counts + 1)).sum(axis=1) for lam in rates], axis=1)


nb_rows = []
for phi in [0.0, 1.0]:
    cells = simulate_cell_counts(overdispersion=phi)
    counts, labels = cells["counts"], cells["labels"]
    types = np.array(cells["rates"].index)
    for n_ref in [1, 2, 3, 5, 10, 20, 40]:
        res = {"MLE": [], "MAP": [], "dead": [], "conf": [], "rmse MLE": [], "rmse MAP": []}
        for rep in range(20):
            pick = np.random.default_rng(rep).permutation(len(labels))
            ref = np.zeros(len(labels), dtype=bool)
            for c in types:
                ref[[i for i in pick if labels[i] == c][:n_ref]] = True
            for name, (pseudo, extra) in {"MLE": (0.0, 0.0), "MAP": (0.5, 1.0)}.items():
                rates = fit_rates(counts[ref], labels[ref], types, pseudo, extra)
                ll = class_log_lik(counts[~ref], rates)
                res[name].append(np.mean(types[ll.argmax(axis=1)] == labels[~ref]))
                res[f"rmse {name}"].append(np.sqrt(np.mean((rates - cells["rates"].to_numpy()) ** 2)))
                if name == "MLE":
                    res["dead"].append(np.mean(np.all(np.isinf(ll), axis=1)))
                else:
                    post = np.exp(ll - ll.max(axis=1, keepdims=True))
                    res["conf"].append(np.mean(post.max(axis=1) / post.sum(axis=1)))
        nb_rows.append({"φ": phi, "reference cells per type": n_ref, "accuracy MLE": np.mean(res["MLE"]),
                        "accuracy MAP": np.mean(res["MAP"]), "cells with P = 0 for every type (MLE)": np.mean(res["dead"]),
                        "mean max posterior (MAP)": np.mean(res["conf"]), "rate RMSE MLE": np.mean(res["rmse MLE"]),
                        "rate RMSE MAP": np.mean(res["rmse MAP"])})
nb_tab = pd.DataFrame(nb_rows)
print(nb_tab.round(3).to_string(index=False))

fig, ax = plt.subplots(figsize=(6.4, 3.5))
for phi, ls in [(0.0, "-"), (1.0, "--")]:
    sub_ = nb_tab[nb_tab["φ"] == phi]
    ax.plot(sub_["reference cells per type"], sub_["accuracy MLE"], "o" + ls, color="tab:red", label=f"MLE rates, φ = {phi:g}")
    ax.plot(sub_["reference cells per type"], sub_["accuracy MAP"], "s" + ls, color="tab:blue", label=f"MAP rates, φ = {phi:g}")
ax.set_xscale("log"); ax.set_xlabel("labelled reference cells per type"); ax.set_ylabel("accuracy on the other cells")
ax.set_ylim(0, 1.03); ax.set_title("a prior rescues the classifier when references are few"); ax.legend(fontsize=7)
plt.tight_layout(); plt.show()
````

🔍 **What just happened?** With Poisson data ($\varphi = 0$) the **MLE** classifier fails badly with few reference cells
(accuracy 0.37 with one cell per type, 0.57 with five): a gene never seen in a type's reference cells gets
$\hat\lambda = 0$, so any test cell with even one molecule of it has probability exactly 0 for that type — with 1–3 reference
cells most test cells get probability 0 for *every* type (column "P = 0 for every type"), and the argmax is meaningless. The
**MAP** rates are never 0 and classify every test cell correctly even from a single reference cell per type. Interestingly,
the MAP rates are *not* closer to the true rates at small $n$ (RMSE 1.79 vs 1.56 with one cell): the prior pulls large marker
rates towards zero, but it removes the zeros — and for classification the zeros are what matter.

With **over-dispersed** counts ($\varphi = 1$, a model violation: real counts vary more than Poisson) the MAP classifier is
still good (0.87 with one reference cell, 0.99 with 40), but it is **over-confident**: its mean maximal posterior is
0.989–0.998 while its accuracy is 0.87–0.99. A naive Bayes model that assumes independent Poisson genes multiplies 40 pieces of
evidence that are each too sharp; its probabilities should not be taken at face value.

## 📌 Key takeaways

**The course in ten sentences.**

1. **Logic and sets** turn a biological question into statements that can be true or false — "responds", "shared",
   "specific" — and a hypothesis that data can reject.
2. **Tuples, products and matrices** describe designs and data: a design is a Cartesian product, a data set a samples ×
   features matrix, a model a matrix equation.
3. **Rank** decides whether a model has a unique answer; dependent columns (the dummy-variable trap, collinear predictors,
   both-allele coding) leave some questions unanswerable.
4. **Eigenvalues and the SVD** find structure: the main axes of variation of a data matrix, the long-run growth of a
   population, the stability of an equilibrium.
5. **Calculus** describes change and finds optima: derivatives for rates and sensitivities, gradients for fitting,
   Jacobians for the stability of dynamical systems.
6. **Probability** models uncertainty, and **conditioning and Bayes' rule** reverse it — from a test result to a diagnosis,
   from data to parameters.
7. **Estimation** chooses parameters: maximum likelihood from the data alone, MAP and empirical Bayes with prior knowledge —
   less variance for some bias.
8. **Every estimate needs its uncertainty**: standard errors, confidence intervals and the CLT make it computable.
9. **Tests and FDR control** decide which effects to report; **power** — planned before the experiment — decides how many
   real effects you will find; a non-significant result is not a negative result.
10. **Honest reporting** separates association from mechanism and snapshot from process, states limitations, and makes every
    number reproducible — and simulation with a known truth is the best way to learn what each method can do.

**In this module:** the infection study found 47, 41 and 28 responding genes in spleen, kidney and liver at FDR 5 % with
precision 0.90–0.96 and recall 0.77–0.86; the intersection recovered 8 of the 12 shared genes ($\approx12r^3$); 4 animals per
group gave power ≈ 0.83, 2 animals almost none; shrinkage lowered the error of the fold changes by ≈ 20 % (16 % on average over 60 repeats) but under-estimated
large effects; co-expression along network edges vanished once infection was conditioned on; and a toy model showed that one
snapshot can catch a log2 fold change anywhere from 0.2 to 2.3.

## ✅ Checkpoint — choosing the right tool

For each situation name the tool, say why, and say what would be a mistake.

1. A pathologist counts 200 cells in a biopsy — T cells, B cells, macrophages, fibroblasts — and asks whether the composition
   matches a reference of 40 : 30 : 20 : 10 %.
   <details><summary>Answer</summary>A χ² goodness-of-fit test on the four counts, expected counts $200\cdot(0.4, 0.3, 0.2, 0.1) = (80, 60, 40, 20)$, 3 degrees of freedom (Module 26). A mistake: four separate binomial tests (multiplicity, and the counts are dependent because they sum to 200), or dividing by the probabilities instead of the expected counts.</details>
2. LDL cholesterol is measured in the same 10 patients before and after a diet.
   <details><summary>Answer</summary>A paired $t$-test on the 10 differences (or a Wilcoxon signed-rank test if outliers are a concern), with a CI for the mean difference (Modules 22, 25). A mistake: a two-sample test, which ignores the pairing and buries the effect under between-patient variation (Part 3c: $p = 0.25$ instead of 0.005).</details>
3. 20,000 genes are measured in 3 treated and 3 control cultures: which genes changed?
   <details><summary>Answer</summary>One test per gene (regression/$t$-test), then Benjamini–Hochberg to control the FDR (Module 26); with only 4 degrees of freedom per gene, empirical-Bayes moderation of variances and shrinkage of fold changes help (Module 24, limma/DESeq2). A mistake: calling every gene with $p<0.05$ (≈ 1,000 false positives expected), or reporting the raw top fold changes (winner's curse).</details>
4. Field biologists know the age-specific survival and fecundity of an endangered population and ask whether it will persist.
   <details><summary>Answer</summary>A Leslie matrix: the dominant eigenvalue is the long-run growth factor ($\lambda<1$: decline), its eigenvector the stable age structure, and sensitivities/elasticities from the left and right eigenvectors rank management options (Modules 12, 13; Part 3b). A mistake: extrapolating the total count of the last few years, which ignores the age structure.</details>
5. A virologist asks whether a newly introduced pathogen will spread in a host population.
   <details><summary>Answer</summary>A dynamical model (e.g. SIR): find the disease-free equilibrium, linearise with the Jacobian, and check whether an eigenvalue has positive real part — equivalently $R_0>1$ (Module 15; Section 8). A mistake: judging from the first few days of counts only, when growth may still be dominated by chance.</details>
6. A variant has been seen 0 times in 50 sequenced genomes. How common is it?
   <details><summary>Answer</summary>A Beta prior on the frequency and the Beta posterior (MAP, posterior mean or an upper credible bound) (Module 24; Problem P26). A mistake: reporting the MLE 0/100 alleles = 0, i.e. "absent".</details>
7. Where are the CpG islands in a 1-Mb region?
   <details><summary>Answer</summary>Two Markov chains (island, background) and log-odds scores in sliding windows, or a two-state HMM (Module 20; Part 3a). A mistake: a window longer than the islands, or scoring single bases independently (which ignores the CG dependence that defines islands).</details>
8. 300 tumour samples × 5,000 genes: what are the main sources of variation?
   <details><summary>Answer</summary>Centre the matrix and take its SVD (PCA); relate the leading components to known factors — subtype, batch, sex, purity — by regression or plots before interpreting them (Module 13; Section 3). A mistake: interpreting PC1 biologically when it is a batch effect, or forgetting to centre.</details>
9. Does fasting insulin depend on BMI once age and glucose are taken into account?
   <details><summary>Answer</summary>Multiple linear regression of insulin on BMI, age and glucose; the BMI coefficient with its $t$-test and CI (Module 27). A mistake: a simple correlation of insulin with BMI (confounded), or a causal claim from observational data.</details>
10. How many mice per group are needed to detect a twofold change in a gene with between-mouse SD 0.6 log2 units?
    <details><summary>Answer</summary>A power calculation: effect $\Delta = 1$ log2 unit, σ = 0.6; the normal approximation gives $n\approx2\big((z_{0.975}+z_{0.8})\sigma/\Delta\big)^2 = 2(2.80\cdot0.6)^2\approx5.7$, and the exact power of the two-sample $t$-test (non-central $t$) is 0.74 with 6 and 0.82 with **7 mice per group** — more if thousands of genes are tested with FDR control (Section 5.4). Verify by simulation (Modules 22, 25). A mistake: choosing $n$ after seeing the data, or ignoring multiple testing when planning.</details>

## 🚀 Bonus — going further

- **Real expression data.** Multi-tissue and infection studies are deposited in the
  [Gene Expression Omnibus (GEO)](https://www.ncbi.nlm.nih.gov/geo/); [GTEx](https://gtexportal.org/) shows how gene expression
  varies across 50+ human tissues. The standard tools for the analysis of Sections 4–6 are the Bioconductor packages limma,
  edgeR and DESeq2 ([bioconductor.org](https://bioconductor.org/)); in Python, [statsmodels](https://www.statsmodels.org/) fits the
  same linear models.
- **Single cells.** The [CELLxGENE](https://cellxgene.cziscience.com/) portal hosts millions of annotated cells; Scanpy and Seurat are
  the standard toolkits. Try the naive Bayes of Part 3d on a real data set and watch over-dispersion at work.
- **Networks and pathways.** [STRING](https://string-db.org/) (protein interactions) and [Reactome](https://reactome.org/) (pathways)
  provide the real versions of the cartoon network of Section 7.
- **Genomes.** The [UCSC Genome Browser](https://genome.ucsc.edu/) has a "CpG Islands" track computed from sequence statistics like
  those of Part 3a — compare your detector's logic with its definition.
- **Populations.** The [COMPADRE and COMADRE databases](https://compadre-db.org/) collect thousands of published matrix population
  models of plants and animals: compute $\lambda$ and elasticities for a real species.
- **Clinical trials.** Read the statistical analysis plan of a registered trial on [ClinicalTrials.gov](https://clinicaltrials.gov/)
  and compare it with the plan you wrote in Part 3c.
